## Penelitian Verifikasi Asal Gambar Setelah Pengeditan Menggunakan AI

### 1. Gambaran Penelitian

Saat ini, gambar dapat diedit menggunakan kecerdasan buatan (AI). Seseorang dapat mengubah latar belakang, menambahkan objek, menghapus bagian tertentu, atau mengubah tampilan sebuah foto.

Perubahan tersebut menimbulkan satu pertanyaan penting:

**Bagaimana cara mengetahui foto asli yang menjadi sumber sebuah gambar setelah gambar tersebut diedit menggunakan AI?**

Penelitian JATIKARYA bertujuan mengembangkan dan menguji cara menemukan asal gambar menggunakan dua teknologi, yaitu TrustMark dan DINOv2.

Hasil pemeriksaan kedua teknologi tersebut akan dibandingkan, kemudian digabungkan untuk mengetahui apakah sistem dapat memberikan keputusan yang lebih tepat dan dapat dipercaya.

### 2. Teknologi yang Digunakan

**TrustMark**

Teknologi untuk memasukkan tanda pengenal tersembunyi ke dalam gambar.

Tanda tersebut digunakan untuk membantu mengenali identitas gambar setelah mengalami perubahan.

**DINOv2**

Teknologi AI yang digunakan untuk membandingkan kemiripan antara dua gambar.

DINOv2 membantu mencari gambar asli yang paling mirip dengan gambar hasil pengeditan.

**Calibrated Fusion**

Cara menggabungkan hasil TrustMark dan DINOv2 menggunakan aturan yang ditentukan berdasarkan data percobaan.

Tujuannya agar sistem tidak sembarangan menentukan asal gambar ketika bukti yang tersedia kurang meyakinkan atau saling bertentangan.

### 3. Empat Metode yang Dibandingkan

| Metode | Cara Kerja |
|---|---|
| TrustMark-only | Memeriksa tanda tersembunyi pada gambar. |
| DINOv2-only | Mencari gambar sumber berdasarkan kemiripan visual. |
| Naive Fusion | Menggunakan TrustMark terlebih dahulu, kemudian DINOv2 jika diperlukan. |
| Calibrated Fusion | Menggabungkan kedua pemeriksaan dengan aturan yang dipelajari dari data percobaan. |

Penelitian ini berfokus pada pengujian **Calibrated Fusion** dan membandingkannya dengan tiga metode lainnya.

```python
# ==================================================
# JATIKARYA NEW - STEP 02
# Sistem Penyimpanan dan Riwayat Percobaan
# ==================================================

import hashlib
import json
import csv
import shutil
import uuid
from pathlib import Path
from datetime import datetime, timezone

# 1. Fungsi untuk memeriksa keutuhan file
def file_hash(path):
    h = hashlib.sha256()

    with open(path, "rb") as f:
        for bagian in iter(
            lambda: f.read(1024 * 1024), b""
        ):
            h.update(bagian)

    return h.hexdigest()


# 2. Membuat percobaan dengan nomor unik
def start_experiment(step, method, parameters=None):

    waktu = datetime.now(timezone.utc)

    run_id = (
        waktu.strftime("%Y%m%dT%H%M%S%fZ")
        + "_" + uuid.uuid4().hex[:8]
    )

    run_dir = ROOT / "02_EXPERIMENTS" / step / run_id

    run_dir.mkdir(parents=True, exist_ok=False)

    for folder in ["inputs", "outputs", "logs"]:
        (run_dir / folder).mkdir()

    metadata = {
        "run_id": run_id,
        "step": step,
        "method": method,
        "parameters": parameters or {},
        "created_at_utc": waktu.isoformat(),
        "status": "CREATED"
    }

    (run_dir / "metadata.json").write_text(
        json.dumps(metadata, indent=2),
        encoding="utf-8"
    )

    return run_dir


# 3. Menyimpan catatan kegiatan percobaan
def record_event(run_dir, event, details=None):

    catatan = {
        "time_utc": datetime.now(
            timezone.utc
        ).isoformat(),
        "event": event,
        "details": details or {}
    }

    with open(
        run_dir / "logs/events.jsonl",
        "a",
        encoding="utf-8"
    ) as f:
        f.write(
            json.dumps(catatan, ensure_ascii=False)
            + "\n"
        )


# 4. Buat percobaan uji penyimpanan
run02 = start_experiment(
    step="STEP_02_STORAGE",
    method="STORAGE_VERIFICATION",
    parameters={
        "purpose": "Memeriksa penyimpanan file"
    }
)

# 5. Buat file pengujian sederhana
file_asli = run02 / "inputs/storage_test.txt"

file_asli.write_text(
    "JATIKARYA NEW - Pengujian sistem penyimpanan.",
    encoding="utf-8"
)

# 6. Salin file ke folder hasil
file_salinan = run02 / "outputs/storage_test.txt"

shutil.copy2(file_asli, file_salinan)

# 7. Periksa file dan kesamaan isinya
tersimpan = file_salinan.is_file()

hash_cocok = (
    file_hash(file_asli) == file_hash(file_salinan)
)

assert tersimpan, "File hasil gagal disimpan"
assert hash_cocok, "Isi file asli dan salinan berbeda"

# 8. Simpan hasil pemeriksaan
hasil = {
    "file_tersimpan": tersimpan,
    "hash_cocok": hash_cocok,
    "sha256": file_hash(file_salinan)
}

hasil_path = run02 / "outputs/verification.json"

hasil_path.write_text(
    json.dumps(hasil, indent=2),
    encoding="utf-8"
)

record_event(
    run02,
    "STORAGE_VERIFIED",
    hasil
)

# 9. Tandai percobaan berhasil
metadata_path = run02 / "metadata.json"

metadata = json.loads(
    metadata_path.read_text(encoding="utf-8")
)

metadata["status"] = "PASS"

metadata_path.write_text(
    json.dumps(metadata, indent=2),
    encoding="utf-8"
)

# 10. Tambahkan ke daftar riwayat percobaan
log_path = ROOT / "04_LOGS/experiment_log.csv"

buat_header = not log_path.exists()

with open(
    log_path, "a", newline="", encoding="utf-8"
) as f:

    writer = csv.writer(f)

    if buat_header:
        writer.writerow([
            "run_id", "step", "method",
            "status", "folder"
        ])

    writer.writerow([
        run02.name,
        "STEP_02_STORAGE",
        "STORAGE_VERIFICATION",
        "PASS",
        str(run02.relative_to(ROOT))
    ])

# 11. Tampilkan hasil
print("=== JATIKARYA NEW - STEP 02 ===")
print("Workspace:", ROOT)
print("Experiment ID:", run02.name)
print("File tersimpan:", tersimpan)
print("Hash cocok:", hash_cocok)
print("Experiment log:", log_path.is_file())
print("Folder percobaan:", run02)
print("STATUS: STORAGE SYSTEM READY")
```

```python
# ==================================================
# JATIKARYA NEW - STEP 03
# Pendaftaran Gambar Sumber Asli
# ==================================================

from google.colab import files
from PIL import Image
from pathlib import Path
from datetime import datetime, timezone
import hashlib
import io
import csv
import json

# 1. Pastikan kita menggunakan proyek yang benar
assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

registry_path = ROOT / "04_LOGS/source_registry.csv"
source_root = ROOT / "01_DATASET/ORIGINAL"

# 2. Unggah satu gambar dari perangkat
print("Silakan pilih satu gambar asli.")
uploaded = files.upload()

assert len(uploaded) == 1, (
    "Pilih tepat satu gambar untuk setiap pendaftaran."
)

filename, image_bytes = next(iter(uploaded.items()))
filename = Path(filename).name

# 3. Periksa apakah file benar-benar gambar
assert len(image_bytes) > 0, "File kosong."

with Image.open(io.BytesIO(image_bytes)) as img:
    image_format = img.format
    width, height = img.size
    img.verify()

assert image_format in ("JPEG", "PNG", "WEBP"), (
    f"Format {image_format} belum didukung."
)

# 4. Buat identitas berdasarkan isi gambar
sha256 = hashlib.sha256(image_bytes).hexdigest()
source_id = "SRC_" + sha256[:16]

extensions = {
    "JPEG": ".jpg",
    "PNG": ".png",
    "WEBP": ".webp"
}

# 5. Periksa daftar gambar yang sudah terdaftar
columns = [
    "source_id",
    "original_filename",
    "image_format",
    "width",
    "height",
    "file_size_bytes",
    "sha256",
    "stored_path",
    "registered_at_utc"
]

registered = []

if registry_path.exists():
    with open(registry_path, newline="", encoding="utf-8") as f:
        reader = csv.DictReader(f)
        assert reader.fieldnames == columns, (
            "Format daftar sumber berbeda. Perlu diperiksa."
        )
        registered = list(reader)

existing = [
    row for row in registered
    if row["sha256"] == sha256
]

# 6. Simpan gambar hanya jika belum terdaftar
if existing:
    assert len(existing) == 1, "Duplikasi pada daftar sumber."

    record = existing[0]
    saved_path = ROOT / record["stored_path"]

    assert saved_path.is_file(), "File sumber lama tidak ditemukan."
    assert file_hash(saved_path) == sha256, (
        "Isi sumber tersimpan tidak cocok."
    )

    source_id = record["source_id"]
    status = "ALREADY REGISTERED"

else:
    assert all(
        row["source_id"] != source_id
        for row in registered
    ), "Konflik Source ID terdeteksi."

    source_dir = source_root / source_id

    assert not source_dir.exists(), (
        "Folder sumber sudah ada tetapi belum tercatat."
    )

    source_dir.mkdir(parents=True)

    saved_path = (
        source_dir / ("original" + extensions[image_format])
    )

    # Simpan isi file asli tanpa pengeditan
    with open(saved_path, "xb") as f:
        f.write(image_bytes)

    assert file_hash(saved_path) == sha256, (
        "File berubah selama penyimpanan."
    )

    record = {
        "source_id": source_id,
        "original_filename": filename,
        "image_format": image_format,
        "width": width,
        "height": height,
        "file_size_bytes": len(image_bytes),
        "sha256": sha256,
        "stored_path": str(saved_path.relative_to(ROOT)),
        "registered_at_utc": datetime.now(
            timezone.utc
        ).isoformat()
    }

    # Simpan informasi gambar
    manifest_path = source_dir / "source_metadata.json"

    with open(manifest_path, "x", encoding="utf-8") as f:
        json.dump(
            record, f,
            indent=2,
            ensure_ascii=False
        )

    # Masukkan ke daftar sumber penelitian
    write_header = not registry_path.exists()

    with open(
        registry_path, "a", newline="", encoding="utf-8"
    ) as f:
        writer = csv.DictWriter(f, fieldnames=columns)

        if write_header:
            writer.writeheader()

        writer.writerow(record)

    status = "SOURCE REGISTERED"

# 7. Periksa hasil akhir
assert saved_path.is_file()
assert file_hash(saved_path) == sha256

print("\n=== JATIKARYA NEW - STEP 03 ===")
print("Nama gambar:", filename)
print("Format gambar:", image_format)
print("Ukuran gambar:", f"{width} x {height} piksel")
print("Source ID:", source_id)
print("Hash cocok:", True)
print("Gambar tersimpan:", saved_path.is_file())
print("Lokasi gambar:", saved_path)
print("Daftar sumber tersedia:", registry_path.is_file())
print("STATUS:", status)

# 8. Tampilkan gambar untuk pemeriksaan visual
print("\nPratinjau gambar sumber:")
display(Image.open(saved_path))
```

```python
# ==================================================
# JATIKARYA NEW - STEP 04B
# TrustMark Otomatis, Aman untuk Pengujian Ulang
# ==================================================

from trustmark import TrustMark
from PIL import Image, ImageOps
from pathlib import Path
from datetime import datetime, timezone
from importlib.metadata import version
import pandas as pd
import torch
import shutil
import json

# 1. Pastikan folder penelitian benar
assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

# 2. Baca daftar gambar yang sudah terdaftar
registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str
)

assert not registry.empty, "Belum ada gambar sumber."
assert registry["source_id"].is_unique

print("=== PILIH GAMBAR SUMBER ===")

for nomor, (_, row) in enumerate(
    registry.iterrows(), start=1
):
    print(
        f"{nomor}. {row['original_filename']} "
        f"({row['source_id']})"
    )

if len(registry) == 1:
    pilihan = 1
    print("Satu gambar tersedia, dipilih otomatis.")
else:
    pilihan = int(input("Masukkan nomor gambar: "))

assert 1 <= pilihan <= len(registry), (
    "Nomor gambar tidak tersedia."
)

record = registry.iloc[pilihan - 1]
SOURCE_ID = record["source_id"]

source_path = ROOT / record["stored_path"]

assert source_path.is_file(), "File sumber tidak ditemukan."
assert file_hash(source_path) == record["sha256"], (
    "Isi gambar sumber tidak cocok dengan catatan."
)

# 3. Konfigurasi TrustMark
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"
MAX_SIZE = 768
TM_VERSION = version("trustmark")

source_hash = record["sha256"]
payload_bits = bin(
    int(source_hash, 16)
)[2:].zfill(256)[:61]

assert len(payload_bits) == 61

# Periksa agar payload tidak sama dengan sumber lain
for _, other in registry.iterrows():
    if other["source_id"] != SOURCE_ID:
        other_payload = bin(
            int(other["sha256"], 16)
        )[2:].zfill(256)[:61]

        assert payload_bits != other_payload, (
            "Konflik kode watermark antar gambar."
        )

# 4. Cari hasil berhasil sebelumnya
step_dir = (
    ROOT / "02_EXPERIMENTS/STEP_04_TRUSTMARK_ENCODE"
)

previous = []

for result_file in step_dir.glob(
    "*/outputs/verification_result.json"
):
    try:
        data = json.loads(result_file.read_text())
        meta = json.loads(
            (
                result_file.parent.parent / "metadata.json"
            ).read_text()
        )

        image_path = ROOT / data["stored_path"]

        valid = (
            data.get("source_id") == SOURCE_ID
            and data.get("source_sha256") == source_hash
            and data.get("payload_bits") == payload_bits
            and data.get("payload_match") is True
            and data.get("trustmark_version") == TM_VERSION
            and meta.get("parameters", {}).get(
                "max_dimension"
            ) == MAX_SIZE
            and image_path.is_file()
            and file_hash(image_path)
                == data.get("watermarked_sha256")
        )

        if valid:
            previous.append((result_file, data))

    except (OSError, KeyError, ValueError, TypeError):
        continue

# 5. Tentukan apakah perlu pengujian baru
run_new = True

if previous:
    previous.sort(key=lambda item: item[0].as_posix())
    old_file, old_result = previous[-1]

    print("\nHasil watermark sebelumnya ditemukan.")
    print("Run ID:", old_result["run_id"])
    print("Status sebelumnya: PASS")

    print("\n1. Gunakan hasil yang sudah ada")
    print("2. Jalankan pengujian ulang")

    action = input("Pilih 1 atau 2 [default 1]: ").strip()

    assert action in ("", "1", "2"), (
        "Pilihan hanya 1 atau 2."
    )

    run_new = action == "2"

if not run_new:
    print("\n=== JATIKARYA NEW - STEP 04B ===")
    print("Source ID:", SOURCE_ID)
    print("Watermark detected:",
          old_result["watermark_detected"])
    print("Payload match:",
          old_result["payload_match"])
    print("Run ID:", old_result["run_id"])
    print("STATUS: EXISTING RESULT VERIFIED")
    print("Tidak ada pengujian baru yang dijalankan.")

else:
    # 6. Buat folder eksperimen baru
    parameters = {
        "source_id": SOURCE_ID,
        "model": "TrustMark-Q",
        "encoding": "BCH_5",
        "device": DEVICE,
        "max_dimension": MAX_SIZE,
        "trustmark_version": TM_VERSION,
        "source_sha256": source_hash
    }

    run_dir = start_experiment(
        step="STEP_04_TRUSTMARK_ENCODE",
        method="TrustMark-Q",
        parameters=parameters
    )

    metadata_path = run_dir / "metadata.json"

    try:
        # 7. Buat salinan kerja
        with Image.open(source_path) as img:
            working_image = ImageOps.exif_transpose(
                img
            ).convert("RGB")

        working_image.thumbnail(
            (MAX_SIZE, MAX_SIZE),
            Image.Resampling.LANCZOS
        )

        prepared_path = (
            run_dir / "inputs/source_prepared.png"
        )

        working_image.save(prepared_path)

        # 8. Siapkan model TrustMark
        print("\nMemproses watermark...")

        tm = TrustMark(
            verbose=False,
            model_type="Q",
            encoding_type=TrustMark.Encoding.BCH_5,
            device=DEVICE
        )

        # 9. Tanam watermark
        encoded = tm.encode(
            working_image,
            payload_bits,
            MODE="binary"
        )

        output_path = (
            run_dir / "outputs/watermarked.png"
        )

        encoded.save(output_path, format="PNG")

        # 10. Baca ulang watermark
        with Image.open(output_path) as img:
            verify_image = img.convert("RGB")

        decoded, detected, schema = tm.decode(
            verify_image,
            MODE="binary"
        )

        payload_match = (
            bool(detected)
            and decoded == payload_bits
            and schema == TrustMark.Encoding.BCH_5
        )

        # 11. Simpan permanen hanya jika valid
        permanent_path = None

        if payload_match:
            wm_dir = (
                ROOT / "01_DATASET/WATERMARKED"
                / SOURCE_ID
            )
            wm_dir.mkdir(
                parents=True,
                exist_ok=True
            )

            permanent_path = (
                wm_dir /
                f"{run_dir.name}_watermarked.png"
            )

            shutil.copy2(output_path, permanent_path)

            assert (
                file_hash(output_path)
                == file_hash(permanent_path)
            )

        # 12. Simpan hasil pemeriksaan
        result = {
            "source_id": SOURCE_ID,
            "run_id": run_dir.name,
            "source_sha256": source_hash,
            "prepared_sha256": file_hash(prepared_path),
            "watermarked_sha256": file_hash(output_path),
            "payload_bits": payload_bits,
            "payload_length": len(payload_bits),
            "watermark_detected": bool(detected),
            "payload_match": bool(payload_match),
            "decoded_payload": decoded,
            "schema": int(schema) if schema is not None else None,
            "trustmark_version": TM_VERSION,
            "device": DEVICE,
            "image_resolution": list(working_image.size),
            "stored_path": (
                str(permanent_path.relative_to(ROOT))
                if permanent_path else None
            ),
            "status": "PASS" if payload_match else "FAIL",
            "timestamp_utc": datetime.now(
                timezone.utc
            ).isoformat()
        }

        result_path = (
            run_dir / "outputs/verification_result.json"
        )

        result_path.write_text(
            json.dumps(result, indent=2),
            encoding="utf-8"
        )

        record_event(
            run_dir,
            "TRUSTMARK_VERIFICATION",
            result
        )

        # 13. Perbarui status percobaan
        metadata = json.loads(
            metadata_path.read_text()
        )

        metadata["status"] = result["status"]
        metadata["completed_at_utc"] = result[
            "timestamp_utc"
        ]

        metadata_path.write_text(
            json.dumps(metadata, indent=2),
            encoding="utf-8"
        )

        # 14. Tampilkan hasil
        print("\n=== JATIKARYA NEW - STEP 04B ===")
        print("Source ID:", SOURCE_ID)
        print("Run ID:", run_dir.name)
        print("Payload length:", len(payload_bits), "bit")
        print("Watermark detected:", bool(detected))
        print("Payload match:", bool(payload_match))
        print("Device:", DEVICE)
        print("Hasil tersimpan:", result_path)
        print("STATUS:", result["status"])

    except Exception as e:
        record_event(
            run_dir,
            "RUN_FAILED",
            {
                "error_type": type(e).__name__,
                "message": str(e)
            }
        )

        metadata = json.loads(
            metadata_path.read_text()
        )

        metadata["status"] = "ERROR"

        metadata_path.write_text(
            json.dumps(metadata, indent=2),
            encoding="utf-8"
        )

        raise

# ==================================================
# STEP 04B - Unduh hasil watermark
# ==================================================

from google.colab import files

if not run_new:
    download_path = ROOT / old_result["stored_path"]
else:
    assert result["payload_match"], (
        "Watermark belum berhasil diverifikasi."
    )
    download_path = permanent_path

assert download_path.is_file(), (
    "File watermark tidak ditemukan."
)

print("Mengunduh gambar ber-watermark...")
print("Nama file:", download_path.name)

files.download(str(download_path))
```

```python
# ==================================================
# JATIKARYA NEW - STEP 04C
# Pemeriksaan Manual Watermark
# Download -> Upload -> Decode
# ==================================================

from google.colab import files
from trustmark import TrustMark
from PIL import Image, ImageOps
from pathlib import Path
from datetime import datetime, timezone
import hashlib
import pandas as pd
import torch
import json
import io
import time
import shutil

# 1. Pastikan data dari STEP 04B tersedia
assert ROOT.name == "JATIKARYA_NEW"
assert "download_path" in globals(), (
    "Jalankan STEP 04B terlebih dahulu."
)
assert "SOURCE_ID" in globals()

reference_path = Path(download_path)

assert reference_path.is_file(), (
    "Gambar watermark sebelumnya tidak ditemukan."
)
assert reference_path.resolve().is_relative_to(
    ROOT.resolve()
), "File berada di luar proyek baru."

# 2. Ambil kode watermark yang seharusnya ada
registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str
)

entry = registry[
    registry["source_id"] == SOURCE_ID
]

assert len(entry) == 1

source_hash = entry.iloc[0]["sha256"]

expected_payload = bin(
    int(source_hash, 16)
)[2:].zfill(256)[:61]

# 3. Upload gambar hasil download
print("=== JATIKARYA NEW - STEP 04C ===")
print("Silakan upload gambar hasil STEP 04B.")

uploaded = files.upload()

assert len(uploaded) == 1, (
    "Pilih tepat satu gambar."
)

filename, image_bytes = next(
    iter(uploaded.items())
)

# 4. Periksa file yang diupload
uploaded_hash = hashlib.sha256(
    image_bytes
).hexdigest()

reference_hash = file_hash(reference_path)

hash_match = uploaded_hash == reference_hash

with Image.open(io.BytesIO(image_bytes)) as img:
    image_format = img.format
    img.verify()

assert image_format in ("PNG", "JPEG", "WEBP"), (
    "Format gambar tidak didukung."
)

with Image.open(io.BytesIO(image_bytes)) as img:
    test_image = ImageOps.exif_transpose(
        img
    ).convert("RGB")

# 5. Buat catatan percobaan baru
run04c = start_experiment(
    step="STEP_04C_MANUAL_VERIFY",
    method="TRUSTMARK_MANUAL_DECODE",
    parameters={
        "source_id": SOURCE_ID,
        "reference_sha256": reference_hash,
        "uploaded_filename": filename,
        "phase": "PILOT"
    }
)

metadata_path = run04c / "metadata.json"

try:
    # Simpan gambar upload tanpa mengubah isinya
    upload_copy = (
        run04c / "inputs/uploaded_image"
    )

    upload_copy = upload_copy.with_suffix(
        {"PNG": ".png",
         "JPEG": ".jpg",
         "WEBP": ".webp"}[image_format]
    )

    with open(upload_copy, "xb") as f:
        f.write(image_bytes)

    assert file_hash(upload_copy) == uploaded_hash

    # Simpan salinan gambar pembanding
    reference_copy = (
        run04c / "inputs/reference_watermarked.png"
    )

    shutil.copy2(reference_path, reference_copy)

    assert file_hash(reference_copy) == reference_hash

    # 6. Baca watermark dari gambar upload
    device = (
        "cuda" if torch.cuda.is_available() else "cpu"
    )

    print("Membaca watermark dari gambar upload...")

    tm04c = TrustMark(
        verbose=False,
        model_type="Q",
        encoding_type=TrustMark.Encoding.BCH_5,
        device=device
    )

    start = time.perf_counter()

    decoded, detected, schema = tm04c.decode(
        test_image,
        MODE="binary"
    )

    decode_seconds = time.perf_counter() - start

    payload_match = (
        bool(detected)
        and decoded == expected_payload
        and schema == TrustMark.Encoding.BCH_5
    )

    # 7. Tentukan status hasil
    if hash_match and payload_match:
        status = "PASS"
    elif payload_match:
        status = "REVIEW"
    else:
        status = "FAIL"

    # 8. Simpan hasil pemeriksaan
    result = {
        "source_id": SOURCE_ID,
        "uploaded_filename": filename,
        "uploaded_sha256": uploaded_hash,
        "reference_sha256": reference_hash,
        "hash_match": bool(hash_match),
        "watermark_detected": bool(detected),
        "payload_match": bool(payload_match),
        "decoded_payload": decoded,
        "decode_seconds": decode_seconds,
        "device": device,
        "status": status,
        "timestamp_utc": datetime.now(
            timezone.utc
        ).isoformat()
    }

    result_path = (
        run04c / "outputs/manual_verification.json"
    )

    result_path.write_text(
        json.dumps(result, indent=2),
        encoding="utf-8"
    )

    record_event(
        run04c,
        "MANUAL_VERIFICATION_COMPLETED",
        result
    )

    metadata = json.loads(
        metadata_path.read_text()
    )
    metadata["status"] = status

    metadata_path.write_text(
        json.dumps(metadata, indent=2),
        encoding="utf-8"
    )

    # 9. Tampilkan hasil
    print("\n=== HASIL STEP 04C ===")
    print("Nama gambar:", filename)
    print("Hash cocok:", hash_match)
    print("Watermark detected:", bool(detected))
    print("Payload match:", bool(payload_match))
    print(
        "Waktu pembacaan:",
        round(decode_seconds, 3), "detik"
    )
    print("Hasil tersimpan:", result_path)
    print("STATUS:", status)

except Exception as e:
    record_event(
        run04c,
        "RUN_FAILED",
        {
            "error_type": type(e).__name__,
            "message": str(e)
        }
    )

    metadata = json.loads(
        metadata_path.read_text()
    )
    metadata["status"] = "ERROR"

    metadata_path.write_text(
        json.dumps(metadata, indent=2),
        encoding="utf-8"
    )
    raise
```

```python
# ==================================================
# JATIKARYA NEW - STEP 04D
# Pemeriksaan TrustMark pada Gambar Asli
# ==================================================

from google.colab import files
from trustmark import TrustMark
from PIL import Image
from pathlib import Path
from datetime import datetime, timezone
import pandas as pd
import hashlib
import io
import torch
import json
import time

# 1. Periksa sumber yang terdaftar
assert ROOT.name == "JATIKARYA_NEW"

registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str
)

entry = registry.loc[
    registry["source_id"] == SOURCE_ID
]

assert len(entry) == 1

source = entry.iloc[0]

expected_payload = bin(
    int(source["sha256"], 16)
)[2:].zfill(256)[:61]

# 2. Unggah gambar asli
print("=== JATIKARYA NEW - STEP 04D ===")
print("Unggah gambar ORIGINAL tanpa watermark.")

uploaded = files.upload()

assert len(uploaded) == 1, (
    "Pilih tepat satu gambar."
)

filename, image_bytes = next(
    iter(uploaded.items())
)

# 3. Pastikan gambar benar-benar sumber asli
uploaded_hash = hashlib.sha256(
    image_bytes
).hexdigest()

assert uploaded_hash == source["sha256"], (
    "File yang diunggah bukan gambar asli "
    "yang terdaftar. Tidak diproses."
)

with Image.open(io.BytesIO(image_bytes)) as img:
    image = img.convert("RGB")

# 4. Buat folder percobaan tersendiri
run04d = start_experiment(
    step="STEP_04D_ORIGINAL_CONTROL",
    method="TRUSTMARK_NEGATIVE_CONTROL",
    parameters={
        "source_id": SOURCE_ID,
        "input_type": "ORIGINAL"
    }
)

try:
    input_path = run04d / "inputs/original.png"

    with open(input_path, "xb") as f:
        f.write(image_bytes)

    assert file_hash(input_path) == uploaded_hash

    # 5. Gunakan model yang sudah tersedia
    if "tm04c" in globals():
        detector = tm04c
    else:
        device = (
            "cuda" if torch.cuda.is_available()
            else "cpu"
        )

        detector = TrustMark(
            verbose=False,
            model_type="Q",
            encoding_type=TrustMark.Encoding.BCH_5,
            device=device
        )

    # 6. Periksa apakah ada watermark
    start = time.perf_counter()

    decoded, detected, schema = detector.decode(
        image,
        MODE="binary"
    )

    duration = time.perf_counter() - start

    payload_match = (
        bool(detected)
        and decoded == expected_payload
        and schema == TrustMark.Encoding.BCH_5
    )

    # 7. Tentukan hasil kontrol
    if not detected:
        status = "PASS"
        meaning = "Tidak ditemukan watermark oleh detector"
    elif payload_match:
        status = "REVIEW"
        meaning = "Payload JATIKARYA terdeteksi pada original"
    else:
        status = "REVIEW"
        meaning = "Ada deteksi, tetapi payload tidak cocok"

    # 8. Simpan hasil
    result = {
        "source_id": SOURCE_ID,
        "input_type": "ORIGINAL",
        "uploaded_filename": filename,
        "uploaded_sha256": uploaded_hash,
        "watermark_detected": bool(detected),
        "payload_match": bool(payload_match),
        "decoded_payload": decoded,
        "decode_seconds": duration,
        "status": status,
        "meaning": meaning,
        "timestamp_utc": datetime.now(
            timezone.utc
        ).isoformat()
    }

    result_path = (
        run04d / "outputs/original_control.json"
    )

    result_path.write_text(
        json.dumps(result, indent=2),
        encoding="utf-8"
    )

    record_event(
        run04d,
        "ORIGINAL_CONTROL_COMPLETED",
        result
    )

    metadata_path = run04d / "metadata.json"
    metadata = json.loads(
        metadata_path.read_text()
    )
    metadata["status"] = status

    metadata_path.write_text(
        json.dumps(metadata, indent=2),
        encoding="utf-8"
    )

    # 9. Tampilkan hasil
    print("\n=== HASIL STEP 04D ===")
    print("Nama gambar:", filename)
    print("Hash original cocok:", True)
    print("Watermark detected:", bool(detected))
    print("Payload match:", bool(payload_match))
    print("Waktu pembacaan:", round(duration, 3), "detik")
    print("Kesimpulan:", meaning)
    print("Hasil tersimpan:", result_path)
    print("STATUS:", status)

except Exception as e:
    record_event(
        run04d,
        "RUN_FAILED",
        {
            "error_type": type(e).__name__,
            "message": str(e)
        }
    )

    metadata_path = run04d / "metadata.json"
    metadata = json.loads(metadata_path.read_text())
    metadata["status"] = "ERROR"
    metadata_path.write_text(
        json.dumps(metadata, indent=2),
        encoding="utf-8"
    )
    raise
```

```python
# ==================================================
# JATIKARYA NEW - STEP 05B
# Pendaftaran Ciri Gambar Asli Menggunakan DINOv2
# TANPA TRUSTMARK
# ==================================================

from PIL import Image, ImageOps
from torchvision import transforms
from datetime import datetime, timezone
import torch
import torch.nn.functional as F
import pandas as pd
import numpy as np
import shutil
import json
import time

# 1. Periksa kesiapan
assert ROOT.name == "JATIKARYA_NEW"
assert "model05a" in globals(), (
    "Jalankan STEP 05A terlebih dahulu."
)
assert getattr(model05a, "embed_dim", None) == 384

# 2. Baca daftar gambar ASLI
registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str
)

assert not registry.empty
assert registry["source_id"].is_unique

print("=== DAFTAR GAMBAR ASLI ===")

for nomor, (_, row) in enumerate(registry.iterrows(), 1):
    print(
        f"{nomor}. {row['original_filename']} "
        f"({row['source_id']})"
    )

if len(registry) == 1:
    pilihan = 1
    print("Satu gambar tersedia, dipilih otomatis.")
else:
    pilihan = int(input("Pilih nomor gambar: "))
    assert 1 <= pilihan <= len(registry)

source = registry.iloc[pilihan - 1]
selected_id = source["source_id"]

original_path = ROOT / source["stored_path"]

# Pastikan mengambil gambar dari folder ORIGINAL
original_folder = (
    ROOT / "01_DATASET/ORIGINAL"
).resolve()

assert original_path.resolve().is_relative_to(
    original_folder
), "Gambar bukan berasal dari folder ORIGINAL."

assert original_path.is_file()
assert file_hash(original_path) == source["sha256"]

# 3. Periksa apakah ciri gambar sudah disimpan
step05b_folder = (
    ROOT / "02_EXPERIMENTS/"
    "STEP_05B_DINOV2_ORIGINAL"
)

MODEL_NAME = "dinov2_vits14"
PREPROCESS_NAME = "Resize518_CenterCrop518_ImageNet"

previous = []

for result_file in step05b_folder.glob(
    "*/outputs/reference_result.json"
):
    try:
        data = json.loads(
            result_file.read_text(encoding="utf-8")
        )

        embedding_path = (
            result_file.parent / "embedding_original.npy"
        )

        if not (
            data.get("source_id") == selected_id
            and data.get("source_sha256") == source["sha256"]
            and data.get("model") == MODEL_NAME
            and data.get("preprocessing") == PREPROCESS_NAME
            and data.get("status") == "PASS"
            and embedding_path.is_file()
            and file_hash(embedding_path)
                == data.get("embedding_sha256")
        ):
            continue

        vector = np.load(
            embedding_path,
            allow_pickle=False
        )

        if (
            vector.shape == (384,)
            and np.isfinite(vector).all()
        ):
            previous.append(
                (result_file, data)
            )

    except (OSError, KeyError, ValueError, TypeError):
        continue

# 4. Gunakan data lama jika sudah tersedia
if previous:
    previous.sort(key=lambda x: x[0].as_posix())

    result_path, result = previous[-1]

    print("\n=== HASIL STEP 05B ===")
    print("Gambar:", source["original_filename"])
    print("Source ID:", selected_id)
    print("Model:", result["model"])
    print("Embedding dimension:", result["embedding_dimension"])
    print("Gambar dari folder ORIGINAL:", True)
    print("Menggunakan TrustMark:", False)
    print("Hasil tersimpan:", result_path)
    print("STATUS: EXISTING ORIGINAL VERIFIED")

else:
    # 5. Buat folder percobaan baru
    device = (
        "cuda" if torch.cuda.is_available()
        else "cpu"
    )

    run05b = start_experiment(
        step="STEP_05B_DINOV2_ORIGINAL",
        method="DINOV2_ORIGINAL_REFERENCE",
        parameters={
            "source_id": selected_id,
            "model": MODEL_NAME,
            "preprocessing": PREPROCESS_NAME,
            "device": device,
            "input_type": "ORIGINAL"
        }
    )

    metadata_path = run05b / "metadata.json"

    try:
        # 6. Simpan salinan file asli
        input_path = (
            run05b / "inputs" /
            ("original" + original_path.suffix)
        )

        shutil.copy2(original_path, input_path)

        assert file_hash(input_path) == source["sha256"]

        # 7. Siapkan model dan cara membaca gambar
        model05a = model05a.to(device).eval()

        preprocessing = transforms.Compose([
            transforms.Resize(518),
            transforms.CenterCrop(518),
            transforms.ToTensor(),
            transforms.Normalize(
                mean=[0.485, 0.456, 0.406],
                std=[0.229, 0.224, 0.225]
            )
        ])

        # 8. Baca ciri gambar asli
        start = time.perf_counter()

        with Image.open(input_path) as img:
            image = ImageOps.exif_transpose(
                img
            ).convert("RGB")

            tensor = preprocessing(
                image
            ).unsqueeze(0).to(device)

        with torch.inference_mode():
            features = model05a(tensor)
            features = F.normalize(
                features, p=2, dim=1
            )

        vector = (
            features.cpu().numpy()[0]
            .astype(np.float32)
        )

        duration = time.perf_counter() - start

        assert vector.shape == (384,)
        assert np.isfinite(vector).all()

        # 9. Simpan ciri gambar
        output = run05b / "outputs"

        embedding_path = (
            output / "embedding_original.npy"
        )

        np.save(embedding_path, vector)

        # 10. Simpan catatan hasil
        result = {
            "source_id": selected_id,
            "original_filename": source["original_filename"],
            "source_sha256": source["sha256"],
            "model": MODEL_NAME,
            "preprocessing": PREPROCESS_NAME,
            "embedding_dimension": 384,
            "embedding_sha256": file_hash(embedding_path),
            "execution_seconds": duration,
            "device": device,
            "input_type": "ORIGINAL",
            "trustmark_used": False,
            "status": "PASS",
            "timestamp_utc": datetime.now(
                timezone.utc
            ).isoformat()
        }

        result_path = (
            output / "reference_result.json"
        )

        result_path.write_text(
            json.dumps(result, indent=2),
            encoding="utf-8"
        )

        record_event(
            run05b,
            "ORIGINAL_FEATURES_COMPLETED",
            result
        )

        metadata = json.loads(
            metadata_path.read_text()
        )

        metadata["status"] = "PASS"

        metadata_path.write_text(
            json.dumps(metadata, indent=2),
            encoding="utf-8"
        )

        print("\n=== HASIL STEP 05B ===")
        print("Gambar:", source["original_filename"])
        print("Source ID:", selected_id)
        print("Model:", MODEL_NAME)
        print("Embedding dimension:", 384)
        print("Gambar dari folder ORIGINAL:", True)
        print("Menggunakan TrustMark:", False)
        print(
            "Waktu membaca gambar:",
            round(duration, 3), "detik"
        )
        print("Hasil tersimpan:", result_path)
        print("STATUS: DINOV2 ORIGINAL READY")

    except Exception as e:
        record_event(
            run05b,
            "RUN_FAILED",
            {
                "error_type": type(e).__name__,
                "message": str(e)
            }
        )

        metadata = json.loads(
            metadata_path.read_text()
        )

        metadata["status"] = "ERROR"

        metadata_path.write_text(
            json.dumps(metadata, indent=2),
            encoding="utf-8"
        )

        raise
```

```python
# ==================================================
# JATIKARYA NEW - STEP 05C
# Upload Manual Gambar Asli dan Uji DINOv2
# TANPA TRUSTMARK
# ==================================================

from google.colab import files
from PIL import Image, ImageOps
from torchvision import transforms
from datetime import datetime, timezone
import torch
import torch.nn.functional as F
import numpy as np
import pandas as pd
import hashlib
import json
import io
import time

# 1. Periksa proyek dan model
assert ROOT.name == "JATIKARYA_NEW"
assert "model05a" in globals(), (
    "Jalankan STEP 05A terlebih dahulu."
)
assert model05a.embed_dim == 384

# 2. Pilih sumber pembanding
registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str
)

assert not registry.empty
assert registry["source_id"].is_unique

if len(registry) == 1:
    source = registry.iloc[0]
    print("Satu sumber tersedia, dipilih otomatis.")
else:
    print("Daftar sumber:")
    for n, (_, row) in enumerate(registry.iterrows(), 1):
        print(n, row["original_filename"], row["source_id"])

    pilihan = int(input("Nomor sumber: "))
    assert 1 <= pilihan <= len(registry)
    source = registry.iloc[pilihan - 1]

selected_id = source["source_id"]

original_path = ROOT / source["stored_path"]

assert original_path.is_file()
assert file_hash(original_path) == source["sha256"]

# 3. Cari embedding asli dari STEP 05B
step05b = (
    ROOT / "02_EXPERIMENTS/STEP_05B_DINOV2_ORIGINAL"
)

candidates = []

for p in step05b.glob("*/outputs/reference_result.json"):
    try:
        data = json.loads(
            p.read_text(encoding="utf-8")
        )
        vector_path = p.parent / "embedding_original.npy"
        metadata = json.loads(
            (p.parent.parent / "metadata.json").read_text()
        )

        if (
            data.get("source_id") == selected_id
            and data.get("source_sha256") == source["sha256"]
            and data.get("model") == "dinov2_vits14"
            and data.get("preprocessing")
                == "Resize518_CenterCrop518_ImageNet"
            and data.get("status") == "PASS"
            and metadata.get("status") == "PASS"
            and vector_path.is_file()
            and file_hash(vector_path)
                == data.get("embedding_sha256")
        ):
            candidates.append((p, vector_path))

    except (OSError, KeyError, ValueError, TypeError):
        continue

assert candidates, (
    "Referensi asli STEP 05B belum ditemukan."
)

result_file, vector_path = sorted(
    candidates, key=lambda x: x[0].as_posix()
)[-1]

reference_vector = np.load(
    vector_path, allow_pickle=False
)

assert reference_vector.shape == (384,)
assert np.isfinite(reference_vector).all()

# 4. Unggah gambar asli
print("\n=== JATIKARYA NEW - STEP 05C ===")
print("Upload gambar ORIGINAL yang terdaftar.")

uploaded = files.upload()

assert len(uploaded) == 1, "Pilih satu gambar saja."

filename, image_bytes = next(iter(uploaded.items()))

uploaded_hash = hashlib.sha256(
    image_bytes
).hexdigest()

assert uploaded_hash == source["sha256"], (
    "File bukan gambar asli yang terdaftar. "
    "Silakan pilih gambar yang benar."
)

with Image.open(io.BytesIO(image_bytes)) as img:
    image_format = img.format
    img.verify()

assert image_format in ("JPEG", "PNG", "WEBP")

# 5. Siapkan pembacaan gambar
device = "cuda" if torch.cuda.is_available() else "cpu"

model05a = model05a.to(device).eval()

preprocess = transforms.Compose([
    transforms.Resize(518),
    transforms.CenterCrop(518),
    transforms.ToTensor(),
    transforms.Normalize(
        mean=[0.485, 0.456, 0.406],
        std=[0.229, 0.224, 0.225]
    )
])

def baca_ciri05c(image):
    image = ImageOps.exif_transpose(
        image
    ).convert("RGB")

    tensor = preprocess(
        image
    ).unsqueeze(0).to(device)

    with torch.inference_mode():
        vector = model05a(tensor)
        vector = F.normalize(vector, p=2, dim=1)

    return vector.cpu().numpy()[0].astype(np.float32)

# 6. Pastikan model cocok dengan referensi
with Image.open(original_path) as img:
    check_vector = baca_ciri05c(img)

assert np.allclose(
    check_vector,
    reference_vector,
    rtol=1e-3,
    atol=1e-3
), "Model tidak konsisten dengan hasil STEP 05B."

# 7. Buat catatan eksperimen baru
run05c = start_experiment(
    step="STEP_05C_DINOV2_MANUAL",
    method="DINOV2_ORIGINAL_COMPARISON",
    parameters={
        "source_id": selected_id,
        "reference_run": result_file.parent.parent.name,
        "input_type": "ORIGINAL"
    }
)

metadata_path = run05c / "metadata.json"

try:
    # 8. Simpan file unggahan
    ext = {
        "JPEG": ".jpg",
        "PNG": ".png",
        "WEBP": ".webp"
    }[image_format]

    upload_path = (
        run05c / "inputs" / ("uploaded_original" + ext)
    )

    with open(upload_path, "xb") as f:
        f.write(image_bytes)

    assert file_hash(upload_path) == uploaded_hash

    # 9. Baca ciri gambar unggahan
    mulai = time.perf_counter()

    with Image.open(upload_path) as img:
        upload_vector = baca_ciri05c(img)

    duration = time.perf_counter() - mulai

    assert upload_vector.shape == (384,)
    assert np.isfinite(upload_vector).all()

    # 10. Hitung kemiripan
    similarity = float(
        np.dot(reference_vector, upload_vector)
    )

    assert -1.001 <= similarity <= 1.001

    # 11. Simpan hasil
    output = run05c / "outputs"

    np.save(
        output / "embedding_uploaded.npy",
        upload_vector
    )

    result = {
        "source_id": selected_id,
        "uploaded_filename": filename,
        "original_hash_match": True,
        "model": "dinov2_vits14",
        "embedding_dimension": 384,
        "cosine_similarity": similarity,
        "execution_seconds": duration,
        "device": device,
        "trustmark_used": False,
        "status": "PASS",
        "timestamp_utc": datetime.now(
            timezone.utc
        ).isoformat()
    }

    result_path = output / "manual_result.json"

    result_path.write_text(
        json.dumps(result, indent=2),
        encoding="utf-8"
    )

    record_event(
        run05c,
        "DINOV2_MANUAL_COMPLETED",
        result
    )

    metadata = json.loads(
        metadata_path.read_text()
    )
    metadata["status"] = "PASS"
    metadata_path.write_text(
        json.dumps(metadata, indent=2),
        encoding="utf-8"
    )

    # 12. Tampilkan hasil
    print("\n=== HASIL STEP 05C ===")
    print("Nama gambar:", filename)
    print("Hash original cocok:", True)
    print("Model:", "DINOv2 ViT-S/14")
    print("Embedding dimension:", 384)
    print("Cosine similarity:", round(similarity, 6))
    print("Waktu pembacaan:", round(duration, 3), "detik")
    print("Menggunakan TrustMark:", False)
    print("Hasil tersimpan:", result_path)
    print("STATUS: PASS")

except Exception as e:
    record_event(
        run05c,
        "RUN_FAILED",
        {
            "error_type": type(e).__name__,
            "message": str(e)
        }
    )

    metadata = json.loads(
        metadata_path.read_text()
    )
    metadata["status"] = "ERROR"
    metadata_path.write_text(
        json.dumps(metadata, indent=2),
        encoding="utf-8"
    )
    raise
```

```python
# ==================================================
# JATIKARYA NEW - STEP 06A
# UPLOAD GAMBAR EDIT OTOMATIS
# Tanpa mengulang gambar yang sudah terdaftar
# ==================================================

from google.colab import files
from pathlib import Path
from PIL import Image
from datetime import datetime, timezone
from IPython.display import display
import pandas as pd
import hashlib
import io
import json
import re

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

# 1. Pilih gambar sumber
registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str
)

assert not registry.empty
assert registry["source_id"].is_unique

print("=== PILIH GAMBAR SUMBER ===")

for i, (_, row) in enumerate(registry.iterrows(), 1):
    print(
        f"{i}. {row['original_filename']} "
        f"({row['source_id']})"
    )

nomor = (
    1 if len(registry) == 1
    else int(input("Nomor sumber: "))
)

assert 1 <= nomor <= len(registry)

sumber = registry.iloc[nomor - 1]

assert file_hash(
    ROOT / sumber["stored_path"]
) == sumber["sha256"]


# 2. Cari hasil watermark yang valid
watermarks = []

folder_wm = (
    ROOT / "02_EXPERIMENTS/STEP_04_TRUSTMARK_ENCODE"
)

for p in folder_wm.glob(
    "*/outputs/verification_result.json"
):
    try:
        data = json.loads(
            p.read_text(encoding="utf-8")
        )

        image = (
            ROOT / data["stored_path"]
        ).resolve()

        if (
            data.get("source_id") == sumber["source_id"]
            and data.get("source_sha256") == sumber["sha256"]
            and data.get("payload_match") is True
            and image.is_relative_to(
                (ROOT / "01_DATASET/WATERMARKED").resolve()
            )
            and image.is_file()
            and file_hash(image) == data["watermarked_sha256"]
        ):
            watermarks.append(
                (p.parent.parent.name, data)
            )

    except (OSError, KeyError, ValueError, TypeError):
        continue

assert watermarks, (
    "Belum ada watermark valid untuk sumber ini."
)

watermarks.sort(key=lambda x: x[0])

print("\n=== PILIH HASIL WATERMARK ===")

for i, (run_id, _) in enumerate(watermarks, 1):
    print(f"{i}. {run_id}")

nomor_wm = (
    1 if len(watermarks) == 1
    else int(input("Nomor watermark: "))
)

assert 1 <= nomor_wm <= len(watermarks)

wm_run, wm_info = watermarks[nomor_wm - 1]
wm_hash = wm_info["watermarked_sha256"]


# 3. Cari gambar edit yang sudah didaftarkan
folder_upload = (
    ROOT / "02_EXPERIMENTS/STEP_06A_UPLOAD_EDITS"
)

sudah_ada = set()

for p in folder_upload.glob(
    "*/outputs/edit_manifest.json"
):
    manifest = json.loads(
        p.read_text(encoding="utf-8")
    )

    if (
        manifest.get("source_id") == sumber["source_id"]
        and manifest.get("parent_reference_sha256") == wm_hash
    ):
        for item in manifest.get("files", []):
            gambar = ROOT / item["stored_path"]

            if (
                gambar.is_file()
                and file_hash(gambar) == item["sha256"]
            ):
                sudah_ada.add(item["sha256"])


# 4. Upload gambar sebanyak yang diinginkan
print("\nPilih gambar hasil edit.")
print("Boleh upload banyak gambar sekaligus.")

uploads = files.upload()

assert uploads, "Belum ada file yang diunggah."

baru = []
duplikat = []

for nama, isi in uploads.items():

    assert Path(nama).name == nama, (
        "Nama file tidak aman."
    )

    with Image.open(io.BytesIO(isi)) as img:
        assert img.format in ("PNG", "JPEG", "WEBP")
        ukuran = img.size
        img.verify()

    sha = hashlib.sha256(isi).hexdigest()

    # Lewati file dengan isi yang sudah terdaftar
    if sha in sudah_ada:
        duplikat.append({
            "nama_file": nama,
            "sha256": sha
        })
        continue

    sudah_ada.add(sha)

    # Baca jenis dan angka edit dari nama file
    label = re.sub(
        r"\s*\(\d+\)$",
        "",
        Path(nama).stem
    ).strip()

    match = re.search(
        r"\s+(-?\d+(?:[.,]\d+)?)$",
        label
    )

    jenis = (
        label[:match.start()].strip()
        if match else label
    )

    nilai = (
        match.group(1)
        if match else "Tidak disebutkan"
    )

    baru.append({
        "nama_file": nama,
        "isi": isi,
        "sha256": sha,
        "jenis_edit": jenis,
        "nilai_dari_nama": nilai,
        "lebar_px": ukuran[0],
        "tinggi_px": ukuran[1]
    })


# 5. Catat file yang dilewati
if duplikat:

    catatan = ROOT / "04_LOGS/step06a_duplikat.csv"

    pd.DataFrame([
        {
            "source_id": sumber["source_id"],
            "nama_file": x["nama_file"],
            "sha256": x["sha256"],
            "waktu_utc": datetime.now(
                timezone.utc
            ).isoformat()
        }
        for x in duplikat
    ]).to_csv(
        catatan,
        mode="a",
        index=False,
        header=not catatan.exists()
    )


# 6. Simpan hanya gambar baru
if baru:

    run = start_experiment(
        "STEP_06A_UPLOAD_EDITS",
        "MANUAL_EDIT_REGISTRATION",
        {
            "source_id": sumber["source_id"],
            "parent_run": wm_run
        }
    )

    tujuan = (
        ROOT / "01_DATASET/MANUAL_EDITS"
        / sumber["source_id"] / run.name
    )

    tujuan.mkdir(
        parents=True,
        exist_ok=False
    )

    rows = []

    try:
        for item in baru:

            lokasi = tujuan / item["nama_file"]

            with lokasi.open("xb") as f:
                f.write(item["isi"])

            assert file_hash(lokasi) == item["sha256"]

            rows.append({
                "source_id": sumber["source_id"],
                "nama_file": item["nama_file"],
                "jenis_edit": item["jenis_edit"],
                "nilai_dari_nama": item["nilai_dari_nama"],
                "lebar_px": item["lebar_px"],
                "tinggi_px": item["tinggi_px"],
                "sha256": item["sha256"],
                "stored_path": str(
                    lokasi.relative_to(ROOT)
                ),
                "parent_type_declared": "WATERMARKED",
                "parent_reference_sha256": wm_hash,
                "parent_run": wm_run
            })

        # 7. Simpan daftar gambar dan asalnya
        manifest = {
            "source_id": sumber["source_id"],
            "parent_type_declared": "WATERMARKED",
            "parent_reference_sha256": wm_hash,
            "parent_run": wm_run,
            "files": rows,
            "created_at_utc": datetime.now(
                timezone.utc
            ).isoformat()
        }

        output = run / "outputs"

        (output / "edit_manifest.json").write_text(
            json.dumps(
                manifest,
                indent=2,
                ensure_ascii=False
            ),
            encoding="utf-8"
        )

        pd.DataFrame(rows).to_csv(
            output / "edit_manifest.csv",
            index=False
        )

        record_event(
            run,
            "MANUAL_EDITS_IMPORTED",
            {"count": len(rows)}
        )

        meta = run / "metadata.json"

        data = json.loads(
            meta.read_text(encoding="utf-8")
        )

        data["status"] = "PASS"

        meta.write_text(
            json.dumps(data, indent=2),
            encoding="utf-8"
        )

    except Exception as exc:

        record_event(
            run,
            "RUN_FAILED",
            {"error": str(exc)}
        )

        meta = run / "metadata.json"

        data = json.loads(
            meta.read_text(encoding="utf-8")
        )

        data["status"] = "ERROR"

        meta.write_text(
            json.dumps(data, indent=2),
            encoding="utf-8"
        )

        raise

    # 8. Tampilkan gambar yang baru didaftarkan
    print("\n=== GAMBAR BARU ===")

    display(
        pd.DataFrame(rows)[[
            "nama_file",
            "jenis_edit",
            "nilai_dari_nama",
            "lebar_px",
            "tinggi_px"
        ]]
    )

    print(
        "Manifest:",
        output / "edit_manifest.json"
    )


# 9. Ringkasan hasil
print("\n=== RINGKASAN STEP 06A ===")

print("Gambar baru:", len(baru))
print("File identik dilewati:", len(duplikat))

print(
    "STATUS:",
    "DATA BARU TERSIMPAN"
    if baru else "TIDAK ADA DATA BARU"
)

print(
    "Selanjutnya jalankan STEP 06B dan 06C "
    "untuk kumpulan baru saja."
)
```

```python
# ==================================================
# JATIKARYA NEW - STEP 06B
# TRUSTMARK-ONLY: SEMUA KUMPULAN GAMBAR EDIT
# Otomatis, tidak mengulang hasil yang sudah ada
# BELUM FUSION
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from importlib.metadata import version
from PIL import Image, ImageOps
import pandas as pd
import hashlib
import json
import time
import torch
from trustmark import TrustMark

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

for f in (file_hash, start_experiment, record_event):
    assert callable(f)

def baca_json(path):
    return json.loads(path.read_text(encoding="utf-8"))

def cek_file(rel, folder, expected_hash):
    path = (ROOT / rel).resolve()
    assert path.is_relative_to((ROOT / folder).resolve()), (
        f"Lokasi salah: {rel}"
    )
    assert path.is_file() and file_hash(path) == expected_hash, (
        f"File tidak cocok: {rel}"
    )
    return path

# 1. Baca sumber dan watermark yang valid
registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str
)

assert not registry.empty
assert registry.source_id.is_unique

sources = {
    r["source_id"]: r
    for _, r in registry.iterrows()
}

for sid, r in sources.items():
    cek_file(
        r["stored_path"],
        "01_DATASET/ORIGINAL",
        r["sha256"]
    )

watermarks = set()

folder_wm = ROOT / "02_EXPERIMENTS/STEP_04_TRUSTMARK_ENCODE"

for p in folder_wm.glob(
    "*/outputs/verification_result.json"
):
    try:
        w = baca_json(p)
        sid = w["source_id"]

        if (
            sid in sources
            and w["payload_match"] is True
            and w["source_sha256"] == sources[sid]["sha256"]
        ):
            cek_file(
                w["stored_path"],
                "01_DATASET/WATERMARKED",
                w["watermarked_sha256"]
            )

            watermarks.add(
                (sid, w["watermarked_sha256"])
            )

    except (KeyError, ValueError, OSError,
            TypeError, AssertionError):
        continue

# 2. Gabungkan semua gambar edit STEP 06A
records = {}
man_by_hash = {}

folder_a = ROOT / "02_EXPERIMENTS/STEP_06A_UPLOAD_EDITS"

for p in sorted(
    folder_a.glob("*/outputs/edit_manifest.json")
):
    m = baca_json(p)

    if m.get("parent_type_declared") != "WATERMARKED":
        continue

    meta = baca_json(
        p.parent.parent / "metadata.json"
    )

    assert meta["status"] == "PASS", (
        f"Kumpulan belum lengkap: {p}"
    )

    sid = m["source_id"]
    wm_sha = m["parent_reference_sha256"]

    assert (sid, wm_sha) in watermarks, (
        f"Watermark induk belum valid: {p}"
    )

    assert len(m["files"]) > 0

    manifest_sha = file_hash(p)

    man_by_hash[manifest_sha] = (
        sid, wm_sha, set()
    )

    for item in m["files"]:
        assert item["source_id"] == sid

        cek_file(
            item["stored_path"],
            "01_DATASET/MANUAL_EDITS",
            item["sha256"]
        )

        key = (
            sid,
            wm_sha,
            item["sha256"]
        )

        man_by_hash[manifest_sha][2].add(
            item["sha256"]
        )

        # Gambar identik cukup dicatat sekali
        records.setdefault(key, {
            "source_id": sid,
            "nama_file": item["nama_file"],
            "jenis_edit": item["jenis_edit"],
            "nilai_dari_nama": item["nilai_dari_nama"],
            "sha256": item["sha256"],
            "parent_reference_sha256": wm_sha,
            "stored_path": item["stored_path"]
        })

assert records, (
    "Belum ada gambar edit ber-watermark."
)

TM_VERSION = version("trustmark")
keys = sorted(records)

dataset_sha = hashlib.sha256(
    json.dumps(keys).encode("utf-8")
).hexdigest()

folder_b = ROOT / "02_EXPERIMENTS/STEP_06B_TRUSTMARK_ONLY"

cache = {}
full_report = None

# 3. Cari hasil pengujian sebelumnya
for sp in sorted(
    folder_b.glob("*/outputs/summary.json")
):
    try:
        info = baca_json(sp)
        csv_path = sp.parent / "trustmark_results.csv"

        if not (
            csv_path.is_file()
            and file_hash(csv_path) == info["results_sha256"]
            and info["trustmark_version"] == TM_VERSION
            and info["error_count"] == 0
        ):
            continue

        rows = pd.read_csv(
            csv_path,
            dtype=str,
            keep_default_na=False
        ).to_dict("records")

        # Jika seluruh kumpulan sudah pernah diuji
        if (
            info.get("dataset_sha256") == dataset_sha
            and len(rows) == len(keys)
        ):
            full_report = (csv_path, info)

        origin = man_by_hash.get(
            info.get("input_manifest_sha256")
        )

        for row in rows:
            sid = row["source_id"]
            sha = row["sha256"]

            parent = row.get(
                "parent_reference_sha256"
            )

            if not parent and origin:
                if (
                    origin[0] == sid
                    and sha in origin[2]
                ):
                    parent = origin[1]

            key = (sid, parent, sha)

            if key not in records or row.get("error"):
                continue

            detected = (
                row["watermark_detected"] == "True"
            )

            match = (
                row["payload_match"] == "True"
            )

            label = row["hasil"]

            expected = bin(
                int(sources[sid]["sha256"], 16)
            )[2:].zfill(256)[:61]

            valid = (
                (
                    label == "KODE_COCOK"
                    and detected
                    and match
                    and row.get("decoded_payload") == expected
                )
                or (
                    label == "KODE_TIDAK_COCOK"
                    and detected
                    and not match
                )
                or (
                    label == "TIDAK_TERDETEKSI"
                    and not detected
                    and not match
                )
            )

            if valid:
                cache[key] = row

    except (KeyError, ValueError, OSError, TypeError):
        continue

# 4. Jika rekap lengkap sudah tersedia
if full_report is not None:

    result_file, summary = full_report
    hasil = pd.read_csv(result_file)

    print(
        "Seluruh hasil sudah tersedia. "
        "Tidak ada pengujian ulang."
    )

else:
    # 5. Buat satu percobaan untuk rekap gabungan
    run = start_experiment(
        "STEP_06B_TRUSTMARK_ONLY",
        "TrustMark-Q-BCH5-Decode",
        {
            "dataset_sha256": dataset_sha,
            "trustmark_version": TM_VERSION
        }
    )

    meta_file = run / "metadata.json"

    try:
        pending = [
            key for key in keys
            if key not in cache
        ]

        print("=== STEP 06B: SEMUA KUMPULAN ===")
        print("Jumlah gambar unik:", len(keys))
        print("Hasil lama digunakan:", len(keys) - len(pending))
        print("Perlu diuji:", len(pending))

        device = (
            "cuda" if torch.cuda.is_available()
            else "cpu"
        )

        detector = None

        # Muat TrustMark hanya jika diperlukan
        if pending:
            detector = TrustMark(
                verbose=False,
                model_type="Q",
                encoding_type=TrustMark.Encoding.BCH_5,
                device=device
            )

        output_rows = []

        # 6. Periksa semua gambar
        for i, key in enumerate(keys, 1):

            item = records[key].copy()

            if key in cache:
                old = cache[key]

                for col in (
                    "watermark_detected",
                    "payload_match",
                    "hasil",
                    "waktu_detik",
                    "decoded_payload",
                    "schema",
                    "error"
                ):
                    item[col] = old.get(col, "")

                item["asal_hasil"] = "ARSIP"

            else:
                sid = key[0]

                expected = bin(
                    int(sources[sid]["sha256"], 16)
                )[2:].zfill(256)[:61]

                try:
                    with Image.open(
                        ROOT / item["stored_path"]
                    ) as im:
                        img = ImageOps.exif_transpose(
                            im
                        ).convert("RGB")

                    start = time.perf_counter()

                    decoded, detected, schema = (
                        detector.decode(
                            img,
                            MODE="binary"
                        )
                    )

                    elapsed = (
                        time.perf_counter() - start
                    )

                    detected = bool(detected)

                    match = (
                        detected
                        and decoded == expected
                        and schema == TrustMark.Encoding.BCH_5
                    )

                    label = (
                        "KODE_COCOK"
                        if match
                        else (
                            "KODE_TIDAK_COCOK"
                            if detected
                            else "TIDAK_TERDETEKSI"
                        )
                    )

                    item.update(
                        watermark_detected=detected,
                        payload_match=bool(match),
                        hasil=label,
                        waktu_detik=round(elapsed, 6),
                        decoded_payload=(
                            str(decoded)
                            if decoded is not None
                            else ""
                        ),
                        schema=(
                            str(schema)
                            if schema is not None
                            else ""
                        ),
                        error=""
                    )

                except Exception as e:
                    item.update(
                        watermark_detected=False,
                        payload_match=False,
                        hasil="ERROR",
                        waktu_detik=None,
                        decoded_payload="",
                        schema="",
                        error=f"{type(e).__name__}: {e}"
                    )

                item["asal_hasil"] = "BARU"

            output_rows.append(item)

            print(
                f"[{i}/{len(keys)}] "
                f"{item['nama_file']}: "
                f"{item['hasil']} "
                f"({item['asal_hasil']})"
            )

        # 7. Simpan seluruh hasil dalam satu tabel
        hasil = pd.DataFrame(output_rows)

        result_file = (
            run / "outputs/trustmark_results.csv"
        )

        hasil.to_csv(
            result_file,
            index=False
        )

        summary = {
            "dataset_sha256": dataset_sha,
            "trustmark_version": TM_VERSION,
            "total_images": len(hasil),
            "from_archive": len(keys) - len(pending),
            "newly_tested": len(pending),
            "matched_count": int(
                (hasil.hasil == "KODE_COCOK").sum()
            ),
            "not_detected_count": int(
                (hasil.hasil == "TIDAK_TERDETEKSI").sum()
            ),
            "mismatch_count": int(
                (hasil.hasil == "KODE_TIDAK_COCOK").sum()
            ),
            "error_count": int(
                (hasil.hasil == "ERROR").sum()
            ),
            "results_sha256": file_hash(result_file),
            "timestamp_utc": datetime.now(
                timezone.utc
            ).isoformat()
        }

        summary_path = run / "outputs/summary.json"

        summary_path.write_text(
            json.dumps(summary, indent=2),
            encoding="utf-8"
        )

        record_event(
            run,
            "TRUSTMARK_ALL_BATCHES_COMPLETED",
            summary
        )

        meta = baca_json(meta_file)

        meta["status"] = (
            "COMPLETED"
            if summary["error_count"] == 0
            else "COMPLETED_WITH_ERRORS"
        )

        meta_file.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

    except Exception as e:
        record_event(
            run,
            "RUN_FAILED",
            {"error": f"{type(e).__name__}: {e}"}
        )

        meta = baca_json(meta_file)
        meta["status"] = "ERROR"

        meta_file.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

        raise

# 8. Tampilkan seluruh hasil
print("\n=== STEP 06B: SELURUH GAMBAR EDIT ===")

display(
    hasil[[
        "nama_file",
        "jenis_edit",
        "nilai_dari_nama",
        "watermark_detected",
        "payload_match",
        "hasil",
        "waktu_detik",
        "asal_hasil"
    ]]
)

print("\nTotal gambar unik:", summary["total_images"])
print("Kode cocok:", summary["matched_count"])
print("Tidak terdeteksi:", summary["not_detected_count"])
print("Kode tidak cocok:", summary["mismatch_count"])
print("Error:", summary["error_count"])
print("Hasil tersimpan:", result_file)
print("STATUS: TRUSTMARK-ONLY, BELUM FUSION")
```

```python
# ==================================================
# JATIKARYA - TES SEMENTARA ROTASI 180
# Tidak menyimpan atau mengubah data penelitian
# ==================================================

from pathlib import Path
from PIL import Image
from trustmark import TrustMark
import pandas as pd
import torch
import json

# 1. Temukan gambar rotate 180.png yang terdaftar
assert ROOT.name == "JATIKARYA_NEW"

matches = []

folder = ROOT / "02_EXPERIMENTS/STEP_06A_UPLOAD_EDITS"

for p in folder.glob("*/outputs/edit_manifest.json"):
    manifest = json.loads(p.read_text(encoding="utf-8"))

    for item in manifest["files"]:
        if item["nama_file"].lower() == "rotate 180.png":
            path = ROOT / item["stored_path"]

            if path.is_file() and file_hash(path) == item["sha256"]:
                matches.append((path, item["source_id"]))

assert len(matches) == 1, (
    "Gambar rotate 180.png tidak ditemukan secara unik."
)

gambar_path, source_id = matches[0]

# 2. Ambil kode watermark dari sumber terdaftar
registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str
)

source = registry.loc[
    registry["source_id"] == source_id
]

assert len(source) == 1

sha = source.iloc[0]["sha256"]
payload_asli = bin(int(sha, 16))[2:].zfill(256)[:61]

# 3. Putar kembali 180 derajat di memori
with Image.open(gambar_path) as img:
    gambar_dikembalikan = (
        img.convert("RGB")
        .transpose(Image.Transpose.ROTATE_180)
    )

# 4. Siapkan pembaca TrustMark
# Gunakan model yang sudah ada jika tersedia
pembaca = globals().get("detector")

if pembaca is None:
    pembaca = TrustMark(
        verbose=False,
        model_type="Q",
        encoding_type=TrustMark.Encoding.BCH_5,
        device="cuda" if torch.cuda.is_available() else "cpu"
    )

# 5. Baca watermark satu kali
decoded, detected, schema = pembaca.decode(
    gambar_dikembalikan,
    MODE="binary"
)

cocok = (
    bool(detected)
    and decoded == payload_asli
    and schema == TrustMark.Encoding.BCH_5
)

# 6. Tampilkan hasil
print("=== TES SEMENTARA ROTASI 180 ===")
print("Gambar:", gambar_path.name)
print("Rotasi tambahan:", "180 derajat")
print("Watermark ditemukan:", bool(detected))
print("Kode cocok:", bool(cocok))

if cocok:
    print("KESIMPULAN: Kode berhasil dipulihkan.")
elif detected:
    print("KESIMPULAN: Ada deteksi, tetapi kode tidak cocok.")
else:
    print("KESIMPULAN: Watermark tetap tidak terdeteksi.")

print("Tidak ada file eksperimen baru yang disimpan.")
```

```python
# ==================================================
# TES SEMENTARA: ROTASI 90 DAN 270 DERAJAT
# Tidak mengubah atau menyimpan file penelitian
# ==================================================

from PIL import Image
from trustmark import TrustMark
import pandas as pd
import json
import torch

assert ROOT.name == "JATIKARYA_NEW"

# 1. Cari gambar ber-watermark yang sudah berhasil
pilihan = []

folder = ROOT / "02_EXPERIMENTS/STEP_04_TRUSTMARK_ENCODE"

for p in sorted(folder.glob(
    "*/outputs/verification_result.json"
)):
    data = json.loads(p.read_text(encoding="utf-8"))

    if data.get("payload_match") is not True:
        continue

    path = ROOT / data["stored_path"]

    if (
        path.is_file()
        and file_hash(path) == data["watermarked_sha256"]
    ):
        pilihan.append((path, data))

assert pilihan, "Gambar watermark tidak ditemukan."

if len(pilihan) == 1:
    nomor = 1
else:
    for i, (path, _) in enumerate(pilihan, 1):
        print(f"{i}. {path.name}")
    nomor = int(input("Pilih gambar: "))

assert 1 <= nomor <= len(pilihan)

path, data = pilihan[nomor - 1]

# 2. Ambil gambar dan kode watermark
with Image.open(path) as img:
    gambar = img.convert("RGB")

payload = data["payload_bits"]

# 3. Siapkan TrustMark
pembaca = TrustMark(
    verbose=False,
    model_type="Q",
    encoding_type=TrustMark.Encoding.BCH_5,
    device="cuda" if torch.cuda.is_available() else "cpu"
)

# 4. Buat rotasi hanya di memori
rotasi_90 = gambar.transpose(
    Image.Transpose.ROTATE_90
)

rotasi_270 = gambar.transpose(
    Image.Transpose.ROTATE_270
)

dikembalikan = rotasi_90.transpose(
    Image.Transpose.ROTATE_270
)

# 5. Periksa watermark masing-masing
percobaan = {
    "Gambar normal": gambar,
    "Rotasi 90 derajat": rotasi_90,
    "Rotasi 270 derajat": rotasi_270,
    "Rotasi 90 lalu dikembalikan": dikembalikan
}

print("=== TES SEMENTARA ORIENTASI GAMBAR ===")

for nama, image in percobaan.items():

    decoded, detected, schema = pembaca.decode(
        image, MODE="binary"
    )

    cocok = (
        bool(detected)
        and decoded == payload
        and schema == TrustMark.Encoding.BCH_5
    )

    print(
        f"{nama}: "
        f"Terdeteksi={bool(detected)}, "
        f"Kode cocok={bool(cocok)}"
    )

print("\nTes selesai. Tidak ada file baru disimpan.")
```

```python
# ==================================================
# JATIKARYA NEW - STEP 06C
# DINOv2-only untuk SEMUA gambar edit
# Otomatis menggunakan hasil lama yang valid
# BELUM FUSION
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from PIL import Image, ImageOps
from torchvision import transforms
import torch
import torch.nn.functional as F
import pandas as pd
import numpy as np
import hashlib
import json
import time

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

assert "model05a" in globals(), (
    "Jalankan STEP 05A terlebih dahulu."
)
assert getattr(model05a, "embed_dim", None) == 384

def read_json(path):
    return json.loads(path.read_text(encoding="utf-8"))

def checked_path(relative, folder, sha):
    path = (ROOT / relative).resolve()
    assert path.is_relative_to((ROOT / folder).resolve())
    assert path.is_file() and file_hash(path) == sha
    return path

# 1. Baca daftar gambar asli
registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str
)

assert registry["source_id"].is_unique

sources = {
    row["source_id"]: row
    for _, row in registry.iterrows()
}

for source in sources.values():
    checked_path(
        source["stored_path"],
        "01_DATASET/ORIGINAL",
        source["sha256"]
    )

# 2. Verifikasi daftar gambar ber-watermark
watermarks = set()

wm_folder = (
    ROOT / "02_EXPERIMENTS/STEP_04_TRUSTMARK_ENCODE"
)

for path in wm_folder.glob(
    "*/outputs/verification_result.json"
):
    try:
        wm = read_json(path)
        sid = wm["source_id"]

        if (
            sid in sources
            and wm.get("payload_match") is True
            and wm["source_sha256"] == sources[sid]["sha256"]
        ):
            checked_path(
                wm["stored_path"],
                "01_DATASET/WATERMARKED",
                wm["watermarked_sha256"]
            )
            watermarks.add(
                (sid, wm["watermarked_sha256"])
            )

    except (OSError, KeyError, ValueError,
            TypeError, AssertionError):
        continue

# 3. Kumpulkan seluruh gambar edit unik
records = {}
manifest_keys = {}

folder06a = (
    ROOT / "02_EXPERIMENTS/STEP_06A_UPLOAD_EDITS"
)

for path in sorted(
    folder06a.glob("*/outputs/edit_manifest.json")
):
    manifest = read_json(path)

    if manifest.get("parent_type_declared") != "WATERMARKED":
        continue

    metadata = read_json(
        path.parent.parent / "metadata.json"
    )

    assert metadata["status"] == "PASS"

    sid = manifest["source_id"]
    wm_sha = manifest["parent_reference_sha256"]

    assert (sid, wm_sha) in watermarks, (
        "Referensi watermark tidak valid."
    )

    origin_keys = set()

    for item in manifest["files"]:
        assert item["source_id"] == sid

        checked_path(
            item["stored_path"],
            "01_DATASET/MANUAL_EDITS",
            item["sha256"]
        )

        key = (sid, wm_sha, item["sha256"])

        records.setdefault(key, item)
        origin_keys.add(key)

    manifest_keys[file_hash(path)] = origin_keys

assert records, "Belum ada gambar edit terdaftar."

keys = sorted(records)

dataset_sha = hashlib.sha256(
    json.dumps(keys, separators=(",", ":")).encode()
).hexdigest()

# 4. Baca ciri gambar asli dari STEP 05B
references = {}

folder05b = (
    ROOT / "02_EXPERIMENTS/STEP_05B_DINOV2_ORIGINAL"
)

for sid in sorted({key[0] for key in keys}):

    found = []

    for path in folder05b.glob(
        "*/outputs/reference_result.json"
    ):
        try:
            info = read_json(path)

            embedding_path = (
                path.parent / "embedding_original.npy"
            )

            if (
                info.get("source_id") == sid
                and info.get("source_sha256")
                    == sources[sid]["sha256"]
                and info.get("model") == "dinov2_vits14"
                and info.get("preprocessing")
                    == "Resize518_CenterCrop518_ImageNet"
                and info.get("status") == "PASS"
                and embedding_path.is_file()
                and file_hash(embedding_path)
                    == info["embedding_sha256"]
            ):
                vector = np.load(
                    embedding_path,
                    allow_pickle=False
                )

                if (
                    vector.shape == (384,)
                    and np.isfinite(vector).all()
                ):
                    found.append((path, info, vector))

        except (OSError, KeyError, ValueError, TypeError):
            continue

    assert found, (
        f"Referensi DINOv2 belum tersedia: {sid}"
    )

    _, info, vector = sorted(
        found, key=lambda x: str(x[0])
    )[-1]

    references[sid] = {
        "vector": vector,
        "hash": info["embedding_sha256"]
    }

reference_hashes = {
    sid: data["hash"]
    for sid, data in references.items()
}

# 5. Cari hasil DINOv2 yang sudah pernah diuji
folder06c = (
    ROOT / "02_EXPERIMENTS/STEP_06C_DINOV2_ONLY"
)

cache = {}
full_report = None

for summary_path in sorted(
    folder06c.glob("*/outputs/summary.json")
):
    try:
        summary = read_json(summary_path)

        csv_path = (
            summary_path.parent / "dinov2_results.csv"
        )

        if not (
            summary.get("model") == "dinov2_vits14"
            and summary.get("preprocessing")
                == "Resize518_CenterCrop518_ImageNet"
            and summary.get("error_count") == 0
            and csv_path.is_file()
            and file_hash(csv_path)
                == summary["results_sha256"]
        ):
            continue

        table = pd.read_csv(
            csv_path,
            dtype=str,
            keep_default_na=False
        )

        if len(table) != summary["total_images"]:
            continue

        is_full = (
            summary.get("dataset_sha256") == dataset_sha
            and summary.get("reference_hashes")
                == reference_hashes
        )

        allowed = (
            set(keys) if is_full
            else manifest_keys.get(
                summary.get("input_manifest_sha256"),
                set()
            )
        )

        if not allowed:
            continue

        if is_full and len(table) == len(keys):
            full_report = (csv_path, summary)

        for _, row in table.iterrows():

            sid = row["source_id"]
            sha = row["image_sha256"]

            if is_full:
                key = (
                    sid,
                    row["parent_reference_sha256"],
                    sha
                )
            else:
                choices = [
                    key for key in allowed
                    if key[0] == sid and key[2] == sha
                ]

                if (
                    len(choices) != 1
                    or summary.get(
                        "reference_embedding_sha256"
                    ) != references[sid]["hash"]
                ):
                    continue

                key = choices[0]

            if (
                key in allowed
                and key in records
                and row.get("status") == "PASS"
                and not row.get("error", "")
                and np.isfinite(
                    float(row["cosine_similarity"])
                )
                and -1.001 <= float(
                    row["cosine_similarity"]
                ) <= 1.001
            ):
                cache[key] = row.to_dict()

    except (OSError, KeyError, ValueError, TypeError):
        continue

# 6. Jika seluruh hasil sudah tersedia
if full_report:

    result_path, summary = full_report
    result_df = pd.read_csv(result_path)

    status = "EXISTING RESULT VERIFIED"

else:
    pending = [
        key for key in keys
        if key not in cache
    ]

    print("=== JATIKARYA NEW - STEP 06C ===")
    print("Total gambar unik:", len(keys))
    print("Hasil lama digunakan:", len(keys) - len(pending))
    print("Gambar perlu diproses:", len(pending))

    # 7. Siapkan model hanya bila ada gambar baru
    if pending:
        device = (
            "cuda" if torch.cuda.is_available()
            else "cpu"
        )

        model05a = model05a.to(device).eval()

        preprocessing = transforms.Compose([
            transforms.Resize(518),
            transforms.CenterCrop(518),
            transforms.ToTensor(),
            transforms.Normalize(
                [0.485, 0.456, 0.406],
                [0.229, 0.224, 0.225]
            )
        ])

        def extract_features(path):

            with Image.open(path) as img:
                image = ImageOps.exif_transpose(
                    img
                ).convert("RGB")

                tensor = preprocessing(
                    image
                ).unsqueeze(0).to(device)

            with torch.inference_mode():
                vector = model05a(tensor)
                vector = F.normalize(
                    vector, p=2, dim=1
                )

            return (
                vector.cpu().numpy()[0]
                .astype(np.float32)
            )

        # Pastikan model sama dengan STEP 05B
        for sid in sorted(
            {key[0] for key in pending}
        ):
            original = checked_path(
                sources[sid]["stored_path"],
                "01_DATASET/ORIGINAL",
                sources[sid]["sha256"]
            )

            check_vector = extract_features(original)

            assert np.allclose(
                check_vector,
                references[sid]["vector"],
                rtol=1e-3,
                atol=1e-3
            ), (
                "Model tidak sesuai dengan STEP 05B."
            )

    # 8. Buat catatan percobaan baru
    run = start_experiment(
        "STEP_06C_DINOV2_ONLY",
        "DINOv2-ViT-S14",
        {
            "dataset_sha256": dataset_sha,
            "reference_hashes": reference_hashes,
            "model": "dinov2_vits14"
        }
    )

    metadata_path = run / "metadata.json"
    rows = []

    try:
        # 9. Periksa semua gambar
        for nomor, key in enumerate(keys, 1):

            item = records[key]
            sid, wm_sha, image_sha = key

            row = {
                "source_id": sid,
                "nama_file": item["nama_file"],
                "jenis_edit": item["jenis_edit"],
                "nilai_dari_nama":
                    item["nilai_dari_nama"],
                "image_sha256": image_sha,
                "parent_reference_sha256": wm_sha
            }

            if key in cache:
                old = cache[key]

                row.update(
                    cosine_similarity=float(
                        old["cosine_similarity"]
                    ),
                    waktu_detik=float(
                        old["waktu_detik"]
                    ),
                    status="PASS",
                    error="",
                    asal_hasil="ARSIP"
                )

            else:
                try:
                    mulai = time.perf_counter()

                    vector = extract_features(
                        ROOT / item["stored_path"]
                    )

                    duration = (
                        time.perf_counter() - mulai
                    )

                    assert vector.shape == (384,)
                    assert np.isfinite(vector).all()

                    similarity = float(
                        np.dot(
                            references[sid]["vector"],
                            vector
                        )
                    )

                    assert -1.001 <= similarity <= 1.001

                    embedding_path = (
                        run / "outputs"
                        / f"embedding_{nomor:03d}.npy"
                    )

                    np.save(embedding_path, vector)

                    row.update(
                        cosine_similarity=similarity,
                        waktu_detik=duration,
                        status="PASS",
                        error="",
                        asal_hasil="BARU",
                        embedding_path=str(
                            embedding_path.relative_to(ROOT)
                        ),
                        embedding_sha256=file_hash(
                            embedding_path
                        )
                    )

                except Exception as e:
                    row.update(
                        cosine_similarity=None,
                        waktu_detik=None,
                        status="ERROR",
                        error=f"{type(e).__name__}: {e}",
                        asal_hasil="BARU"
                    )

            rows.append(row)

            print(
                f"[{nomor}/{len(keys)}] "
                f"{row['nama_file']}: "
                f"{row['status']} "
                f"({row['asal_hasil']})"
            )

        # 10. Simpan seluruh hasil
        result_df = pd.DataFrame(rows)

        result_path = (
            run / "outputs/dinov2_results.csv"
        )

        result_df.to_csv(
            result_path,
            index=False
        )

        error_count = int(
            (result_df["status"] == "ERROR").sum()
        )

        summary = {
            "dataset_sha256": dataset_sha,
            "reference_hashes": reference_hashes,
            "model": "dinov2_vits14",
            "preprocessing":
                "Resize518_CenterCrop518_ImageNet",
            "total_images": len(result_df),
            "error_count": error_count,
            "from_archive": len(keys) - len(pending),
            "newly_tested": len(pending),
            "results_sha256": file_hash(result_path),
            "timestamp_utc": datetime.now(
                timezone.utc
            ).isoformat()
        }

        (
            run / "outputs/summary.json"
        ).write_text(
            json.dumps(summary, indent=2),
            encoding="utf-8"
        )

        record_event(
            run,
            "DINOV2_ALL_BATCHES_COMPLETED",
            summary
        )

        metadata = read_json(metadata_path)

        metadata["status"] = (
            "COMPLETED"
            if error_count == 0
            else "COMPLETED_WITH_ERRORS"
        )

        metadata_path.write_text(
            json.dumps(metadata, indent=2),
            encoding="utf-8"
        )

        status = metadata["status"]

    except Exception as e:
        record_event(
            run,
            "RUN_FAILED",
            {"error": f"{type(e).__name__}: {e}"}
        )

        metadata = read_json(metadata_path)
        metadata["status"] = "ERROR"

        metadata_path.write_text(
            json.dumps(metadata, indent=2),
            encoding="utf-8"
        )

        raise

# 11. Tampilkan tabel seluruh gambar
print("\n=== HASIL STEP 06C: DINOV2-ONLY ===")

display(
    result_df[[
        "nama_file",
        "jenis_edit",
        "nilai_dari_nama",
        "cosine_similarity",
        "waktu_detik",
        "status",
        "asal_hasil"
    ]]
)

print("\nJumlah gambar:", len(result_df))
print("Error:", summary["error_count"])
print("Hasil tersimpan:", result_path)
print("STATUS:", status, "| BELUM FUSION")
```

```python
# ==================================================
# JATIKARYA NEW - STEP 06D
# Tabel Perbandingan Semua Gambar
# TRUSTMARK + DINOV2, TANPA FUSION
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
import pandas as pd
import hashlib
import json
import shutil

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

assert all(
    callable(globals().get(n))
    for n in ("file_hash", "start_experiment", "record_event")
)

# 1. Kumpulkan seluruh gambar dari STEP 06A
edit_root = (ROOT / "01_DATASET/MANUAL_EDITS").resolve()
folder_a = ROOT / "02_EXPERIMENTS/STEP_06A_UPLOAD_EDITS"

images = {}

for p in sorted(folder_a.glob("*/outputs/edit_manifest.json")):

    manifest = json.loads(p.read_text(encoding="utf-8"))

    meta = json.loads(
        (p.parent.parent / "metadata.json").read_text(
            encoding="utf-8"
        )
    )

    if (
        manifest.get("parent_type_declared") != "WATERMARKED"
        or meta.get("status") != "PASS"
    ):
        continue

    for x in manifest["files"]:

        assert x["source_id"] == manifest["source_id"]

        path = (ROOT / x["stored_path"]).resolve()

        assert path.is_relative_to(edit_root)
        assert path.is_file()
        assert file_hash(path) == x["sha256"]

        key = (
            x["source_id"],
            manifest["parent_reference_sha256"],
            x["sha256"]
        )

        # File identik dihitung satu kali
        if key not in images:

            images[key] = {
                "source_id": key[0],
                "parent_reference_sha256": key[1],
                "sha256": key[2],
                "nama_file": x["nama_file"],
                "jenis_edit": x["jenis_edit"],
                "nilai_dari_nama": x["nilai_dari_nama"],
                "stored_path": x["stored_path"]
            }

assert images, "Belum ada gambar dari STEP 06A."

keys = [
    "source_id",
    "parent_reference_sha256",
    "sha256"
]

expected = set(images)


# 2. Cari hasil lengkap dari STEP 06B dan 06C
def hasil_lengkap(step, filename, rename=None):

    folder = ROOT / "02_EXPERIMENTS" / step
    ditemukan = []

    for summary_file in folder.glob("*/outputs/summary.json"):

        try:
            info = json.loads(
                summary_file.read_text(encoding="utf-8")
            )

            meta = json.loads(
                (
                    summary_file.parent.parent / "metadata.json"
                ).read_text(encoding="utf-8")
            )

            csv_path = summary_file.parent / filename

            if (
                meta.get("status") != "COMPLETED"
                or int(info.get("error_count", -1)) != 0
                or int(info.get("total_images", -1)) != len(expected)
                or not csv_path.is_file()
                or file_hash(csv_path) != info.get("results_sha256")
            ):
                continue

            df = pd.read_csv(
                csv_path,
                dtype=str,
                keep_default_na=False
            )

            if rename:
                df = df.rename(columns=rename)

            if not set(keys + ["nama_file"]).issubset(df.columns):
                continue

            found_keys = list(
                map(tuple, df[keys].to_numpy())
            )

            if (
                len(found_keys) != len(expected)
                or set(found_keys) != expected
            ):
                continue

            if (
                step == "STEP_06C_DINOV2_ONLY"
                and not df["status"].eq("PASS").all()
            ):
                continue

            if (
                step == "STEP_06B_TRUSTMARK_ONLY"
                and not df["hasil"].isin([
                    "KODE_COCOK",
                    "TIDAK_TERDETEKSI",
                    "KODE_TIDAK_COCOK"
                ]).all()
            ):
                continue

            ditemukan.append((
                summary_file.parent.parent.name,
                csv_path,
                df
            ))

        except (OSError, KeyError, TypeError, ValueError):
            continue

    assert ditemukan, (
        f"Hasil lengkap {step} belum tersedia."
    )

    return sorted(
        ditemukan, key=lambda x: x[0]
    )[-1]


# 3. Ambil hasil TrustMark dan DINOv2
run_b, csv_b, tm = hasil_lengkap(
    "STEP_06B_TRUSTMARK_ONLY",
    "trustmark_results.csv"
)

run_c, csv_c, di = hasil_lengkap(
    "STEP_06C_DINOV2_ONLY",
    "dinov2_results.csv",
    {"image_sha256": "sha256"}
)


# 4. Cocokkan kedua metode berdasarkan isi file
base = pd.DataFrame(images.values())

tm_cols = keys + [
    "watermark_detected",
    "payload_match",
    "hasil",
    "waktu_detik",
    "asal_hasil"
]

di_cols = keys + [
    "cosine_similarity",
    "waktu_detik",
    "asal_hasil"
]

assert set(tm_cols).issubset(tm.columns)
assert set(di_cols).issubset(di.columns)

combined = base.merge(
    tm[tm_cols].rename(columns={
        "hasil": "hasil_trustmark",
        "waktu_detik": "waktu_trustmark",
        "asal_hasil": "asal_trustmark"
    }),
    on=keys,
    validate="one_to_one"
).merge(
    di[di_cols].rename(columns={
        "cosine_similarity": "kemiripan_dinov2",
        "waktu_detik": "waktu_dinov2",
        "asal_hasil": "asal_dinov2"
    }),
    on=keys,
    validate="one_to_one"
)

assert len(combined) == len(expected)

for col in (
    "kemiripan_dinov2",
    "waktu_trustmark",
    "waktu_dinov2"
):
    combined[col] = pd.to_numeric(
        combined[col],
        errors="raise"
    )

assert combined["kemiripan_dinov2"].between(
    -1.001, 1.001
).all()

combined = combined.sort_values(
    ["source_id", "nama_file"]
).reset_index(drop=True)


# 5. Periksa apakah tabel gabungan sudah dibuat
fingerprint = hashlib.sha256(
    (
        file_hash(csv_b)
        + "|"
        + file_hash(csv_c)
    ).encode("utf-8")
).hexdigest()

folder_d = (
    ROOT / "02_EXPERIMENTS/"
    "STEP_06D_INDIVIDUAL_COMPARISON"
)

old = []

for s in folder_d.glob(
    "*/outputs/comparison_all_summary.json"
):
    try:
        summary = json.loads(
            s.read_text(encoding="utf-8")
        )

        csv_path = (
            s.parent / "individual_comparison_all.csv"
        )

        if (
            summary.get("input_fingerprint") == fingerprint
            and csv_path.is_file()
            and file_hash(csv_path)
                == summary.get("result_sha256")
        ):
            old.append((
                s.parent.parent.name,
                csv_path
            ))

    except (OSError, ValueError, KeyError):
        continue


# 6. Gunakan hasil lama atau buat tabel baru
if old:

    result_path = sorted(old)[-1][1]

    combined = pd.read_csv(result_path)

    status = "EXISTING COMPARISON VERIFIED"

else:

    run_d = start_experiment(
        "STEP_06D_INDIVIDUAL_COMPARISON",
        "SIDE_BY_SIDE_NO_FUSION",
        {
            "trustmark_run": run_b,
            "dinov2_run": run_c,
            "input_fingerprint": fingerprint
        }
    )

    meta_path = run_d / "metadata.json"

    try:
        # Simpan salinan kedua hasil sebagai bukti
        for src, name in (
            (csv_b, "trustmark_results.csv"),
            (csv_c, "dinov2_results.csv")
        ):

            dst = run_d / "inputs" / name

            shutil.copy2(src, dst)

            assert file_hash(src) == file_hash(dst)

        result_path = (
            run_d / "outputs/individual_comparison_all.csv"
        )

        combined.to_csv(
            result_path,
            index=False
        )

        summary = {
            "input_fingerprint": fingerprint,
            "trustmark_results_sha256": file_hash(csv_b),
            "dinov2_results_sha256": file_hash(csv_c),
            "result_sha256": file_hash(result_path),
            "total_images": len(combined),
            "fusion_used": False,
            "created_at_utc": datetime.now(
                timezone.utc
            ).isoformat()
        }

        (
            run_d / "outputs/comparison_all_summary.json"
        ).write_text(
            json.dumps(summary, indent=2),
            encoding="utf-8"
        )

        record_event(
            run_d,
            "INDIVIDUAL_COMPARISON_COMPLETED",
            summary
        )

        meta = json.loads(
            meta_path.read_text(encoding="utf-8")
        )

        meta["status"] = "COMPLETED"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

        status = "COMPARISON COMPLETED"

    except Exception as e:

        record_event(
            run_d,
            "RUN_FAILED",
            {"error": f"{type(e).__name__}: {e}"}
        )

        meta = json.loads(
            meta_path.read_text(encoding="utf-8")
        )

        meta["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

        raise


# 7. Tampilkan tabel dengan nama kolom sederhana
print("=== JATIKARYA NEW - STEP 06D ===")
print("PERBANDINGAN SELURUH GAMBAR\n")

table = combined[[
    "nama_file",
    "jenis_edit",
    "hasil_trustmark",
    "kemiripan_dinov2",
    "waktu_trustmark",
    "waktu_dinov2"
]].copy()

table.columns = [
    "Nama gambar",
    "Jenis edit",
    "Hasil TrustMark",
    "Kemiripan DINOv2",
    "TrustMark (detik)",
    "DINOv2 (detik)"
]

for col in (
    "Kemiripan DINOv2",
    "TrustMark (detik)",
    "DINOv2 (detik)"
):
    table[col] = pd.to_numeric(
        table[col]
    ).round(6)

display(table)

# 8. Ringkasan penelitian
print("\nJumlah gambar unik:", len(combined))

print(
    "Kode cocok:",
    int(
        (combined["hasil_trustmark"] == "KODE_COCOK").sum()
    )
)

print(
    "Tidak ditemukan:",
    int(
        (
            combined["hasil_trustmark"]
            == "TIDAK_TERDETEKSI"
        ).sum()
    )
)

print(
    "Kode tidak cocok:",
    int(
        (
            combined["hasil_trustmark"]
            == "KODE_TIDAK_COCOK"
        ).sum()
    )
)

print(
    "Kemiripan DINOv2 terendah:",
    round(
        float(combined["kemiripan_dinov2"].min()),
        6
    )
)

print("File tersimpan:", result_path)
print("STATUS:", status, "| BELUM FUSION")
```

```python
# ==================================================
# JATIKARYA NEW - STEP 07A
# UPLOAD -> PERIKSA -> PRATINJAU -> SIMPAN
# ==================================================

from google.colab import files
from PIL import Image, ImageOps
from IPython.display import display
from pathlib import Path
from datetime import datetime, timezone
import pandas as pd
import hashlib
import io
import json
import csv
import shutil
import os

# 1. Periksa lokasi proyek
assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

assert all(
    callable(globals().get(f))
    for f in (
        "file_hash",
        "start_experiment",
        "record_event"
    )
)

registry_path = ROOT / "04_LOGS/source_registry.csv"

columns = [
    "source_id", "original_filename",
    "image_format", "width", "height",
    "file_size_bytes", "sha256",
    "stored_path", "registered_at_utc"
]

registry = pd.read_csv(
    registry_path,
    dtype=str,
    keep_default_na=False
)

assert list(registry.columns) == columns
assert registry["source_id"].is_unique

# 2. Upload gambar baru
print("=== STEP 07A: UPLOAD FOTO ASLI ===")

uploaded = files.upload()

assert uploaded, "Belum ada file dipilih."

# 3. Periksa semua gambar terlebih dahulu
checked = []

seen = set(registry["sha256"])
existing = set(registry["sha256"])
ids = set(registry["source_id"])

for filename, content in uploaded.items():

    assert filename == Path(filename).name
    assert content, f"File kosong: {filename}"

    extension = Path(filename).suffix.lower()

    assert extension in (
        ".jpg", ".jpeg", ".png", ".webp"
    ), f"Ekstensi tidak didukung: {filename}"

    with Image.open(io.BytesIO(content)) as img:

        fmt = img.format

        assert fmt in (
            "JPEG", "MPO", "PNG", "WEBP"
        ), f"Format belum didukung: {fmt}"

        frames = getattr(img, "n_frames", 1)

        # Gambar utama pada file MPO
        img.seek(0)

        width, height = img.size

        preview = ImageOps.exif_transpose(
            img
        ).convert("RGB")

        preview.thumbnail((240, 240))

    sha = hashlib.sha256(content).hexdigest()
    sid = "SRC_" + sha[:16]

    assert sid not in ids or sha in existing, (
        "Konflik Source ID."
    )

    status = (
        "SUDAH TERDAFTAR"
        if sha in seen
        else "BARU"
    )

    seen.add(sha)

    checked.append({
        "filename": filename,
        "raw": content,
        "image_format": fmt,
        "n_frames": frames,
        "width": width,
        "height": height,
        "sha256": sha,
        "source_id": sid,
        "suffix": extension,
        "status": status,
        "preview": preview
    })

# 4. Tampilkan hasil sebelum menyimpan
print("\n=== HASIL PEMERIKSAAN GAMBAR ===")

display(pd.DataFrame([
    {
        "Nama file": r["filename"],
        "Format": r["image_format"],
        "Jumlah frame": r["n_frames"],
        "Lebar": r["width"],
        "Tinggi": r["height"],
        "Source ID": r["source_id"],
        "Status": r["status"]
    }
    for r in checked
]))

print("\n=== PRATINJAU GAMBAR ===")

for r in checked:
    print(r["filename"], "|", r["status"])
    display(r["preview"])

new = [
    r for r in checked
    if r["status"] == "BARU"
]

print("\nGambar baru:", len(new))
print("Gambar identik dilewati:",
      len(checked) - len(new))

# 5. Simpan hanya setelah dikonfirmasi
if not new:

    print(
        "STATUS: SEMUA SUDAH TERDAFTAR"
    )
    print("Tidak ada file yang diubah.")

else:

    confirm = input(
        "\nKetik SIMPAN jika gambar sudah benar: "
    ).strip()

    if confirm != "SIMPAN":

        print(
            "STATUS: BELUM DISIMPAN"
        )

    else:

        for r in new:
            folder = (
                ROOT / "01_DATASET/ORIGINAL"
                / r["source_id"]
            )

            assert not folder.exists(), (
                f"Folder sumber sudah ada: {folder}"
            )

        run = start_experiment(
            "STEP_07A_MULTI_SOURCE",
            "SOURCE_REGISTRATION",
            {
                "count": len(new),
                "format_policy": "RAW_PRESERVED"
            }
        )

        meta_path = run / "metadata.json"
        rows = []

        try:
            # Cadangkan daftar sumber sebelumnya
            shutil.copy2(
                registry_path,
                run / "inputs/source_registry_before.csv"
            )

            for r in new:

                folder = (
                    ROOT / "01_DATASET/ORIGINAL"
                    / r["source_id"]
                )

                folder.mkdir(
                    parents=True,
                    exist_ok=False
                )

                dest = (
                    folder /
                    ("original" + r["suffix"])
                )

                # Simpan file asli tanpa mengubahnya
                with dest.open("xb") as f:
                    f.write(r["raw"])

                assert file_hash(dest) == r["sha256"]

                record = {
                    "source_id": r["source_id"],
                    "original_filename": r["filename"],
                    "image_format": r["image_format"],
                    "width": r["width"],
                    "height": r["height"],
                    "file_size_bytes": len(r["raw"]),
                    "sha256": r["sha256"],
                    "stored_path": str(
                        dest.relative_to(ROOT)
                    ),
                    "registered_at_utc":
                        datetime.now(
                            timezone.utc
                        ).isoformat()
                }

                # Simpan informasi gambar
                metadata = {
                    **record,
                    "n_frames": r["n_frames"],
                    "processing_frame_index": 0
                }

                (
                    folder / "source_metadata.json"
                ).write_text(
                    json.dumps(
                        metadata,
                        ensure_ascii=False,
                        indent=2
                    ),
                    encoding="utf-8"
                )

                rows.append(record)

            # 6. Perbarui daftar sumber
            tmp_path = (
                run / "outputs/source_registry_next.csv"
            )

            with tmp_path.open(
                "x",
                encoding="utf-8",
                newline=""
            ) as f:

                writer = csv.DictWriter(
                    f,
                    fieldnames=columns
                )

                writer.writeheader()

                writer.writerows(
                    registry.to_dict("records") + rows
                )

            os.replace(
                tmp_path,
                registry_path
            )

            # 7. Simpan catatan pendaftaran
            result = {
                "registered": rows,
                "MPO_frame_used": 0
            }

            (
                run / "outputs/registration.json"
            ).write_text(
                json.dumps(
                    result,
                    ensure_ascii=False,
                    indent=2
                ),
                encoding="utf-8"
            )

            record_event(
                run,
                "SOURCE_REGISTRATION_COMPLETED",
                {"new_count": len(rows)}
            )

            meta = json.loads(
                meta_path.read_text(encoding="utf-8")
            )

            meta["status"] = "PASS"

            meta_path.write_text(
                json.dumps(meta, indent=2),
                encoding="utf-8"
            )

            print("\n=== HASIL STEP 07A ===")
            print("Foto baru disimpan:", len(rows))
            print(
                "Total foto sumber:",
                len(pd.read_csv(registry_path))
            )
            print(
                "STATUS: SOURCE REGISTRATION COMPLETED"
            )

        except Exception as exc:

            record_event(
                run,
                "RUN_FAILED",
                {"error": str(exc)}
            )

            meta = json.loads(
                meta_path.read_text(encoding="utf-8")
            )

            meta["status"] = "ERROR"

            meta_path.write_text(
                json.dumps(meta, indent=2),
                encoding="utf-8"
            )

            raise
```

```python
# ==================================================
# JATIKARYA NEW - STEP 07B
# TRUSTMARK UNTUK SEMUA SUMBER BARU
# Skip hasil lama, verifikasi sebelum registrasi
# ==================================================

from pathlib import Path
from PIL import Image, ImageOps
from trustmark import TrustMark
from importlib.metadata import version
from datetime import datetime, timezone
from IPython.display import display
import pandas as pd
import torch
import json
import shutil

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

VERSION = version("trustmark")
MAX_SIDE = 768

registry_path = ROOT / "04_LOGS/source_registry.csv"

registry = pd.read_csv(
    registry_path,
    dtype=str,
    keep_default_na=False
)

assert registry.source_id.is_unique
assert not registry.empty

folder_wm = (
    ROOT / "02_EXPERIMENTS/STEP_04_TRUSTMARK_ENCODE"
)

dataset_wm = (
    ROOT / "01_DATASET/WATERMARKED"
)

# 1. Pembuat payload yang sama dengan STEP 04B
def make_payload(sha):
    return bin(int(sha, 16))[2:].zfill(256)[:61]

# 2. Cari watermark lama yang valid
def find_existing(source_id, source_sha):

    expected = make_payload(source_sha)

    reports = sorted(
        folder_wm.glob(
            "*/outputs/verification_result.json"
        ),
        reverse=True
    )

    for report in reports:

        try:
            data = json.loads(
                report.read_text(encoding="utf-8")
            )

            if not (
                data.get("source_id") == source_id
                and data.get("source_sha256") == source_sha
                and data.get("payload_bits") == expected
                and data.get("payload_match") is True
                and data.get("watermark_detected") is True
            ):
                continue

            meta = json.loads(
                (
                    report.parent.parent / "metadata.json"
                ).read_text(encoding="utf-8")
            )

            if meta.get("status") not in (
                "PASS", "COMPLETED"
            ):
                continue

            path = (
                ROOT / data["stored_path"]
            ).resolve()

            assert path.is_relative_to(
                dataset_wm.resolve()
            )

            if (
                path.is_file()
                and file_hash(path)
                    == data["watermarked_sha256"]
            ):
                return data

        except (
            KeyError, OSError, ValueError,
            TypeError, AssertionError
        ):
            continue

    return None

# 3. Periksa seluruh sumber sebelum memproses
sources = []
pending = []
results = []

for _, row in registry.iterrows():

    sid = row["source_id"]
    sha = row["sha256"]
    path = (
        ROOT / row["stored_path"]
    ).resolve()

    assert path.is_relative_to(
        (ROOT / "01_DATASET/ORIGINAL").resolve()
    )

    assert path.is_file(), (
        f"File sumber tidak ditemukan: {sid}"
    )

    assert file_hash(path) == sha, (
        f"SHA-256 sumber tidak cocok: {sid}"
    )

    # Pastikan file benar-benar dapat dibaca
    with Image.open(path) as img:
        img.seek(0)
        img.load()
        fmt = img.format
        frames = getattr(img, "n_frames", 1)

    source = {
        "source_id": sid,
        "filename": row["original_filename"],
        "sha256": sha,
        "path": path,
        "format": fmt,
        "frames": frames
    }

    sources.append(source)

    existing = find_existing(sid, sha)

    if existing:

        results.append({
            "nama_file": source["filename"],
            "source_id": sid,
            "format": fmt,
            "jumlah_frame": frames,
            "status": "ARSIP_VALID",
            "kode_cocok": True,
            "run_id": "ARSIP"
        })

    else:
        pending.append(source)

print("=== STEP 07B: PEMERIKSAAN SUMBER ===")
print("Total sumber:", len(sources))
print("Watermark valid dari arsip:",
      len(sources) - len(pending))
print("Sumber yang perlu diproses:", len(pending))

# 4. Siapkan TrustMark hanya jika ada sumber baru
if pending:

    device = (
        "cuda" if torch.cuda.is_available()
        else "cpu"
    )

    tm07b = TrustMark(
        verbose=False,
        model_type="Q",
        encoding_type=TrustMark.Encoding.BCH_5,
        device=device
    )

    print("Perangkat:", device)
    print("Versi TrustMark:", VERSION)

# 5. Tanam watermark pada sumber baru
for number, source in enumerate(pending, 1):

    sid = source["source_id"]
    sha = source["sha256"]
    payload = make_payload(sha)

    print(
        f"\n[{number}/{len(pending)}] "
        f"{source['filename']}"
    )

    run = start_experiment(
        "STEP_04_TRUSTMARK_ENCODE",
        "TrustMark-Q-BCH5",
        {
            "source_id": sid,
            "source_sha256": sha,
            "max_side": MAX_SIDE,
            "source_frame_index": 0,
            "trustmark_version": VERSION
        }
    )

    meta_path = run / "metadata.json"

    try:
        # Baca gambar utama, termasuk MPO
        with Image.open(source["path"]) as img:

            img.seek(0)

            work = ImageOps.exif_transpose(
                img
            ).convert("RGB")

        # Batasi ukuran tanpa mengubah gambar asli
        work.thumbnail(
            (MAX_SIDE, MAX_SIDE),
            Image.Resampling.LANCZOS
        )

        prepared_path = (
            run / "inputs/source_prepared.png"
        )

        work.save(prepared_path, format="PNG")

        prepared_sha = file_hash(prepared_path)

        # Tanam watermark
        watermarked = tm07b.encode(
            work,
            payload,
            MODE="binary"
        )

        # Simpan kandidat di folder percobaan dulu
        candidate = (
            run / "outputs/watermarked_candidate.png"
        )

        watermarked.save(candidate, format="PNG")

        candidate_sha = file_hash(candidate)

        # Verifikasi dari file hasil, bukan memori
        with Image.open(candidate) as img:

            check_image = img.convert("RGB")

            decoded, detected, schema = (
                tm07b.decode(
                    check_image,
                    MODE="binary"
                )
            )

        detected = bool(detected)

        match = (
            detected
            and decoded == payload
            and schema == TrustMark.Encoding.BCH_5
        )

        stored_path = None

        # Hanya simpan ke dataset jika cocok
        if match:

            destination_dir = dataset_wm / sid

            destination_dir.mkdir(
                parents=True,
                exist_ok=True
            )

            destination = (
                destination_dir /
                f"{run.name}_watermarked.png"
            )

            with candidate.open("rb") as src:
                with destination.open("xb") as dst:
                    shutil.copyfileobj(src, dst)

            assert file_hash(destination) == candidate_sha

            stored_path = str(
                destination.relative_to(ROOT)
            )

        # Simpan bukti pengujian
        report = {
            "source_id": sid,
            "source_sha256": sha,
            "source_original_filename":
                source["filename"],
            "source_original_format":
                source["format"],
            "source_n_frames":
                source["frames"],
            "source_frame_index": 0,
            "prepared_sha256": prepared_sha,
            "prepared_width": work.width,
            "prepared_height": work.height,
            "payload_bits": payload,
            "model_variant": "Q",
            "encoding_type": "BCH_5",
            "trustmark_version": VERSION,
            "watermark_detected": detected,
            "payload_match": bool(match),
            "decoded_payload": (
                str(decoded)
                if decoded is not None else None
            ),
            "decoded_schema": str(schema),
            "watermarked_sha256": candidate_sha,
            "stored_path": stored_path,
            "created_at_utc": datetime.now(
                timezone.utc
            ).isoformat()
        }

        report_path = (
            run / "outputs/verification_result.json"
        )

        report_path.write_text(
            json.dumps(
                report,
                indent=2,
                ensure_ascii=False
            ),
            encoding="utf-8"
        )

        record_event(
            run,
            "TRUSTMARK_ENCODING_CHECKED",
            {
                "source_id": sid,
                "watermark_detected": detected,
                "payload_match": bool(match)
            }
        )

        meta = json.loads(
            meta_path.read_text(encoding="utf-8")
        )

        meta["status"] = (
            "PASS"
            if match
            else "FAILED_VERIFICATION"
        )

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

        status = (
            "BARU_PASS"
            if match
            else "BARU_GAGAL_VERIFIKASI"
        )

        results.append({
            "nama_file": source["filename"],
            "source_id": sid,
            "format": source["format"],
            "jumlah_frame": source["frames"],
            "status": status,
            "kode_cocok": bool(match),
            "run_id": run.name
        })

        print("Watermark ditemukan:", detected)
        print("Kode cocok:", bool(match))
        print("Status:", status)

    except Exception as exc:

        record_event(
            run,
            "RUN_FAILED",
            {
                "error_type": type(exc).__name__,
                "error": str(exc)
            }
        )

        meta = json.loads(
            meta_path.read_text(encoding="utf-8")
        )

        meta["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

        results.append({
            "nama_file": source["filename"],
            "source_id": sid,
            "format": source["format"],
            "jumlah_frame": source["frames"],
            "status": "ERROR",
            "kode_cocok": False,
            "run_id": run.name
        })

        print(
            "ERROR:",
            type(exc).__name__,
            str(exc)
        )

# 6. Rekap keseluruhan
summary_df = pd.DataFrame(results)

summary_df = summary_df.sort_values(
    "nama_file"
).reset_index(drop=True)

print("\n=== HASIL STEP 07B ===")

display(summary_df[[
    "nama_file",
    "source_id",
    "format",
    "jumlah_frame",
    "status",
    "kode_cocok"
]])

print("Total sumber:", len(summary_df))
print("Kode berhasil diverifikasi:",
      int(summary_df["kode_cocok"].sum()))
print("Dari arsip:",
      int((summary_df.status == "ARSIP_VALID").sum()))
print("Berhasil baru:",
      int((summary_df.status == "BARU_PASS").sum()))
print("Gagal atau error:",
      int((~summary_df.kode_cocok).sum()))

print(
    "STATUS:",
    "STEP 07B PASS"
    if summary_df.kode_cocok.all()
    else "STEP 07B PERLU DITINJAU"
)
```

```python
# ==================================================
# JATIKARYA NEW - STEP 07C
# REFERENSI DINOV2 UNTUK SELURUH SUMBER
# Menggunakan embedding lama yang valid
# Tidak menggunakan TrustMark atau Fusion
# ==================================================

from pathlib import Path
from PIL import Image, ImageOps
from torchvision import transforms
from datetime import datetime, timezone
import torch
import torch.nn.functional as F
import numpy as np
import pandas as pd
import json
import time

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

assert all(
    callable(globals().get(x))
    for x in (
        "file_hash",
        "start_experiment",
        "record_event"
    )
)

assert "model05a" in globals(), (
    "Jalankan STEP 05A terlebih dahulu."
)

assert getattr(model05a, "embed_dim", None) == 384, (
    "Model harus DINOv2 ViT-S/14."
)

STEP = "STEP_05B_DINOV2_ORIGINAL"
MODEL = "dinov2_vits14"
PREPROC = "Resize518_CenterCrop518_ImageNet"

REF_DIR = ROOT / "02_EXPERIMENTS" / STEP
ORIG_DIR = (ROOT / "01_DATASET/ORIGINAL").resolve()

registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str,
    keep_default_na=False
)

assert not registry.empty
assert registry["source_id"].is_unique
assert registry["sha256"].is_unique

# 1. Cari embedding lama yang valid
def valid_reference(source_id, source_sha):

    paths = sorted(
        REF_DIR.glob(
            "*/outputs/reference_result.json"
        ),
        reverse=True
    )

    for path in paths:

        try:
            info = json.loads(
                path.read_text(encoding="utf-8")
            )

            embedding_file = (
                path.parent / "embedding_original.npy"
            )

            if not (
                info.get("source_id") == source_id
                and info.get("source_sha256") == source_sha
                and info.get("model") == MODEL
                and info.get("preprocessing") == PREPROC
                and info.get("embedding_dimension") == 384
                and info.get("status") == "PASS"
                and embedding_file.is_file()
                and file_hash(embedding_file)
                    == info.get("embedding_sha256")
            ):
                continue

            vector = np.load(
                embedding_file,
                allow_pickle=False
            )

            if (
                vector.shape == (384,)
                and np.issubdtype(
                    vector.dtype, np.floating
                )
                and np.isfinite(vector).all()
                and np.isclose(
                    np.linalg.norm(vector),
                    1.0,
                    atol=1e-3
                )
            ):
                return path, info

        except (
            OSError, ValueError,
            KeyError, TypeError
        ):
            continue

    return None

# 2. Periksa sembilan sumber
sources = []
pending = []
results = []

for _, row in registry.iterrows():

    sid = row["source_id"]
    sha = row["sha256"]

    path = (
        ROOT / row["stored_path"]
    ).resolve()

    assert path.is_relative_to(ORIG_DIR)
    assert path.is_file()
    assert file_hash(path) == sha

    # Untuk MPO, baca gambar utama (frame 0)
    with Image.open(path) as img:
        img.seek(0)
        img.load()

    source = {
        "source_id": sid,
        "filename": row["original_filename"],
        "sha256": sha,
        "path": path
    }

    sources.append(source)

    old = valid_reference(sid, sha)

    if old:

        results.append({
            "source_id": sid,
            "nama_file": source["filename"],
            "status": "ARSIP_VALID",
            "dimensi": 384
        })

    else:
        pending.append(source)

print("=== STEP 07C: PEMERIKSAAN SUMBER ===")
print("Total sumber:", len(sources))
print("Embedding valid dari arsip:", len(results))
print("Perlu diproses:", len(pending))

# 3. Siapkan DINOv2 jika ada sumber baru
if pending:

    device = (
        "cuda"
        if torch.cuda.is_available()
        else "cpu"
    )

    model05a = model05a.to(device).eval()

    preprocessing = transforms.Compose([
        transforms.Resize(518),
        transforms.CenterCrop(518),
        transforms.ToTensor(),
        transforms.Normalize(
            [0.485, 0.456, 0.406],
            [0.229, 0.224, 0.225]
        )
    ])

    def get_embedding(path):

        with Image.open(path) as img:

            img.seek(0)

            image = ImageOps.exif_transpose(
                img
            ).convert("RGB")

            tensor = preprocessing(
                image
            ).unsqueeze(0).to(device)

        start = time.perf_counter()

        with torch.inference_mode():

            output = model05a(tensor)

            output = F.normalize(
                output,
                p=2,
                dim=1
            )

        duration = (
            time.perf_counter() - start
        )

        vector = (
            output[0]
            .detach()
            .cpu()
            .numpy()
            .astype(np.float32)
        )

        assert vector.shape == (384,)
        assert np.isfinite(vector).all()

        assert np.isclose(
            np.linalg.norm(vector),
            1.0,
            atol=1e-3
        )

        return vector, duration

    # 4. Pastikan model masih konsisten
    # dengan referensi STEP 05B sebelumnya
    example = next(
        (
            s for s in sources
            if valid_reference(
                s["source_id"], s["sha256"]
            )
        ),
        None
    )

    if example:

        previous = valid_reference(
            example["source_id"],
            example["sha256"]
        )

        old_vector = np.load(
            previous[0].parent
            / "embedding_original.npy",
            allow_pickle=False
        )

        new_vector, _ = get_embedding(
            example["path"]
        )

        assert np.allclose(
            new_vector,
            old_vector,
            atol=1e-3,
            rtol=1e-3
        ), (
            "Model atau preprocessing berbeda "
            "dengan STEP 05B. Proses dihentikan."
        )

# 5. Proses sumber yang belum ada embedding
for number, source in enumerate(pending, 1):

    sid = source["source_id"]

    run = start_experiment(
        STEP,
        "DINOV2_ORIGINAL_REFERENCE",
        {
            "source_id": sid,
            "source_sha256": source["sha256"],
            "model": MODEL,
            "preprocessing": PREPROC,
            "source_frame_index": 0
        }
    )

    meta_path = run / "metadata.json"

    try:

        vector, duration = get_embedding(
            source["path"]
        )

        embedding_file = (
            run / "outputs/embedding_original.npy"
        )

        np.save(
            embedding_file,
            vector
        )

        assert np.array_equal(
            np.load(
                embedding_file,
                allow_pickle=False
            ),
            vector
        )

        report = {
            "source_id": sid,
            "source_sha256": source["sha256"],
            "original_filename": source["filename"],
            "model": MODEL,
            "preprocessing": PREPROC,
            "embedding_dimension": 384,
            "embedding_sha256": file_hash(
                embedding_file
            ),
            "source_frame_index": 0,
            "execution_seconds": round(
                duration, 6
            ),
            "status": "PASS",
            "created_at_utc": datetime.now(
                timezone.utc
            ).isoformat()
        }

        (
            run / "outputs/reference_result.json"
        ).write_text(
            json.dumps(
                report,
                indent=2,
                ensure_ascii=False
            ),
            encoding="utf-8"
        )

        record_event(
            run,
            "DINOV2_SOURCE_REFERENCE_PASS",
            {
                "source_id": sid,
                "embedding_sha256":
                    report["embedding_sha256"]
            }
        )

        meta = json.loads(
            meta_path.read_text(encoding="utf-8")
        )

        meta["status"] = "PASS"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

        results.append({
            "source_id": sid,
            "nama_file": source["filename"],
            "status": "BARU_PASS",
            "dimensi": 384
        })

        print(
            f"[{number}/{len(pending)}] "
            f"{source['filename']}: BARU_PASS"
        )

    except Exception as exc:

        record_event(
            run,
            "DINOV2_SOURCE_REFERENCE_ERROR",
            {
                "source_id": sid,
                "error": (
                    f"{type(exc).__name__}: {exc}"
                )
            }
        )

        meta = json.loads(
            meta_path.read_text(encoding="utf-8")
        )

        meta["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

        results.append({
            "source_id": sid,
            "nama_file": source["filename"],
            "status": "ERROR",
            "dimensi": None
        })

        print(
            f"[{number}/{len(pending)}] "
            f"{source['filename']}: ERROR ({exc})"
        )

# 6. Tampilkan hasil seluruh sumber
summary = pd.DataFrame(results)

summary = summary.sort_values(
    "nama_file"
).reset_index(drop=True)

print("\n=== HASIL STEP 07C ===")

display(
    summary[[
        "nama_file",
        "source_id",
        "dimensi",
        "status"
    ]]
)

valid_count = int(
    summary["status"].isin([
        "ARSIP_VALID",
        "BARU_PASS"
    ]).sum()
)

print("Total sumber:", len(summary))
print("Embedding valid:", valid_count)
print(
    "Embedding baru:",
    int(
        (summary["status"] == "BARU_PASS").sum()
    )
)
print(
    "Error:",
    int(
        (summary["status"] == "ERROR").sum()
    )
)

print(
    "STATUS:",
    "STEP 07C PASS"
    if valid_count == len(summary)
    else "STEP 07C PERLU DITINJAU"
)
```

```python
# ==========================================================
# JATIKARYA NEW - STEP 08A
# MEMBUAT GAMBAR EDIT + DOWNLOAD OTOMATIS
#
# 9 sumber x 8 transformasi = 72 gambar
# Tidak mengubah gambar asli atau watermark induk
# Tidak mengulangi batch yang sudah valid
# ==========================================================

from pathlib import Path
from datetime import datetime, timezone
from collections import defaultdict
from io import BytesIO
from PIL import (
    Image, ImageEnhance, ImageFilter,
    ImageOps, ImageDraw, ImageFont
)
from google.colab import files
import PIL
import pandas as pd
import hashlib
import json
import math
import zipfile
import os

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

assert all(
    callable(globals().get(x))
    for x in (
        "file_hash",
        "start_experiment",
        "record_event"
    )
)

# 1. Daftar transformasi
OPS08 = [
    ("brightness_075", "brightness", 0.75),
    ("contrast_130", "contrast", 1.30),
    ("blur_radius_2", "blur", 2.0),
    ("crop_center_080", "crop", 0.80),
    ("resize_050", "resize", 0.50),
    ("flip_horizontal", "flip", None),
    ("rotate_90_ccw", "rotate90", 90),
    ("rotate_180", "rotate180", 180)
]


def ubah08(img, jenis, nilai):

    if jenis == "brightness":
        return ImageEnhance.Brightness(
            img
        ).enhance(nilai)

    if jenis == "contrast":
        return ImageEnhance.Contrast(
            img
        ).enhance(nilai)

    if jenis == "blur":
        return img.filter(
            ImageFilter.GaussianBlur(
                radius=nilai
            )
        )

    if jenis == "crop":
        w = max(1, round(img.width * nilai))
        h = max(1, round(img.height * nilai))

        x = (img.width - w) // 2
        y = (img.height - h) // 2

        return img.crop((x, y, x + w, y + h))

    if jenis == "resize":
        return img.resize(
            (
                max(1, round(img.width * nilai)),
                max(1, round(img.height * nilai))
            ),
            Image.Resampling.LANCZOS
        )

    if jenis == "flip":
        return img.transpose(
            Image.Transpose.FLIP_LEFT_RIGHT
        )

    if jenis == "rotate90":
        return img.transpose(
            Image.Transpose.ROTATE_90
        )

    if jenis == "rotate180":
        return img.transpose(
            Image.Transpose.ROTATE_180
        )

    raise ValueError(
        "Transformasi tidak diketahui: " + jenis
    )


def baca08(path):
    return json.loads(
        path.read_text(encoding="utf-8")
    )


# 2. Baca daftar sumber asli
reg08 = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str,
    keep_default_na=False
)

assert len(reg08) > 0
assert reg08.source_id.is_unique
assert reg08.sha256.is_unique

wm_dir08 = (
    ROOT / "01_DATASET/WATERMARKED"
).resolve()

edit_dir08 = (
    ROOT / "01_DATASET/AUTO_EDITS"
).resolve()

reports08 = sorted(
    (
        ROOT / "02_EXPERIMENTS/"
        "STEP_04_TRUSTMARK_ENCODE"
    ).glob("*/outputs/verification_result.json"),
    reverse=True
)

sources08 = []

# 3. Cari watermark yang sudah valid
for _, src in reg08.iterrows():

    sid = src["source_id"]
    sha = src["sha256"]

    ori = (
        ROOT / src["stored_path"]
    ).resolve()

    assert ori.is_relative_to(
        (ROOT / "01_DATASET/ORIGINAL").resolve()
    )

    assert ori.is_file()
    assert file_hash(ori) == sha, (
        f"Sumber tidak valid: {sid}"
    )

    payload = bin(
        int(sha, 16)
    )[2:].zfill(256)[:61]

    found = None

    for rp in reports08:
        try:
            d = baca08(rp)

            if (
                d.get("source_id") != sid
                or d.get("source_sha256") != sha
                or d.get("payload_bits") != payload
                or d.get("payload_match") is not True
                or d.get("watermark_detected") is not True
            ):
                continue

            metadata = baca08(
                rp.parent.parent / "metadata.json"
            )

            if metadata.get("status") not in (
                "PASS", "COMPLETED"
            ):
                continue

            p = (
                ROOT / d["stored_path"]
            ).resolve()

            if (
                p.is_relative_to(wm_dir08)
                and p.is_file()
                and file_hash(p)
                    == d["watermarked_sha256"]
            ):
                found = (
                    p,
                    d["watermarked_sha256"]
                )
                break

        except (
            OSError, KeyError, ValueError,
            TypeError, AssertionError
        ):
            continue

    assert found, (
        f"Watermark belum valid: {sid}. "
        "Periksa STEP 07B."
    )

    sources08.append({
        "source_id": sid,
        "source_filename":
            src["original_filename"],
        "source_sha256": sha,
        "wm_path": found[0],
        "wm_sha256": found[1]
    })

sources08.sort(
    key=lambda x: x["source_id"]
)

print("=== JATIKARYA NEW - STEP 08A ===")
print("Jumlah sumber:", len(sources08))
print("Transformasi per sumber:", len(OPS08))
print(
    "Target gambar edit:",
    len(sources08) * len(OPS08)
)


# 4. Buat identitas konfigurasi eksperimen
config08 = {
    "workflow": "STEP08A_v1",
    "pillow": PIL.__version__,
    "operations": OPS08,
    "parents": [
        (
            s["source_id"],
            s["source_sha256"],
            s["wm_sha256"]
        )
        for s in sources08
    ]
}

fp08 = hashlib.sha256(
    json.dumps(
        config08,
        sort_keys=True
    ).encode()
).hexdigest()

exp08 = (
    ROOT / "02_EXPERIMENTS/"
    "STEP_08A_AUTOMATED_EDITS"
)

run08 = None
manifest08 = None


# 5. Periksa apakah batch sudah pernah dibuat
for mp in sorted(
    exp08.glob("*/outputs/auto_edit_manifest.json"),
    reverse=True
):

    try:
        m = baca08(mp)

        metadata = baca08(
            mp.parent.parent / "metadata.json"
        )

        if (
            metadata.get("status") != "PASS"
            or m.get("experiment_fingerprint") != fp08
            or len(m.get("files", []))
                != len(sources08) * len(OPS08)
        ):
            continue

        identities = {
            (
                x["source_id"],
                x["transformation_id"]
            )
            for x in m["files"]
        }

        assert len(identities) == len(m["files"])

        valid = all(
            (
                ROOT / x["stored_path"]
            ).resolve().is_relative_to(edit_dir08)
            and (
                ROOT / x["stored_path"]
            ).is_file()
            and file_hash(
                ROOT / x["stored_path"]
            ) == x["sha256"]
            for x in m["files"]
        )

        if valid:
            run08 = mp.parent.parent
            manifest08 = m
            break

    except (
        OSError, ValueError, KeyError,
        TypeError, AssertionError
    ):
        continue


# 6. Buat semua gambar edit jika belum tersedia
if manifest08 is None:

    run08 = start_experiment(
        "STEP_08A_AUTOMATED_EDITS",
        "PIL_DETERMINISTIC_TRANSFORMS",
        {
            "experiment_fingerprint": fp08,
            "source_count": len(sources08),
            "operations_per_source": len(OPS08)
        }
    )

    meta08 = run08 / "metadata.json"
    output08 = []

    try:

        for i, s in enumerate(sources08, 1):

            # Gunakan gambar BER-WATERMARK
            with Image.open(s["wm_path"]) as im:
                base = im.convert("RGB")
                base.load()

            target = (
                ROOT / "01_DATASET/AUTO_EDITS"
                / s["source_id"] / run08.name
            )

            target.mkdir(
                parents=True,
                exist_ok=False
            )

            for op_id, op, val in OPS08:

                edited = ubah08(
                    base, op, val
                )

                dest = target / f"{op_id}.png"

                with dest.open("xb") as handle:
                    edited.save(
                        handle,
                        format="PNG"
                    )

                output08.append({
                    "source_id":
                        s["source_id"],
                    "source_filename":
                        s["source_filename"],
                    "source_sha256":
                        s["source_sha256"],
                    "parent_type":
                        "WATERMARKED",
                    "parent_reference_sha256":
                        s["wm_sha256"],
                    "transformation_id":
                        op_id,
                    "transformation":
                        op,
                    "parameter_value":
                        val,
                    "width":
                        edited.width,
                    "height":
                        edited.height,
                    "sha256":
                        file_hash(dest),
                    "stored_path":
                        str(dest.relative_to(ROOT))
                })

            print(
                f"[{i}/{len(sources08)}] "
                f"{s['source_filename']}: "
                "8 transformasi dibuat"
            )

        assert len(output08) == (
            len(sources08) * len(OPS08)
        )

        # 7. Simpan catatan transformasi
        manifest08 = {
            "schema_version": 1,
            "experiment_fingerprint": fp08,
            "source_count": len(sources08),
            "operation_count": len(OPS08),
            "image_count": len(output08),
            "pillow_version": PIL.__version__,
            "operations": OPS08,
            "created_at_utc":
                datetime.now(
                    timezone.utc
                ).isoformat(),
            "files": output08
        }

        csv08 = (
            run08 / "outputs/auto_edit_manifest.csv"
        )

        pd.DataFrame(output08).to_csv(
            csv08,
            index=False
        )

        manifest08["csv_sha256"] = file_hash(
            csv08
        )

        jp08 = (
            run08 / "outputs/auto_edit_manifest.json"
        )

        jp08.write_text(
            json.dumps(
                manifest08,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

        record_event(
            run08,
            "AUTO_EDITS_CREATED",
            {
                "image_count": len(output08),
                "manifest_sha256": file_hash(jp08)
            }
        )

        metadata = baca08(meta08)
        metadata["status"] = "PASS"

        meta08.write_text(
            json.dumps(
                metadata,
                indent=2
            ),
            encoding="utf-8"
        )

    except Exception as e:

        record_event(
            run08,
            "RUN_FAILED",
            {
                "type": type(e).__name__,
                "message": str(e)
            }
        )

        metadata = baca08(meta08)
        metadata["status"] = "ERROR"

        meta08.write_text(
            json.dumps(metadata, indent=2),
            encoding="utf-8"
        )

        raise

else:

    print(
        "Batch sebelumnya valid. "
        "Gambar tidak dibuat ulang."
    )


# 8. Kelompokkan gambar untuk download
by08 = defaultdict(list)

for e in manifest08["files"]:

    p = (
        ROOT / e["stored_path"]
    ).resolve()

    assert p.is_relative_to(edit_dir08)
    assert p.is_file()
    assert file_hash(p) == e["sha256"]

    by08[e["source_id"]].append((e, p))

assert set(by08) == {
    s["source_id"] for s in sources08
}

# 9. Siapkan font lembar perbandingan
try:
    font08 = ImageFont.truetype(
        "DejaVuSans.ttf", 17
    )
except OSError:
    font08 = ImageFont.load_default()


# 10. Buat ZIP berisi semua hasil
zip08 = Path("/content") / (
    f"JATIKARYA_STEP08A_{run08.name}_REVIEW.zip"
)

part08 = zip08.with_suffix(".part")

try:

    with zipfile.ZipFile(
        part08,
        "w",
        compression=zipfile.ZIP_DEFLATED,
        compresslevel=3
    ) as z:

        z.writestr(
            "manifest_STEP08A.json",
            json.dumps(
                manifest08,
                ensure_ascii=False,
                indent=2
            )
        )

        z.writestr(
            "PETUNJUK.txt",
            "PERBANDINGAN/: 9 lembar ringkasan.\n"
            "SUMBER/: gambar edit PNG resolusi penuh.\n"
            "Buka PNG pada zoom 100% untuk "
            "menilai blur secara visual.\n"
            "Gambar induk adalah gambar watermarked.\n"
        )

        for i, s in enumerate(sources08, 1):

            sid = s["source_id"]

            # Masukkan gambar sebelum diedit
            z.write(
                s["wm_path"],
                (
                    f"SUMBER/{sid}/"
                    "00_sebelum_edit_watermarked.png"
                )
            )

            cards = [
                (
                    "SEBELUM EDIT (watermarked)",
                    s["wm_path"]
                )
            ]

            # Masukkan 8 hasil edit resolusi penuh
            for e, p in sorted(
                by08[sid],
                key=lambda v: v[0]["transformation_id"]
            ):

                z.write(
                    p,
                    (
                        f"SUMBER/{sid}/EDITS/"
                        f"{e['transformation_id']}.png"
                    )
                )

                val = e["parameter_value"]

                label = (
                    e["transformation_id"]
                    if val is None
                    else (
                        f"{e['transformation_id']} "
                        f"({val})"
                    )
                )

                cards.append((label, p))

            # 11. Buat satu lembar perbandingan
            W = 380
            H = 305
            HEAD = 76

            sheet = Image.new(
                "RGB",
                (
                    W * 3,
                    HEAD + H * math.ceil(
                        len(cards) / 3
                    )
                ),
                "white"
            )

            draw = ImageDraw.Draw(sheet)

            draw.text(
                (16, 12),
                sid,
                fill="black",
                font=font08
            )

            draw.text(
                (16, 40),
                s["source_filename"][:72],
                fill="black",
                font=font08
            )

            for k, (label, p) in enumerate(cards):

                x = (k % 3) * W
                y = HEAD + (k // 3) * H

                with Image.open(p) as im:

                    tile = ImageOps.contain(
                        im.convert("RGB"),
                        (350, 255),
                        Image.Resampling.LANCZOS
                    )

                sheet.paste(
                    tile,
                    (
                        x + (W - tile.width) // 2,
                        y + (260 - tile.height) // 2
                    )
                )

                draw.text(
                    (x + 12, y + 268),
                    label[:40],
                    fill="black",
                    font=font08
                )

            buffer = BytesIO()

            sheet.save(
                buffer,
                "JPEG",
                quality=92
            )

            z.writestr(
                (
                    f"PERBANDINGAN/"
                    f"{sid}_perbandingan.jpg"
                ),
                buffer.getvalue()
            )

            print(
                f"ZIP [{i}/{len(sources08)}] "
                f"{sid}: selesai"
            )

    os.replace(part08, zip08)

except Exception:
    part08.unlink(missing_ok=True)
    raise


# 12. Ringkasan dan download otomatis
print("\n=== STEP 08A SELESAI ===")
print("Jumlah sumber:", len(sources08))

print(
    "Transformasi per sumber:",
    len(OPS08)
)

print(
    "Gambar edit tersedia:",
    len(manifest08["files"])
)

print(
    "Status: STEP 08A PASS"
)

print(
    "Catatan: belum diuji TrustMark/DINOv2"
)

print(
    "ZIP:",
    zip08.name,
    "|",
    round(
        zip08.stat().st_size / 1024**2,
        2
    ),
    "MB"
)

files.download(str(zip08))
```

```python
# ==================================================
# JATIKARYA NEW - STEP 08B
# PENGUJIAN TRUSTMARK-ONLY
# Seluruh gambar dari STEP 08A
# Tidak menjalankan DINOv2 atau Fusion
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from importlib.metadata import version
from PIL import Image
from trustmark import TrustMark
from IPython.display import display
import pandas as pd
import torch
import json
import time

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

assert all(
    callable(globals().get(x))
    for x in (
        "file_hash",
        "start_experiment",
        "record_event"
    )
)

def read08b(path):
    return json.loads(
        path.read_text(encoding="utf-8")
    )


# 1. Ambil batch STEP 08A terbaru
folder_a = (
    ROOT / "02_EXPERIMENTS/"
    "STEP_08A_AUTOMATED_EDITS"
)

options = []

for mp in sorted(
    folder_a.glob(
        "*/outputs/auto_edit_manifest.json"
    ),
    reverse=True
):

    try:
        meta = read08b(
            mp.parent.parent / "metadata.json"
        )

        if meta.get("status") == "PASS":

            m = read08b(mp)

            if isinstance(m.get("files"), list) and m["files"]:
                options.append((mp, m))

    except (
        OSError, ValueError,
        TypeError, KeyError
    ):
        continue

assert options, (
    "STEP 08A belum memiliki manifest PASS."
)

manifest_path, manifest = options[0]

manifest_sha = file_hash(
    manifest_path
)


# 2. Periksa identitas semua gambar
registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str,
    keep_default_na=False
)

assert registry["source_id"].is_unique
assert registry["sha256"].is_unique

source_hashes = dict(zip(
    registry["source_id"],
    registry["sha256"]
))

edit_root = (
    ROOT / "01_DATASET/AUTO_EDITS"
).resolve()

watermark_root = (
    ROOT / "01_DATASET/WATERMARKED"
).resolve()

entries = sorted(
    manifest["files"],
    key=lambda x: (
        x["source_id"],
        x["transformation_id"]
    )
)

assert len(entries) == (
    int(manifest["source_count"])
    * int(manifest["operation_count"])
)

assert len(entries) == int(
    manifest.get("image_count", len(entries))
)

assert len({
    (
        e["source_id"],
        e["transformation_id"]
    )
    for e in entries
}) == len(entries)

assert len({
    e["source_id"]
    for e in entries
}) == int(manifest["source_count"])


# 3. Verifikasi seluruh gambar sebelum pengujian
parents_checked = set()

for e in entries:

    sid = e["source_id"]

    assert sid in source_hashes
    assert (
        e["source_sha256"]
        == source_hashes[sid]
    )
    assert e["parent_type"] == "WATERMARKED"

    path = (
        ROOT / e["stored_path"]
    ).resolve()

    assert path.is_relative_to(edit_root)
    assert path.is_file()

    assert file_hash(path) == e["sha256"], (
        f"Hash berubah: {path.name}"
    )

    parent = (
        sid,
        e["parent_reference_sha256"]
    )

    if parent not in parents_checked:

        parent_dir = (
            watermark_root / sid
        ).resolve()

        assert parent_dir.is_relative_to(
            watermark_root
        )

        assert any(
            file_hash(w) == parent[1]
            for w in parent_dir.glob("*.png")
        ), (
            f"Watermark induk tidak ditemukan: {sid}"
        )

        parents_checked.add(parent)


# 4. Gunakan hasil lama jika sudah lengkap
version_tm = version("trustmark")

folder_b = (
    ROOT / "02_EXPERIMENTS/"
    "STEP_08B_TRUSTMARK_ONLY"
)

expected_keys = {
    (
        e["source_id"],
        e["transformation_id"],
        e["sha256"]
    )
    for e in entries
}

existing = None

for sp in sorted(
    folder_b.glob("*/outputs/summary.json"),
    reverse=True
):

    try:
        summary = read08b(sp)

        csv_path = (
            sp.parent / "trustmark_results.csv"
        )

        meta = read08b(
            sp.parent.parent / "metadata.json"
        )

        if not (
            meta.get("status") == "COMPLETED"
            and summary.get("input_manifest_sha256")
                == manifest_sha
            and summary.get("trustmark_version")
                == version_tm
            and summary.get("error_count") == 0
            and csv_path.is_file()
            and file_hash(csv_path)
                == summary.get("results_sha256")
        ):
            continue

        old = pd.read_csv(
            csv_path,
            dtype=str,
            keep_default_na=False
        )

        old_keys = {
            (
                row["source_id"],
                row["transformation_id"],
                row["image_sha256"]
            )
            for _, row in old.iterrows()
        }

        if (
            len(old) == len(entries)
            and old_keys == expected_keys
            and old["hasil"].isin([
                "KODE_COCOK",
                "TIDAK_TERDETEKSI",
                "KODE_TIDAK_COCOK"
            ]).all()
        ):
            existing = (
                old,
                summary,
                csv_path
            )
            break

    except (
        OSError, ValueError,
        TypeError, KeyError
    ):
        continue


# 5. Mulai pengujian jika belum tersedia
if existing:

    result08, summary08, csv08 = existing

    print(
        "Hasil STEP 08B sudah tersedia. "
        "Tidak menjalankan pengujian ulang."
    )

else:

    print("=== JATIKARYA NEW - STEP 08B ===")

    print("Jumlah sumber:", len({
        e["source_id"] for e in entries
    }))

    print("Jumlah gambar edit:", len(entries))
    print("Metode: TRUSTMARK-ONLY")

    # Gunakan konfigurasi yang sama dengan STEP 07B
    detector08 = TrustMark(
        verbose=False,
        model_type="Q",
        encoding_type=TrustMark.Encoding.BCH_5,
        device=(
            "cuda"
            if torch.cuda.is_available()
            else "cpu"
        )
    )

    run08b = start_experiment(
        "STEP_08B_TRUSTMARK_ONLY",
        "TRUSTMARK_Q_BCH5",
        {
            "input_manifest_sha256": manifest_sha,
            "trustmark_version": version_tm,
            "total_images": len(entries)
        }
    )

    meta_path = run08b / "metadata.json"
    rows = []

    try:

        # 6. Periksa watermark pada setiap gambar
        for i, e in enumerate(entries, 1):

            sid = e["source_id"]
            source_sha = source_hashes[sid]

            payload = bin(
                int(source_sha, 16)
            )[2:].zfill(256)[:61]

            item = {
                "source_id": sid,
                "source_filename":
                    e["source_filename"],
                "transformation_id":
                    e["transformation_id"],
                "parameter_value":
                    e["parameter_value"],
                "source_sha256": source_sha,
                "parent_reference_sha256":
                    e["parent_reference_sha256"],
                "image_sha256": e["sha256"],
                "stored_path": e["stored_path"],
                "watermark_detected": False,
                "payload_match": False,
                "hasil": "ERROR",
                "decoded_payload": "",
                "decoded_schema": "",
                "waktu_detik": None,
                "error": ""
            }

            try:

                # Buka gambar hasil edit
                with Image.open(
                    ROOT / e["stored_path"]
                ) as img:

                    image = img.convert("RGB")
                    image.load()

                start_time = time.perf_counter()

                decoded, detected, schema = (
                    detector08.decode(
                        image,
                        MODE="binary"
                    )
                )

                duration = (
                    time.perf_counter()
                    - start_time
                )

                detected = bool(detected)

                matched = (
                    detected
                    and decoded == payload
                    and schema
                        == TrustMark.Encoding.BCH_5
                )

                if matched:
                    label = "KODE_COCOK"

                elif detected:
                    label = "KODE_TIDAK_COCOK"

                else:
                    label = "TIDAK_TERDETEKSI"

                item.update(
                    watermark_detected=detected,
                    payload_match=bool(matched),
                    hasil=label,
                    decoded_payload=(
                        str(decoded)
                        if decoded is not None
                        else ""
                    ),
                    decoded_schema=(
                        str(schema)
                        if schema is not None
                        else ""
                    ),
                    waktu_detik=round(
                        duration, 6
                    )
                )

            except Exception as exc:

                item["error"] = (
                    f"{type(exc).__name__}: {exc}"
                )

            rows.append(item)

            print(
                f"[{i:02d}/{len(entries)}] "
                f"{e['source_filename']} / "
                f"{e['transformation_id']}: "
                f"{item['hasil']}"
            )

        # 7. Simpan hasil pengujian
        result08 = pd.DataFrame(rows)

        csv08 = (
            run08b / "outputs/trustmark_results.csv"
        )

        result08.to_csv(
            csv08,
            index=False
        )

        summary08 = {
            "input_manifest_sha256":
                manifest_sha,
            "input_manifest_path":
                str(manifest_path.relative_to(ROOT)),
            "trustmark_version":
                version_tm,
            "total_images":
                len(result08),
            "matched_count":
                int(
                    (result08.hasil == "KODE_COCOK")
                    .sum()
                ),
            "not_detected_count":
                int(
                    (
                        result08.hasil
                        == "TIDAK_TERDETEKSI"
                    ).sum()
                ),
            "mismatch_count":
                int(
                    (
                        result08.hasil
                        == "KODE_TIDAK_COCOK"
                    ).sum()
                ),
            "error_count":
                int(
                    (result08.hasil == "ERROR")
                    .sum()
                ),
            "results_sha256":
                file_hash(csv08),
            "finished_at_utc":
                datetime.now(
                    timezone.utc
                ).isoformat()
        }

        (
            run08b / "outputs/summary.json"
        ).write_text(
            json.dumps(
                summary08,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

        record_event(
            run08b,
            "TRUSTMARK_ONLY_08B_COMPLETED",
            summary08
        )

        meta = read08b(meta_path)

        meta["status"] = (
            "COMPLETED"
            if summary08["error_count"] == 0
            else "COMPLETED_WITH_ERRORS"
        )

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

    except Exception as exc:

        record_event(
            run08b,
            "RUN_FAILED",
            {
                "error":
                    f"{type(exc).__name__}: {exc}"
            }
        )

        meta = read08b(meta_path)
        meta["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

        raise


# 8. Ringkasan per jenis transformasi
print("\n=== RINGKASAN STEP 08B ===")

summary_table = pd.crosstab(
    result08["transformation_id"],
    result08["hasil"]
)

summary_table = summary_table.reindex(
    columns=[
        "KODE_COCOK",
        "TIDAK_TERDETEKSI",
        "KODE_TIDAK_COCOK",
        "ERROR"
    ],
    fill_value=0
)

display(summary_table)

print("\nTotal gambar:", summary08["total_images"])
print("Kode cocok:", summary08["matched_count"])
print(
    "Tidak terdeteksi:",
    summary08["not_detected_count"]
)
print(
    "Kode tidak cocok:",
    summary08["mismatch_count"]
)
print("Error:", summary08["error_count"])
print("File hasil:", csv08)
print("STATUS: TRUSTMARK-ONLY | BELUM FUSION")
```

```python
# ========================================================
# JATIKARYA NEW - STEP 08C
# DINOv2-ONLY: PENCARIAN SUMBER DARI 9 KANDIDAT
# Semua gambar edit STEP 08A
# BELUM FUSION
# ========================================================

from pathlib import Path
from datetime import datetime, timezone
from PIL import Image, ImageOps
from torchvision import transforms
from IPython.display import display
import torch
import torch.nn.functional as F
import numpy as np
import pandas as pd
import json
import hashlib
import time

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

assert all(
    callable(globals().get(n))
    for n in (
        "file_hash",
        "start_experiment",
        "record_event"
    )
)

MODEL = "dinov2_vits14"
PREPROC = "Resize518_CenterCrop518_ImageNet"

REF_STEP = (
    ROOT / "02_EXPERIMENTS/STEP_05B_DINOV2_ORIGINAL"
)

EDIT_STEP = (
    ROOT / "02_EXPERIMENTS/STEP_08A_AUTOMATED_EDITS"
)

RESULT_STEP = (
    ROOT / "02_EXPERIMENTS/STEP_08C_DINOV2_RETRIEVAL"
)

ORIG_ROOT = (
    ROOT / "01_DATASET/ORIGINAL"
).resolve()

EDIT_ROOT = (
    ROOT / "01_DATASET/AUTO_EDITS"
).resolve()


def read_json08c(path):
    return json.loads(
        path.read_text(encoding="utf-8")
    )


# 1. Baca dan validasi seluruh sumber asli
registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str,
    keep_default_na=False
)

assert not registry.empty
assert registry.source_id.is_unique
assert registry.sha256.is_unique

sources = {}

for _, row in registry.iterrows():

    path = (
        ROOT / row["stored_path"]
    ).resolve()

    assert path.is_relative_to(ORIG_ROOT)
    assert path.is_file()
    assert file_hash(path) == row["sha256"]

    sources[row["source_id"]] = {
        "original_path": path,
        "original_sha256": row["sha256"],
        "original_filename": row["original_filename"]
    }

candidate_ids = sorted(sources)


# 2. Baca hasil STEP 08A
batch = None

for mp in sorted(
    EDIT_STEP.glob(
        "*/outputs/auto_edit_manifest.json"
    ),
    reverse=True
):

    try:
        manifest = read_json08c(mp)

        metadata = read_json08c(
            mp.parent.parent / "metadata.json"
        )

        if (
            metadata.get("status") == "PASS"
            and manifest.get("files")
        ):
            batch = (mp, manifest)
            break

    except (
        OSError, ValueError,
        KeyError, TypeError
    ):
        continue

assert batch, (
    "STEP 08A belum memiliki manifest PASS."
)

manifest_path, manifest = batch

entries = sorted(
    manifest["files"],
    key=lambda x: (
        x["source_id"],
        x["transformation_id"]
    )
)

assert len(entries) == (
    int(manifest["source_count"])
    * int(manifest["operation_count"])
)

assert len(entries) == int(
    manifest.get("image_count", len(entries))
)

assert {
    e["source_id"] for e in entries
} == set(candidate_ids), (
    "Daftar sumber berbeda dari STEP 08A."
)

expected_keys = set()

for e in entries:

    sid = e["source_id"]

    assert (
        e["source_sha256"]
        == sources[sid]["original_sha256"]
    )

    assert e["parent_type"] == "WATERMARKED"

    path = (
        ROOT / e["stored_path"]
    ).resolve()

    assert path.is_relative_to(EDIT_ROOT)
    assert path.is_file()
    assert file_hash(path) == e["sha256"]

    key = (
        sid,
        e["transformation_id"],
        e["sha256"]
    )

    assert key not in expected_keys

    expected_keys.add(key)


# 3. Ambil 9 embedding referensi dari foto ASLI
references = {}
reference_proofs = []

ref_reports = sorted(
    REF_STEP.glob(
        "*/outputs/reference_result.json"
    ),
    reverse=True
)

for sid in candidate_ids:

    found = None

    for rp in ref_reports:

        try:
            info = read_json08c(rp)

            f = (
                rp.parent / "embedding_original.npy"
            )

            if not (
                info.get("source_id") == sid
                and info.get("source_sha256")
                    == sources[sid]["original_sha256"]
                and info.get("model") == MODEL
                and info.get("preprocessing") == PREPROC
                and info.get("embedding_dimension") == 384
                and info.get("status") == "PASS"
                and f.is_file()
                and file_hash(f)
                    == info.get("embedding_sha256")
            ):
                continue

            vector = np.load(
                f,
                allow_pickle=False
            )

            if not (
                vector.shape == (384,)
                and np.issubdtype(
                    vector.dtype,
                    np.floating
                )
                and np.isfinite(vector).all()
                and np.isclose(
                    np.linalg.norm(vector),
                    1.0,
                    atol=1e-3
                )
            ):
                continue

            found = (
                vector.astype(np.float32),
                info["embedding_sha256"]
            )

            break

        except (
            OSError, ValueError,
            KeyError, TypeError
        ):
            continue

    assert found, (
        f"Embedding sumber {sid} tidak ditemukan. "
        "Periksa STEP 07C."
    )

    references[sid] = found[0]

    reference_proofs.append({
        "source_id": sid,
        "original_sha256":
            sources[sid]["original_sha256"],
        "embedding_sha256": found[1]
    })


# 4. Identitas dataset dan embedding
refs_hash = hashlib.sha256(
    json.dumps(
        reference_proofs,
        sort_keys=True,
        separators=(",", ":")
    ).encode()
).hexdigest()

manifest_hash = file_hash(manifest_path)


# 5. Cari hasil sebelumnya agar tidak diuji ulang
previous = None

for sp in sorted(
    RESULT_STEP.glob("*/outputs/summary.json"),
    reverse=True
):

    try:
        summary = read_json08c(sp)

        metadata = read_json08c(
            sp.parent.parent / "metadata.json"
        )

        result_csv = (
            sp.parent / "dinov2_retrieval_results.csv"
        )

        ranking_csv = (
            sp.parent / "all_candidate_rankings.csv"
        )

        if not (
            metadata.get("status") == "COMPLETED"
            and summary.get("input_manifest_sha256")
                == manifest_hash
            and summary.get("reference_fingerprint")
                == refs_hash
            and summary.get("model") == MODEL
            and summary.get("preprocessing") == PREPROC
            and summary.get("error_count") == 0
            and result_csv.is_file()
            and ranking_csv.is_file()
            and file_hash(result_csv)
                == summary.get("results_sha256")
            and file_hash(ranking_csv)
                == summary.get("rankings_sha256")
        ):
            continue

        old = pd.read_csv(
            result_csv,
            dtype={
                "source_id": str,
                "transformation_id": str,
                "image_sha256": str
            }
        )

        actual_keys = {
            (
                r.source_id,
                r.transformation_id,
                r.image_sha256
            )
            for r in old.itertuples()
        }

        if (
            len(old) == len(entries)
            and actual_keys == expected_keys
            and old["status"].eq("PASS").all()
            and int(
                summary.get("ranking_count", -1)
            ) == len(entries) * len(candidate_ids)
        ):
            previous = (
                old,
                summary,
                result_csv,
                ranking_csv
            )
            break

    except (
        OSError, ValueError,
        KeyError, TypeError
    ):
        continue


# 6. Gunakan hasil arsip atau jalankan DINOv2
if previous is not None:

    result_df, summary, result_csv, ranking_csv = (
        previous
    )

    print(
        "Hasil STEP 08C valid dari arsip. "
        "Tidak menghitung ulang."
    )

else:

    assert "model05a" in globals(), (
        "Model DINOv2 belum tersedia. "
        "Jalankan STEP 05A terlebih dahulu."
    )

    assert getattr(
        model05a, "embed_dim", None
    ) == 384

    device = (
        "cuda" if torch.cuda.is_available()
        else "cpu"
    )

    model05a = model05a.to(device).eval()

    preprocessing = transforms.Compose([
        transforms.Resize(518),
        transforms.CenterCrop(518),
        transforms.ToTensor(),
        transforms.Normalize(
            [0.485, 0.456, 0.406],
            [0.229, 0.224, 0.225]
        )
    ])

    def embedding08c(path):

        with Image.open(path) as img:

            img.seek(0)

            rgb = ImageOps.exif_transpose(
                img
            ).convert("RGB")

            tensor = preprocessing(
                rgb
            ).unsqueeze(0).to(device)

        with torch.inference_mode():

            vector = F.normalize(
                model05a(tensor),
                p=2,
                dim=1
            )

        output = (
            vector[0]
            .detach()
            .cpu()
            .numpy()
            .astype(np.float32)
        )

        assert output.shape == (384,)
        assert np.isfinite(output).all()

        return output


    # 7. Pastikan model sama dengan STEP 07C
    test_sid = candidate_ids[0]

    check_vector = embedding08c(
        sources[test_sid]["original_path"]
    )

    assert np.allclose(
        check_vector,
        references[test_sid],
        atol=2e-3,
        rtol=2e-3
    ), (
        "Model atau preprocessing tidak sesuai "
        "dengan STEP 07C."
    )

    ref_matrix = np.stack([
        references[sid]
        for sid in candidate_ids
    ])

    assert ref_matrix.shape == (
        len(candidate_ids), 384
    )

    print("=== STEP 08C: DINOV2-ONLY ===")
    print("Jumlah gambar:", len(entries))
    print("Kandidat sumber:", len(candidate_ids))
    print("Perangkat:", device)


    # 8. Buat eksperimen baru
    run08c = start_experiment(
        "STEP_08C_DINOV2_RETRIEVAL",
        "DINOV2_MULTI_SOURCE_RETRIEVAL",
        {
            "input_manifest_sha256": manifest_hash,
            "reference_fingerprint": refs_hash,
            "model": MODEL,
            "preprocessing": PREPROC,
            "candidate_count": len(candidate_ids)
        }
    )

    meta_path = run08c / "metadata.json"

    result_rows = []
    ranking_rows = []

    try:

        # 9. Bandingkan setiap gambar edit
        # dengan SEMBILAN sumber sekaligus
        for i, e in enumerate(entries, 1):

            sid = e["source_id"]

            row = {
                "source_id": sid,
                "source_filename":
                    e["source_filename"],
                "transformation_id":
                    e["transformation_id"],
                "image_sha256": e["sha256"],
                "candidate_count":
                    len(candidate_ids),
                "status": "ERROR",
                "error": ""
            }

            try:

                start = time.perf_counter()

                vector = embedding08c(
                    ROOT / e["stored_path"]
                )

                # Cosine similarity terhadap 9 sumber
                similarities = (
                    ref_matrix @ vector
                )

                assert np.isfinite(
                    similarities
                ).all()

                assert np.all(
                    (-1.001 <= similarities)
                    & (similarities <= 1.001)
                )

                # Urutkan dari similarity tertinggi
                order = sorted(
                    range(len(candidate_ids)),
                    key=lambda j: (
                        -float(similarities[j]),
                        candidate_ids[j]
                    )
                )

                ranked_ids = [
                    candidate_ids[j]
                    for j in order
                ]

                true_index = candidate_ids.index(sid)

                true_rank = (
                    ranked_ids.index(sid) + 1
                )

                best_index = order[0]
                second_index = order[1]

                highest_other = max(
                    float(similarities[j])
                    for j in range(len(similarities))
                    if j != true_index
                )

                duration = (
                    time.perf_counter() - start
                )

                row.update({
                    "source_prediksi_top1":
                        candidate_ids[best_index],
                    "source_benar_rank":
                        true_rank,
                    "top1_benar":
                        candidate_ids[best_index] == sid,
                    "kemiripan_sumber_benar":
                        float(
                            similarities[true_index]
                        ),
                    "kemiripan_top1":
                        float(
                            similarities[best_index]
                        ),
                    "kemiripan_top2":
                        float(
                            similarities[second_index]
                        ),
                    "margin_top1_top2":
                        float(
                            similarities[best_index]
                            - similarities[second_index]
                        ),
                    "margin_benar_vs_pesaing":
                        float(
                            similarities[true_index]
                        ) - highest_other,
                    "waktu_detik":
                        round(duration, 6),
                    "status": "PASS",
                    "error": ""
                })

                # Simpan seluruh sembilan peringkat
                for rank, j in enumerate(order, 1):

                    ranking_rows.append({
                        "source_id_gambar_edit": sid,
                        "transformation_id":
                            e["transformation_id"],
                        "image_sha256": e["sha256"],
                        "peringkat": rank,
                        "source_id_kandidat":
                            candidate_ids[j],
                        "cosine_similarity":
                            float(similarities[j]),
                        "kandidat_adalah_sumber_benar":
                            candidate_ids[j] == sid
                    })

            except Exception as exc:

                row["error"] = (
                    f"{type(exc).__name__}: {exc}"
                )

            result_rows.append(row)

            print(
                f"[{i:02d}/{len(entries)}] "
                f"{e['transformation_id']} "
                f"({sid}): {row['status']}"
                + (
                    f", peringkat sumber benar="
                    f"{row['source_benar_rank']}"
                    if row["status"] == "PASS"
                    else f", {row['error']}"
                )
            )


        # 10. Simpan hasil
        result_df = pd.DataFrame(result_rows)

        rankings_df = pd.DataFrame(
            ranking_rows
        )

        result_csv = (
            run08c / "outputs/"
            "dinov2_retrieval_results.csv"
        )

        ranking_csv = (
            run08c / "outputs/"
            "all_candidate_rankings.csv"
        )

        result_df.to_csv(
            result_csv,
            index=False
        )

        rankings_df.to_csv(
            ranking_csv,
            index=False
        )

        errors = int(
            (
                result_df["status"] == "ERROR"
            ).sum()
        )

        ok = result_df[
            result_df["status"] == "PASS"
        ]

        summary = {
            "input_manifest_sha256": manifest_hash,
            "reference_fingerprint": refs_hash,
            "model": MODEL,
            "preprocessing": PREPROC,
            "total_images": len(entries),
            "candidate_count": len(candidate_ids),
            "successful_inferences": len(ok),
            "top1_correct_count":
                int(ok["top1_benar"].sum())
                if len(ok) else 0,
            "top3_correct_count":
                int(
                    (
                        ok["source_benar_rank"] <= 3
                    ).sum()
                ) if len(ok) else 0,
            "error_count": errors,
            "ranking_count": len(rankings_df),
            "results_sha256":
                file_hash(result_csv),
            "rankings_sha256":
                file_hash(ranking_csv),
            "finished_at_utc":
                datetime.now(
                    timezone.utc
                ).isoformat(),
            "fusion_used": False,
            "score_threshold_used": False
        }

        (
            run08c / "outputs/summary.json"
        ).write_text(
            json.dumps(
                summary,
                indent=2,
                ensure_ascii=False
            ),
            encoding="utf-8"
        )

        record_event(
            run08c,
            "DINOV2_RETRIEVAL_COMPLETED",
            summary
        )

        metadata = read_json08c(meta_path)

        metadata["status"] = (
            "COMPLETED"
            if errors == 0
            else "COMPLETED_WITH_ERRORS"
        )

        meta_path.write_text(
            json.dumps(
                metadata,
                indent=2
            ),
            encoding="utf-8"
        )

    except Exception as exc:

        record_event(
            run08c,
            "RUN_FAILED",
            {
                "error":
                    f"{type(exc).__name__}: {exc}"
            }
        )

        metadata = read_json08c(meta_path)
        metadata["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(
                metadata,
                indent=2
            ),
            encoding="utf-8"
        )

        raise


# 11. Tampilkan hasil keseluruhan
print(
    "\n=== HASIL STEP 08C: DINOV2-ONLY ==="
)

print(
    "Gambar diuji:",
    summary["total_images"]
)

print(
    "Jumlah kandidat sumber:",
    summary["candidate_count"]
)

print(
    "Berhasil dihitung:",
    summary["successful_inferences"]
)

print(
    "Sumber benar pada peringkat 1:",
    summary["top1_correct_count"]
)

print(
    "Sumber benar dalam 3 teratas:",
    summary["top3_correct_count"]
)

print(
    "Error:",
    summary["error_count"]
)


# 12. Ringkasan berdasarkan jenis transformasi
ok = result_df.loc[
    result_df["status"] == "PASS"
].copy()

if not ok.empty:

    ok["top1_benar"] = (
        ok["top1_benar"]
        .astype(str)
        .str.lower()
        .eq("true")
    )

    report = ok.groupby(
        "transformation_id",
        as_index=False
    ).agg(
        jumlah=("top1_benar", "size"),
        top1_benar=("top1_benar", "sum"),
        rerata_similarity_sumber_benar=(
            "kemiripan_sumber_benar",
            "mean"
        )
    )

    report[
        "rerata_similarity_sumber_benar"
    ] = report[
        "rerata_similarity_sumber_benar"
    ].round(6)

    display(report)

    # 13. Tunjukkan gambar yang salah sumber
    print(
        "\n=== KASUS SALAH IDENTIFIKASI ==="
    )

    wrong = ok[
        ~ok["top1_benar"]
    ][[
        "source_filename",
        "transformation_id",
        "source_id",
        "source_prediksi_top1",
        "source_benar_rank",
        "kemiripan_sumber_benar"
    ]]

    if wrong.empty:
        print(
            "Tidak ada salah identifikasi "
            "pada batch ini."
        )
    else:
        display(wrong)


print("\nHasil:", result_csv)
print("Seluruh peringkat:", ranking_csv)
print("STATUS: DINOV2-ONLY | BELUM FUSION")
```

```python
# ==================================================
# JATIKARYA NEW - DETAIL HASIL DINOV2 STEP 08C
# Membaca hasil tersimpan, tanpa pengujian ulang
# ==================================================

from pathlib import Path
from IPython.display import display
import pandas as pd
import json

assert ROOT.name == "JATIKARYA_NEW"

folder = (
    ROOT / "02_EXPERIMENTS/"
    "STEP_08C_DINOV2_RETRIEVAL"
)

valid = []

for summary_path in sorted(
    folder.glob("*/outputs/summary.json"),
    reverse=True
):
    try:
        info = json.loads(
            summary_path.read_text(encoding="utf-8")
        )

        meta = json.loads(
            (
                summary_path.parent.parent
                / "metadata.json"
            ).read_text(encoding="utf-8")
        )

        hasil_path = (
            summary_path.parent
            / "dinov2_retrieval_results.csv"
        )

        ranking_path = (
            summary_path.parent
            / "all_candidate_rankings.csv"
        )

        if (
            meta.get("status") == "COMPLETED"
            and info.get("error_count") == 0
            and hasil_path.is_file()
            and ranking_path.is_file()
            and file_hash(hasil_path)
                == info.get("results_sha256")
            and file_hash(ranking_path)
                == info.get("rankings_sha256")
        ):
            valid.append(
                (hasil_path, ranking_path)
            )

    except (OSError, ValueError, KeyError, TypeError):
        continue

assert valid, "Belum ada hasil STEP 08C yang valid."

hasil_path, ranking_path = valid[0]

hasil = pd.read_csv(hasil_path)
ranking = pd.read_csv(ranking_path)

assert len(hasil) == 72, (
    "Jumlah gambar bukan 72."
)

assert len(ranking) == 648, (
    "Jumlah perbandingan bukan 648."
)

# 1. Tabel lengkap 72 gambar edit
print("=== 72 HASIL PER GAMBAR EDIT ===")

detail = hasil[[
    "source_filename",
    "source_id",
    "transformation_id",
    "kemiripan_sumber_benar",
    "source_prediksi_top1",
    "source_benar_rank",
    "top1_benar"
]].copy()

detail = detail.sort_values(
    ["source_id", "transformation_id"]
)

display(detail.round(6))

# 2. Matriks kemiripan 9 sumber x 8 transformasi
print("\n=== MATRIKS SIMILARITY: 9 SUMBER x 8 EDIT ===")

matriks = detail.pivot(
    index=["source_id", "source_filename"],
    columns="transformation_id",
    values="kemiripan_sumber_benar"
)

display(matriks.round(6))

# 3. Perbandingan terhadap seluruh kandidat sumber
print("\n=== HASIL SEMBILAN KANDIDAT ===")

print("Total gambar edit:", len(hasil))
print("Total skor similarity:", len(ranking))

# Contoh sembilan kandidat untuk satu gambar edit
contoh = ranking[
    (ranking["source_id_gambar_edit"]
     == detail.iloc[0]["source_id"])
    &
    (ranking["transformation_id"]
     == detail.iloc[0]["transformation_id"])
].sort_values("peringkat")

print("\nContoh peringkat kandidat satu gambar:")
display(contoh)

print("\nSeluruh 648 skor tersimpan di:")
print(ranking_path)

print("\nStatus: READ-ONLY, TIDAK ADA PENGUJIAN ULANG")
```

```python
# ==========================================================
# JATIKARYA NEW - STEP 08D
# PERBANDINGAN TRUSTMARK-ONLY DAN DINOV2-ONLY
# Semua gambar STEP 08A, tanpa Fusion
# ==========================================================

from pathlib import Path
from datetime import datetime, timezone
from IPython.display import display
import pandas as pd
import json
import hashlib

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

assert all(callable(globals().get(n)) for n in (
    "file_hash", "start_experiment", "record_event"
))

KEY = [
    "source_id",
    "transformation_id",
    "image_sha256"
]

STEP08A = (
    ROOT / "02_EXPERIMENTS/STEP_08A_AUTOMATED_EDITS"
)

OUT08D = (
    ROOT / "02_EXPERIMENTS/"
    "STEP_08D_INDIVIDUAL_COMPARISON"
)

EDIT_ROOT = (
    ROOT / "01_DATASET/AUTO_EDITS"
).resolve()


def jread08d(path):
    return json.loads(
        path.read_text(encoding="utf-8")
    )


# 1. Cari manifest STEP 08A yang valid
manifest_file = None

for p in sorted(
    STEP08A.glob(
        "*/outputs/auto_edit_manifest.json"
    ),
    reverse=True
):
    try:
        m = jread08d(p)
        meta = jread08d(
            p.parent.parent / "metadata.json"
        )

        if (
            meta.get("status") == "PASS"
            and m.get("files")
            and len(m["files"])
                == int(m["source_count"])
                * int(m["operation_count"])
        ):
            manifest_file = p
            manifest = m
            break

    except (
        OSError, ValueError,
        TypeError, KeyError
    ):
        continue

assert manifest_file is not None, (
    "Manifest STEP 08A valid belum ditemukan."
)

manifest_sha = file_hash(manifest_file)

base = pd.DataFrame(
    manifest["files"]
).rename(
    columns={"sha256": "image_sha256"}
)

assert not base.duplicated(KEY).any()

assert len(base) == int(
    manifest.get("image_count", len(base))
)

# Pastikan seluruh gambar masih sama
for row in base.itertuples(index=False):

    p = (
        ROOT / row.stored_path
    ).resolve()

    assert p.is_relative_to(EDIT_ROOT)
    assert p.is_file()
    assert file_hash(p) == row.image_sha256

    assert row.parent_type == "WATERMARKED"

expected = set(
    map(tuple, base[KEY].to_numpy())
)


# 2. Cari hasil pengujian yang cocok
def load_result08d(
    step,
    csv_name,
    sha_key,
    extra_file=None,
    extra_sha_key=None
):

    folder = (
        ROOT / "02_EXPERIMENTS" / step
    )

    reports = sorted(
        folder.glob("*/outputs/summary.json"),
        reverse=True
    )

    for sp in reports:

        try:
            s = jread08d(sp)

            meta = jread08d(
                sp.parent.parent / "metadata.json"
            )

            f = sp.parent / csv_name

            if not (
                meta.get("status") == "COMPLETED"
                and s.get("input_manifest_sha256")
                    == manifest_sha
                and int(
                    s.get("error_count", -1)
                ) == 0
                and f.is_file()
                and file_hash(f) == s.get(sha_key)
            ):
                continue

            if extra_file:

                extra = sp.parent / extra_file

                if (
                    not extra.is_file()
                    or file_hash(extra)
                        != s.get(extra_sha_key)
                ):
                    continue

            frame = pd.read_csv(
                f,
                dtype=str,
                keep_default_na=False
            )

            if (
                not set(KEY).issubset(
                    frame.columns
                )
                or len(frame) != len(base)
            ):
                continue

            found = set(
                map(
                    tuple,
                    frame[KEY].to_numpy()
                )
            )

            if (
                frame.duplicated(KEY).any()
                or found != expected
            ):
                continue

            return frame, f

        except (
            OSError, ValueError,
            TypeError, KeyError,
            AssertionError
        ):
            continue

    raise RuntimeError(
        f"Hasil {step} tidak cocok atau belum lengkap."
    )


# 3. Baca hasil TrustMark dan DINOv2
tm, tm_file = load_result08d(
    "STEP_08B_TRUSTMARK_ONLY",
    "trustmark_results.csv",
    "results_sha256"
)

di, di_file = load_result08d(
    "STEP_08C_DINOV2_RETRIEVAL",
    "dinov2_retrieval_results.csv",
    "results_sha256",
    "all_candidate_rankings.csv",
    "rankings_sha256"
)


# 4. Periksa integritas hasil
assert tm["hasil"].isin([
    "KODE_COCOK",
    "TIDAK_TERDETEKSI",
    "KODE_TIDAK_COCOK"
]).all()

assert di["status"].eq("PASS").all()

base_index = base.set_index(KEY)
tm_index = pd.MultiIndex.from_frame(tm[KEY])

assert tm["source_sha256"].eq(
    base_index.loc[
        tm_index, "source_sha256"
    ].to_numpy()
).all()

assert tm["parent_reference_sha256"].eq(
    base_index.loc[
        tm_index, "parent_reference_sha256"
    ].to_numpy()
).all()

rank = pd.to_numeric(
    di["source_benar_rank"],
    errors="raise"
)

assert rank.between(
    1,
    int(di["candidate_count"].iloc[0])
).all()

rank1 = (
    di["source_prediksi_top1"]
    .eq(di["source_id"])
)

assert rank1.equals(rank.eq(1))

assert rank1.equals(
    di["top1_benar"].str.lower().eq("true")
)


# 5. Gabungkan hasil pada gambar yang sama
cmp = base.merge(
    tm[
        KEY + [
            "hasil",
            "watermark_detected",
            "payload_match",
            "waktu_detik"
        ]
    ].rename(
        columns={
            "hasil": "hasil_trustmark",
            "waktu_detik": "waktu_trustmark_detik"
        }
    ),
    on=KEY,
    how="inner",
    validate="one_to_one"
).merge(
    di[
        KEY + [
            "source_prediksi_top1",
            "source_benar_rank",
            "top1_benar",
            "kemiripan_sumber_benar",
            "kemiripan_top1",
            "margin_top1_top2",
            "waktu_detik"
        ]
    ].rename(
        columns={
            "waktu_detik": "waktu_dinov2_detik"
        }
    ),
    on=KEY,
    how="inner",
    validate="one_to_one"
)

assert len(cmp) == len(base)

cmp["dinov2_top1_benar"] = (
    cmp["source_prediksi_top1"]
    .eq(cmp["source_id"])
)

cmp["trustmark_kode_cocok"] = (
    cmp["hasil_trustmark"]
    .eq("KODE_COCOK")
)

# Kategori hanya untuk membandingkan dua metode
# BUKAN keputusan Fusion
def kategori08d(row):

    if row["trustmark_kode_cocok"]:
        trustmark = "TM cocok"

    elif row["hasil_trustmark"] == "TIDAK_TERDETEKSI":
        trustmark = "TM tak terdeteksi"

    else:
        trustmark = "TM kode tidak cocok"

    dinov2 = (
        "DINO Top-1 benar"
        if row["dinov2_top1_benar"]
        else "DINO Top-1 salah"
    )

    return trustmark + " | " + dinov2


cmp["kategori_perbandingan"] = cmp.apply(
    kategori08d,
    axis=1
)

numeric_columns = [
    "source_benar_rank",
    "kemiripan_sumber_benar",
    "kemiripan_top1",
    "margin_top1_top2",
    "waktu_trustmark_detik",
    "waktu_dinov2_detik"
]

for column in numeric_columns:

    cmp[column] = pd.to_numeric(
        cmp[column],
        errors="raise"
    )

cmp = cmp.sort_values(
    KEY
).reset_index(drop=True)


# 6. Sidik jari input perbandingan
fingerprint = hashlib.sha256(
    json.dumps(
        {
            "step": "STEP08D_v1",
            "manifest_sha256": manifest_sha,
            "trustmark_results_sha256":
                file_hash(tm_file),
            "dinov2_results_sha256":
                file_hash(di_file)
        },
        sort_keys=True
    ).encode()
).hexdigest()


# 7. Gunakan arsip jika tabel sama sudah dibuat
comparison_file = None

for p in sorted(
    OUT08D.glob(
        "*/outputs/comparison_summary.json"
    ),
    reverse=True
):

    try:
        s = jread08d(p)

        meta = jread08d(
            p.parent.parent / "metadata.json"
        )

        csv = (
            p.parent / "individual_comparison.csv"
        )

        if (
            meta.get("status") == "COMPLETED"
            and s.get("input_fingerprint")
                == fingerprint
            and csv.is_file()
            and file_hash(csv)
                == s.get("result_sha256")
        ):
            comparison_file = csv
            break

    except (
        OSError, ValueError,
        KeyError, TypeError
    ):
        continue


# 8. Simpan tabel baru jika belum ada
if comparison_file is None:

    run = start_experiment(
        "STEP_08D_INDIVIDUAL_COMPARISON",
        "SIDE_BY_SIDE_NO_FUSION",
        {
            "input_fingerprint": fingerprint
        }
    )

    meta_path = run / "metadata.json"

    try:
        comparison_file = (
            run / "outputs/individual_comparison.csv"
        )

        cmp.to_csv(
            comparison_file,
            index=False
        )

        summary = {
            "input_fingerprint": fingerprint,
            "input_manifest_sha256": manifest_sha,
            "trustmark_results_sha256":
                file_hash(tm_file),
            "dinov2_results_sha256":
                file_hash(di_file),
            "result_sha256":
                file_hash(comparison_file),
            "total_images": len(cmp),
            "fusion_used": False,
            "created_at_utc":
                datetime.now(
                    timezone.utc
                ).isoformat()
        }

        (
            run / "outputs/comparison_summary.json"
        ).write_text(
            json.dumps(
                summary,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

        record_event(
            run,
            "INDIVIDUAL_COMPARISON_COMPLETED",
            summary
        )

        metadata = jread08d(meta_path)

        metadata["status"] = "COMPLETED"

        meta_path.write_text(
            json.dumps(
                metadata,
                indent=2
            ),
            encoding="utf-8"
        )

        print(
            "Perbandingan baru berhasil disimpan."
        )

    except Exception as exc:

        record_event(
            run,
            "RUN_FAILED",
            {"error": str(exc)}
        )

        metadata = jread08d(meta_path)
        metadata["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(
                metadata,
                indent=2
            ),
            encoding="utf-8"
        )

        raise

else:

    print(
        "Perbandingan lama valid. "
        "Tidak membuat ulang."
    )


# 9. Ringkasan per transformasi
report = cmp.groupby(
    "transformation_id",
    as_index=False
).agg(
    gambar=(
        "image_sha256",
        "size"
    ),
    trustmark_kode_cocok=(
        "trustmark_kode_cocok",
        "sum"
    ),
    dinov2_top1_benar=(
        "dinov2_top1_benar",
        "sum"
    ),
    rata_kemiripan_dinov2=(
        "kemiripan_sumber_benar",
        "mean"
    ),
    rata_margin_top1_top2=(
        "margin_top1_top2",
        "mean"
    )
)

report["rata_kemiripan_dinov2"] = (
    report["rata_kemiripan_dinov2"].round(6)
)

report["rata_margin_top1_top2"] = (
    report["rata_margin_top1_top2"].round(6)
)

print(
    "\n=== STEP 08D: PERBANDINGAN INDIVIDUAL ==="
)

display(report)

print("\n=== KATEGORI HASIL ===")

display(
    cmp["kategori_perbandingan"]
    .value_counts()
    .rename_axis("Kategori")
    .reset_index(name="Jumlah")
)

print("\nJumlah gambar:", len(cmp))
print("Hasil tersimpan:", comparison_file)
print("STATUS: SELESAI | BELUM FUSION")
```

```python
# ======================================================
# JATIKARYA NEW - STEP 08E
# CROPPING TERKONTROL: 5 POSISI, 9 SUMBER
# TrustMark-only, belum Fusion
# ======================================================

from pathlib import Path
from datetime import datetime, timezone
from importlib.metadata import version
from PIL import Image, ImageChops
from trustmark import TrustMark
from IPython.display import display
import pandas as pd
import hashlib
import json
import time
import torch

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

def baca08e(p):
    return json.loads(p.read_text(encoding="utf-8"))

orig_root = (ROOT / "01_DATASET/ORIGINAL").resolve()
wm_root = (ROOT / "01_DATASET/WATERMARKED").resolve()
edit_root = (ROOT / "01_DATASET/AUTO_EDITS").resolve()
crop_root = (ROOT / "01_DATASET/CROP_POSITION_EDITS").resolve()

folder_a = ROOT / "02_EXPERIMENTS/STEP_08A_AUTOMATED_EDITS"
folder_b = ROOT / "02_EXPERIMENTS/STEP_08B_TRUSTMARK_ONLY"
folder_e = ROOT / "02_EXPERIMENTS/STEP_08E_CROP_POSITION"

rasio = 0.80

# 1. Baca manifest STEP 08A yang telah lulus
batch = None

for p in sorted(
    folder_a.glob("*/outputs/auto_edit_manifest.json"),
    reverse=True
):
    try:
        m = baca08e(p)
        meta = baca08e(p.parent.parent / "metadata.json")

        if meta.get("status") == "PASS" and m.get("files"):
            batch = (p, m)
            break
    except (OSError, ValueError, KeyError):
        continue

assert batch, "STEP 08A PASS belum ditemukan."

manifest_path, manifest = batch
manifest_sha = file_hash(manifest_path)

# 2. Periksa seluruh sumber
reg = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str,
    keep_default_na=False
)

assert reg.source_id.is_unique

sources = {}

for _, s in reg.iterrows():
    p = (ROOT / s["stored_path"]).resolve()

    assert p.is_relative_to(orig_root)
    assert p.is_file()
    assert file_hash(p) == s["sha256"]

    sources[s["source_id"]] = s

# 3. Temukan gambar crop tengah sebelumnya
centers = {}

for e in manifest["files"]:

    if e["transformation_id"] != "crop_center_080":
        continue

    sid = e["source_id"]

    assert sid in sources
    assert e["source_sha256"] == sources[sid]["sha256"]
    assert sid not in centers

    p = (ROOT / e["stored_path"]).resolve()

    assert p.is_relative_to(edit_root)
    assert p.is_file()
    assert file_hash(p) == e["sha256"]

    centers[sid] = e

assert set(centers) == set(sources), (
    "Crop tengah belum lengkap untuk semua sumber."
)

# 4. Cari watermark induk sesuai STEP 08A
parents = {}

for sid, e in centers.items():

    matches = [
        p for p in sorted(
            (wm_root / sid).glob("*.png")
        )
        if p.is_file()
        and file_hash(p) == e["parent_reference_sha256"]
    ]

    assert matches, (
        f"Watermark induk tidak ditemukan: {sid}"
    )

    parents[sid] = matches[0]

# 5. Gunakan hasil crop tengah dari STEP 08B
tm_version = version("trustmark")
center_cache = None

for sp in sorted(
    folder_b.glob("*/outputs/summary.json"),
    reverse=True
):
    try:
        summary = baca08e(sp)
        csv = sp.parent / "trustmark_results.csv"

        meta = baca08e(
            sp.parent.parent / "metadata.json"
        )

        if not (
            meta.get("status") == "COMPLETED"
            and summary.get("input_manifest_sha256")
                == manifest_sha
            and summary.get("trustmark_version")
                == tm_version
            and summary.get("error_count") == 0
            and csv.is_file()
            and file_hash(csv)
                == summary.get("results_sha256")
        ):
            continue

        df = pd.read_csv(
            csv, dtype=str, keep_default_na=False
        )

        crop_df = df[
            df["transformation_id"] == "crop_center_080"
        ]

        valid = {}

        for _, row in crop_df.iterrows():

            sid = row["source_id"]

            if (
                sid not in centers
                or sid in valid
                or row["image_sha256"]
                    != centers[sid]["sha256"]
            ):
                continue

            if row["hasil"] not in (
                "KODE_COCOK",
                "TIDAK_TERDETEKSI",
                "KODE_TIDAK_COCOK"
            ):
                continue

            valid[sid] = row.to_dict()

        if set(valid) == set(sources):
            center_cache = valid
            break

    except (OSError, ValueError, KeyError, TypeError):
        continue

assert center_cache is not None, (
    "Hasil crop tengah STEP 08B tidak ditemukan."
)

# 6. Identitas konfigurasi
config = {
    "version": "STEP08E_POSITION_V1",
    "ratio": rasio,
    "trustmark_version": tm_version,
    "sources": [
        (
            sid,
            centers[sid]["sha256"],
            centers[sid]["parent_reference_sha256"]
        )
        for sid in sorted(sources)
    ]
}

fingerprint = hashlib.sha256(
    json.dumps(config, sort_keys=True).encode()
).hexdigest()

# 7. Periksa apakah STEP 08E sudah selesai
existing = None

for sp in sorted(
    folder_e.glob("*/outputs/summary.json"),
    reverse=True
):
    try:
        summary = baca08e(sp)
        meta = baca08e(
            sp.parent.parent / "metadata.json"
        )

        csv = sp.parent / "crop_position_results.csv"

        if not (
            meta.get("status") == "COMPLETED"
            and summary.get("input_fingerprint")
                == fingerprint
            and summary.get("total_images")
                == len(sources) * 5
            and summary.get("error_count") == 0
            and csv.is_file()
            and file_hash(csv)
                == summary.get("results_sha256")
        ):
            continue

        old = pd.read_csv(
            csv, dtype=str, keep_default_na=False
        )

        assert len(old) == len(sources) * 5

        assert len(
            old[["source_id", "posisi"]].drop_duplicates()
        ) == len(old)

        files_valid = True

        for _, row in old.iterrows():

            p = (ROOT / row["stored_path"]).resolve()

            if not (
                (
                    p.is_relative_to(crop_root)
                    or p.is_relative_to(edit_root)
                )
                and p.is_file()
                and file_hash(p) == row["image_sha256"]
            ):
                files_valid = False
                break

        if files_valid:
            existing = (old, csv)
            break

    except (
        OSError, AssertionError,
        ValueError, KeyError, TypeError
    ):
        continue

# 8. Proses crop yang belum ada
if existing:

    hasil08e, hasil_file = existing

    print(
        "STEP 08E sudah tersedia. "
        "Tidak membuat atau menguji ulang."
    )

else:

    print("=== STEP 08E: CROPPING TERKONTROL ===")
    print("Jumlah sumber:", len(sources))
    print("Posisi per sumber: 5")
    print("Hasil lama dipakai:", len(sources))
    print("Gambar baru diuji:", len(sources) * 4)

    detector08e = TrustMark(
        verbose=False,
        model_type="Q",
        encoding_type=TrustMark.Encoding.BCH_5,
        device=(
            "cuda" if torch.cuda.is_available()
            else "cpu"
        )
    )

    run08e = start_experiment(
        "STEP_08E_CROP_POSITION",
        "CONTROLLED_CROP_TRUSTMARK_ONLY",
        {
            "input_fingerprint": fingerprint,
            "retained_fraction": rasio
        }
    )

    meta_path = run08e / "metadata.json"
    rows = []

    try:

        for sid in sorted(sources):

            source = sources[sid]

            with Image.open(parents[sid]) as img:
                base = img.convert("RGB")
                base.load()

            w, h = base.size

            cw = max(1, round(w * rasio))
            ch = max(1, round(h * rasio))

            cx = (w - cw) // 2
            cy = (h - ch) // 2

            # Nama posisi menunjukkan bagian
            # yang lebih banyak dibuang.
            positions = {
                "tengah": (cx, cy),
                "buang_kiri": (w - cw, cy),
                "buang_kanan": (0, cy),
                "buang_atas": (cx, h - ch),
                "buang_bawah": (cx, 0)
            }

            payload = bin(
                int(source["sha256"], 16)
            )[2:].zfill(256)[:61]

            for posisi, (x, y) in positions.items():

                crop_img = base.crop((
                    x, y, x + cw, y + ch
                ))

                if posisi == "tengah":

                    old = center_cache[sid]

                    path = (
                        ROOT / centers[sid]["stored_path"]
                    ).resolve()

                    # Pastikan crop tengah memang
                    # identik dengan STEP 08A.
                    with Image.open(path) as ref:
                        same = ImageChops.difference(
                            crop_img,
                            ref.convert("RGB")
                        ).getbbox() is None

                    assert same, (
                        "Crop tengah berbeda dari STEP 08A."
                    )

                    status = old["hasil"]
                    detected = (
                        old["watermark_detected"].lower()
                        == "true"
                    )

                    matched = (
                        old["payload_match"].lower()
                        == "true"
                    )

                    duration = float(
                        old["waktu_detik"]
                    )

                    sha = centers[sid]["sha256"]
                    origin = "ARSIP_08B"

                else:

                    folder = (
                        crop_root / sid / run08e.name
                    )

                    folder.mkdir(
                        parents=True,
                        exist_ok=True
                    )

                    path = folder / f"{posisi}.png"

                    with path.open("xb") as f:
                        crop_img.save(
                            f, format="PNG"
                        )

                    sha = file_hash(path)

                    # Baca ulang file hasil crop
                    with Image.open(path) as img:
                        image_to_test = img.convert("RGB")

                    start = time.perf_counter()

                    decoded, flag, schema = (
                        detector08e.decode(
                            image_to_test,
                            MODE="binary"
                        )
                    )

                    duration = (
                        time.perf_counter() - start
                    )

                    detected = bool(flag)

                    matched = bool(
                        detected
                        and decoded == payload
                        and schema
                            == TrustMark.Encoding.BCH_5
                    )

                    if matched:
                        status = "KODE_COCOK"
                    elif detected:
                        status = "KODE_TIDAK_COCOK"
                    else:
                        status = "TIDAK_TERDETEKSI"

                    origin = "BARU_08E"

                rows.append({
                    "source_id": sid,
                    "source_filename":
                        source["original_filename"],
                    "posisi": posisi,
                    "persen_lebar_dipertahankan": 80,
                    "persen_tinggi_dipertahankan": 80,
                    "crop_box_px":
                        str([x, y, x + cw, y + ch]),
                    "parent_reference_sha256":
                        centers[sid][
                            "parent_reference_sha256"
                        ],
                    "image_sha256": sha,
                    "stored_path":
                        str(path.relative_to(ROOT)),
                    "watermark_detected": detected,
                    "payload_match": matched,
                    "hasil": status,
                    "waktu_detik":
                        round(duration, 6),
                    "asal_hasil": origin
                })

            print(
                source["original_filename"],
                ": 5 posisi diperiksa"
            )

        # 9. Simpan hasil penelitian
        hasil08e = pd.DataFrame(rows)

        assert len(hasil08e) == len(sources) * 5

        hasil_file = (
            run08e
            / "outputs/crop_position_results.csv"
        )

        hasil08e.to_csv(
            hasil_file,
            index=False
        )

        summary = {
            "input_fingerprint": fingerprint,
            "total_images": len(hasil08e),
            "newly_tested": len(sources) * 4,
            "reused_center": len(sources),
            "error_count": 0,
            "results_sha256": file_hash(hasil_file),
            "created_at_utc":
                datetime.now(
                    timezone.utc
                ).isoformat(),
            "fusion_used": False
        }

        (
            run08e / "outputs/summary.json"
        ).write_text(
            json.dumps(summary, indent=2),
            encoding="utf-8"
        )

        record_event(
            run08e,
            "CONTROLLED_CROP_COMPLETE",
            summary
        )

        metadata = baca08e(meta_path)
        metadata["status"] = "COMPLETED"

        meta_path.write_text(
            json.dumps(metadata, indent=2),
            encoding="utf-8"
        )

    except Exception as exc:

        record_event(
            run08e,
            "RUN_FAILED",
            {
                "error":
                    f"{type(exc).__name__}: {exc}"
            }
        )

        metadata = baca08e(meta_path)
        metadata["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(metadata, indent=2),
            encoding="utf-8"
        )

        raise

# 10. Tampilkan hasil per posisi
print("\n=== HASIL STEP 08E ===")

tabel = pd.crosstab(
    hasil08e["posisi"],
    hasil08e["hasil"]
).reindex(
    index=[
        "tengah",
        "buang_kiri",
        "buang_kanan",
        "buang_atas",
        "buang_bawah"
    ],
    columns=[
        "KODE_COCOK",
        "TIDAK_TERDETEKSI",
        "KODE_TIDAK_COCOK"
    ],
    fill_value=0
)

display(tabel)

print("Total gambar crop:", len(hasil08e))

print(
    "Hasil lama digunakan:",
    sum(hasil08e["asal_hasil"] == "ARSIP_08B")
)

print(
    "Hasil pengujian baru:",
    sum(hasil08e["asal_hasil"] == "BARU_08E")
)

print("Hasil tersimpan:", hasil_file)
print("STATUS: TRUSTMARK-ONLY | BELUM FUSION")
```

```python
# ========================================================
# JATIKARYA NEW - STEP 08F
# DINOV2-ONLY: CROPPING 5 POSISI, 9 SUMBER
# 9 hasil lama + 36 pengujian baru
# BELUM FUSION
# ========================================================

from pathlib import Path
from datetime import datetime, timezone
from PIL import Image, ImageOps
from torchvision import transforms
from IPython.display import display
import torch
import torch.nn.functional as F
import pandas as pd
import numpy as np
import hashlib
import json
import time

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"
assert all(
    callable(globals().get(n))
    for n in (
        "file_hash",
        "start_experiment",
        "record_event"
    )
)

BASE = ROOT / "02_EXPERIMENTS"
MODEL = "dinov2_vits14"
PREPROC = "Resize518_CenterCrop518_ImageNet"

POSITIONS = [
    "tengah",
    "buang_kiri",
    "buang_kanan",
    "buang_atas",
    "buang_bawah"
]


def read08f(path):
    return json.loads(
        path.read_text(encoding="utf-8")
    )


# 1. Baca sumber dan hasil cropping STEP 08E
reg = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str,
    keep_default_na=False
)

assert reg.source_id.is_unique
assert reg.sha256.is_unique

sources = {
    r.source_id: r
    for r in reg.itertuples(index=False)
}

assert len(sources) == 9, (
    "Jumlah kandidat berubah. "
    "Periksa rancangan evaluasi."
)

crop_root = (
    ROOT / "01_DATASET/CROP_POSITION_EDITS"
).resolve()

edit_root = (
    ROOT / "01_DATASET/AUTO_EDITS"
).resolve()

orig_root = (
    ROOT / "01_DATASET/ORIGINAL"
).resolve()

for r in sources.values():

    p = (
        ROOT / r.stored_path
    ).resolve()

    assert p.is_relative_to(orig_root)
    assert p.is_file()
    assert file_hash(p) == r.sha256


# 2. Cari hasil STEP 08E yang lengkap
crop_selected = None

for sp in sorted(
    (
        BASE / "STEP_08E_CROP_POSITION"
    ).glob("*/outputs/summary.json"),
    reverse=True
):

    try:
        s = read08f(sp)

        meta = read08f(
            sp.parent.parent / "metadata.json"
        )

        csv = (
            sp.parent / "crop_position_results.csv"
        )

        if (
            meta.get("status") == "COMPLETED"
            and s.get("error_count") == 0
            and s.get("total_images") == 45
            and csv.is_file()
            and file_hash(csv) == s.get("results_sha256")
        ):
            crop_selected = csv
            break

    except (
        OSError, ValueError,
        KeyError, TypeError
    ):
        continue

assert crop_selected, (
    "STEP 08E lengkap belum ditemukan."
)

cropping = pd.read_csv(
    crop_selected,
    dtype=str,
    keep_default_na=False
)

assert len(cropping) == 45

assert not cropping.duplicated([
    "source_id", "posisi"
]).any()

assert set(cropping.source_id) == set(sources)
assert set(cropping.posisi) == set(POSITIONS)

for sid, group in cropping.groupby("source_id"):
    assert set(group.posisi) == set(POSITIONS)

# Pastikan file crop tidak berubah
for row in cropping.itertuples(index=False):

    p = (
        ROOT / row.stored_path
    ).resolve()

    assert (
        p.is_relative_to(crop_root)
        or p.is_relative_to(edit_root)
    )

    assert p.is_file()
    assert file_hash(p) == row.image_sha256


# 3. Ambil embedding sembilan sumber asli
refs = {}
proofs = []

reports = sorted(
    (
        BASE / "STEP_05B_DINOV2_ORIGINAL"
    ).glob("*/outputs/reference_result.json"),
    reverse=True
)

for sid in sorted(sources):

    found = None

    for rp in reports:

        try:
            d = read08f(rp)

            npy = (
                rp.parent / "embedding_original.npy"
            )

            if not (
                d.get("source_id") == sid
                and d.get("source_sha256")
                    == sources[sid].sha256
                and d.get("model") == MODEL
                and d.get("preprocessing") == PREPROC
                and d.get("embedding_dimension") == 384
                and d.get("status") == "PASS"
                and npy.is_file()
                and file_hash(npy)
                    == d.get("embedding_sha256")
            ):
                continue

            v = np.load(
                npy,
                allow_pickle=False
            )

            if (
                v.shape == (384,)
                and np.isfinite(v).all()
                and np.isclose(
                    np.linalg.norm(v),
                    1,
                    atol=1e-3
                )
            ):
                found = (
                    v.astype(np.float32),
                    d["embedding_sha256"]
                )
                break

        except (
            OSError, ValueError,
            TypeError, KeyError
        ):
            continue

    assert found is not None, (
        f"Embedding sumber belum valid: {sid}"
    )

    refs[sid] = found[0]

    proofs.append({
        "source_id": sid,
        "original_sha256":
            sources[sid].sha256,
        "embedding_sha256": found[1]
    })

ref_fp = hashlib.sha256(
    json.dumps(
        proofs,
        sort_keys=True,
        separators=(",", ":")
    ).encode()
).hexdigest()


# 4. Ambil crop tengah dari STEP 08C
prior = None

for sp in sorted(
    (
        BASE / "STEP_08C_DINOV2_RETRIEVAL"
    ).glob("*/outputs/summary.json"),
    reverse=True
):

    try:
        info = read08f(sp)

        meta = read08f(
            sp.parent.parent / "metadata.json"
        )

        result_file = (
            sp.parent / "dinov2_retrieval_results.csv"
        )

        rank_file = (
            sp.parent / "all_candidate_rankings.csv"
        )

        if (
            meta.get("status") == "COMPLETED"
            and info.get("error_count") == 0
            and info.get("reference_fingerprint") == ref_fp
            and info.get("model") == MODEL
            and info.get("preprocessing") == PREPROC
            and result_file.is_file()
            and rank_file.is_file()
            and file_hash(result_file)
                == info.get("results_sha256")
            and file_hash(rank_file)
                == info.get("rankings_sha256")
        ):
            prior = (
                pd.read_csv(
                    result_file,
                    dtype=str,
                    keep_default_na=False
                ),
                pd.read_csv(
                    rank_file,
                    dtype=str,
                    keep_default_na=False
                )
            )
            break

    except (
        OSError, ValueError,
        KeyError, TypeError
    ):
        continue

assert prior is not None, (
    "Arsip STEP 08C belum ditemukan."
)

old_result, old_rank = prior

center = cropping[
    cropping.posisi == "tengah"
]

assert len(center) == len(sources)

center_cache = {}

for c in center.itertuples(index=False):

    matches = old_result[
        (old_result.source_id == c.source_id)
        & (
            old_result.transformation_id
            == "crop_center_080"
        )
        & (
            old_result.image_sha256
            == c.image_sha256
        )
        & (old_result.status == "PASS")
    ]

    assert len(matches) == 1, (
        f"Crop tengah tidak cocok: {c.source_id}"
    )

    ranks = old_rank[
        (
            old_rank.source_id_gambar_edit
            == c.source_id
        )
        & (
            old_rank.transformation_id
            == "crop_center_080"
        )
        & (
            old_rank.image_sha256
            == c.image_sha256
        )
    ]

    assert len(ranks) == len(sources)

    assert set(
        ranks.source_id_kandidat
    ) == set(sources)

    center_cache[c.source_id] = (
        matches.iloc[0],
        ranks
    )


# 5. Cari hasil STEP 08F lama
output_folder = (
    BASE / "STEP_08F_DINOV2_CROP_RETRIEVAL"
)

input_hash = file_hash(crop_selected)

previous = None

for sp in sorted(
    output_folder.glob("*/outputs/summary.json"),
    reverse=True
):

    try:
        info = read08f(sp)

        meta = read08f(
            sp.parent.parent / "metadata.json"
        )

        csv1 = (
            sp.parent / "crop_dinov2_results.csv"
        )

        csv2 = (
            sp.parent / "crop_all_candidate_rankings.csv"
        )

        if not (
            meta.get("status") == "COMPLETED"
            and info.get("input_crop_sha256")
                == input_hash
            and info.get("reference_fingerprint")
                == ref_fp
            and info.get("model") == MODEL
            and info.get("preprocessing") == PREPROC
            and csv1.is_file()
            and csv2.is_file()
            and file_hash(csv1)
                == info.get("results_sha256")
            and file_hash(csv2)
                == info.get("rankings_sha256")
        ):
            continue

        x = pd.read_csv(
            csv1,
            dtype=str,
            keep_default_na=False
        )

        y = pd.read_csv(
            csv2,
            dtype=str,
            keep_default_na=False
        )

        x_keys = set(zip(
            x.source_id,
            x.posisi,
            x.image_sha256
        ))

        expected_keys = set(zip(
            cropping.source_id,
            cropping.posisi,
            cropping.image_sha256
        ))

        if (
            len(x) == 45
            and len(y) == 405
            and not x.duplicated([
                "source_id", "posisi"
            ]).any()
            and x_keys == expected_keys
            and x.status.eq("PASS").all()
        ):
            previous = (x, info)
            break

    except (
        OSError, ValueError,
        TypeError, KeyError
    ):
        continue


# 6. Jalankan pengujian yang belum ada
if previous is not None:

    hasil08f, summary08f = previous

    print(
        "STEP 08F sudah tersedia. "
        "Tidak menghitung ulang."
    )

else:

    assert (
        "model05a" in globals()
        and getattr(
            model05a, "embed_dim", None
        ) == 384
    ), "Jalankan STEP 05A."

    device = (
        "cuda"
        if torch.cuda.is_available()
        else "cpu"
    )

    model05a = model05a.to(device).eval()

    process = transforms.Compose([
        transforms.Resize(518),
        transforms.CenterCrop(518),
        transforms.ToTensor(),
        transforms.Normalize(
            [0.485, 0.456, 0.406],
            [0.229, 0.224, 0.225]
        )
    ])

    def embed08f(path):

        with Image.open(path) as im:

            im.seek(0)

            rgb = ImageOps.exif_transpose(
                im
            ).convert("RGB")

            tensor = process(
                rgb
            ).unsqueeze(0).to(device)

        with torch.inference_mode():

            v = F.normalize(
                model05a(tensor),
                p=2,
                dim=1
            )

        out = (
            v[0]
            .cpu()
            .numpy()
            .astype(np.float32)
        )

        assert out.shape == (384,)
        assert np.isfinite(out).all()

        return out


    # Periksa konsistensi model
    test_sid = sorted(sources)[0]

    test = embed08f(
        ROOT / sources[test_sid].stored_path
    )

    assert np.allclose(
        test,
        refs[test_sid],
        atol=2e-3,
        rtol=2e-3
    ), (
        "Model atau preprocessing berbeda "
        "dari referensi."
    )

    ids = sorted(sources)

    matrix = np.stack([
        refs[sid] for sid in ids
    ])

    print("=== STEP 08F: DINOV2 CROPPING ===")
    print("Total gambar:", len(cropping))
    print("Hasil lama:", len(center_cache))
    print("Gambar diuji baru:", 36)
    print("Kandidat sumber:", len(ids))

    run = start_experiment(
        "STEP_08F_DINOV2_CROP_RETRIEVAL",
        "DINOV2_CROP_MULTI_SOURCE",
        {
            "input_crop_sha256": input_hash,
            "reference_fingerprint": ref_fp
        }
    )

    meta_path = run / "metadata.json"

    output_rows = []
    rankings = []

    try:

        for i, c in enumerate(
            cropping.itertuples(index=False), 1
        ):

            if c.posisi == "tengah":

                old, old_ranks = (
                    center_cache[c.source_id]
                )

                item = {
                    "source_id": c.source_id,
                    "source_filename": c.source_filename,
                    "posisi": c.posisi,
                    "image_sha256": c.image_sha256,
                    "hasil_trustmark": c.hasil,
                    "prediksi_top1":
                        old.source_prediksi_top1,
                    "sumber_benar_rank":
                        int(old.source_benar_rank),
                    "top1_benar":
                        str(old.top1_benar).lower()
                        == "true",
                    "similarity_sumber_benar":
                        float(
                            old.kemiripan_sumber_benar
                        ),
                    "similarity_top1":
                        float(old.kemiripan_top1),
                    "margin_top1_top2":
                        float(old.margin_top1_top2),
                    "waktu_detik":
                        float(old.waktu_detik),
                    "status": "PASS",
                    "asal_hasil": "ARSIP_08C"
                }

                for rr in old_ranks.itertuples(
                    index=False
                ):

                    rankings.append({
                        "source_id_gambar_edit":
                            c.source_id,
                        "posisi": c.posisi,
                        "image_sha256":
                            c.image_sha256,
                        "peringkat":
                            int(rr.peringkat),
                        "source_id_kandidat":
                            rr.source_id_kandidat,
                        "cosine_similarity":
                            float(rr.cosine_similarity)
                    })

            else:

                start = time.perf_counter()

                v = embed08f(
                    ROOT / c.stored_path
                )

                scores = matrix @ v

                assert np.isfinite(scores).all()
                assert np.all(
                    np.abs(scores) <= 1.001
                )

                order = sorted(
                    range(len(ids)),
                    key=lambda j: (
                        -float(scores[j]),
                        ids[j]
                    )
                )

                truth = ids.index(c.source_id)

                true_rank = (
                    order.index(truth) + 1
                )

                item = {
                    "source_id": c.source_id,
                    "source_filename": c.source_filename,
                    "posisi": c.posisi,
                    "image_sha256": c.image_sha256,
                    "hasil_trustmark": c.hasil,
                    "prediksi_top1":
                        ids[order[0]],
                    "sumber_benar_rank":
                        true_rank,
                    "top1_benar":
                        true_rank == 1,
                    "similarity_sumber_benar":
                        float(scores[truth]),
                    "similarity_top1":
                        float(scores[order[0]]),
                    "margin_top1_top2":
                        float(
                            scores[order[0]]
                            - scores[order[1]]
                        ),
                    "waktu_detik":
                        round(
                            time.perf_counter()
                            - start, 6
                        ),
                    "status": "PASS",
                    "asal_hasil": "BARU_08F"
                }

                for rank, j in enumerate(
                    order, 1
                ):

                    rankings.append({
                        "source_id_gambar_edit":
                            c.source_id,
                        "posisi": c.posisi,
                        "image_sha256":
                            c.image_sha256,
                        "peringkat": rank,
                        "source_id_kandidat":
                            ids[j],
                        "cosine_similarity":
                            float(scores[j])
                    })

            output_rows.append(item)

            label = (
                "BENAR"
                if item["top1_benar"]
                else "SALAH"
            )

            print(
                f"[{i:02d}/45] "
                f"{c.source_filename} / "
                f"{c.posisi}: Top-1 {label} "
                f"({item['asal_hasil']})"
            )

        # 7. Simpan seluruh hasil
        hasil08f = pd.DataFrame(
            output_rows
        )

        ranks08f = pd.DataFrame(
            rankings
        )

        assert len(hasil08f) == 45
        assert len(ranks08f) == 405

        out_csv = (
            run / "outputs/crop_dinov2_results.csv"
        )

        rank_csv = (
            run / "outputs/"
            "crop_all_candidate_rankings.csv"
        )

        hasil08f.to_csv(
            out_csv,
            index=False
        )

        ranks08f.to_csv(
            rank_csv,
            index=False
        )

        summary08f = {
            "input_crop_sha256": input_hash,
            "reference_fingerprint": ref_fp,
            "model": MODEL,
            "preprocessing": PREPROC,
            "source_count": len(ids),
            "total_images": len(hasil08f),
            "reused_center_count": 9,
            "newly_tested_count": 36,
            "top1_correct_count": int(
                hasil08f.top1_benar.sum()
            ),
            "error_count": 0,
            "results_sha256":
                file_hash(out_csv),
            "rankings_sha256":
                file_hash(rank_csv),
            "created_at_utc":
                datetime.now(
                    timezone.utc
                ).isoformat(),
            "fusion_used": False
        }

        (
            run / "outputs/summary.json"
        ).write_text(
            json.dumps(
                summary08f,
                indent=2
            ),
            encoding="utf-8"
        )

        record_event(
            run,
            "DINOV2_CROP_RETRIEVAL_COMPLETED",
            summary08f
        )

        meta = read08f(meta_path)
        meta["status"] = "COMPLETED"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

    except Exception as e:

        record_event(
            run,
            "RUN_FAILED",
            {
                "error":
                    f"{type(e).__name__}: {e}"
            }
        )

        meta = read08f(meta_path)
        meta["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

        raise


# 8. Tampilkan ringkasan hasil per posisi
hasil08f["top1_benar"] = (
    hasil08f.top1_benar.astype(str)
    .str.lower()
    .eq("true")
)

hasil08f["similarity_sumber_benar"] = (
    pd.to_numeric(
        hasil08f.similarity_sumber_benar
    )
)

report = hasil08f.groupby(
    "posisi",
    as_index=False
).agg(
    gambar=("source_id", "size"),
    trustmark_cocok=(
        "hasil_trustmark",
        lambda x: int(
            (x == "KODE_COCOK").sum()
        )
    ),
    dinov2_top1_benar=(
        "top1_benar", "sum"
    ),
    rerata_similarity=(
        "similarity_sumber_benar",
        "mean"
    )
)

report["rerata_similarity"] = (
    report.rerata_similarity.round(6)
)

report["posisi"] = pd.Categorical(
    report.posisi,
    POSITIONS,
    ordered=True
)

report = report.sort_values(
    "posisi"
)

print("\n=== STEP 08F: HASIL PER POSISI ===")

display(report)

print(
    "\n=== GAMBAR DENGAN WATERMARK "
    "TIDAK COCOK / TIDAK TERDETEKSI ==="
)

display(
    hasil08f[
        hasil08f.hasil_trustmark
        != "KODE_COCOK"
    ][[
        "source_filename",
        "posisi",
        "hasil_trustmark",
        "top1_benar",
        "similarity_sumber_benar",
        "sumber_benar_rank"
    ]]
)

print("Total crop:", len(hasil08f))

print(
    "DINOv2 Top-1 benar:",
    int(hasil08f.top1_benar.sum())
)

print(
    "STATUS: DINOV2-ONLY | BELUM FUSION"
)
```

```python
# ==================================================
# JATIKARYA NEW - STEP 09A
# DOWNLOAD 9 FOTO WATERMARKED UNTUK ai
# Read-only: tidak mengubah data penelitian
# ==================================================

from pathlib import Path
from google.colab import files
from PIL import Image
import pandas as pd
import json
import re
import zipfile

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"
assert callable(file_hash)

source_dir = (
    ROOT / "01_DATASET/ORIGINAL"
).resolve()

wm_dir = (
    ROOT / "01_DATASET/WATERMARKED"
).resolve()

report_dir = (
    ROOT / "02_EXPERIMENTS/STEP_04_TRUSTMARK_ENCODE"
)

registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str,
    keep_default_na=False
)

assert not registry.empty
assert registry.source_id.is_unique
assert registry.sha256.is_unique

reports = sorted(
    report_dir.glob(
        "*/outputs/verification_result.json"
    ),
    reverse=True
)

selected = []

# 1. Cari foto WM yang valid untuk setiap sumber
for row in registry.itertuples(index=False):

    orig = (
        ROOT / row.stored_path
    ).resolve()

    assert orig.is_relative_to(source_dir)
    assert orig.is_file()
    assert file_hash(orig) == row.sha256, (
        f"File asli berubah: {row.source_id}"
    )

    expected_payload = bin(
        int(row.sha256, 16)
    )[2:].zfill(256)[:61]

    match = None

    for rp in reports:

        try:
            r = json.loads(
                rp.read_text(encoding="utf-8")
            )

            if not (
                r.get("source_id") == row.source_id
                and r.get("source_sha256") == row.sha256
                and r.get("payload_bits") == expected_payload
                and r.get("payload_match") is True
                and r.get("watermark_detected") is True
            ):
                continue

            meta = json.loads(
                (
                    rp.parent.parent / "metadata.json"
                ).read_text(encoding="utf-8")
            )

            if meta.get("status") not in (
                "PASS", "COMPLETED"
            ):
                continue

            wm = (
                ROOT / r["stored_path"]
            ).resolve()

            if not (
                wm.is_relative_to(wm_dir)
                and wm.is_file()
                and file_hash(wm)
                    == r["watermarked_sha256"]
            ):
                continue

            with Image.open(wm) as im:
                assert im.format == "PNG"
                im.verify()

            match = (
                wm,
                r,
                rp.parent.parent.name
            )
            break

        except (
            OSError, ValueError, TypeError,
            KeyError, AssertionError
        ):
            continue

    if match is None:
        raise RuntimeError(
            f"Watermark valid tidak ditemukan: "
            f"{row.original_filename} "
            f"({row.source_id}). "
            "Periksa STEP 07B."
        )

    wm, result, run_id = match

    # Nama mudah dikenali dan tetap unik
    short_name = re.sub(
        r"[^A-Za-z0-9_-]+",
        "_",
        Path(row.original_filename).stem
    ).strip("_")[:45] or "foto"

    zip_name = (
        f"FOTO_WM/{row.source_id}"
        f"__{short_name}__WM.png"
    )

    selected.append({
        "source_id": row.source_id,
        "nama_foto_asli": row.original_filename,
        "file_wm": wm,
        "zip_name": zip_name,
        "sha256_wm": result["watermarked_sha256"],
        "run_trustmark": run_id,
        "frame_sumber": result.get(
            "source_frame_index", 0
        )
    })

# 2. Pastikan semua sumber memiliki WM valid
assert len(selected) == len(registry)
assert len({
    x["zip_name"] for x in selected
}) == len(selected)

# 3. Buat ZIP sementara di Colab
zip_path = Path("/content") / (
    "JATIKARYA_STEP09_FOTO_WM_UNTUK_GEMINI.zip"
)

manifest = [
    {
        k: v
        for k, v in s.items()
        if k != "file_wm"
    }
    for s in selected
]

with zipfile.ZipFile(
    zip_path,
    mode="w",
    compression=zipfile.ZIP_DEFLATED,
    compresslevel=4
) as z:

    for s in selected:
        z.write(
            s["file_wm"],
            s["zip_name"]
        )

    z.writestr(
        "DAFTAR_FOTO.json",
        json.dumps(
            manifest,
            ensure_ascii=False,
            indent=2
        )
    )

    z.writestr(
        "PETUNJUK.txt",
        "JATIKARYA - FOTO WM UNTUK GEMINI\n\n"
        "Gunakan gambar dari folder FOTO_WM.\n"
        "Buat 3 edit terpisah per sumber:\n"
        "1. Object removal\n"
        "2. Object replacement\n"
        "3. Background modification\n\n"
        "Catat prompt dan model Gemini.\n"
        "Jangan menanam watermark ulang.\n"
    )

# 4. Verifikasi integritas ZIP
with zipfile.ZipFile(zip_path) as z:

    assert z.testzip() is None

    assert len([
        n for n in z.namelist()
        if n.startswith("FOTO_WM/")
    ]) == len(selected)

# 5. Tampilkan hasil dan download otomatis
print("=== FOTO WATERMARKED SIAP UNTUK GEMINI ===")
print("Total sumber:", len(registry))
print("Watermark valid:", len(selected))
print("File ZIP:", zip_path.name)
print(
    "Ukuran:",
    round(zip_path.stat().st_size / 1024**2, 2),
    "MB"
)
print("Data penelitian tidak diubah.")

files.download(str(zip_path))
```

```python
# ==================================================
# JATIKARYA NEW - STEP 09B
# PENDAFTARAN 54 GAMBAR GENERATIF
# ChatGPT + Gemini | 3 edit x 9 sumber
# Tidak menguji ulang atau menanam watermark
# ==================================================

from google.colab import files
from PIL import Image
from IPython.display import display
from pathlib import Path
from io import BytesIO
from datetime import datetime, timezone
import pandas as pd
import hashlib
import json
import re

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

# 1. Pemetaan berdasarkan urutan 9 foto Anda
PETA09 = {
    1: ("SRC_5e7d8e04e7382bab", "Kopi Kopinako"),
    2: ("SRC_6f8175d77476728a", "Kadal terrarium"),
    3: ("SRC_9b694fca4505129b", "Kopi JAOO"),
    4: ("SRC_33feba5735b719cb", "Kopi latte"),
    5: ("SRC_211b9e8b868839c4", "Jalan raya"),
    6: ("SRC_af9637872435c21c", "Kolam ikan"),
    7: ("SRC_b09b075066fd85e4", "Point Coffee"),
    8: ("SRC_ce8fb67e7fe5f142", "Gate stasiun"),
    9: ("SRC_efc9dea6ceecac17", "Piring makanan")
}

REGEX09 = re.compile(
    r"^(removal|replacement|background)"
    r"[ _-]+(gpt|chatgpt|gemini)"
    r"[ _-]+(0?[1-9])"
    r"\.(png|jpg|jpeg|webp)$",
    re.IGNORECASE
)

registry09 = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str,
    keep_default_na=False
)

assert registry09.source_id.is_unique

assert set(registry09.source_id) == {
    sid for sid, _ in PETA09.values()
}, "Daftar sembilan sumber tidak sesuai."

source09 = {
    r.source_id: r
    for r in registry09.itertuples(index=False)
}

# 2. Tampilkan pemetaan untuk diperiksa
print("=== PEMETAAN NOMOR FOTO ===")

display(pd.DataFrame([
    {
        "Nomor": f"{n:02d}",
        "Foto": nama,
        "Source ID": sid
    }
    for n, (sid, nama) in PETA09.items()
]))

# 3. Cari induk watermarked yang valid
wm_root09 = (
    ROOT / "01_DATASET/WATERMARKED"
).resolve()

reports09 = sorted(
    (
        ROOT / "02_EXPERIMENTS/"
        "STEP_04_TRUSTMARK_ENCODE"
    ).glob("*/outputs/verification_result.json"),
    reverse=True
)

parent09 = {}

for sid, _ in PETA09.values():

    sha = source09[sid].sha256
    payload = bin(int(sha, 16))[2:].zfill(256)[:61]

    for rp in reports09:

        try:
            result = json.loads(
                rp.read_text(encoding="utf-8")
            )

            meta = json.loads(
                (
                    rp.parent.parent / "metadata.json"
                ).read_text(encoding="utf-8")
            )

            if not (
                result.get("source_id") == sid
                and result.get("source_sha256") == sha
                and result.get("payload_bits") == payload
                and result.get("payload_match") is True
                and result.get("watermark_detected") is True
                and meta.get("status") in (
                    "PASS", "COMPLETED"
                )
            ):
                continue

            wm = (
                ROOT / result["stored_path"]
            ).resolve()

            if (
                wm.is_relative_to(wm_root09)
                and wm.is_file()
                and file_hash(wm)
                    == result["watermarked_sha256"]
            ):
                parent09[sid] = (
                    result["watermarked_sha256"]
                )
                break

        except (
            OSError, ValueError,
            KeyError, TypeError
        ):
            continue

    assert sid in parent09, (
        f"Watermark induk belum valid: {sid}"
    )

# 4. Upload 54 gambar sekaligus
print("\n=== UPLOAD HASIL GENERATIF ===")
print("Pilih 54 gambar ChatGPT dan Gemini.")

uploaded09 = files.upload()

assert uploaded09, "Belum ada gambar diunggah."

# 5. Baca nama, format, dan hash setiap gambar
items09 = {}

alias_model = {
    "gpt": "chatgpt",
    "chatgpt": "chatgpt",
    "gemini": "gemini"
}

for filename, content in uploaded09.items():

    match = REGEX09.fullmatch(
        Path(filename).name
    )

    assert match, (
        f"Nama file belum sesuai: {filename}"
    )

    edit, model, number, ext = match.groups()

    edit = edit.lower()
    model = alias_model[model.lower()]
    number = int(number)

    key = (number, model, edit)

    assert key not in items09, (
        f"Duplikat kombinasi: {filename}"
    )

    with Image.open(BytesIO(content)) as img:
        actual_format = img.format
        width, height = img.size
        img.verify()

    assert actual_format in (
        "PNG", "JPEG", "WEBP"
    ), f"Format belum didukung: {filename}"

    sid = PETA09[number][0]

    items09[key] = {
        "source_number": number,
        "source_id": sid,
        "source_filename":
            source09[sid].original_filename,
        "provider": model,
        "edit_type": edit,
        "uploaded_filename": filename,
        "actual_format": actual_format,
        "width": width,
        "height": height,
        "sha256":
            hashlib.sha256(content).hexdigest(),
        "parent_type_declared": "WATERMARKED",
        "parent_reference_sha256": parent09[sid],
        "lineage_evidence":
            "USER_DECLARED_NOT_INDEPENDENTLY_VERIFIED",
        "prompt_id":
            f"PROMPT_{number:02d}_{edit}",
        "provider_model_version": "NOT_RECORDED",
        "_bytes": content
    }

# 6. Periksa kelengkapan 54 kombinasi
expected09 = {
    (n, ai, edit)
    for n in PETA09
    for ai in ("chatgpt", "gemini")
    for edit in (
        "removal", "replacement", "background"
    )
}

missing09 = sorted(expected09 - set(items09))

print("\n=== HASIL PEMERIKSAAN ===")
print("Gambar diterima:", len(items09))
print("Target:", len(expected09))

if missing09:
    print("Kombinasi yang belum tersedia:")
    for n, ai, edit in missing09:
        print(f"{n:02d} | {ai} | {edit}")

assert not missing09, (
    "Batch belum lengkap. Belum ada data disimpan."
)

# 7. Periksa hasil yang sudah pernah terdaftar
EXP09 = (
    ROOT / "02_EXPERIMENTS/"
    "STEP_09B_GENERATIVE_REGISTRATION"
)

DATA09 = (
    ROOT / "01_DATASET/GENERATIVE_EDITS"
).resolve()

existing09 = {}

for mp in EXP09.glob(
    "*/outputs/generative_manifest.json"
):

    meta = json.loads(
        (
            mp.parent.parent / "metadata.json"
        ).read_text(encoding="utf-8")
    )

    if meta.get("status") != "PASS":
        continue

    old = json.loads(
        mp.read_text(encoding="utf-8")
    )

    for row in old["files"]:

        key = (
            int(row["source_number"]),
            row["provider"],
            row["edit_type"]
        )

        p = (
            ROOT / row["stored_path"]
        ).resolve()

        assert p.is_relative_to(DATA09)
        assert p.is_file()
        assert file_hash(p) == row["sha256"]

        assert (
            key not in existing09
            or existing09[key]["sha256"]
                == row["sha256"]
        ), f"Konflik hasil lama: {key}"

        existing09[key] = row

for key, item in items09.items():

    if key in existing09:
        assert (
            existing09[key]["sha256"]
            == item["sha256"]
        ), (
            "File untuk kombinasi "
            f"{key} berbeda dari arsip. "
            "Data lama tidak akan ditimpa."
        )

# 8. Pratinjau kelengkapan
preview09 = pd.DataFrame([
    {
        "Nomor": f"{n:02d}",
        "AI": ai,
        "Jenis edit": edit,
        "Format asli":
            items09[(n, ai, edit)]["actual_format"]
    }
    for n, ai, edit in sorted(expected09)
])

print("\n=== RINGKASAN UPLOAD ===")

display(
    preview09.groupby(
        ["Nomor", "AI"]
    ).size().unstack(fill_value=0)
)

print("\nJumlah hasil:", len(preview09))

new_keys09 = [
    key for key in sorted(expected09)
    if key not in existing09
]

print("Belum terdaftar:", len(new_keys09))
print("Sudah terdaftar:", len(existing09))

# 9. Simpan setelah konfirmasi
if not new_keys09:

    print(
        "STATUS: SEMUA GAMBAR SUDAH TERDAFTAR"
    )

else:

    confirm09 = input(
        "\nKetik SIMPAN untuk melanjutkan: "
    ).strip()

    if confirm09 != "SIMPAN":

        print("STATUS: BELUM DISIMPAN")

    else:

        run09 = start_experiment(
            "STEP_09B_GENERATIVE_REGISTRATION",
            "AI_EDIT_OUTPUT_REGISTRATION",
            {
                "uploaded_count": len(items09),
                "new_count": len(new_keys09)
            }
        )

        meta_path09 = run09 / "metadata.json"
        saved09 = []

        try:

            for key in new_keys09:

                item = items09[key]

                suffix = {
                    "PNG": ".png",
                    "JPEG": ".jpg",
                    "WEBP": ".webp"
                }[item["actual_format"]]

                dest = (
                    ROOT
                    / "01_DATASET/GENERATIVE_EDITS"
                    / item["source_id"]
                    / item["provider"]
                    / item["edit_type"]
                    / run09.name
                    / (
                        f"{item['source_number']:02d}_"
                        f"{item['edit_type']}_"
                        f"{item['provider']}{suffix}"
                    )
                )

                dest.parent.mkdir(
                    parents=True,
                    exist_ok=True
                )

                # Simpan byte asli tanpa konversi
                with dest.open("xb") as f:
                    f.write(item["_bytes"])

                assert file_hash(dest) == (
                    item["sha256"]
                )

                record = {
                    k: v for k, v in item.items()
                    if k != "_bytes"
                }

                record["stored_path"] = str(
                    dest.relative_to(ROOT)
                )

                record["registered_at_utc"] = (
                    datetime.now(
                        timezone.utc
                    ).isoformat()
                )

                saved09.append(record)

            manifest09 = {
                "schema_version": 1,
                "expected_total": 54,
                "new_files": len(saved09),
                "previously_registered":
                    len(existing09),
                "created_at_utc":
                    datetime.now(
                        timezone.utc
                    ).isoformat(),
                "files": saved09
            }

            json_path09 = (
                run09 / "outputs/"
                "generative_manifest.json"
            )

            csv_path09 = (
                run09 / "outputs/"
                "generative_manifest.csv"
            )

            json_path09.write_text(
                json.dumps(
                    manifest09,
                    ensure_ascii=False,
                    indent=2
                ),
                encoding="utf-8"
            )

            pd.DataFrame(saved09).to_csv(
                csv_path09,
                index=False
            )

            record_event(
                run09,
                "GENERATIVE_FILES_REGISTERED",
                {
                    "new_files": len(saved09),
                    "manifest_sha256":
                        file_hash(json_path09),
                    "manifest_csv_sha256":
                        file_hash(csv_path09)
                }
            )

            meta = json.loads(
                meta_path09.read_text(
                    encoding="utf-8"
                )
            )

            meta["status"] = "PASS"

            meta_path09.write_text(
                json.dumps(
                    meta, indent=2
                ),
                encoding="utf-8"
            )

            print("\n=== STEP 09B SELESAI ===")
            print(
                "Gambar baru terdaftar:",
                len(saved09)
            )
            print(
                "Total terdaftar:",
                len(saved09) + len(existing09)
            )
            print("Manifest:", json_path09)
            print("STATUS: PASS | BELUM DIUJI")

        except Exception as exc:

            record_event(
                run09,
                "RUN_FAILED",
                {"error": str(exc)}
            )

            meta = json.loads(
                meta_path09.read_text(
                    encoding="utf-8"
                )
            )

            meta["status"] = "ERROR"

            meta_path09.write_text(
                json.dumps(meta, indent=2),
                encoding="utf-8"
            )

            raise
```

```python
# ==================================================
# JATIKARYA NEW - STEP 09C
# TRUSTMARK-ONLY PADA 54 GAMBAR GENERATIF
# ChatGPT vs Gemini | Removal, Replacement, Background
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from importlib.metadata import version
from PIL import Image, ImageOps
from trustmark import TrustMark
from IPython.display import display
import pandas as pd
import torch
import json
import time

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"
assert all(callable(globals().get(n)) for n in (
    "file_hash", "start_experiment", "record_event"
))

def read09c(p):
    return json.loads(p.read_text(encoding="utf-8"))

exp = ROOT / "02_EXPERIMENTS"
data_root = (ROOT / "01_DATASET/GENERATIVE_EDITS").resolve()
wm_root = (ROOT / "01_DATASET/WATERMARKED").resolve()
original_root = (ROOT / "01_DATASET/ORIGINAL").resolve()

# 1. Verifikasi sembilan sumber asli
registry = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str,
    keep_default_na=False
)

assert len(registry) == 9 and registry.source_id.is_unique
sources = registry.set_index("source_id").to_dict("index")

for sid, s in sources.items():
    original = (ROOT / s["stored_path"]).resolve()
    assert original.is_relative_to(original_root)
    assert original.is_file()
    assert file_hash(original) == s["sha256"]

# 2. Validasi watermark induk
valid_parents = set()

for p in (exp / "STEP_04_TRUSTMARK_ENCODE").glob(
    "*/outputs/verification_result.json"
):
    try:
        d = read09c(p)
        status = read09c(
            p.parent.parent / "metadata.json"
        ).get("status")

        sid = d["source_id"]

        if sid not in sources or status not in (
            "PASS", "COMPLETED"
        ):
            continue

        expected = bin(
            int(sources[sid]["sha256"], 16)
        )[2:].zfill(256)[:61]

        if (
            d.get("source_sha256") != sources[sid]["sha256"]
            or d.get("payload_bits") != expected
            or d.get("watermark_detected") is not True
            or d.get("payload_match") is not True
        ):
            continue

        parent = (ROOT / d["stored_path"]).resolve()

        if (
            parent.is_relative_to(wm_root)
            and parent.is_file()
            and file_hash(parent) == d["watermarked_sha256"]
        ):
            valid_parents.add((
                sid, d["watermarked_sha256"]
            ))

    except (
        OSError, KeyError, ValueError,
        TypeError, AssertionError
    ):
        continue

# 3. Ambil 54 gambar dari STEP 09B
selected = None

for path in sorted(
    (exp / "STEP_09B_GENERATIVE_REGISTRATION").glob(
        "*/outputs/generative_manifest.json"
    ),
    reverse=True
):
    try:
        m = read09c(path)
        meta = read09c(
            path.parent.parent / "metadata.json"
        )

        if (
            meta.get("status") == "PASS"
            and len(m.get("files", [])) == 54
        ):
            selected = (path, m)
            break

    except (OSError, ValueError, KeyError, TypeError):
        continue

assert selected, "Manifest STEP 09B lengkap belum ditemukan."

manifest_path, manifest = selected

entries = sorted(
    manifest["files"],
    key=lambda x: (
        int(x["source_number"]),
        x["provider"],
        x["edit_type"]
    )
)

expected_keys = {
    (n, ai, op)
    for n in range(1, 10)
    for ai in ("chatgpt", "gemini")
    for op in ("removal", "replacement", "background")
}

actual_keys = {
    (
        int(x["source_number"]),
        x["provider"],
        x["edit_type"]
    )
    for x in entries
}

assert len(entries) == 54
assert actual_keys == expected_keys

# Setiap nomor harus mengarah ke satu Source ID
number_source = {
    n: {
        x["source_id"]
        for x in entries
        if int(x["source_number"]) == n
    }
    for n in range(1, 10)
}

assert all(len(v) == 1 for v in number_source.values())
assert {
    next(iter(v)) for v in number_source.values()
} == set(sources)

# 4. Periksa seluruh file sebelum pengujian
for x in entries:
    sid = x["source_id"]

    assert sid in sources
    assert (
        sid, x["parent_reference_sha256"]
    ) in valid_parents, (
        f"Watermark induk tidak valid: {sid}"
    )

    path = (ROOT / x["stored_path"]).resolve()

    assert path.is_relative_to(data_root)
    assert path.is_file()
    assert file_hash(path) == x["sha256"], (
        f"File berubah: {x['uploaded_filename']}"
    )

    with Image.open(path) as im:
        assert im.format == x["actual_format"]
        im.verify()

manifest_sha = file_hash(manifest_path)
version_tm = version("trustmark")

results_dir = exp / "STEP_09C_TRUSTMARK_GENERATIVE"

keys = {
    (
        x["source_id"],
        x["provider"],
        x["edit_type"],
        x["sha256"]
    )
    for x in entries
}

# 5. Cari hasil lengkap dari arsip
existing = None

for sp in sorted(
    results_dir.glob("*/outputs/summary.json"),
    reverse=True
):
    try:
        s = read09c(sp)
        meta = read09c(
            sp.parent.parent / "metadata.json"
        )
        csv_path = sp.parent / "trustmark_generatif.csv"

        if not (
            meta.get("status") == "COMPLETED"
            and s.get("input_manifest_sha256") == manifest_sha
            and s.get("trustmark_version") == version_tm
            and s.get("error_count") == 0
            and s.get("total_images") == 54
            and csv_path.is_file()
            and file_hash(csv_path) == s.get("results_sha256")
        ):
            continue

        df = pd.read_csv(
            csv_path,
            dtype=str,
            keep_default_na=False
        )

        found = {
            (
                r.source_id,
                r.provider,
                r.edit_type,
                r.image_sha256
            )
            for r in df.itertuples(index=False)
        }

        if (
            len(df) == 54
            and found == keys
            and df["hasil"].isin([
                "KODE_COCOK",
                "TIDAK_TERDETEKSI",
                "KODE_TIDAK_COCOK"
            ]).all()
        ):
            existing = (df, s, csv_path)
            break

    except (OSError, ValueError, KeyError, TypeError):
        continue

# 6. Jalankan TrustMark jika belum tersedia
if existing:
    result, summary, result_path = existing

    print(
        "Hasil STEP 09C valid dari arsip. "
        "Tidak menghitung ulang."
    )

else:
    detector = TrustMark(
        verbose=False,
        model_type="Q",
        encoding_type=TrustMark.Encoding.BCH_5,
        device=(
            "cuda" if torch.cuda.is_available()
            else "cpu"
        )
    )

    run = start_experiment(
        "STEP_09C_TRUSTMARK_GENERATIVE",
        "TRUSTMARK_Q_BCH5_ONLY",
        {
            "input_manifest_sha256": manifest_sha,
            "trustmark_version": version_tm
        }
    )

    meta_path = run / "metadata.json"
    rows = []

    try:
        for i, x in enumerate(entries, 1):
            sid = x["source_id"]

            expected = bin(
                int(sources[sid]["sha256"], 16)
            )[2:].zfill(256)[:61]

            row = {
                "source_number": int(x["source_number"]),
                "source_id": sid,
                "provider": x["provider"],
                "edit_type": x["edit_type"],
                "uploaded_filename": x["uploaded_filename"],
                "image_sha256": x["sha256"],
                "parent_reference_sha256":
                    x["parent_reference_sha256"],
                "actual_format": x["actual_format"],
                "watermark_detected": False,
                "payload_match": False,
                "hasil": "ERROR",
                "decoded_payload": "",
                "decoded_schema": "",
                "waktu_detik": None,
                "error": ""
            }

            try:
                with Image.open(
                    ROOT / x["stored_path"]
                ) as im:
                    image = ImageOps.exif_transpose(
                        im
                    ).convert("RGB")

                t0 = time.perf_counter()

                decoded, detected, schema = detector.decode(
                    image,
                    MODE="binary"
                )

                elapsed = time.perf_counter() - t0
                detected = bool(detected)

                matched = bool(
                    detected
                    and decoded == expected
                    and schema == TrustMark.Encoding.BCH_5
                )

                label = (
                    "KODE_COCOK"
                    if matched
                    else (
                        "KODE_TIDAK_COCOK"
                        if detected
                        else "TIDAK_TERDETEKSI"
                    )
                )

                row.update(
                    watermark_detected=detected,
                    payload_match=matched,
                    hasil=label,
                    decoded_payload=(
                        str(decoded)
                        if decoded is not None else ""
                    ),
                    decoded_schema=(
                        str(schema)
                        if schema is not None else ""
                    ),
                    waktu_detik=round(elapsed, 6)
                )

            except Exception as e:
                row["error"] = (
                    f"{type(e).__name__}: {e}"
                )

            rows.append(row)

            print(
                f"[{i:02d}/54] "
                f"{x['provider']} | "
                f"{x['edit_type']} | "
                f"{int(x['source_number']):02d} "
                f"= {row['hasil']}"
            )

        # 7. Simpan hasil dan audit
        result = pd.DataFrame(rows)

        result_path = (
            run / "outputs/trustmark_generatif.csv"
        )

        result.to_csv(result_path, index=False)

        summary = {
            "input_manifest_sha256": manifest_sha,
            "trustmark_version": version_tm,
            "total_images": len(result),
            "matched_count": int(
                result.hasil.eq("KODE_COCOK").sum()
            ),
            "not_detected_count": int(
                result.hasil.eq("TIDAK_TERDETEKSI").sum()
            ),
            "mismatch_count": int(
                result.hasil.eq("KODE_TIDAK_COCOK").sum()
            ),
            "error_count": int(
                result.hasil.eq("ERROR").sum()
            ),
            "results_sha256": file_hash(result_path),
            "finished_at_utc":
                datetime.now(timezone.utc).isoformat(),
            "fusion_used": False
        }

        (
            run / "outputs/summary.json"
        ).write_text(
            json.dumps(summary, indent=2),
            encoding="utf-8"
        )

        record_event(
            run,
            "TRUSTMARK_GENERATIVE_COMPLETED",
            summary
        )

        meta = read09c(meta_path)
        meta["status"] = (
            "COMPLETED"
            if summary["error_count"] == 0
            else "COMPLETED_WITH_ERRORS"
        )

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

    except Exception as e:
        record_event(
            run,
            "RUN_FAILED",
            {"error": f"{type(e).__name__}: {e}"}
        )

        meta = read09c(meta_path)
        meta["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )
        raise

# 8. Ringkasan berdasarkan AI dan jenis edit
print("\n=== STEP 09C: HASIL TRUSTMARK GENERATIF ===")

counts = pd.crosstab(
    index=[
        result["provider"],
        result["edit_type"]
    ],
    columns=result["hasil"]
)

counts = counts.reindex(
    columns=[
        "KODE_COCOK",
        "TIDAK_TERDETEKSI",
        "KODE_TIDAK_COCOK",
        "ERROR"
    ],
    fill_value=0
)

display(counts)

print("Total gambar:", len(result))
print(
    "Kode cocok:",
    int(result.hasil.eq("KODE_COCOK").sum())
)
print(
    "Tidak terdeteksi:",
    int(result.hasil.eq("TIDAK_TERDETEKSI").sum())
)
print(
    "Kode tidak cocok:",
    int(result.hasil.eq("KODE_TIDAK_COCOK").sum())
)
print(
    "Error:",
    int(result.hasil.eq("ERROR").sum())
)

print("Hasil tersimpan:", result_path)
print("STATUS: TRUSTMARK-ONLY | BELUM FUSION")
```

```python
# ========================================================
# JATIKARYA NEW - STEP 09D
# DINOV2-ONLY: 54 GAMBAR GENERATIF x 9 SUMBER
# ChatGPT vs Gemini, belum Fusion
# ========================================================

from pathlib import Path
from datetime import datetime, timezone
from PIL import Image, ImageOps
from torchvision import transforms
from IPython.display import display
import torch
import torch.nn.functional as F
import numpy as np
import pandas as pd
import hashlib
import json
import time

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"
assert all(callable(globals().get(n)) for n in (
    "file_hash", "start_experiment", "record_event"
))

BASE = ROOT / "02_EXPERIMENTS"
MODEL = "dinov2_vits14"
PREPROC = "Resize518_CenterCrop518_ImageNet"
EDIT_ROOT = (ROOT / "01_DATASET/GENERATIVE_EDITS").resolve()
ORIG_ROOT = (ROOT / "01_DATASET/ORIGINAL").resolve()

def jread09d(p):
    return json.loads(p.read_text(encoding="utf-8"))

# 1. Validasi sembilan sumber asli
reg = pd.read_csv(
    ROOT / "04_LOGS/source_registry.csv",
    dtype=str,
    keep_default_na=False
)

assert len(reg) == 9
assert reg.source_id.is_unique
assert reg.sha256.is_unique

sources = {}

for r in reg.itertuples(index=False):
    p = (ROOT / r.stored_path).resolve()
    assert p.is_relative_to(ORIG_ROOT)
    assert p.is_file()
    assert file_hash(p) == r.sha256

    sources[r.source_id] = {
        "sha": r.sha256,
        "path": p,
        "filename": r.original_filename
    }

ids = sorted(sources)

# 2. Ambil embedding DINOv2 yang sudah valid
refs = {}
ref_proofs = []

ref_reports = sorted(
    (BASE / "STEP_05B_DINOV2_ORIGINAL").glob(
        "*/outputs/reference_result.json"
    ),
    reverse=True
)

for sid in ids:
    found = None

    for p in ref_reports:
        try:
            d = jread09d(p)
            npy = p.parent / "embedding_original.npy"

            if not (
                d.get("source_id") == sid
                and d.get("source_sha256") == sources[sid]["sha"]
                and d.get("model") == MODEL
                and d.get("preprocessing") == PREPROC
                and d.get("embedding_dimension") == 384
                and d.get("status") == "PASS"
                and npy.is_file()
                and file_hash(npy) == d.get("embedding_sha256")
            ):
                continue

            v = np.load(npy, allow_pickle=False)

            if (
                v.shape == (384,)
                and np.issubdtype(v.dtype, np.floating)
                and np.isfinite(v).all()
                and np.isclose(np.linalg.norm(v), 1, atol=1e-3)
            ):
                found = (
                    v.astype(np.float32),
                    d["embedding_sha256"]
                )
                break

        except (OSError, KeyError, ValueError, TypeError):
            continue

    assert found is not None, (
        f"Embedding asli belum valid: {sid}"
    )

    refs[sid] = found[0]
    ref_proofs.append({
        "source_id": sid,
        "source_sha256": sources[sid]["sha"],
        "embedding_sha256": found[1]
    })

ref_matrix = np.stack([refs[sid] for sid in ids])

# 3. Baca manifest STEP 09B
selected = None

for p in sorted(
    (BASE / "STEP_09B_GENERATIVE_REGISTRATION").glob(
        "*/outputs/generative_manifest.json"
    ),
    reverse=True
):
    try:
        m = jread09d(p)
        meta = jread09d(p.parent.parent / "metadata.json")

        if (
            meta.get("status") == "PASS"
            and len(m.get("files", [])) == 54
        ):
            selected = (p, m)
            break

    except (OSError, KeyError, ValueError, TypeError):
        continue

assert selected, "Manifest STEP 09B tidak ditemukan."

mp, manifest = selected

entries = sorted(
    manifest["files"],
    key=lambda x: (
        int(x["source_number"]),
        x["provider"],
        x["edit_type"]
    )
)

expected = {
    (n, ai, op)
    for n in range(1, 10)
    for ai in ("chatgpt", "gemini")
    for op in ("removal", "replacement", "background")
}

actual = {
    (
        int(x["source_number"]),
        x["provider"],
        x["edit_type"]
    )
    for x in entries
}

assert len(entries) == len(actual) == 54
assert actual == expected
assert len({x["source_id"] for x in entries}) == len(ids)

for n in range(1, 10):
    assert len({
        x["source_id"] for x in entries
        if int(x["source_number"]) == n
    }) == 1

# Pastikan file yang diuji sama dengan file terdaftar
for x in entries:
    assert x["source_id"] in sources
    assert x["parent_type_declared"] == "WATERMARKED"

    f = (ROOT / x["stored_path"]).resolve()

    assert f.is_relative_to(EDIT_ROOT)
    assert f.is_file()
    assert file_hash(f) == x["sha256"], (
        f"File berbeda: {x['uploaded_filename']}"
    )

manifest_sha = file_hash(mp)

ref_sha = hashlib.sha256(
    json.dumps(ref_proofs, sort_keys=True).encode()
).hexdigest()

fingerprint = hashlib.sha256(
    json.dumps({
        "version": 1,
        "manifest": manifest_sha,
        "references": ref_sha,
        "model": MODEL,
        "preprocessing": PREPROC
    }, sort_keys=True).encode()
).hexdigest()

expected_rows = {
    (
        x["source_id"],
        x["provider"],
        x["edit_type"],
        x["sha256"]
    )
    for x in entries
}

# 4. Cari arsip jika sudah pernah dijalankan
out_dir = BASE / "STEP_09D_DINOV2_GENERATIVE"
existing = None

for sp in sorted(
    out_dir.glob("*/outputs/summary.json"),
    reverse=True
):
    try:
        s = jread09d(sp)
        meta = jread09d(
            sp.parent.parent / "metadata.json"
        )

        f1 = sp.parent / "dinov2_generatif.csv"
        f2 = sp.parent / "all_candidate_rankings.csv"

        if not (
            meta.get("status") == "COMPLETED"
            and s.get("fingerprint") == fingerprint
            and s.get("error_count") == 0
            and f1.is_file() and f2.is_file()
            and file_hash(f1) == s.get("results_sha256")
            and file_hash(f2) == s.get("rankings_sha256")
        ):
            continue

        a = pd.read_csv(
            f1, dtype=str, keep_default_na=False
        )
        b = pd.read_csv(
            f2, dtype=str, keep_default_na=False
        )

        akeys = {
            (r.source_id, r.provider, r.edit_type,
             r.image_sha256)
            for r in a.itertuples(index=False)
        }

        bkeys = {
            (r.source_id_gambar_edit, r.provider,
             r.edit_type, r.image_sha256)
            for r in b.itertuples(index=False)
        }

        if (
            len(a) == 54
            and len(b) == 54 * len(ids)
            and akeys == expected_rows
            and bkeys == expected_rows
            and a.status.eq("PASS").all()
        ):
            existing = (a, s, f1, f2)
            break

    except (
        OSError, KeyError, ValueError,
        TypeError, AttributeError
    ):
        continue

# 5. Jalankan DINOv2 bila belum tersedia
if existing:
    result, summary, output_csv, ranking_csv = existing

    print(
        "STEP 09D ditemukan dalam arsip. "
        "Tidak menghitung ulang."
    )

else:
    assert (
        "model05a" in globals()
        and getattr(model05a, "embed_dim", None) == 384
    ), "Jalankan STEP 05A terlebih dahulu."

    device = (
        "cuda" if torch.cuda.is_available()
        else "cpu"
    )

    model05a = model05a.to(device).eval()

    prep = transforms.Compose([
        transforms.Resize(518),
        transforms.CenterCrop(518),
        transforms.ToTensor(),
        transforms.Normalize(
            [0.485, 0.456, 0.406],
            [0.229, 0.224, 0.225]
        )
    ])

    def get_embedding09d(p):
        with Image.open(p) as img:
            img.seek(0)
            rgb = ImageOps.exif_transpose(
                img
            ).convert("RGB")

            inp = prep(rgb).unsqueeze(0).to(device)

        with torch.inference_mode():
            v = F.normalize(
                model05a(inp),
                p=2,
                dim=1
            )

        out = (
            v[0].detach().cpu().numpy()
            .astype(np.float32)
        )

        assert out.shape == (384,)
        assert np.isfinite(out).all()

        return out

    # Periksa konsistensi model dengan STEP 07C
    sid_check = ids[0]

    assert np.allclose(
        get_embedding09d(
            sources[sid_check]["path"]
        ),
        refs[sid_check],
        atol=2e-3,
        rtol=2e-3
    ), (
        "Model berbeda dari referensi STEP 07C."
    )

    print("=== STEP 09D: DINOV2 GENERATIF ===")
    print("Gambar:", len(entries))
    print("Kandidat sumber:", len(ids))
    print("Perangkat:", device)

    run = start_experiment(
        "STEP_09D_DINOV2_GENERATIVE",
        "DINOV2_MULTI_SOURCE_ONLY",
        {
            "fingerprint": fingerprint,
            "total_images": len(entries),
            "candidate_count": len(ids)
        }
    )

    meta_path = run / "metadata.json"

    rows = []
    all_rankings = []

    try:
        for i, x in enumerate(entries, 1):
            sid = x["source_id"]

            row = {
                "source_number": int(x["source_number"]),
                "source_id": sid,
                "source_filename": x["source_filename"],
                "provider": x["provider"],
                "edit_type": x["edit_type"],
                "uploaded_filename": x["uploaded_filename"],
                "image_sha256": x["sha256"],
                "status": "ERROR",
                "error": ""
            }

            try:
                t0 = time.perf_counter()

                v = get_embedding09d(
                    ROOT / x["stored_path"]
                )

                # Bandingkan dengan sembilan sumber asli
                scores = ref_matrix @ v

                assert np.isfinite(scores).all()
                assert np.all(np.abs(scores) <= 1.001)

                order = sorted(
                    range(len(ids)),
                    key=lambda j: (
                        -float(scores[j]),
                        ids[j]
                    )
                )

                true_idx = ids.index(sid)
                true_rank = order.index(true_idx) + 1

                elapsed = time.perf_counter() - t0

                competitor = max(
                    float(scores[j])
                    for j in range(len(ids))
                    if j != true_idx
                )

                row.update({
                    "source_prediksi_top1": ids[order[0]],
                    "sumber_benar_rank": true_rank,
                    "top1_benar": true_rank == 1,
                    "similarity_sumber_benar":
                        float(scores[true_idx]),
                    "similarity_top1":
                        float(scores[order[0]]),
                    "margin_top1_top2":
                        float(
                            scores[order[0]]
                            - scores[order[1]]
                        ),
                    "margin_benar_vs_pesaing":
                        float(scores[true_idx]) - competitor,
                    "waktu_detik": round(elapsed, 6),
                    "status": "PASS"
                })

                for rank, j in enumerate(order, 1):
                    all_rankings.append({
                        "source_id_gambar_edit": sid,
                        "provider": x["provider"],
                        "edit_type": x["edit_type"],
                        "image_sha256": x["sha256"],
                        "peringkat": rank,
                        "source_id_kandidat": ids[j],
                        "cosine_similarity":
                            float(scores[j]),
                        "kandidat_sumber_benar":
                            ids[j] == sid
                    })

            except Exception as exc:
                row["error"] = (
                    f"{type(exc).__name__}: {exc}"
                )

            rows.append(row)

            print(
                f"[{i:02d}/54] "
                f"{x['provider']} | "
                f"{x['edit_type']} | "
                f"{int(x['source_number']):02d}: "
                f"{row['status']}"
                + (
                    f" (rank {row['sumber_benar_rank']})"
                    if row["status"] == "PASS"
                    else f" ({row['error']})"
                )
            )

        # 6. Simpan hasil dan 486 skor kandidat
        result = pd.DataFrame(rows)
        ranking = pd.DataFrame(all_rankings)

        output_csv = (
            run / "outputs/dinov2_generatif.csv"
        )

        ranking_csv = (
            run / "outputs/all_candidate_rankings.csv"
        )

        result.to_csv(output_csv, index=False)
        ranking.to_csv(ranking_csv, index=False)

        good = result[
            result.status == "PASS"
        ]

        summary = {
            "fingerprint": fingerprint,
            "input_manifest_sha256": manifest_sha,
            "reference_fingerprint": ref_sha,
            "model": MODEL,
            "preprocessing": PREPROC,
            "total_images": len(result),
            "successful_inferences": len(good),
            "candidate_count": len(ids),
            "top1_correct_count":
                int(good.top1_benar.sum()),
            "error_count":
                int(result.status.eq("ERROR").sum()),
            "ranking_count": len(ranking),
            "results_sha256": file_hash(output_csv),
            "rankings_sha256": file_hash(ranking_csv),
            "fusion_used": False,
            "created_at_utc":
                datetime.now(
                    timezone.utc
                ).isoformat()
        }

        (
            run / "outputs/summary.json"
        ).write_text(
            json.dumps(
                summary,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

        record_event(
            run,
            "DINOV2_GENERATIVE_COMPLETED",
            summary
        )

        meta = jread09d(meta_path)

        meta["status"] = (
            "COMPLETED"
            if summary["error_count"] == 0
            else "COMPLETED_WITH_ERRORS"
        )

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

    except Exception as exc:
        record_event(
            run,
            "RUN_FAILED",
            {"error": str(exc)}
        )

        meta = jread09d(meta_path)
        meta["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

        raise

# 7. Ringkasan per AI dan jenis edit
print("\n=== HASIL STEP 09D: DINOV2-ONLY ===")

valid = result[
    result.status.eq("PASS")
].copy()

valid["top1_benar"] = (
    valid.top1_benar.astype(str)
    .str.lower().eq("true")
)

valid["similarity_sumber_benar"] = pd.to_numeric(
    valid.similarity_sumber_benar
)

report = valid.groupby(
    ["provider", "edit_type"],
    as_index=False
).agg(
    gambar=("image_sha256", "size"),
    top1_benar=("top1_benar", "sum"),
    rata_similarity_sumber_benar=(
        "similarity_sumber_benar",
        "mean"
    )
)

report["rata_similarity_sumber_benar"] = (
    report["rata_similarity_sumber_benar"]
    .round(6)
)

display(report)

# 8. Tampilkan kasus salah identifikasi
wrong = valid.loc[
    ~valid.top1_benar,
    [
        "source_filename",
        "provider",
        "edit_type",
        "source_id",
        "source_prediksi_top1",
        "sumber_benar_rank",
        "similarity_sumber_benar"
    ]
]

print("\n=== KASUS SALAH IDENTIFIKASI ===")

if len(wrong):
    display(wrong)
else:
    print(
        "Tidak ada salah identifikasi "
        "pada hasil yang berhasil diproses."
    )

print("\nTotal gambar:", len(result))
print(
    "Top-1 benar:",
    int(valid.top1_benar.sum())
)
print("Error:", summary["error_count"])
print("Hasil:", output_csv)
print("Seluruh skor kandidat:", ranking_csv)
print("STATUS: DINOV2-ONLY | BELUM FUSION")
```

```python
# ==================================================
# JATIKARYA NEW - STEP 09E
# VALIDASI DAN PERBANDINGAN 54 GAMBAR GENERATIF
# TrustMark-only vs DINOv2-only | BELUM FUSION
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from IPython.display import display
import pandas as pd
import json
import hashlib

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

EXP = ROOT / "02_EXPERIMENTS"

# Gunakan run yang telah selesai dan dilaporkan
p_manifest = (
    EXP / "STEP_09B_GENERATIVE_REGISTRATION"
    / "20261107T185128500014Z_d37967ff"
    / "outputs/generative_manifest.json"
)

p_tm = (
    EXP / "STEP_09C_TRUSTMARK_GENERATIVE"
    / "20261107T185640103435Z_3d710573"
    / "outputs/trustmark_generatif.csv"
)

p_dino = (
    EXP / "STEP_09D_DINOV2_GENERATIVE"
    / "20261107T190219357254Z_ef5bc870"
    / "outputs/dinov2_generatif.csv"
)

def baca09e(p):
    return json.loads(p.read_text(encoding="utf-8"))

def cek_run09e(path, manifest_sha, expected_status):
    assert path.is_file()

    meta = baca09e(
        path.parent.parent / "metadata.json"
    )
    summary = baca09e(
        path.parent / "summary.json"
    )

    assert meta["status"] == expected_status
    assert summary["input_manifest_sha256"] == manifest_sha
    assert summary["results_sha256"] == file_hash(path)
    assert int(summary["error_count"]) == 0
    assert int(summary["total_images"]) == 54

    return pd.read_csv(
        path, dtype=str, keep_default_na=False
    )

# 1. Baca manifest dan periksa gambar
assert p_manifest.is_file()
manifest = baca09e(p_manifest)
assert len(manifest["files"]) == 54

manifest_sha = file_hash(p_manifest)

base = pd.DataFrame(
    manifest["files"]
).rename(columns={"sha256": "image_sha256"})

KEY = [
    "source_id",
    "provider",
    "edit_type",
    "image_sha256"
]

assert not base.duplicated(KEY).any()

allowed = (
    ROOT / "01_DATASET/GENERATIVE_EDITS"
).resolve()

for r in base.itertuples(index=False):
    path = (ROOT / r.stored_path).resolve()

    assert path.is_relative_to(allowed)
    assert path.is_file()
    assert file_hash(path) == r.image_sha256

expected = set(
    map(tuple, base[KEY].to_numpy())
)

# 2. Baca hasil kedua metode
tm = cek_run09e(
    p_tm, manifest_sha, "COMPLETED"
)
di = cek_run09e(
    p_dino, manifest_sha, "COMPLETED"
)

for df in (tm, di):
    assert len(df) == 54
    assert not df.duplicated(KEY).any()

    actual = set(
        map(tuple, df[KEY].to_numpy())
    )
    assert actual == expected

assert tm["hasil"].isin([
    "KODE_COCOK",
    "TIDAK_TERDETEKSI",
    "KODE_TIDAK_COCOK"
]).all()

assert di["status"].eq("PASS").all()

# Verifikasi arsip 486 skor kandidat DINOv2
rank_path = (
    p_dino.parent / "all_candidate_rankings.csv"
)

rank_summary = baca09e(
    p_dino.parent / "summary.json"
)

assert rank_path.is_file()
assert file_hash(rank_path) == rank_summary["rankings_sha256"]
assert int(rank_summary["ranking_count"]) == 486

# 3. Cocokkan berdasarkan identitas file
tm = tm[KEY + [
    "parent_reference_sha256",
    "hasil",
    "payload_match"
]].rename(columns={
    "hasil": "hasil_trustmark",
    "parent_reference_sha256": "parent_sha_tm"
})

di = di[KEY + [
    "source_prediksi_top1",
    "sumber_benar_rank",
    "similarity_sumber_benar",
    "margin_top1_top2"
]]

hasil = base.merge(
    tm, on=KEY, validate="one_to_one"
).merge(
    di, on=KEY, validate="one_to_one"
)

assert len(hasil) == 54

assert hasil["parent_reference_sha256"].eq(
    hasil["parent_sha_tm"]
).all()

# 4. Kategori perbandingan, bukan Fusion
hasil["trustmark_cocok"] = (
    hasil["hasil_trustmark"] == "KODE_COCOK"
)

hasil["dinov2_benar"] = (
    hasil["source_prediksi_top1"]
    == hasil["source_id"]
)

hasil["rank_benar"] = pd.to_numeric(
    hasil["sumber_benar_rank"]
)

assert hasil["dinov2_benar"].eq(
    hasil["rank_benar"] == 1
).all()

assert hasil["trustmark_cocok"].eq(
    hasil["payload_match"].str.lower() == "true"
).all()

hasil["kategori"] = ""

hasil.loc[
    hasil.trustmark_cocok & hasil.dinov2_benar,
    "kategori"
] = "KEDUA_BERHASIL"

hasil.loc[
    hasil.trustmark_cocok & ~hasil.dinov2_benar,
    "kategori"
] = "TM_COCOK_DINOV2_SALAH"

hasil.loc[
    ~hasil.trustmark_cocok & hasil.dinov2_benar,
    "kategori"
] = "TM_TIDAK_COCOK_DINOV2_BENAR"

hasil.loc[
    ~hasil.trustmark_cocok & ~hasil.dinov2_benar,
    "kategori"
] = "KEDUA_BELUM_BERHASIL"

assert hasil["kategori"].ne("").all()

# 5. Identitas hasil untuk mencegah duplikasi
fingerprint = hashlib.sha256(
    json.dumps({
        "manifest": manifest_sha,
        "trustmark": file_hash(p_tm),
        "dinov2": file_hash(p_dino)
    }, sort_keys=True).encode()
).hexdigest()

folder_out = (
    EXP / "STEP_09E_GENERATIVE_COMPARISON"
)

saved = None

for sp in folder_out.glob("*/outputs/summary.json"):
    try:
        old = baca09e(sp)
        meta = baca09e(
            sp.parent.parent / "metadata.json"
        )
        csv = sp.parent / "paired_comparison.csv"

        if (
            meta.get("status") == "COMPLETED"
            and old.get("fingerprint") == fingerprint
            and csv.is_file()
            and file_hash(csv) == old["results_sha256"]
        ):
            saved = csv
            break
    except (OSError, ValueError, KeyError):
        continue

if saved is None:
    run = start_experiment(
        "STEP_09E_GENERATIVE_COMPARISON",
        "SIDE_BY_SIDE_NO_FUSION",
        {"fingerprint": fingerprint}
    )

    meta_path = run / "metadata.json"

    try:
        saved = run / "outputs/paired_comparison.csv"
        hasil.to_csv(saved, index=False)

        summary = {
            "fingerprint": fingerprint,
            "total_images": 54,
            "counts": hasil["kategori"]
                .value_counts().to_dict(),
            "results_sha256": file_hash(saved),
            "fusion_used": False,
            "created_at_utc":
                datetime.now(timezone.utc).isoformat()
        }

        (
            run / "outputs/summary.json"
        ).write_text(
            json.dumps(summary, indent=2),
            encoding="utf-8"
        )

        record_event(
            run,
            "GENERATIVE_COMPARISON_COMPLETE",
            summary
        )

        meta = baca09e(meta_path)
        meta["status"] = "COMPLETED"
        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

    except Exception as exc:
        record_event(
            run, "RUN_FAILED", {"error": str(exc)}
        )

        meta = baca09e(meta_path)
        meta["status"] = "ERROR"
        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )
        raise

# 6. Tampilkan hasil
print("=== STEP 09E: PERBANDINGAN GENERATIF ===")

display(
    hasil.groupby(
        ["provider", "edit_type"],
        as_index=False
    ).agg(
        gambar=("image_sha256", "size"),
        trustmark_cocok=("trustmark_cocok", "sum"),
        dinov2_benar=("dinov2_benar", "sum")
    )
)

print("\n=== KATEGORI KEBERHASILAN ===")

display(
    hasil["kategori"].value_counts()
    .rename_axis("Kategori")
    .reset_index(name="Jumlah")
)

print("\n=== KASUS DINOV2 SALAH ===")

display(
    hasil.loc[
        ~hasil["dinov2_benar"],
        [
            "source_number",
            "provider",
            "edit_type",
            "hasil_trustmark",
            "source_id",
            "source_prediksi_top1",
            "sumber_benar_rank",
            "similarity_sumber_benar"
        ]
    ]
)

print("\nFile:", saved)
print("STATUS: SELESAI | BELUM FUSION")
```

```python
# ==================================================
# JATIKARYA NEW - STEP 10A
# NAIVE FUSION: TRUSTMARK FIRST, ELSE DINOV2
# Pilot 54 gambar generatif, belum kalibrasi
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from IPython.display import display
import pandas as pd
import hashlib
import json

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"

assert all(callable(globals().get(n)) for n in (
    "file_hash", "start_experiment", "record_event"
))

EXP = ROOT / "02_EXPERIMENTS"

# 1. Gunakan hasil yang sudah diverifikasi
P_E = (
    EXP / "STEP_09E_GENERATIVE_COMPARISON"
    / "20261107T190858942693Z_594ec718"
    / "outputs/paired_comparison.csv"
)

P_TM = (
    EXP / "STEP_09C_TRUSTMARK_GENERATIVE"
    / "20261107T185640103435Z_3d710573"
    / "outputs/trustmark_generatif.csv"
)

P_REG = ROOT / "04_LOGS/source_registry.csv"

def read10(p):
    return json.loads(p.read_text(encoding="utf-8"))

for p in (P_E, P_TM, P_REG):
    assert p.is_file(), f"File tidak ditemukan: {p}"

for p in (P_E, P_TM):
    info = read10(p.parent / "summary.json")
    meta = read10(
        p.parent.parent / "metadata.json"
    )

    assert meta["status"] == "COMPLETED"
    assert info["results_sha256"] == file_hash(p)

pairs = pd.read_csv(
    P_E, dtype=str, keep_default_na=False
)

tm = pd.read_csv(
    P_TM, dtype=str, keep_default_na=False
)

registry = pd.read_csv(
    P_REG, dtype=str, keep_default_na=False
)

assert len(pairs) == len(tm) == 54
assert len(registry) == 9
assert registry.source_id.is_unique
assert registry.sha256.is_unique

KEY = [
    "source_id",
    "provider",
    "edit_type",
    "image_sha256"
]

assert not pairs.duplicated(KEY).any()
assert not tm.duplicated(KEY).any()

# 2. Gabungkan hasil dengan payload asli TrustMark
data = pairs.merge(
    tm[KEY + [
        "watermark_detected",
        "decoded_payload",
        "hasil"
    ]],
    on=KEY,
    validate="one_to_one"
)

assert len(data) == 54
assert (
    data["hasil"] == data["hasil_trustmark"]
).all()

# 3. Buat katalog kode watermark
# Berdasarkan SHA-256 sumber, bukan label gambar edit
catalog = {}

for r in registry.itertuples(index=False):
    payload = bin(
        int(r.sha256, 16)
    )[2:].zfill(256)[:61]

    assert payload not in catalog, (
        "Terjadi benturan payload antar sumber."
    )

    catalog[payload] = r.source_id

valid_sources = set(registry.source_id)

assert set(data.source_id) == valid_sources
assert data.source_prediksi_top1.isin(
    valid_sources
).all()

# 4. Tentukan Naive Fusion
# TIDAK menggunakan source_id sebenarnya
# untuk memilih prediksi.
rows = []

for r in data.itertuples(index=False):

    detected = (
        str(r.watermark_detected).lower() == "true"
    )

    tm_source = (
        catalog.get(str(r.decoded_payload))
        if detected else None
    )

    dino_source = r.source_prediksi_top1

    if tm_source is not None:
        predicted = tm_source
        decision = "TRUSTMARK_PAYLOAD"
    else:
        predicted = dino_source
        decision = "DINOV2_TOP1"

    conflict = (
        tm_source is not None
        and tm_source != dino_source
    )

    # Ground truth dipakai HANYA untuk evaluasi.
    rows.append({
        "source_number": r.source_number,
        "source_id_sebenarnya": r.source_id,
        "provider": r.provider,
        "edit_type": r.edit_type,
        "image_sha256": r.image_sha256,
        "hasil_trustmark": r.hasil_trustmark,
        "source_dari_watermark": tm_source or "",
        "source_top1_dinov2": dino_source,
        "source_prediksi_naive": predicted,
        "dasar_keputusan": decision,
        "konflik_tm_dinov2": conflict,
        "dinov2_benar": (
            dino_source == r.source_id
        ),
        "naive_benar": (
            predicted == r.source_id
        )
    })

hasil10 = pd.DataFrame(rows)

assert len(hasil10) == 54

# 5. Identitas percobaan agar tidak membuat duplikat
fingerprint = hashlib.sha256(
    json.dumps({
        "rule": "TM_PAYLOAD_FIRST_ELSE_DINO_V1",
        "comparison": file_hash(P_E),
        "trustmark": file_hash(P_TM),
        "registry": file_hash(P_REG)
    }, sort_keys=True).encode()
).hexdigest()

folder10 = EXP / "STEP_10A_NAIVE_FUSION_PILOT"

saved = None

for sp in sorted(
    folder10.glob("*/outputs/summary.json"),
    reverse=True
):
    try:
        info = read10(sp)
        meta = read10(
            sp.parent.parent / "metadata.json"
        )
        f = (
            sp.parent / "naive_fusion_results.csv"
        )

        if (
            meta.get("status") == "COMPLETED"
            and info.get("fingerprint") == fingerprint
            and f.is_file()
            and file_hash(f) == info["results_sha256"]
        ):
            saved = f
            break

    except (OSError, ValueError, KeyError):
        continue

# 6. Simpan hasil baru jika belum tersedia
if saved is None:

    run10 = start_experiment(
        "STEP_10A_NAIVE_FUSION_PILOT",
        "TRUSTMARK_FIRST_ELSE_DINOV2",
        {"fingerprint": fingerprint}
    )

    meta_path = run10 / "metadata.json"

    try:
        saved = (
            run10 / "outputs/naive_fusion_results.csv"
        )

        hasil10.to_csv(saved, index=False)

        summary = {
            "fingerprint": fingerprint,
            "total_images": len(hasil10),
            "tm_payload_mapped": int(
                hasil10.source_dari_watermark
                .ne("").sum()
            ),
            "conflict_count": int(
                hasil10.konflik_tm_dinov2.sum()
            ),
            "dinov2_correct": int(
                hasil10.dinov2_benar.sum()
            ),
            "naive_correct": int(
                hasil10.naive_benar.sum()
            ),
            "results_sha256": file_hash(saved),
            "pilot_only": True,
            "created_at_utc":
                datetime.now(timezone.utc).isoformat()
        }

        (
            run10 / "outputs/summary.json"
        ).write_text(
            json.dumps(summary, indent=2),
            encoding="utf-8"
        )

        record_event(
            run10,
            "NAIVE_FUSION_PILOT_COMPLETED",
            summary
        )

        meta = read10(meta_path)
        meta["status"] = "COMPLETED"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

    except Exception as exc:

        record_event(
            run10,
            "RUN_FAILED",
            {"error": str(exc)}
        )

        meta = read10(meta_path)
        meta["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

        raise

else:
    print("Hasil arsip digunakan kembali.")

# 7. Ringkasan pengujian
print("\n=== STEP 10A: NAIVE FUSION ===")

print("Gambar:", len(hasil10))

print(
    "Payload TrustMark terpetakan:",
    int(hasil10.source_dari_watermark.ne("").sum())
)

print(
    "Konflik kedua metode:",
    int(hasil10.konflik_tm_dinov2.sum())
)

print(
    "DINOv2 Top-1 benar:",
    int(hasil10.dinov2_benar.sum())
)

print(
    "Naive Fusion benar:",
    int(hasil10.naive_benar.sum())
)

print("\n=== KONFLIK KEDUA METODE ===")

display(
    hasil10.loc[
        hasil10.konflik_tm_dinov2,
        [
            "source_number",
            "provider",
            "edit_type",
            "source_dari_watermark",
            "source_top1_dinov2",
            "source_prediksi_naive",
            "naive_benar"
        ]
    ]
)

print("\n=== KASUS NAIVE FUSION SALAH ===")

display(
    hasil10.loc[
        ~hasil10.naive_benar,
        [
            "source_number",
            "provider",
            "edit_type",
            "source_id_sebenarnya",
            "source_prediksi_naive",
            "dasar_keputusan"
        ]
    ]
)

print("\nFile:", saved)
print("STATUS: PILOT | BELUM CALIBRATED FUSION")
```

```python
# ==================================================
# JATIKARYA NEW - STEP 10B
# AUDIT SKOR & PROTOKOL CALIBRATED FUSION
# Belum melakukan fitting atau penentuan threshold
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from IPython.display import display
import pandas as pd
import numpy as np
import hashlib
import json

assert ROOT.name == "JATIKARYA_NEW"
assert all(callable(globals().get(n)) for n in (
    "file_hash", "start_experiment", "record_event"
))

E = ROOT / "02_EXPERIMENTS"

P = {
    "tm": E / (
        "STEP_09C_TRUSTMARK_GENERATIVE/"
        "20261107T185640103435Z_3d710573/"
        "outputs/trustmark_generatif.csv"
    ),
    "di": E / (
        "STEP_09D_DINOV2_GENERATIVE/"
        "20261107T190219357254Z_ef5bc870/"
        "outputs/dinov2_generatif.csv"
    ),
    "ranking": E / (
        "STEP_09D_DINOV2_GENERATIVE/"
        "20261107T190219357254Z_ef5bc870/"
        "outputs/all_candidate_rankings.csv"
    ),
    "pair": E / (
        "STEP_09E_GENERATIVE_COMPARISON/"
        "20261107T190858942693Z_594ec718/"
        "outputs/paired_comparison.csv"
    ),
    "naive": E / (
        "STEP_10A_NAIVE_FUSION_PILOT/"
        "20261107T191315251795Z_5f097337/"
        "outputs/naive_fusion_results.csv"
    )
}

KEY = [
    "source_id",
    "provider",
    "edit_type",
    "image_sha256"
]

def jload(p):
    return json.loads(
        p.read_text(encoding="utf-8")
    )

# 1. Validasi integritas hasil sebelumnya
for typ in ("tm", "di", "pair", "naive"):
    p = P[typ]

    assert p.is_file(), f"File tidak ada: {p}"

    summ = jload(
        p.parent / "summary.json"
    )

    meta = jload(
        p.parent.parent / "metadata.json"
    )

    assert meta["status"] == "COMPLETED"
    assert summ["results_sha256"] == file_hash(p)

assert P["ranking"].is_file()

assert file_hash(P["ranking"]) == jload(
    P["di"].parent / "summary.json"
)["rankings_sha256"]

D = {
    name: pd.read_csv(
        path,
        dtype=str,
        keep_default_na=False
    )
    for name, path in P.items()
}

D["naive"] = D["naive"].rename(
    columns={"source_id_sebenarnya": "source_id"}
)

assert all(
    len(D[k]) == 54
    for k in ("tm", "di", "pair", "naive")
)

assert len(D["ranking"]) == 486

expected_keys = set(
    map(tuple, D["pair"][KEY].to_numpy())
)

for k in ("tm", "di", "pair", "naive"):
    assert not D[k].duplicated(KEY).any()

    actual_keys = set(
        map(tuple, D[k][KEY].to_numpy())
    )

    assert actual_keys == expected_keys, (
        f"Identitas gambar berbeda: {k}"
    )

pair = D["pair"].copy()
naive = D["naive"].copy()
rank = D["ranking"].copy()

# 2. Pastikan hasil sesuai checkpoint 09E
expected_categories = {
    "TM_TIDAK_COCOK_DINOV2_BENAR": 40,
    "KEDUA_BERHASIL": 12,
    "TM_COCOK_DINOV2_SALAH": 1,
    "KEDUA_BELUM_BERHASIL": 1
}

assert (
    pair["kategori"].value_counts().to_dict()
    == expected_categories
), "Hasil STEP 09E berbeda dari checkpoint."

# 3. Verifikasi 486 similarity
rank["peringkat"] = pd.to_numeric(
    rank["peringkat"],
    errors="raise"
).astype(int)

rank["cosine_similarity"] = pd.to_numeric(
    rank["cosine_similarity"],
    errors="raise"
)

assert np.isfinite(
    rank["cosine_similarity"]
).all()

assert rank["cosine_similarity"].between(
    -1.0001, 1.0001
).all()

GROUP = [
    "source_id_gambar_edit",
    "provider",
    "edit_type",
    "image_sha256"
]

assert (
    rank.groupby(GROUP).size() == 9
).all()

assert len(rank.groupby(GROUP)) == 54

assert rank.groupby(GROUP)["peringkat"].apply(
    lambda x: set(x) == set(range(1, 10))
).all()

catalog = set(
    pd.read_csv(
        ROOT / "04_LOGS/source_registry.csv",
        dtype=str
    )["source_id"]
)

assert len(catalog) == 9

assert rank.groupby(GROUP)[
    "source_id_kandidat"
].apply(
    lambda x: set(x) == catalog
).all()

# 4. Ambil skor kandidat peringkat 1 dan 2
best = rank[
    rank["peringkat"] == 1
].rename(columns={
    "source_id_gambar_edit": "source_id",
    "source_id_kandidat": "top1_terhitung",
    "cosine_similarity": "skor_top1_terhitung"
})[
    KEY + [
        "top1_terhitung",
        "skor_top1_terhitung"
    ]
]

second = rank[
    rank["peringkat"] == 2
].rename(columns={
    "source_id_gambar_edit": "source_id",
    "cosine_similarity": "skor_top2_terhitung"
})[
    KEY + ["skor_top2_terhitung"]
]

audit = pair.merge(
    best,
    on=KEY,
    validate="one_to_one"
).merge(
    second,
    on=KEY,
    validate="one_to_one"
)

audit = audit.merge(
    naive[
        KEY + [
            "source_dari_watermark",
            "source_prediksi_naive",
            "konflik_tm_dinov2",
            "naive_benar"
        ]
    ],
    on=KEY,
    validate="one_to_one"
)

assert len(audit) == 54

assert audit["top1_terhitung"].eq(
    audit["source_prediksi_top1"]
).all()

# 5. Hitung fitur yang relevan untuk kalibrasi
audit["s1"] = audit[
    "skor_top1_terhitung"
].astype(float)

audit["s2"] = audit[
    "skor_top2_terhitung"
].astype(float)

audit["margin"] = (
    audit["s1"] - audit["s2"]
)

assert (audit["margin"] >= -1e-6).all()

assert np.allclose(
    audit["margin"],
    pd.to_numeric(
        audit["margin_top1_top2"]
    ),
    atol=1e-5
)

audit["dino_benar_audit"] = (
    audit["source_prediksi_top1"]
    == audit["source_id"]
)

audit["naive_benar_audit"] = (
    audit["source_prediksi_naive"]
    == audit["source_id"]
)

audit["tm_terpetakan"] = (
    audit["source_dari_watermark"] != ""
)

audit["konflik_audit"] = (
    audit["tm_terpetakan"]
    & (
        audit["source_dari_watermark"]
        != audit["source_prediksi_top1"]
    )
)

# 6. Double-check terhadap hasil sebelumnya
assert audit["tm_terpetakan"].sum() == 13
assert audit["dino_benar_audit"].sum() == 52
assert audit["naive_benar_audit"].sum() == 53
assert audit["konflik_audit"].sum() == 1

assert audit["tm_terpetakan"].eq(
    audit["hasil_trustmark"] == "KODE_COCOK"
).all()

assert audit["konflik_audit"].eq(
    audit["konflik_tm_dinov2"]
    .str.lower().eq("true")
).all()

assert audit["naive_benar_audit"].eq(
    audit["naive_benar"]
    .str.lower().eq("true")
).all()

# 7. Susun laporan audit
cols = [
    "source_number",
    "source_id",
    "provider",
    "edit_type",
    "image_sha256",
    "hasil_trustmark",
    "source_dari_watermark",
    "source_prediksi_top1",
    "source_prediksi_naive",
    "s1",
    "s2",
    "margin",
    "dino_benar_audit",
    "naive_benar_audit",
    "konflik_audit",
    "kategori"
]

report = audit[cols].sort_values(
    ["source_number", "provider", "edit_type"]
)

# 8. Rancangan metode: belum fitting
protocol = {
    "status": "DRAFT_NOT_TRAINED",

    "objective": (
        "Closed-set image provenance: "
        "identify the registered source of "
        "a generatively edited image."
    ),

    "inference_inputs": [
        "TrustMark decoded payload mapped to catalog",
        "DINOv2 top-1 source",
        "DINOv2 top-1 cosine similarity",
        "DINOv2 top-1 minus top-2 margin"
    ],

    "ground_truth_use": (
        "Evaluation only. Never use the true "
        "source to select a prediction."
    ),

    "candidate_rules": [
        (
            "If mapped TrustMark and DINOv2 agree, "
            "accept their source."
        ),
        (
            "If they disagree, flag conflict. "
            "Choose the final action using "
            "development data, not test data."
        ),
        (
            "If TrustMark is unavailable, use "
            "calibrated DINOv2 reliability. "
            "Abstain when evidence is insufficient."
        )
    ],

    "calibration": (
        "Fit and select thresholds on "
        "development data with source-disjoint "
        "splits. Evaluate on untouched test "
        "sources whose references are present "
        "in the fixed catalog."
    ),

    "evaluation_metrics": [
        "Top-1 identification accuracy",
        "Coverage",
        "Selective error rate",
        "Abstention rate",
        "Conflict count",
        "Brier score or calibration error"
    ],

    "limitations": [
        "Only 9 independent source images",
        "Only 2 DINOv2 errors in pilot",
        "No threshold estimated in this step",
        "No calibrated probability estimated",
        "No final performance claim"
    ]
}

# 9. Fingerprint untuk menghindari arsip duplikat
fingerprint = hashlib.sha256(
    json.dumps({
        k: file_hash(v)
        for k, v in P.items()
    }, sort_keys=True).encode()
).hexdigest()

output_root = (
    E / "STEP_10B_CALIBRATION_PROTOCOL"
)

reused = None

for s in sorted(
    output_root.glob("*/outputs/summary.json"),
    reverse=True
):
    try:
        z = jload(s)
        meta = jload(
            s.parent.parent / "metadata.json"
        )

        a = s.parent / "feature_audit.csv"
        b = s.parent / "protocol.json"

        if (
            meta.get("status") == "COMPLETED"
            and z.get("fingerprint") == fingerprint
            and a.is_file()
            and b.is_file()
            and file_hash(a) == z.get("audit_sha256")
            and file_hash(b) == z.get("protocol_sha256")
        ):
            reused = (a, b)
            break

    except (KeyError, OSError, ValueError):
        continue

# 10. Simpan laporan baru jika belum tersedia
if reused:
    f_audit, f_protocol = reused
    print("Arsip STEP 10B valid. Tidak membuat duplikat.")

else:
    run = start_experiment(
        "STEP_10B_CALIBRATION_PROTOCOL",
        "READ_ONLY_SCORE_AUDIT",
        {"fingerprint": fingerprint}
    )

    f_audit = (
        run / "outputs/feature_audit.csv"
    )

    f_protocol = (
        run / "outputs/protocol.json"
    )

    meta_path = run / "metadata.json"

    try:
        f_audit.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        report.to_csv(
            f_audit,
            index=False
        )

        f_protocol.write_text(
            json.dumps(
                protocol,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

        summary = {
            "fingerprint": fingerprint,
            "images": 54,
            "candidate_scores": 486,
            "dino_correct": 52,
            "naive_correct": 53,
            "tm_mapped": 13,
            "conflicts": 1,
            "audit_sha256": file_hash(f_audit),
            "protocol_sha256": file_hash(f_protocol),
            "status": "AUDIT_ONLY_NOT_CALIBRATED",
            "created_at_utc":
                datetime.now(
                    timezone.utc
                ).isoformat()
        }

        (
            run / "outputs/summary.json"
        ).write_text(
            json.dumps(
                summary,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

        record_event(
            run,
            "CALIBRATION_PROTOCOL_AUDITED",
            summary
        )

        meta = jload(meta_path)
        meta["status"] = "COMPLETED"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )

    except Exception:
        meta = jload(meta_path)
        meta["status"] = "ERROR"

        meta_path.write_text(
            json.dumps(meta, indent=2),
            encoding="utf-8"
        )
        raise

# 11. Tampilkan hasil audit
print("\n=== STEP 10B: AUDIT SKOR ===")
print("Gambar:", len(audit))
print("Skor kandidat:", len(rank))
print("TrustMark terpetakan:", int(audit.tm_terpetakan.sum()))
print("Konflik:", int(audit.konflik_audit.sum()))

print("\n=== DISTRIBUSI SKOR DINOV2 ===")
print("Deskriptif saja, bukan ambang kalibrasi.")

display(
    audit.groupby("dino_benar_audit")[
        ["s1", "margin"]
    ].agg([
        "count", "min", "median", "max"
    ]).round(5)
)

print("\n=== KASUS DINOV2 SALAH ===")

display(
    report.loc[
        ~report["dino_benar_audit"],
        [
            "source_number",
            "provider",
            "edit_type",
            "hasil_trustmark",
            "s1",
            "s2",
            "margin",
            "source_prediksi_naive",
            "naive_benar_audit"
        ]
    ]
)

print("\n=== LOKASI HASIL ===")
print("Audit:", f_audit)
print("Protokol:", f_protocol)
print(
    "STATUS: AUDIT SELESAI | "
    "BELUM CALIBRATED FUSION"
)
```

```python
# ==================================================
# JATIKARYA NEW - RESTORE CONFIGURATION
# Mengembalikan ROOT dan helper eksperimen
# Tanpa mengulang STEP 01-10B
# ==================================================

from google.colab import drive
from pathlib import Path
from datetime import datetime, timezone
import hashlib
import json
import uuid

# 1. Hubungkan Google Drive
drive.mount("/content/drive")

# 2. Kembalikan lokasi proyek
ROOT = Path(
    "/content/drive/MyDrive/"
    "PROJECT JURNAL - JATIKARYA/"
    "JATIKARYA_NEW"
)

assert ROOT.is_dir(), (
    "Folder JATIKARYA_NEW tidak ditemukan."
)

assert (
    ROOT / "04_LOGS/source_registry.csv"
).is_file(), "Registry sumber sebelumnya tidak ditemukan."

# 3. Kembalikan helper SHA-256 jika belum ada
if not callable(globals().get("file_hash")):

    def file_hash(path):
        h = hashlib.sha256()

        with open(path, "rb") as f:
            for chunk in iter(
                lambda: f.read(1024 * 1024),
                b""
            ):
                h.update(chunk)

        return h.hexdigest()

# 4. Kembalikan helper eksperimen
if not callable(globals().get("start_experiment")):

    def start_experiment(step, method, parameters=None):

        stamp = datetime.now(
            timezone.utc
        ).strftime("%Y%m%dT%H%M%S%fZ")

        run_id = (
            f"{stamp}_{uuid.uuid4().hex[:8]}"
        )

        run = (
            ROOT / "02_EXPERIMENTS"
            / step / run_id
        )

        run.mkdir(parents=True, exist_ok=False)
        (run / "outputs").mkdir()
        (run / "logs").mkdir()

        metadata = {
            "run_id": run_id,
            "step": step,
            "method": method,
            "parameters": parameters or {},
            "status": "RUNNING",
            "created_at_utc": datetime.now(
                timezone.utc
            ).isoformat()
        }

        (run / "metadata.json").write_text(
            json.dumps(
                metadata,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

        return run

# 5. Kembalikan helper audit log
if not callable(globals().get("record_event")):

    def record_event(run, event, details=None):

        record = {
            "timestamp_utc": datetime.now(
                timezone.utc
            ).isoformat(),
            "event": event,
            "details": details or {}
        }

        log_path = Path(run) / "logs/events.jsonl"
        log_path.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        with log_path.open(
            "a", encoding="utf-8"
        ) as f:
            f.write(
                json.dumps(
                    record,
                    ensure_ascii=False
                ) + "\n"
            )

# 6. Verifikasi
print("=== KONFIGURASI JATIKARYA ===")
print("ROOT:", ROOT)
print("Folder ditemukan:", ROOT.is_dir())
print(
    "Registry ditemukan:",
    (ROOT / "04_LOGS/source_registry.csv").is_file()
)
print("Helper SHA-256:", callable(file_hash))
print(
    "Helper eksperimen:",
    callable(start_experiment)
)
print("Helper audit:", callable(record_event))
print("STATUS: SIAP MELANJUTKAN STEP 11A")
```

```python
# ==================================================
# JATIKARYA NEW - STEP 11A
# UPLOAD LANGSUNG 40 FOTO SUMBER BARU
# 20 CALIBRATION + 20 FINAL_TEST
# TANPA ZIP, WATERMARK, ATAU EDIT AI
# ==================================================

from pathlib import Path
from io import BytesIO
from datetime import datetime, timezone
from google.colab import files
from IPython.display import display
from PIL import Image, ImageOps, ImageDraw
import pandas as pd
import hashlib
import json

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"
assert all(callable(globals().get(k)) for k in (
    "file_hash", "start_experiment", "record_event"
))

# 1. Pastikan data pilot tetap utuh
pilot_path = ROOT / "04_LOGS/source_registry.csv"

pilot = pd.read_csv(
    pilot_path,
    dtype=str,
    keep_default_na=False
)

assert len(pilot) == 9
assert pilot.source_id.is_unique
assert pilot.sha256.is_unique

# 2. Upload 40 foto langsung
print("Pilih seluruh 40 foto asli sekaligus.")
print("Gunakan Ctrl + A saat memilih file.")

uploaded11a = files.upload()

assert len(uploaded11a) == 40, (
    f"Diterima {len(uploaded11a)} gambar. "
    "Diperlukan tepat 40 gambar."
)

# 3. Validasi gambar dan identitas
items = []
known_sha = set(pilot.sha256)
known_id = set(pilot.source_id)
seen_names = set()

for filename, raw in sorted(uploaded11a.items()):

    assert Path(filename).name == filename
    assert filename.casefold() not in seen_names
    seen_names.add(filename.casefold())

    assert filename.lower().endswith(
        (".jpg", ".jpeg", ".png", ".webp")
    ), f"Ekstensi tidak didukung: {filename}"

    sha = hashlib.sha256(raw).hexdigest()
    sid = "SRC_" + sha[:16]

    assert sha not in known_sha, (
        f"Foto duplikat: {filename}"
    )
    assert sid not in known_id, (
        f"Source ID bentrok: {filename}"
    )

    known_sha.add(sha)
    known_id.add(sid)

    with Image.open(BytesIO(raw)) as img:
        fmt = img.format
        width, height = img.size
        frames = getattr(img, "n_frames", 1)
        img.seek(0)
        img.load()

    assert fmt in ("JPEG", "MPO", "PNG", "WEBP")

    items.append({
        "source_id": sid,
        "original_filename": filename,
        "sha256": sha,
        "image_format": fmt,
        "frames": frames,
        "processing_frame_index": 0,
        "width": width,
        "height": height,
        "_raw": raw
    })

assert len(items) == 40

# 4. Nomor stabil untuk referensi pengeditan AI
items.sort(
    key=lambda x: (
        x["original_filename"].casefold(),
        x["sha256"]
    )
)

for no, item in enumerate(items, 1):
    item["source_number"] = f"{no:02d}"

# 5. Pembagian deterministik 20 + 20
split_order = sorted(
    items,
    key=lambda x: hashlib.sha256(
        ("JATIKARYA11A_V1_" + x["sha256"]).encode()
    ).hexdigest()
)

for idx, item in enumerate(split_order):
    item["split"] = (
        "CALIBRATION"
        if idx < 20
        else "FINAL_TEST"
    )

for item in items:
    item["stored_path"] = str(
        Path("01_DATASET/INDEPENDENT_ORIGINALS")
        / item["split"]
        / item["source_id"]
        / item["original_filename"]
    )

# 6. Pratinjau seluruh sumber
preview = pd.DataFrame([
    {
        k: v for k, v in item.items()
        if k != "_raw"
    }
    for item in items
])

print("\n=== PRATINJAU STEP 11A ===")
print("Total:", len(preview))
print(preview["split"].value_counts().to_string())

display(preview[[
    "source_number",
    "original_filename",
    "source_id",
    "split",
    "image_format",
    "frames",
    "width",
    "height"
]])

# Pratinjau visual agar bisa mengecek foto serupa
CW, CH, NC = 180, 165, 5
sheet = Image.new(
    "RGB",
    (CW * NC, CH * 8),
    "white"
)
draw = ImageDraw.Draw(sheet)

for i, item in enumerate(items):
    with Image.open(BytesIO(item["_raw"])) as img:
        img.seek(0)
        rgb = ImageOps.exif_transpose(
            img
        ).convert("RGB")

        thumb = ImageOps.contain(
            rgb, (CW - 12, CH - 45)
        )

    x = (i % NC) * CW
    y = (i // NC) * CH

    sheet.paste(
        thumb,
        (
            x + (CW - thumb.width) // 2,
            y
        )
    )

    draw.text(
        (x + 5, y + CH - 38),
        f"{item['source_number']} | {item['split']}",
        fill="black"
    )

    draw.text(
        (x + 5, y + CH - 22),
        item["original_filename"][:22],
        fill="black"
    )

display(sheet)

# 7. Cek apakah batch sudah pernah terdaftar
fingerprint = hashlib.sha256(
    "\n".join(
        sorted(x["sha256"] for x in items)
    ).encode()
).hexdigest()

step_dir = (
    ROOT / "02_EXPERIMENTS/"
    "STEP_11A_INDEPENDENT_SOURCE_SPLIT"
)

existing_manifest = None

for p in sorted(
    step_dir.glob(
        "*/outputs/source_split_manifest.csv"
    ),
    reverse=True
):
    try:
        summary = json.loads(
            (p.parent / "summary.json").read_text(
                encoding="utf-8"
            )
        )
        meta = json.loads(
            (p.parent.parent / "metadata.json").read_text(
                encoding="utf-8"
            )
        )

        if not (
            meta.get("status") == "PASS"
            and summary.get("dataset_fingerprint")
                == fingerprint
            and summary.get("manifest_sha256")
                == file_hash(p)
        ):
            continue

        old = pd.read_csv(
            p, dtype=str, keep_default_na=False
        )

        if (
            len(old) == 40
            and set(old.sha256)
                == {x["sha256"] for x in items}
            and all(
                (ROOT / r.stored_path).is_file()
                and file_hash(ROOT / r.stored_path)
                    == r.sha256
                for r in old.itertuples(index=False)
            )
        ):
            existing_manifest = p
            break

    except (OSError, ValueError, KeyError):
        continue

# 8. Konfirmasi sebelum menyimpan
if existing_manifest:

    print("\nDataset sudah terdaftar sebelumnya.")
    print("Tidak menyimpan duplikat.")
    print("Manifest:", existing_manifest)

else:

    print(
        "\nPeriksa pembagian dan foto yang "
        "terlihat sangat mirip sebelum menyimpan."
    )

    approval = input(
        "Ketik SIMPAN untuk menyimpan 40 sumber: "
    ).strip()

    if approval != "SIMPAN":
        print("DIBATALKAN: Tidak disimpan.")

    else:
        run = start_experiment(
            "STEP_11A_INDEPENDENT_SOURCE_SPLIT",
            "DIRECT_IMAGE_SOURCE_REGISTRATION",
            {
                "dataset_fingerprint": fingerprint,
                "source_count": 40,
                "calibration_count": 20,
                "final_test_count": 20
            }
        )

        meta_path = run / "metadata.json"

        try:
            allowed = (
                ROOT / "01_DATASET/"
                "INDEPENDENT_ORIGINALS"
            ).resolve()

            # Simpan byte asli tanpa konversi
            for item in items:
                dst = (
                    ROOT / item["stored_path"]
                ).resolve()

                assert dst.is_relative_to(allowed)
                dst.parent.mkdir(
                    parents=True,
                    exist_ok=True
                )

                if dst.exists():
                    assert file_hash(dst) == item["sha256"]
                else:
                    with dst.open("xb") as f:
                        f.write(item["_raw"])

                assert file_hash(dst) == item["sha256"]

            out = run / "outputs"
            out.mkdir(parents=True, exist_ok=True)

            manifest_path = (
                out / "source_split_manifest.csv"
            )

            preview.to_csv(manifest_path, index=False)

            protocol = {
                "status": "SPLIT_LOCKED",
                "sources": 40,
                "calibration": 20,
                "final_test": 20,
                "source_disjoint": True,
                "split_rule": (
                    "SHA256(JATIKARYA11A_V1_ + "
                    "source_sha256)"
                ),
                "input_policy": "DIRECT_IMAGES",
                "preserve_original_bytes": True,
                "mpo_processing_frame": 0,
                "pilot_unchanged": True,
                "dataset_fingerprint": fingerprint,
                "manifest_sha256":
                    file_hash(manifest_path),
                "created_at_utc":
                    datetime.now(
                        timezone.utc
                    ).isoformat()
            }

            protocol_path = (
                out / "split_protocol.json"
            )

            protocol_path.write_text(
                json.dumps(
                    protocol,
                    indent=2,
                    ensure_ascii=False
                ),
                encoding="utf-8"
            )

            summary = {
                "dataset_fingerprint": fingerprint,
                "manifest_sha256":
                    file_hash(manifest_path),
                "protocol_sha256":
                    file_hash(protocol_path),
                "total_sources": 40,
                "calibration": 20,
                "final_test": 20
            }

            (
                out / "summary.json"
            ).write_text(
                json.dumps(summary, indent=2),
                encoding="utf-8"
            )

            record_event(
                run,
                "SOURCES_REGISTERED",
                summary
            )

            meta = json.loads(
                meta_path.read_text(encoding="utf-8")
            )
            meta["status"] = "PASS"

            meta_path.write_text(
                json.dumps(meta, indent=2),
                encoding="utf-8"
            )

            print("\n=== STEP 11A SELESAI ===")
            print("STATUS: PASS")
            print("Total sumber:", 40)
            print("CALIBRATION:", 20)
            print("FINAL_TEST:", 20)
            print("Manifest:", manifest_path)
            print("Tidak ada ZIP yang dibuat.")

        except Exception as exc:
            meta = json.loads(
                meta_path.read_text(encoding="utf-8")
            )
            meta["status"] = "ERROR"

            meta_path.write_text(
                json.dumps(meta, indent=2),
                encoding="utf-8"
            )

            record_event(
                run,
                "RUN_FAILED",
                {"error": str(exc)}
            )
            raise
```

```python
# ====================================================
# JATIKARYA NEW - STEP 11B
# 40 SOURCE: TRUSTMARK Q BCH_5 + DINOV2 ViT-S/14
# Tidak mengubah data pilot, tidak membuat ZIP
# ====================================================

from pathlib import Path
from datetime import datetime, timezone
from importlib.metadata import version
import importlib.util, subprocess, sys
import hashlib, json, time
import numpy as np
import pandas as pd
import torch
import torch.nn.functional as F
from PIL import Image, ImageOps
from torchvision import transforms

assert ROOT.name == "JATIKARYA_NEW"
assert ROOT.parent.name == "PROJECT JURNAL - JATIKARYA"
assert all(callable(globals().get(k)) for k in (
    "file_hash", "start_experiment", "record_event"
))

# 1. Siapkan TrustMark versi eksperimen sebelumnya
if importlib.util.find_spec("trustmark") is None:
    subprocess.check_call([
        sys.executable, "-m", "pip",
        "install", "trustmark==0.9.2"
    ])

from trustmark import TrustMark

assert version("trustmark") == "0.9.2", (
    "Versi TrustMark berbeda dari pilot."
)

# 2. Baca manifest STEP 11A yang sudah PASS
MANIFEST = (
    ROOT / "02_EXPERIMENTS"
    / "STEP_11A_INDEPENDENT_SOURCE_SPLIT"
    / "20261108T082931610974Z_ae5c16c6"
    / "outputs/source_split_manifest.csv"
)

assert MANIFEST.is_file()

def read_json(path):
    return json.loads(
        path.read_text(encoding="utf-8")
    )

meta11a = read_json(
    MANIFEST.parent.parent / "metadata.json"
)

summary11a = read_json(
    MANIFEST.parent / "summary.json"
)

assert meta11a["status"] == "PASS"
assert summary11a["manifest_sha256"] == file_hash(MANIFEST)

sources = pd.read_csv(
    MANIFEST, dtype=str, keep_default_na=False
)

assert len(sources) == 40
assert sources.source_id.is_unique
assert sources.sha256.is_unique
assert sources.source_number.is_unique
assert sources.stored_path.is_unique
assert sources.split.value_counts().to_dict() == {
    "CALIBRATION": 20,
    "FINAL_TEST": 20
}

# 3. Validasi keaslian 40 sumber dan data pilot
pilot_file = ROOT / "04_LOGS/source_registry.csv"

pilot = pd.read_csv(
    pilot_file, dtype=str, keep_default_na=False
)

assert len(pilot) == 9
assert set(pilot.sha256).isdisjoint(
    set(sources.sha256)
)

pilot_hash = file_hash(pilot_file)

original_root = (
    ROOT / "01_DATASET/INDEPENDENT_ORIGINALS"
).resolve()

for r in sources.itertuples(index=False):
    src = (ROOT / r.stored_path).resolve()

    assert src.is_relative_to(original_root)
    assert src.is_file()
    assert file_hash(src) == r.sha256
    assert r.source_id == "SRC_" + r.sha256[:16]
    assert r.processing_frame_index == "0"

# Cek benturan payload termasuk sembilan sumber pilot
all_sha = list(pilot.sha256) + list(sources.sha256)

all_payloads = [
    bin(int(s, 16))[2:].zfill(256)[:61]
    for s in all_sha
]

assert len(set(all_payloads)) == 49, (
    "Ada benturan payload. Hentikan encoding."
)

# 4. Inisialisasi DINOv2 sesuai eksperimen pilot
DEVICE = (
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

if (
    "model05a" in globals()
    and getattr(model05a, "embed_dim", None) == 384
):
    dino = model05a
else:
    dino = torch.hub.load(
        "facebookresearch/dinov2",
        "dinov2_vits14",
        pretrained=True,
        trust_repo=True
    )

assert getattr(dino, "embed_dim", None) == 384

model05a = dino.to(DEVICE).eval()

preprocess = transforms.Compose([
    transforms.Resize(518),
    transforms.CenterCrop(518),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

def embedding_source(path):
    with Image.open(path) as img:
        img.seek(0)
        rgb = ImageOps.exif_transpose(
            img
        ).convert("RGB")

        x = preprocess(rgb).unsqueeze(0).to(DEVICE)

    with torch.inference_mode():
        emb = F.normalize(
            model05a(x),
            p=2,
            dim=1
        )

    arr = (
        emb[0].detach().cpu().numpy()
        .astype(np.float32)
    )

    assert arr.shape == (384,)
    assert np.isfinite(arr).all()
    assert np.isclose(
        np.linalg.norm(arr), 1, atol=1e-3
    )

    return arr

# 5. Verifikasi model terhadap embedding pilot
proof_found = False

ref_dir = (
    ROOT / "02_EXPERIMENTS"
    / "STEP_05B_DINOV2_ORIGINAL"
)

for rp in sorted(
    ref_dir.glob("*/outputs/reference_result.json"),
    reverse=True
):
    try:
        ref = read_json(rp)
        sid = ref.get("source_id")

        if not (
            sid in set(pilot.source_id)
            and ref.get("model") == "dinov2_vits14"
            and ref.get("preprocessing")
                == "Resize518_CenterCrop518_ImageNet"
        ):
            continue

        old_emb_path = (
            rp.parent / "embedding_original.npy"
        )

        if not (
            old_emb_path.is_file()
            and file_hash(old_emb_path)
                == ref.get("embedding_sha256")
        ):
            continue

        old_emb = np.load(
            old_emb_path,
            allow_pickle=False
        ).reshape(-1)

        old_source = pilot.loc[
            pilot.source_id == sid
        ].iloc[0]

        new_emb = embedding_source(
            ROOT / old_source.stored_path
        )

        assert np.allclose(
            old_emb, new_emb,
            atol=2e-3, rtol=2e-3
        ), "Model DINOv2 tidak sesuai referensi pilot."

        proof_found = True
        break

    except (OSError, ValueError, KeyError, TypeError):
        continue

assert proof_found, (
    "Referensi DINOv2 pilot tidak ditemukan "
    "atau tidak sesuai. Proses dihentikan."
)

# 6. Inisialisasi TrustMark
tm = TrustMark(
    verbose=False,
    model_type="Q",
    encoding_type=TrustMark.Encoding.BCH_5,
    device=DEVICE
)

# 7. Fingerprint dan checkpoint
config = {
    "input_manifest_sha256": file_hash(MANIFEST),
    "trustmark_version": version("trustmark"),
    "trustmark_mode": "Q_BCH5_BINARY_61",
    "dinov2_model": "dinov2_vits14",
    "preprocessing": "Resize518_CenterCrop518_ImageNet",
    "wm_max_dimension": 768,
    "source_frame": 0
}

fingerprint = hashlib.sha256(
    json.dumps(config, sort_keys=True).encode()
).hexdigest()

STEP = "STEP_11B_WM_DINOV2_BASELINES"
step_root = ROOT / "02_EXPERIMENTS" / STEP

run = None

for mp in sorted(
    step_root.glob("*/metadata.json"),
    reverse=True
):
    try:
        old = read_json(mp)

        if (
            old.get("parameters", {}).get("fingerprint")
            == fingerprint
        ):
            run = mp.parent
            break
    except (OSError, ValueError, KeyError, TypeError):
        continue

if run is None:
    run = start_experiment(
        STEP,
        "TRUSTMARK_AND_DINOV2_SOURCE_BASELINES",
        {
            "fingerprint": fingerprint,
            **config
        }
    )
else:
    print("Run sebelumnya ditemukan. Melanjutkan checkpoint.")

out = run / "outputs"
records_dir = out / "records"
emb_dir = out / "embeddings"

records_dir.mkdir(parents=True, exist_ok=True)
emb_dir.mkdir(parents=True, exist_ok=True)

# Semua 40 gambar WM dalam satu folder, tanpa ZIP
wm_folder = (
    ROOT / "01_DATASET/INDEPENDENT_WATERMARKED"
    / run.name
)

wm_folder.mkdir(parents=True, exist_ok=True)

# 8. Validasi checkpoint sebelum menggunakan kembali
def valid_record(rec, row):
    try:
        if not (
            rec["status"] == "PASS"
            and rec["source_id"] == row.source_id
            and rec["source_sha256"] == row.sha256
            and rec["source_number"] == row.source_number
            and rec["split"] == row.split
            and rec["payload_match"] is True
            and rec["embedding_dimension"] == 384
        ):
            return False

        wm = (ROOT / rec["watermarked_path"]).resolve()
        em = (ROOT / rec["embedding_path"]).resolve()

        if not (
            wm.is_relative_to(wm_folder.resolve())
            and em.is_relative_to(emb_dir.resolve())
        ):
            return False

        if (
            file_hash(wm) != rec["watermarked_sha256"]
            or file_hash(em) != rec["embedding_sha256"]
        ):
            return False

        v = np.load(em, allow_pickle=False)

        return bool(
            v.shape == (384,)
            and np.isfinite(v).all()
            and np.isclose(
                np.linalg.norm(v), 1, atol=1e-3
            )
        )

    except (
        OSError, ValueError,
        KeyError, TypeError
    ):
        return False

# 9. Proses setiap sumber
results = []
errors = []

print("\n=== STEP 11B: PROSES 40 SUMBER ===")
print("Device:", DEVICE)

for i, r in enumerate(sources.itertuples(index=False), 1):

    record_path = (
        records_dir / f"{r.source_id}.json"
    )

    if record_path.is_file():
        try:
            cached = read_json(record_path)

            if valid_record(cached, r):
                results.append(cached)
                print(
                    f"[{i:02d}/40] {r.source_number} "
                    "PASS (checkpoint)"
                )
                continue

        except (OSError, ValueError):
            pass

    try:
        started = time.perf_counter()
        original = ROOT / r.stored_path

        payload = bin(
            int(r.sha256, 16)
        )[2:].zfill(256)[:61]

        # A. Encoding watermark
        wm_path = (
            wm_folder
            / f"{r.source_number}_{r.source_id}_WM.png"
        )

        if not wm_path.exists():
            with Image.open(original) as img:
                img.seek(0)
                rgb = ImageOps.exif_transpose(
                    img
                ).convert("RGB")

            work = rgb.copy()
            work.thumbnail(
                (768, 768),
                Image.Resampling.LANCZOS
            )

            wm_image = tm.encode(
                work,
                payload,
                MODE="binary"
            )

            assert isinstance(wm_image, Image.Image)

            decoded, detected, schema = tm.decode(
                wm_image,
                MODE="binary"
            )

            assert (
                detected
                and decoded == payload
                and schema == TrustMark.Encoding.BCH_5
            ), "Watermark gagal diverifikasi setelah encode."

            with wm_path.open("xb") as f:
                wm_image.save(f, format="PNG")

        # B. Verifikasi setelah disimpan
        with Image.open(wm_path) as check:
            assert check.format == "PNG"

            wm_width, wm_height = check.size

            decoded, detected, schema = tm.decode(
                check.convert("RGB"),
                MODE="binary"
            )

        assert (
            detected
            and decoded == payload
            and schema == TrustMark.Encoding.BCH_5
        ), "Watermark tersimpan tidak valid."

        wm_sha = file_hash(wm_path)

        # C. Embedding dari foto asli
        emb_path = emb_dir / f"{r.source_id}.npy"

        if not emb_path.exists():
            embedding = embedding_source(original)

            with emb_path.open("xb") as f:
                np.save(
                    f, embedding,
                    allow_pickle=False
                )

        embedding = np.load(
            emb_path,
            allow_pickle=False
        )

        assert embedding.shape == (384,)
        assert np.isfinite(embedding).all()
        assert np.isclose(
            np.linalg.norm(embedding),
            1,
            atol=1e-3
        )

        emb_sha = file_hash(emb_path)

        # D. Catatan lengkap
        record = {
            "source_number": r.source_number,
            "source_id": r.source_id,
            "split": r.split,
            "source_filename": r.original_filename,
            "source_sha256": r.sha256,
            "source_path": r.stored_path,
            "parent_type": "ORIGINAL",
            "watermarked_path": str(
                wm_path.relative_to(ROOT)
            ),
            "watermarked_sha256": wm_sha,
            "payload_bits": payload,
            "watermark_detected": True,
            "payload_match": True,
            "watermarked_width": wm_width,
            "watermarked_height": wm_height,
            "embedding_path": str(
                emb_path.relative_to(ROOT)
            ),
            "embedding_sha256": emb_sha,
            "embedding_dimension": 384,
            "duration_seconds": round(
                time.perf_counter() - started, 3
            ),
            "status": "PASS"
        }

        record_path.write_text(
            json.dumps(
                record,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

        record_event(
            run,
            "SOURCE_BASELINE_PASS",
            {
                "source_id": r.source_id,
                "watermarked_sha256": wm_sha,
                "embedding_sha256": emb_sha
            }
        )

        results.append(record)

        print(
            f"[{i:02d}/40] "
            f"{r.source_number} | {r.split}: PASS"
        )

    except Exception as exc:
        error = {
            "source_id": r.source_id,
            "error": f"{type(exc).__name__}: {exc}"
        }

        errors.append(error)

        record_event(
            run, "SOURCE_BASELINE_ERROR", error
        )

        print(
            f"[{i:02d}/40] "
            f"{r.source_number}: ERROR | {exc}"
        )

# 10. Simpan manifest dan hasil final
result_path = (
    out / "source_baseline_manifest.csv"
)

pd.DataFrame(results).to_csv(
    result_path,
    index=False
)

summary = {
    "fingerprint": fingerprint,
    "input_manifest_sha256": file_hash(MANIFEST),
    "trustmark_version": version("trustmark"),
    "device": DEVICE,
    "expected_sources": 40,
    "passed_sources": len(results),
    "calibration_passed": sum(
        x["split"] == "CALIBRATION"
        for x in results
    ),
    "final_test_passed": sum(
        x["split"] == "FINAL_TEST"
        for x in results
    ),
    "error_count": len(errors),
    "errors": errors,
    "manifest_sha256": file_hash(result_path),
    "status": (
        "PASS"
        if len(results) == 40 and not errors
        else "INCOMPLETE"
    ),
    "finished_at_utc": datetime.now(
        timezone.utc
    ).isoformat()
}

(out / "summary.json").write_text(
    json.dumps(
        summary,
        ensure_ascii=False,
        indent=2
    ),
    encoding="utf-8"
)

meta_path = run / "metadata.json"
meta = read_json(meta_path)

meta["status"] = summary["status"]

meta_path.write_text(
    json.dumps(meta, indent=2),
    encoding="utf-8"
)

record_event(
    run,
    "STEP_11B_CHECKPOINT",
    summary
)

# 11. Pastikan data pilot tidak berubah
assert file_hash(pilot_file) == pilot_hash

print("\n=== STEP 11B: RINGKASAN ===")
print("STATUS:", summary["status"])
print("WM + Embedding valid:", len(results), "/40")
print(
    "CALIBRATION:",
    summary["calibration_passed"], "/20"
)
print(
    "FINAL_TEST:",
    summary["final_test_passed"], "/20"
)
print("Error:", len(errors))
print("Folder WM:", wm_folder)
print("Manifest:", result_path)
print("Tidak ada ZIP yang dibuat.")

if summary["status"] != "PASS":
    print(
        "Belum siap untuk pengeditan AI. "
        "Periksa error terlebih dahulu."
    )
```

```python
# =====================================================
# JATIKARYA NEW
# DOWNLOAD 40 FOTO WATERMARKED HASIL STEP 11B
# Tidak mengubah gambar, tidak mengulang eksperimen
# =====================================================

from google.colab import drive, files
from pathlib import Path
import pandas as pd
import hashlib
import json
import zipfile

# 1. Hubungkan Google Drive bila diperlukan
if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

ROOT = Path(
    "/content/drive/MyDrive/"
    "PROJECT JURNAL - JATIKARYA/JATIKARYA_NEW"
)

assert ROOT.is_dir(), "Folder proyek tidak ditemukan."

def sha256_file(path):
    h = hashlib.sha256()
    with open(path, "rb") as f:
        for chunk in iter(
            lambda: f.read(1024 * 1024), b""
        ):
            h.update(chunk)
    return h.hexdigest()

# 2. Manifest STEP 11B yang sudah PASS
run_id = "20261108T083639291045Z_5876dadc"

run_dir = (
    ROOT / "02_EXPERIMENTS"
    / "STEP_11B_WM_DINOV2_BASELINES"
    / run_id
)

manifest_path = (
    run_dir / "outputs/source_baseline_manifest.csv"
)

summary_path = run_dir / "outputs/summary.json"
metadata_path = run_dir / "metadata.json"

assert manifest_path.is_file()
assert summary_path.is_file()
assert metadata_path.is_file()

summary = json.loads(
    summary_path.read_text(encoding="utf-8")
)
metadata = json.loads(
    metadata_path.read_text(encoding="utf-8")
)

assert summary["status"] == "PASS"
assert metadata["status"] == "PASS"
assert summary["passed_sources"] == 40
assert summary["error_count"] == 0
assert (
    sha256_file(manifest_path)
    == summary["manifest_sha256"]
)

df = pd.read_csv(
    manifest_path,
    dtype=str,
    keep_default_na=False
)

assert len(df) == 40
assert df["source_id"].is_unique
assert df["source_number"].is_unique
assert df["status"].eq("PASS").all()
assert df["payload_match"].str.lower().eq("true").all()

assert df["split"].value_counts().to_dict() == {
    "CALIBRATION": 20,
    "FINAL_TEST": 20
}

# 3. Verifikasi file WM satu per satu
wm_root = (
    ROOT / "01_DATASET/INDEPENDENT_WATERMARKED"
    / run_id
).resolve()

files_to_download = []

for row in df.itertuples(index=False):
    path = (ROOT / row.watermarked_path).resolve()

    assert path.is_relative_to(wm_root)
    assert path.is_file(), f"File tidak ditemukan: {path}"

    assert sha256_file(path) == row.watermarked_sha256, (
        f"Hash berbeda: {row.source_number}"
    )

    files_to_download.append((
        path,
        path.name
    ))

assert len(files_to_download) == 40
assert len({
    name for _, name in files_to_download
}) == 40

# 4. Buat arsip sementara untuk download
zip_path = Path(
    "/content/JATIKARYA_40_FOTO_WM_STEP11B.zip"
)

with zipfile.ZipFile(
    zip_path,
    "w",
    compression=zipfile.ZIP_DEFLATED,
    compresslevel=4
) as z:
    for path, filename in files_to_download:
        z.write(
            path,
            arcname=f"FOTO_WM/{filename}"
        )

    z.write(
        manifest_path,
        arcname="source_baseline_manifest.csv"
    )

# 5. Periksa ZIP sebelum diunduh
with zipfile.ZipFile(zip_path) as z:
    assert z.testzip() is None

    image_count = sum(
        name.startswith("FOTO_WM/")
        and name.endswith(".png")
        for name in z.namelist()
    )

    assert image_count == 40

print("=== DOWNLOAD FOTO WM JATIKARYA ===")
print("STATUS: SIAP DOWNLOAD")
print("Foto WM tervalidasi:", image_count)
print("CALIBRATION:", 20)
print("FINAL_TEST:", 20)
print(
    "Ukuran ZIP:",
    round(zip_path.stat().st_size / 1024**2, 2),
    "MB"
)
print("Lokasi:", zip_path)
print("Data penelitian tidak diubah.")

# 6. Download ke laptop
files.download(str(zip_path))
```

```python
# ==================================================
# JATIKARYA NEW - STEP 11D-A
# UPLOAD 120 CHATGPT, KEMUDIAN 120 GEMINI
# Tanpa rename | Simpan asli | Buat galeri audit
# ==================================================

from pathlib import Path
from io import BytesIO
from datetime import datetime, timezone
from google.colab import drive, files
from IPython.display import display
from PIL import Image, ImageOps, ImageDraw
import pandas as pd
import hashlib
import json
import uuid

# Ganti menjadi "gemini" untuk batch kedua
PROVIDER = "chatgpt"

assert PROVIDER in ("chatgpt", "gemini")

# 1. Hubungkan Google Drive
if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

ROOT = Path(
    "/content/drive/MyDrive/"
    "PROJECT JURNAL - JATIKARYA/JATIKARYA_NEW"
)

assert ROOT.is_dir(), "Folder proyek tidak ditemukan."

# 2. Periksa 40 sumber ber-watermark
RUN11B = "20261108T083639291045Z_5876dadc"

MAN11B = (
    ROOT / "02_EXPERIMENTS"
    / "STEP_11B_WM_DINOV2_BASELINES"
    / RUN11B
    / "outputs/source_baseline_manifest.csv"
)

assert MAN11B.is_file()

summary = json.loads(
    (MAN11B.parent / "summary.json").read_text(
        encoding="utf-8"
    )
)

assert summary["status"] == "PASS"
assert summary["passed_sources"] == 40

baseline = pd.read_csv(
    MAN11B,
    dtype=str,
    keep_default_na=False
)

assert len(baseline) == 40
assert baseline.source_id.is_unique
assert baseline.source_number.is_unique

assert set(baseline.source_number) == {
    f"{i:02d}" for i in range(1, 41)
}

# 3. Folder penyimpanan
BASE = (
    ROOT / "02_EXPERIMENTS"
    / "STEP_11D_GENERATIVE_INTAKE"
)

STAGE = (
    ROOT / "01_DATASET"
    / "GENERATIVE_EDITS_STAGING"
)

INDEX = BASE / "staging_index.csv"

FIELDS = [
    "upload_id",
    "original_filename",
    "sha256",
    "actual_format",
    "width",
    "height",
    "stored_path",
    "provider_hint",
    "provider_verified",
    "edit_type",
    "source_number",
    "source_id",
    "label_status"
]

def sha_file(p):
    h = hashlib.sha256()

    with p.open("rb") as f:
        for part in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(part)

    return h.hexdigest()

# 4. Baca arsip upload sebelumnya
if INDEX.exists():

    old = pd.read_csv(
        INDEX,
        dtype=str,
        keep_default_na=False
    )

    assert set(old.columns) == set(FIELDS)
    old = old[FIELDS]

    assert old.upload_id.is_unique

    for r in old.itertuples(index=False):

        p = (ROOT / r.stored_path).resolve()

        assert p.is_relative_to(STAGE.resolve())
        assert p.is_file()
        assert sha_file(p) == r.sha256, (
            f"Arsip berubah: {r.upload_id}"
        )

else:
    old = pd.DataFrame(columns=FIELDS)

old_provider = old[
    old.provider_hint == PROVIDER
]

assert len(old_provider) <= 120

print("=== STEP 11D: UPLOAD GENERATIF ===")
print("AI:", PROVIDER.upper())
print(
    "Sudah tersimpan:",
    len(old_provider), "/120"
)
print(
    "Pilih gambar hasil AI. "
    "Boleh sekaligus atau bertahap."
)

# 5. Upload langsung
uploaded = files.upload()

assert uploaded, "Belum ada gambar dipilih."

seen = set(zip(
    old.original_filename,
    old.sha256,
    old.provider_hint
))

seq = 1 + max(
    [
        int(x[1:])
        for x in old.upload_id
        if x.startswith("G")
        and x[1:].isdigit()
    ] or [0]
)

records = []
raws = {}

for name, raw in sorted(
    uploaded.items(),
    key=lambda x: x[0].casefold()
):

    assert Path(name).name == name

    assert len(raw) <= 60 * 1024**2, (
        f"File terlalu besar: {name}"
    )

    digest = hashlib.sha256(raw).hexdigest()

    key = (name, digest, PROVIDER)

    if key in seen:
        print("Sudah ada, dilewati:", name)
        continue

    # Jangan mengubah file jika pernah tercatat
    # sebagai milik AI lain atau belum diketahui.
    conflict = old.loc[
        (old.original_filename == name)
        & (old.sha256 == digest)
        & (old.provider_hint != PROVIDER)
    ]

    assert conflict.empty, (
        f"File sudah terdaftar pada batch lain: {name}"
    )

    with Image.open(BytesIO(raw)) as im:
        fmt = im.format
        width, height = im.size
        im.verify()

    assert fmt in ("PNG", "JPEG", "WEBP"), (
        f"Format belum didukung: {name} ({fmt})"
    )

    uid = f"G{seq:03d}"
    seq += 1

    dest = STAGE / uid / name

    records.append({
        "upload_id": uid,
        "original_filename": name,
        "sha256": digest,
        "actual_format": fmt,
        "width": str(width),
        "height": str(height),
        "stored_path": str(
            dest.relative_to(ROOT)
        ),
        "provider_hint": PROVIDER,
        "provider_verified": "",
        "edit_type": "",
        "source_number": "",
        "source_id": "",
        "label_status": "BELUM_DIPETAKAN"
    })

    raws[uid] = raw
    seen.add(key)

assert len(old_provider) + len(records) <= 120, (
    "Jumlah gambar AI ini melebihi 120."
)

assert len(old) + len(records) <= 240, (
    "Total gambar melebihi 240."
)

# 6. Pratinjau
print("\n=== HASIL PEMERIKSAAN ===")
print("File baru:", len(records))
print(
    "Total", PROVIDER.upper(), ":",
    len(old_provider) + len(records), "/120"
)

if records:

    display(
        pd.DataFrame(records)[[
            "upload_id",
            "original_filename",
            "actual_format"
        ]].head(20)
    )

    approval = input(
        "\nKetik SIMPAN untuk menyimpan "
        "tanpa rename: "
    ).strip()

else:
    approval = ""

# 7. Simpan setelah persetujuan
if approval == "SIMPAN":

    BASE.mkdir(parents=True, exist_ok=True)

    for r in records:

        p = ROOT / r["stored_path"]
        p.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        if p.exists():
            assert sha_file(p) == r["sha256"]
        else:
            with p.open("xb") as f:
                f.write(raws[r["upload_id"]])

        assert sha_file(p) == r["sha256"]

    merged = pd.concat(
        [
            old,
            pd.DataFrame(
                records,
                columns=FIELDS
            )
        ],
        ignore_index=True
    )

    assert merged.upload_id.is_unique

    tmp = INDEX.with_name(
        INDEX.name + ".tmp"
    )

    merged.to_csv(tmp, index=False)
    tmp.replace(INDEX)

    batch_id = (
        datetime.now(timezone.utc)
        .strftime("%Y%m%dT%H%M%S%fZ")
        + "_" + uuid.uuid4().hex[:8]
    )

    BATCH = BASE / batch_id
    BATCH.mkdir(
        parents=True,
        exist_ok=False
    )

    pd.DataFrame(
        records,
        columns=FIELDS
    ).to_csv(
        BATCH / "batch.csv",
        index=False
    )

    metadata = {
        "provider_declared_by_user": PROVIDER,
        "new_images": len(records),
        "total_provider": int(
            (merged.provider_hint == PROVIDER).sum()
        ),
        "total_all": len(merged),
        "index_sha256": sha_file(INDEX),
        "status": "STAGED_NOT_LABELED",
        "created_at_utc":
            datetime.now(timezone.utc).isoformat()
    }

    (BATCH / "metadata.json").write_text(
        json.dumps(
            metadata,
            indent=2
        ),
        encoding="utf-8"
    )

    # 8. Galeri 40 sumber asli WM
    gallery = Image.new(
        "RGB",
        (1000, 8 * 162),
        "white"
    )

    draw = ImageDraw.Draw(gallery)

    reference = baseline.sort_values(
        "source_number"
    )

    for j, r in enumerate(
        reference.itertuples(index=False)
    ):

        with Image.open(
            ROOT / r.watermarked_path
        ) as im:

            thumb = ImageOps.contain(
                im.convert("RGB"),
                (186, 128)
            )

        x = (j % 5) * 200
        y = (j // 5) * 162

        gallery.paste(
            thumb,
            (
                x + (200 - thumb.width) // 2,
                y
            )
        )

        draw.text(
            (x + 5, y + 132),
            f"SUMBER {r.source_number} | {r.split[:3]}",
            fill="black"
        )

    gallery.save(
        BATCH / "REFERENSI_40_FOTO.jpg",
        quality=90
    )

    # 9. Galeri hasil generatif
    subset = merged[
        merged.provider_hint == PROVIDER
    ].copy()

    subset["_num"] = (
        subset.upload_id.str[1:].astype(int)
    )

    subset = subset.sort_values("_num")

    for begin in range(0, len(subset), 30):

        page = subset.iloc[
            begin:begin + 30
        ]

        board = Image.new(
            "RGB",
            (1200, 5 * 170),
            "white"
        )

        draw = ImageDraw.Draw(board)

        for j, r in enumerate(
            page.itertuples(index=False)
        ):

            with Image.open(
                ROOT / r.stored_path
            ) as im:

                thumb = ImageOps.contain(
                    ImageOps.exif_transpose(
                        im
                    ).convert("RGB"),
                    (190, 125)
                )

            x = (j % 6) * 200
            y = (j // 6) * 170

            board.paste(
                thumb,
                (
                    x + (200 - thumb.width) // 2,
                    y
                )
            )

            draw.text(
                (x + 5, y + 128),
                r.upload_id,
                fill="black"
            )

            draw.text(
                (x + 5, y + 145),
                r.original_filename[:26],
                fill="black"
            )

        board.save(
            BATCH / (
                f"{PROVIDER}_GALERI_"
                f"{begin // 30 + 1:02d}.jpg"
            ),
            quality=88
        )

        display(board)

    # 10. Template pemetaan identitas
    label = subset[FIELDS].copy()

    label["source_number"] = ""
    label["source_id"] = ""
    label["edit_type"] = ""
    label["label_status"] = "BELUM_DIPETAKAN"

    label.to_csv(
        BATCH / f"TEMPLATE_PEMETAAN_{PROVIDER}.csv",
        index=False
    )

    print("\n=== UPLOAD TERSIMPAN ===")
    print("AI:", PROVIDER.upper())
    print("Gambar AI ini:", len(subset), "/120")
    print("Total semua AI:", len(merged), "/240")
    print("Indeks:", INDEX)
    print("Galeri dan template:", BATCH)
    print("STATUS: STAGED_NOT_LABELED")
    print("Nama asli tidak diubah.")

else:
    print("Tidak ada file baru disimpan.")
```

```python

# ==================================================
# JATIKARYA NEW - STEP 11D-A
# UPLOAD 120 CHATGPT, KEMUDIAN 120 GEMINI
# Tanpa rename | Simpan asli | Buat galeri audit
# ==================================================

from pathlib import Path
from io import BytesIO
from datetime import datetime, timezone
from google.colab import drive, files
from IPython.display import display
from PIL import Image, ImageOps, ImageDraw
import pandas as pd
import hashlib
import json
import uuid

# Ganti menjadi "gemini" untuk batch kedua
PROVIDER = "gemini"

assert PROVIDER in ("chatgpt", "gemini")

# 1. Hubungkan Google Drive
if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

ROOT = Path(
    "/content/drive/MyDrive/"
    "PROJECT JURNAL - JATIKARYA/JATIKARYA_NEW"
)

assert ROOT.is_dir(), "Folder proyek tidak ditemukan."

# 2. Periksa 40 sumber ber-watermark
RUN11B = "20261108T083639291045Z_5876dadc"

MAN11B = (
    ROOT / "02_EXPERIMENTS"
    / "STEP_11B_WM_DINOV2_BASELINES"
    / RUN11B
    / "outputs/source_baseline_manifest.csv"
)

assert MAN11B.is_file()

summary = json.loads(
    (MAN11B.parent / "summary.json").read_text(
        encoding="utf-8"
    )
)

assert summary["status"] == "PASS"
assert summary["passed_sources"] == 40

baseline = pd.read_csv(
    MAN11B,
    dtype=str,
    keep_default_na=False
)

assert len(baseline) == 40
assert baseline.source_id.is_unique
assert baseline.source_number.is_unique

assert set(baseline.source_number) == {
    f"{i:02d}" for i in range(1, 41)
}

# 3. Folder penyimpanan
BASE = (
    ROOT / "02_EXPERIMENTS"
    / "STEP_11D_GENERATIVE_INTAKE"
)

STAGE = (
    ROOT / "01_DATASET"
    / "GENERATIVE_EDITS_STAGING"
)

INDEX = BASE / "staging_index.csv"

FIELDS = [
    "upload_id",
    "original_filename",
    "sha256",
    "actual_format",
    "width",
    "height",
    "stored_path",
    "provider_hint",
    "provider_verified",
    "edit_type",
    "source_number",
    "source_id",
    "label_status"
]

def sha_file(p):
    h = hashlib.sha256()

    with p.open("rb") as f:
        for part in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(part)

    return h.hexdigest()

# 4. Baca arsip upload sebelumnya
if INDEX.exists():

    old = pd.read_csv(
        INDEX,
        dtype=str,
        keep_default_na=False
    )

    assert set(old.columns) == set(FIELDS)
    old = old[FIELDS]

    assert old.upload_id.is_unique

    for r in old.itertuples(index=False):

        p = (ROOT / r.stored_path).resolve()

        assert p.is_relative_to(STAGE.resolve())
        assert p.is_file()
        assert sha_file(p) == r.sha256, (
            f"Arsip berubah: {r.upload_id}"
        )

else:
    old = pd.DataFrame(columns=FIELDS)

old_provider = old[
    old.provider_hint == PROVIDER
]

assert len(old_provider) <= 120

print("=== STEP 11D: UPLOAD GENERATIF ===")
print("AI:", PROVIDER.upper())
print(
    "Sudah tersimpan:",
    len(old_provider), "/120"
)
print(
    "Pilih gambar hasil AI. "
    "Boleh sekaligus atau bertahap."
)

# 5. Upload langsung
uploaded = files.upload()

assert uploaded, "Belum ada gambar dipilih."

seen = set(zip(
    old.original_filename,
    old.sha256,
    old.provider_hint
))

seq = 1 + max(
    [
        int(x[1:])
        for x in old.upload_id
        if x.startswith("G")
        and x[1:].isdigit()
    ] or [0]
)

records = []
raws = {}

for name, raw in sorted(
    uploaded.items(),
    key=lambda x: x[0].casefold()
):

    assert Path(name).name == name

    assert len(raw) <= 60 * 1024**2, (
        f"File terlalu besar: {name}"
    )

    digest = hashlib.sha256(raw).hexdigest()

    key = (name, digest, PROVIDER)

    if key in seen:
        print("Sudah ada, dilewati:", name)
        continue

    # Jangan mengubah file jika pernah tercatat
    # sebagai milik AI lain atau belum diketahui.
    conflict = old.loc[
        (old.original_filename == name)
        & (old.sha256 == digest)
        & (old.provider_hint != PROVIDER)
    ]

    assert conflict.empty, (
        f"File sudah terdaftar pada batch lain: {name}"
    )

    with Image.open(BytesIO(raw)) as im:
        fmt = im.format
        width, height = im.size
        im.verify()

    assert fmt in ("PNG", "JPEG", "WEBP"), (
        f"Format belum didukung: {name} ({fmt})"
    )

    uid = f"G{seq:03d}"
    seq += 1

    dest = STAGE / uid / name

    records.append({
        "upload_id": uid,
        "original_filename": name,
        "sha256": digest,
        "actual_format": fmt,
        "width": str(width),
        "height": str(height),
        "stored_path": str(
            dest.relative_to(ROOT)
        ),
        "provider_hint": PROVIDER,
        "provider_verified": "",
        "edit_type": "",
        "source_number": "",
        "source_id": "",
        "label_status": "BELUM_DIPETAKAN"
    })

    raws[uid] = raw
    seen.add(key)

assert len(old_provider) + len(records) <= 120, (
    "Jumlah gambar AI ini melebihi 120."
)

assert len(old) + len(records) <= 240, (
    "Total gambar melebihi 240."
)

# 6. Pratinjau
print("\n=== HASIL PEMERIKSAAN ===")
print("File baru:", len(records))
print(
    "Total", PROVIDER.upper(), ":",
    len(old_provider) + len(records), "/120"
)

if records:

    display(
        pd.DataFrame(records)[[
            "upload_id",
            "original_filename",
            "actual_format"
        ]].head(20)
    )

    approval = input(
        "\nKetik SIMPAN untuk menyimpan "
        "tanpa rename: "
    ).strip()

else:
    approval = ""

# 7. Simpan setelah persetujuan
if approval == "SIMPAN":

    BASE.mkdir(parents=True, exist_ok=True)

    for r in records:

        p = ROOT / r["stored_path"]
        p.parent.mkdir(
            parents=True,
            exist_ok=True
        )

        if p.exists():
            assert sha_file(p) == r["sha256"]
        else:
            with p.open("xb") as f:
                f.write(raws[r["upload_id"]])

        assert sha_file(p) == r["sha256"]

    merged = pd.concat(
        [
            old,
            pd.DataFrame(
                records,
                columns=FIELDS
            )
        ],
        ignore_index=True
    )

    assert merged.upload_id.is_unique

    tmp = INDEX.with_name(
        INDEX.name + ".tmp"
    )

    merged.to_csv(tmp, index=False)
    tmp.replace(INDEX)

    batch_id = (
        datetime.now(timezone.utc)
        .strftime("%Y%m%dT%H%M%S%fZ")
        + "_" + uuid.uuid4().hex[:8]
    )

    BATCH = BASE / batch_id
    BATCH.mkdir(
        parents=True,
        exist_ok=False
    )

    pd.DataFrame(
        records,
        columns=FIELDS
    ).to_csv(
        BATCH / "batch.csv",
        index=False
    )

    metadata = {
        "provider_declared_by_user": PROVIDER,
        "new_images": len(records),
        "total_provider": int(
            (merged.provider_hint == PROVIDER).sum()
        ),
        "total_all": len(merged),
        "index_sha256": sha_file(INDEX),
        "status": "STAGED_NOT_LABELED",
        "created_at_utc":
            datetime.now(timezone.utc).isoformat()
    }

    (BATCH / "metadata.json").write_text(
        json.dumps(
            metadata,
            indent=2
        ),
        encoding="utf-8"
    )

    # 8. Galeri 40 sumber asli WM
    gallery = Image.new(
        "RGB",
        (1000, 8 * 162),
        "white"
    )

    draw = ImageDraw.Draw(gallery)

    reference = baseline.sort_values(
        "source_number"
    )

    for j, r in enumerate(
        reference.itertuples(index=False)
    ):

        with Image.open(
            ROOT / r.watermarked_path
        ) as im:

            thumb = ImageOps.contain(
                im.convert("RGB"),
                (186, 128)
            )

        x = (j % 5) * 200
        y = (j // 5) * 162

        gallery.paste(
            thumb,
            (
                x + (200 - thumb.width) // 2,
                y
            )
        )

        draw.text(
            (x + 5, y + 132),
            f"SUMBER {r.source_number} | {r.split[:3]}",
            fill="black"
        )

    gallery.save(
        BATCH / "REFERENSI_40_FOTO.jpg",
        quality=90
    )

    # 9. Galeri hasil generatif
    subset = merged[
        merged.provider_hint == PROVIDER
    ].copy()

    subset["_num"] = (
        subset.upload_id.str[1:].astype(int)
    )

    subset = subset.sort_values("_num")

    for begin in range(0, len(subset), 30):

        page = subset.iloc[
            begin:begin + 30
        ]

        board = Image.new(
            "RGB",
            (1200, 5 * 170),
            "white"
        )

        draw = ImageDraw.Draw(board)

        for j, r in enumerate(
            page.itertuples(index=False)
        ):

            with Image.open(
                ROOT / r.stored_path
            ) as im:

                thumb = ImageOps.contain(
                    ImageOps.exif_transpose(
                        im
                    ).convert("RGB"),
                    (190, 125)
                )

            x = (j % 6) * 200
            y = (j // 6) * 170

            board.paste(
                thumb,
                (
                    x + (200 - thumb.width) // 2,
                    y
                )
            )

            draw.text(
                (x + 5, y + 128),
                r.upload_id,
                fill="black"
            )

            draw.text(
                (x + 5, y + 145),
                r.original_filename[:26],
                fill="black"
            )

        board.save(
            BATCH / (
                f"{PROVIDER}_GALERI_"
                f"{begin // 30 + 1:02d}.jpg"
            ),
            quality=88
        )

        display(board)

    # 10. Template pemetaan identitas
    label = subset[FIELDS].copy()

    label["source_number"] = ""
    label["source_id"] = ""
    label["edit_type"] = ""
    label["label_status"] = "BELUM_DIPETAKAN"

    label.to_csv(
        BATCH / f"TEMPLATE_PEMETAAN_{PROVIDER}.csv",
        index=False
    )

    print("\n=== UPLOAD TERSIMPAN ===")
    print("AI:", PROVIDER.upper())
    print("Gambar AI ini:", len(subset), "/120")
    print("Total semua AI:", len(merged), "/240")
    print("Indeks:", INDEX)
    print("Galeri dan template:", BATCH)
    print("STATUS: STAGED_NOT_LABELED")
    print("Nama asli tidak diubah.")

else:
    print("Tidak ada file baru disimpan.")
```

```python
# ==================================================
# JATIKARYA NEW - STEP 11D FINAL
# UPLOAD ULANG 240 GAMBAR YANG SUDAH DI-RENAME
# TANPA CSV / PEMETAAN MANUAL / ZIP
# ==================================================

from pathlib import Path
from io import BytesIO
from datetime import datetime, timezone
from collections import Counter
from google.colab import drive, files
from PIL import Image
from IPython.display import display
import pandas as pd
import hashlib
import json
import re
import uuid

# RUN 1: chatgpt
# RUN 2: gemini
PROVIDER = "chatgpt"

assert PROVIDER in ("chatgpt", "gemini")

# 1. Hubungkan Drive
if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

ROOT = Path(
    "/content/drive/MyDrive/"
    "PROJECT JURNAL - JATIKARYA/JATIKARYA_NEW"
)

assert ROOT.is_dir(), "Folder JATIKARYA_NEW tidak ditemukan."

# 2. Lokasi sumber yang sudah valid
REF = (
    ROOT / "02_EXPERIMENTS"
    / "STEP_11B_WM_DINOV2_BASELINES"
    / "20261108T083639291045Z_5876dadc"
    / "outputs/source_baseline_manifest.csv"
)

# Folder baru, terpisah dari staging lama
BASE = (
    ROOT / "02_EXPERIMENTS"
    / "STEP_11D_FINAL_REUPLOAD"
)

DATA = (
    ROOT / "01_DATASET"
    / "GENERATIVE_EDITS_FINAL"
)

INDEX = BASE / "final_upload_index.csv"

FINAL = (
    BASE / "generative_final_manifest_240.csv"
)

REPORT = BASE / "final_manifest_evidence.json"

FIELDS = [
    "provider",
    "source_number",
    "source_id",
    "split",
    "edit_type",
    "original_filename",
    "sha256",
    "actual_format",
    "width",
    "height",
    "stored_path",
    "source_filename",
    "parent_watermarked_sha256",
    "label_evidence"
]

def sha_file(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for part in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(part)
    return h.hexdigest()

# 3. Periksa 40 sumber STEP 11B
assert REF.is_file()

summary_ref = json.loads(
    (REF.parent / "summary.json").read_text(
        encoding="utf-8"
    )
)

meta_ref = json.loads(
    (REF.parent.parent / "metadata.json").read_text(
        encoding="utf-8"
    )
)

assert summary_ref["status"] == "PASS"
assert meta_ref["status"] == "PASS"
assert summary_ref["manifest_sha256"] == sha_file(REF)

refs = pd.read_csv(
    REF, dtype=str, keep_default_na=False
)

assert len(refs) == 40
assert refs.source_number.is_unique

assert set(refs.source_number) == {
    f"{n:02d}" for n in range(1, 41)
}

assert Counter(refs.split) == Counter({
    "CALIBRATION": 20,
    "FINAL_TEST": 20
})

ref_map = refs.set_index(
    "source_number"
).to_dict("index")

# 4. Baca progres upload final
if INDEX.is_file():

    old = pd.read_csv(
        INDEX, dtype=str, keep_default_na=False
    )

    assert list(old.columns) == FIELDS

else:
    old = pd.DataFrame(columns=FIELDS)

KEY = ["provider", "source_number", "edit_type"]

assert not old.duplicated(KEY).any()
assert old.provider.isin(
    ["chatgpt", "gemini"]
).all()

# Verifikasi file yang sudah tersimpan
for r in old.itertuples(index=False):

    assert r.source_number in ref_map

    ref = ref_map[r.source_number]

    assert r.source_id == ref["source_id"]
    assert r.split == ref["split"]

    p = (ROOT / r.stored_path).resolve()

    assert p.is_relative_to(DATA.resolve())
    assert p.is_file()
    assert sha_file(p) == r.sha256, (
        f"File tersimpan berubah: {r.original_filename}"
    )

current = int(
    (old.provider == PROVIDER).sum()
)

assert current <= 120
assert len(old) <= 240

print("=== STEP 11D FINAL ===")
print("AI:", PROVIDER.upper())
print("Sudah tersimpan:", current, "/120")
print("Total:", len(old), "/240")

new_records = []
raw_data = {}

# 5. Upload gambar yang sudah di-rename
if current < 120:

    print("\nPilih gambar hasil", PROVIDER.upper())
    print("Boleh upload sekaligus atau bertahap.")

    uploaded = files.upload()

    assert uploaded, "Tidak ada gambar diunggah."

    pattern = re.compile(
        r"^(removal|replacement|background)"
        r"[ _-]+(gpt|chatgpt|gemini)"
        r"[ _-]+(\d{1,2})"
        r"\.(png|jpg|jpeg|webp)$",
        re.IGNORECASE
    )

    existing_keys = {
        (r.provider, r.source_number, r.edit_type): r
        for r in old.itertuples(index=False)
    }

    seen = set()

    for filename, raw in uploaded.items():

        assert Path(filename).name == filename

        match = pattern.fullmatch(filename)

        assert match, (
            f"Nama file tidak sesuai: {filename}"
        )

        edit, model, number, extension = match.groups()

        model = (
            "chatgpt"
            if model.lower() in ("gpt", "chatgpt")
            else "gemini"
        )

        assert model == PROVIDER, (
            f"File dari AI berbeda: {filename}"
        )

        number = int(number)

        assert 1 <= number <= 40, (
            f"Nomor sumber tidak valid: {filename}"
        )

        source_number = f"{number:02d}"
        edit = edit.lower()

        key = (model, source_number, edit)

        assert key not in seen, (
            f"Kombinasi ganda: {filename}"
        )

        seen.add(key)

        digest = hashlib.sha256(raw).hexdigest()

        # Format asli gambar
        with Image.open(BytesIO(raw)) as img:
            fmt = img.format
            width, height = img.size
            img.verify()

        assert fmt in (
            "PNG", "JPEG", "WEBP"
        ), f"Format tidak didukung: {filename}"

        # Abaikan file yang sudah terdaftar dan sama
        if key in existing_keys:

            assert existing_keys[key].sha256 == digest, (
                f"File {filename} berbeda dari "
                "versi yang sudah tersimpan."
            )

            print("Sudah ada:", filename)
            continue

        ref = ref_map[source_number]

        dest = (
            DATA / model / filename
        )

        new_records.append({
            "provider": model,
            "source_number": source_number,
            "source_id": ref["source_id"],
            "split": ref["split"],
            "edit_type": edit,
            "original_filename": filename,
            "sha256": digest,
            "actual_format": fmt,
            "width": str(width),
            "height": str(height),
            "stored_path": str(
                dest.relative_to(ROOT)
            ),
            "source_filename": ref["source_filename"],
            "parent_watermarked_sha256":
                ref["watermarked_sha256"],
            "label_evidence":
                "USER_RENAMED_NOT_VISUALLY_VERIFIED"
        })

        raw_data[filename] = raw

    assert current + len(new_records) <= 120

    combined = pd.concat(
        [
            old,
            pd.DataFrame(
                new_records,
                columns=FIELDS
            )
        ],
        ignore_index=True
    )

    assert not combined.duplicated(KEY).any()

    # 6. Pratinjau
    print("\n=== VALIDASI UPLOAD ===")
    print("File baru:", len(new_records))
    print("Total setelah upload:", len(combined))

    display(pd.crosstab(
        combined.provider,
        combined.edit_type
    ))

    if new_records:

        display(
            pd.DataFrame(new_records)[[
                "original_filename",
                "source_number",
                "edit_type",
                "split"
            ]].head(15)
        )

        approval = input(
            "\nKetik SIMPAN untuk menyimpan: "
        ).strip()

        if approval != "SIMPAN":
            raise RuntimeError(
                "Penyimpanan dibatalkan."
            )

        # 7. Simpan tanpa mengubah gambar
        for row in new_records:

            dest = (
                ROOT / row["stored_path"]
            ).resolve()

            assert dest.is_relative_to(DATA.resolve())

            dest.parent.mkdir(
                parents=True,
                exist_ok=True
            )

            if dest.exists():

                assert sha_file(dest) == row["sha256"]

            else:

                with dest.open("xb") as f:
                    f.write(
                        raw_data[row["original_filename"]]
                    )

            assert sha_file(dest) == row["sha256"]

        # 8. Perbarui indeks final
        BASE.mkdir(
            parents=True,
            exist_ok=True
        )

        combined = combined[FIELDS].sort_values(
            KEY
        ).reset_index(drop=True)

        tmp = INDEX.with_name(
            INDEX.name + ".tmp"
        )

        combined.to_csv(
            tmp,
            index=False
        )

        tmp.replace(INDEX)

        old = combined

        # Catat audit setiap batch
        audit_id = (
            datetime.now(timezone.utc)
            .strftime("%Y%m%dT%H%M%S%fZ")
            + "_" + uuid.uuid4().hex[:8]
        )

        audit_folder = BASE / "upload_events"
        audit_folder.mkdir(
            parents=True,
            exist_ok=True
        )

        audit = {
            "provider": PROVIDER,
            "new_count": len(new_records),
            "total_registered": len(old),
            "baseline_sha256": sha_file(REF),
            "index_sha256": sha_file(INDEX),
            "created_at_utc":
                datetime.now(timezone.utc).isoformat()
        }

        (
            audit_folder / f"{audit_id}.json"
        ).write_text(
            json.dumps(
                audit,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

else:
    print("120 gambar AI ini sudah tersimpan.")

# 9. Periksa jumlah akhir
n_gpt = int(
    (old.provider == "chatgpt").sum()
)

n_gemini = int(
    (old.provider == "gemini").sum()
)

print("\n=== RINGKASAN UPLOAD ===")
print("ChatGPT:", n_gpt, "/120")
print("Gemini:", n_gemini, "/120")
print("Total:", len(old), "/240")

# 10. Finalisasi jika semua 240 telah tersedia
if len(old) == 240:

    expected = {
        (ai, f"{n:02d}", edit)
        for ai in ("chatgpt", "gemini")
        for n in range(1, 41)
        for edit in (
            "removal",
            "replacement",
            "background"
        )
    }

    actual = set(
        map(
            tuple,
            old[KEY].to_numpy()
        )
    )

    assert actual == expected
    assert len(old) == len(actual) == 240

    # Double-check seluruh file final
    for r in old.itertuples(index=False):

        p = (ROOT / r.stored_path).resolve()

        assert p.is_relative_to(DATA.resolve())
        assert p.is_file()
        assert sha_file(p) == r.sha256

    final_df = old[FIELDS].sort_values(
        KEY
    ).reset_index(drop=True)

    # Jangan timpa manifest yang berbeda
    if FINAL.exists():

        prev = pd.read_csv(
            FINAL,
            dtype=str,
            keep_default_na=False
        )

        assert prev.equals(final_df), (
            "Manifest final lama berbeda."
        )

    else:

        with FINAL.open(
            "x",
            encoding="utf-8",
            newline=""
        ) as f:
            final_df.to_csv(
                f,
                index=False
            )

    evidence = {
        "status":
            "READY_FOR_TESTING_WITH_USER_DECLARED_LABELS",
        "total_images": 240,
        "chatgpt_images": 120,
        "gemini_images": 120,
        "baseline_manifest_sha256": sha_file(REF),
        "final_manifest_sha256": sha_file(FINAL),
        "label_note": (
            "Source ID and edit type are derived "
            "from user-assigned filenames, not "
            "independently verified from pixels."
        ),
        "created_at_utc":
            datetime.now(timezone.utc).isoformat()
    }

    if REPORT.exists():

        previous = json.loads(
            REPORT.read_text(encoding="utf-8")
        )

        assert (
            previous["final_manifest_sha256"]
            == evidence["final_manifest_sha256"]
        )

    else:

        REPORT.write_text(
            json.dumps(
                evidence,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

    print("\n=== STEP 11D FINAL SELESAI ===")
    print("STATUS:", evidence["status"])
    print("Total gambar valid: 240/240")
    print("Manifest:", FINAL)
    print("SHA256:", sha_file(FINAL))

else:

    print("\nSTATUS: BELUM LENGKAP")
    print(
        "Jalankan kembali untuk batch berikutnya."
    )
```

```python
# ==================================================
# JATIKARYA NEW - STEP 11D FINAL
# UPLOAD ULANG 240 GAMBAR YANG SUDAH DI-RENAME
# TANPA CSV / PEMETAAN MANUAL / ZIP
# ==================================================

from pathlib import Path
from io import BytesIO
from datetime import datetime, timezone
from collections import Counter
from google.colab import drive, files
from PIL import Image
from IPython.display import display
import pandas as pd
import hashlib
import json
import re
import uuid

# RUN 1: chatgpt
# RUN 2: gemini
PROVIDER = "gemini"

assert PROVIDER in ("chatgpt", "gemini")

# 1. Hubungkan Drive
if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

ROOT = Path(
    "/content/drive/MyDrive/"
    "PROJECT JURNAL - JATIKARYA/JATIKARYA_NEW"
)

assert ROOT.is_dir(), "Folder JATIKARYA_NEW tidak ditemukan."

# 2. Lokasi sumber yang sudah valid
REF = (
    ROOT / "02_EXPERIMENTS"
    / "STEP_11B_WM_DINOV2_BASELINES"
    / "20261108T083639291045Z_5876dadc"
    / "outputs/source_baseline_manifest.csv"
)

# Folder baru, terpisah dari staging lama
BASE = (
    ROOT / "02_EXPERIMENTS"
    / "STEP_11D_FINAL_REUPLOAD"
)

DATA = (
    ROOT / "01_DATASET"
    / "GENERATIVE_EDITS_FINAL"
)

INDEX = BASE / "final_upload_index.csv"

FINAL = (
    BASE / "generative_final_manifest_240.csv"
)

REPORT = BASE / "final_manifest_evidence.json"

FIELDS = [
    "provider",
    "source_number",
    "source_id",
    "split",
    "edit_type",
    "original_filename",
    "sha256",
    "actual_format",
    "width",
    "height",
    "stored_path",
    "source_filename",
    "parent_watermarked_sha256",
    "label_evidence"
]

def sha_file(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for part in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(part)
    return h.hexdigest()

# 3. Periksa 40 sumber STEP 11B
assert REF.is_file()

summary_ref = json.loads(
    (REF.parent / "summary.json").read_text(
        encoding="utf-8"
    )
)

meta_ref = json.loads(
    (REF.parent.parent / "metadata.json").read_text(
        encoding="utf-8"
    )
)

assert summary_ref["status"] == "PASS"
assert meta_ref["status"] == "PASS"
assert summary_ref["manifest_sha256"] == sha_file(REF)

refs = pd.read_csv(
    REF, dtype=str, keep_default_na=False
)

assert len(refs) == 40
assert refs.source_number.is_unique

assert set(refs.source_number) == {
    f"{n:02d}" for n in range(1, 41)
}

assert Counter(refs.split) == Counter({
    "CALIBRATION": 20,
    "FINAL_TEST": 20
})

ref_map = refs.set_index(
    "source_number"
).to_dict("index")

# 4. Baca progres upload final
if INDEX.is_file():

    old = pd.read_csv(
        INDEX, dtype=str, keep_default_na=False
    )

    assert list(old.columns) == FIELDS

else:
    old = pd.DataFrame(columns=FIELDS)

KEY = ["provider", "source_number", "edit_type"]

assert not old.duplicated(KEY).any()
assert old.provider.isin(
    ["chatgpt", "gemini"]
).all()

# Verifikasi file yang sudah tersimpan
for r in old.itertuples(index=False):

    assert r.source_number in ref_map

    ref = ref_map[r.source_number]

    assert r.source_id == ref["source_id"]
    assert r.split == ref["split"]

    p = (ROOT / r.stored_path).resolve()

    assert p.is_relative_to(DATA.resolve())
    assert p.is_file()
    assert sha_file(p) == r.sha256, (
        f"File tersimpan berubah: {r.original_filename}"
    )

current = int(
    (old.provider == PROVIDER).sum()
)

assert current <= 120
assert len(old) <= 240

print("=== STEP 11D FINAL ===")
print("AI:", PROVIDER.upper())
print("Sudah tersimpan:", current, "/120")
print("Total:", len(old), "/240")

new_records = []
raw_data = {}

# 5. Upload gambar yang sudah di-rename
if current < 120:

    print("\nPilih gambar hasil", PROVIDER.upper())
    print("Boleh upload sekaligus atau bertahap.")

    uploaded = files.upload()

    assert uploaded, "Tidak ada gambar diunggah."

    pattern = re.compile(
        r"^(removal|replacement|background)"
        r"[ _-]+(gpt|chatgpt|gemini)"
        r"[ _-]+(\d{1,2})"
        r"\.(png|jpg|jpeg|webp)$",
        re.IGNORECASE
    )

    existing_keys = {
        (r.provider, r.source_number, r.edit_type): r
        for r in old.itertuples(index=False)
    }

    seen = set()

    for filename, raw in uploaded.items():

        assert Path(filename).name == filename

        match = pattern.fullmatch(filename)

        assert match, (
            f"Nama file tidak sesuai: {filename}"
        )

        edit, model, number, extension = match.groups()

        model = (
            "chatgpt"
            if model.lower() in ("gpt", "chatgpt")
            else "gemini"
        )

        assert model == PROVIDER, (
            f"File dari AI berbeda: {filename}"
        )

        number = int(number)

        assert 1 <= number <= 40, (
            f"Nomor sumber tidak valid: {filename}"
        )

        source_number = f"{number:02d}"
        edit = edit.lower()

        key = (model, source_number, edit)

        assert key not in seen, (
            f"Kombinasi ganda: {filename}"
        )

        seen.add(key)

        digest = hashlib.sha256(raw).hexdigest()

        # Format asli gambar
        with Image.open(BytesIO(raw)) as img:
            fmt = img.format
            width, height = img.size
            img.verify()

        assert fmt in (
            "PNG", "JPEG", "WEBP"
        ), f"Format tidak didukung: {filename}"

        # Abaikan file yang sudah terdaftar dan sama
        if key in existing_keys:

            assert existing_keys[key].sha256 == digest, (
                f"File {filename} berbeda dari "
                "versi yang sudah tersimpan."
            )

            print("Sudah ada:", filename)
            continue

        ref = ref_map[source_number]

        dest = (
            DATA / model / filename
        )

        new_records.append({
            "provider": model,
            "source_number": source_number,
            "source_id": ref["source_id"],
            "split": ref["split"],
            "edit_type": edit,
            "original_filename": filename,
            "sha256": digest,
            "actual_format": fmt,
            "width": str(width),
            "height": str(height),
            "stored_path": str(
                dest.relative_to(ROOT)
            ),
            "source_filename": ref["source_filename"],
            "parent_watermarked_sha256":
                ref["watermarked_sha256"],
            "label_evidence":
                "USER_RENAMED_NOT_VISUALLY_VERIFIED"
        })

        raw_data[filename] = raw

    assert current + len(new_records) <= 120

    combined = pd.concat(
        [
            old,
            pd.DataFrame(
                new_records,
                columns=FIELDS
            )
        ],
        ignore_index=True
    )

    assert not combined.duplicated(KEY).any()

    # 6. Pratinjau
    print("\n=== VALIDASI UPLOAD ===")
    print("File baru:", len(new_records))
    print("Total setelah upload:", len(combined))

    display(pd.crosstab(
        combined.provider,
        combined.edit_type
    ))

    if new_records:

        display(
            pd.DataFrame(new_records)[[
                "original_filename",
                "source_number",
                "edit_type",
                "split"
            ]].head(15)
        )

        approval = input(
            "\nKetik SIMPAN untuk menyimpan: "
        ).strip()

        if approval != "SIMPAN":
            raise RuntimeError(
                "Penyimpanan dibatalkan."
            )

        # 7. Simpan tanpa mengubah gambar
        for row in new_records:

            dest = (
                ROOT / row["stored_path"]
            ).resolve()

            assert dest.is_relative_to(DATA.resolve())

            dest.parent.mkdir(
                parents=True,
                exist_ok=True
            )

            if dest.exists():

                assert sha_file(dest) == row["sha256"]

            else:

                with dest.open("xb") as f:
                    f.write(
                        raw_data[row["original_filename"]]
                    )

            assert sha_file(dest) == row["sha256"]

        # 8. Perbarui indeks final
        BASE.mkdir(
            parents=True,
            exist_ok=True
        )

        combined = combined[FIELDS].sort_values(
            KEY
        ).reset_index(drop=True)

        tmp = INDEX.with_name(
            INDEX.name + ".tmp"
        )

        combined.to_csv(
            tmp,
            index=False
        )

        tmp.replace(INDEX)

        old = combined

        # Catat audit setiap batch
        audit_id = (
            datetime.now(timezone.utc)
            .strftime("%Y%m%dT%H%M%S%fZ")
            + "_" + uuid.uuid4().hex[:8]
        )

        audit_folder = BASE / "upload_events"
        audit_folder.mkdir(
            parents=True,
            exist_ok=True
        )

        audit = {
            "provider": PROVIDER,
            "new_count": len(new_records),
            "total_registered": len(old),
            "baseline_sha256": sha_file(REF),
            "index_sha256": sha_file(INDEX),
            "created_at_utc":
                datetime.now(timezone.utc).isoformat()
        }

        (
            audit_folder / f"{audit_id}.json"
        ).write_text(
            json.dumps(
                audit,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

else:
    print("120 gambar AI ini sudah tersimpan.")

# 9. Periksa jumlah akhir
n_gpt = int(
    (old.provider == "chatgpt").sum()
)

n_gemini = int(
    (old.provider == "gemini").sum()
)

print("\n=== RINGKASAN UPLOAD ===")
print("ChatGPT:", n_gpt, "/120")
print("Gemini:", n_gemini, "/120")
print("Total:", len(old), "/240")

# 10. Finalisasi jika semua 240 telah tersedia
if len(old) == 240:

    expected = {
        (ai, f"{n:02d}", edit)
        for ai in ("chatgpt", "gemini")
        for n in range(1, 41)
        for edit in (
            "removal",
            "replacement",
            "background"
        )
    }

    actual = set(
        map(
            tuple,
            old[KEY].to_numpy()
        )
    )

    assert actual == expected
    assert len(old) == len(actual) == 240

    # Double-check seluruh file final
    for r in old.itertuples(index=False):

        p = (ROOT / r.stored_path).resolve()

        assert p.is_relative_to(DATA.resolve())
        assert p.is_file()
        assert sha_file(p) == r.sha256

    final_df = old[FIELDS].sort_values(
        KEY
    ).reset_index(drop=True)

    # Jangan timpa manifest yang berbeda
    if FINAL.exists():

        prev = pd.read_csv(
            FINAL,
            dtype=str,
            keep_default_na=False
        )

        assert prev.equals(final_df), (
            "Manifest final lama berbeda."
        )

    else:

        with FINAL.open(
            "x",
            encoding="utf-8",
            newline=""
        ) as f:
            final_df.to_csv(
                f,
                index=False
            )

    evidence = {
        "status":
            "READY_FOR_TESTING_WITH_USER_DECLARED_LABELS",
        "total_images": 240,
        "chatgpt_images": 120,
        "gemini_images": 120,
        "baseline_manifest_sha256": sha_file(REF),
        "final_manifest_sha256": sha_file(FINAL),
        "label_note": (
            "Source ID and edit type are derived "
            "from user-assigned filenames, not "
            "independently verified from pixels."
        ),
        "created_at_utc":
            datetime.now(timezone.utc).isoformat()
    }

    if REPORT.exists():

        previous = json.loads(
            REPORT.read_text(encoding="utf-8")
        )

        assert (
            previous["final_manifest_sha256"]
            == evidence["final_manifest_sha256"]
        )

    else:

        REPORT.write_text(
            json.dumps(
                evidence,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

    print("\n=== STEP 11D FINAL SELESAI ===")
    print("STATUS:", evidence["status"])
    print("Total gambar valid: 240/240")
    print("Manifest:", FINAL)
    print("SHA256:", sha_file(FINAL))

else:

    print("\nSTATUS: BELUM LENGKAP")
    print(
        "Jalankan kembali untuk batch berikutnya."
    )
```

```python
# ==================================================
# JATIKARYA NEW - STEP 11E
# TRUSTMARK-ONLY: 240 GAMBAR GENERATIF
# 40 SUMBER | 2 AI | 3 JENIS EDIT
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from importlib.metadata import version
from PIL import Image, ImageOps
from google.colab import drive
import subprocess
import sys
import hashlib
import json
import time
import pandas as pd
import torch

# 1. Sambungkan Google Drive
if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

ROOT = Path(
    "/content/drive/MyDrive/"
    "PROJECT JURNAL - JATIKARYA/JATIKARYA_NEW"
)

BASE = ROOT / "02_EXPERIMENTS"

MAN = (
    BASE / "STEP_11D_FINAL_REUPLOAD"
    / "generative_final_manifest_240.csv"
)

REF = (
    BASE / "STEP_11B_WM_DINOV2_BASELINES"
    / "20261108T083639291045Z_5876dadc"
    / "outputs/source_baseline_manifest.csv"
)

DATA = (
    ROOT / "01_DATASET/GENERATIVE_EDITS_FINAL"
).resolve()

assert ROOT.is_dir()
assert MAN.is_file()
assert REF.is_file()

def sha_file(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for chunk in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(chunk)
    return h.hexdigest()

def read_json(path):
    return json.loads(
        Path(path).read_text(encoding="utf-8")
    )

# 2. Baca 40 referensi dan 240 gambar
ref_info = read_json(
    REF.parent / "summary.json"
)

assert ref_info["status"] == "PASS"
assert sha_file(REF) == ref_info["manifest_sha256"]

idx = pd.read_csv(
    MAN,
    dtype=str,
    keep_default_na=False
)

refs = pd.read_csv(
    REF,
    dtype=str,
    keep_default_na=False
)

assert len(idx) == 240
assert len(refs) == 40
assert refs.source_number.is_unique

assert idx.provider.value_counts().to_dict() == {
    "chatgpt": 120,
    "gemini": 120
}

assert idx.split.value_counts().to_dict() == {
    "CALIBRATION": 120,
    "FINAL_TEST": 120
}

ref_map = refs.set_index(
    "source_number"
).to_dict("index")

KEY = [
    "provider",
    "source_number",
    "edit_type"
]

assert not idx.duplicated(KEY).any()

expected = {
    (ai, f"{n:02d}", edit)
    for ai in ("chatgpt", "gemini")
    for n in range(1, 41)
    for edit in (
        "removal",
        "replacement",
        "background"
    )
}

assert set(
    map(tuple, idx[KEY].to_numpy())
) == expected

# 3. Double-check identitas dan integritas file
print("=== VALIDASI 240 GAMBAR ===")

for r in idx.itertuples(index=False):
    assert r.source_number in ref_map

    src = ref_map[r.source_number]

    assert r.source_id == src["source_id"]
    assert r.split == src["split"]

    assert (
        r.parent_watermarked_sha256
        == src["watermarked_sha256"]
    )

    path = (ROOT / r.stored_path).resolve()

    assert path.is_relative_to(DATA)
    assert path.is_file()

    assert sha_file(path) == r.sha256, (
        f"Hash berbeda: {r.original_filename}"
    )

print("240/240 gambar dan identitas: VALID")

# 4. Siapkan TrustMark yang sama seperti pilot
try:
    tm_version = version("trustmark")
except Exception:
    subprocess.check_call([
        sys.executable,
        "-m", "pip", "install",
        "trustmark==0.9.2"
    ])
    tm_version = version("trustmark")

assert tm_version == "0.9.2", (
    f"Versi TrustMark berbeda: {tm_version}"
)

from trustmark import TrustMark

# 5. Buat identitas eksperimen
config = {
    "input_manifest_sha256": sha_file(MAN),
    "reference_manifest_sha256": sha_file(REF),
    "trustmark_version": tm_version,
    "model_type": "Q",
    "encoding": "BCH_5",
    "mode": "binary",
    "orientation": "EXIF_TRANSPOSE_RGB_FRAME0"
}

fingerprint = hashlib.sha256(
    json.dumps(config, sort_keys=True).encode()
).hexdigest()

RUN = (
    BASE / "STEP_11E_TRUSTMARK_GENERATIVE"
    / ("RUN_" + fingerprint[:20])
)

OUT = RUN / "outputs"
RECORDS = OUT / "records"

RECORDS.mkdir(
    parents=True,
    exist_ok=True
)

config_path = RUN / "config.json"

if config_path.exists():
    assert read_json(config_path) == config
else:
    config_path.write_text(
        json.dumps(config, indent=2),
        encoding="utf-8"
    )

# 6. Siapkan katalog payload dari sumber asli
# Bukan dari prediksi DINOv2
payload_map = {}

for r in refs.itertuples(index=False):
    payload = bin(
        int(r.source_sha256, 16)
    )[2:].zfill(256)[:61]

    assert payload not in payload_map

    payload_map[payload] = r.source_id

assert len(payload_map) == 40

DEVICE = (
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

tm = TrustMark(
    verbose=False,
    model_type="Q",
    encoding_type=TrustMark.Encoding.BCH_5,
    device=DEVICE
)

print("\n=== MULAI TRUSTMARK-ONLY ===")
print("Device:", DEVICE)
print("Total:", len(idx))

# 7. Jalankan 240 pengujian
outcomes = []

for i, r in enumerate(
    idx.sort_values(KEY).itertuples(index=False),
    1
):
    uid = (
        f"{r.provider}_"
        f"{r.source_number}_"
        f"{r.edit_type}"
    )

    record_path = (
        RECORDS / f"{uid}.json"
    )

    # Gunakan checkpoint jika valid
    if record_path.exists():

        record = read_json(record_path)

        assert (
            record["image_sha256"] == r.sha256
            and record["source_id"] == r.source_id
            and record["provider"] == r.provider
            and record["edit_type"] == r.edit_type
        ), f"Checkpoint konflik: {uid}"

    else:

        start = time.perf_counter()

        record = {
            "provider": r.provider,
            "source_number": r.source_number,
            "source_id": r.source_id,
            "split": r.split,
            "edit_type": r.edit_type,
            "filename": r.original_filename,
            "image_sha256": r.sha256,
            "parent_watermarked_sha256":
                r.parent_watermarked_sha256,
            "detected": False,
            "payload_match": False,
            "decoded_payload": "",
            "decoded_schema": "",
            "mapped_source_id": "",
            "status": "ERROR",
            "error": ""
        }

        try:
            with Image.open(
                ROOT / r.stored_path
            ) as im:

                im.seek(0)

                rgb = ImageOps.exif_transpose(
                    im
                ).convert("RGB")

            decoded, detected, schema = tm.decode(
                rgb,
                MODE="binary"
            )

            detected = bool(detected)

            schema_ok = (
                schema == TrustMark.Encoding.BCH_5
            )

            decoded_str = (
                str(decoded)
                if decoded is not None
                else ""
            )

            mapped = (
                payload_map.get(decoded_str, "")
                if detected and schema_ok
                else ""
            )

            source_sha = ref_map[
                r.source_number
            ]["source_sha256"]

            expected_payload = bin(
                int(source_sha, 16)
            )[2:].zfill(256)[:61]

            matched = bool(
                detected
                and schema_ok
                and decoded_str == expected_payload
            )

            if not detected:
                status = "TIDAK_TERDETEKSI"

            elif not schema_ok:
                status = "SCHEMA_TIDAK_SESUAI"

            elif matched:
                status = "KODE_COCOK"

            elif mapped:
                status = "KODE_SUMBER_LAIN"

            else:
                status = "KODE_TIDAK_COCOK"

            record.update({
                "detected": detected,
                "payload_match": matched,
                "decoded_payload": decoded_str,
                "decoded_schema": str(schema),
                "mapped_source_id": mapped,
                "status": status
            })

        except Exception as exc:
            record["error"] = (
                f"{type(exc).__name__}: {exc}"
            )

        record["seconds"] = round(
            time.perf_counter() - start,
            5
        )

        # Simpan checkpoint segera
        temp = record_path.with_suffix(
            ".json.tmp"
        )

        temp.write_text(
            json.dumps(
                record,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

        temp.replace(record_path)

    outcomes.append(record)

    if (
        i % 10 == 0
        or record["status"] == "ERROR"
    ):
        print(
            f"[{i}/240] "
            f"{r.provider} | "
            f"{r.source_number} | "
            f"{r.edit_type}: "
            f"{record['status']}"
        )

# 8. Simpan hasil keseluruhan
result = pd.DataFrame(outcomes)

assert len(result) == 240

csv = (
    OUT / "trustmark_240_results.csv"
)

result.to_csv(
    csv,
    index=False
)

counts = result.status.value_counts().to_dict()

errors = int(
    result.status.eq("ERROR").sum()
)

summary = {
    "fingerprint": fingerprint,
    "total_images": len(result),
    "matched_count": int(
        result.status.eq("KODE_COCOK").sum()
    ),
    "not_detected_count": int(
        result.status.eq(
            "TIDAK_TERDETEKSI"
        ).sum()
    ),
    "error_count": errors,
    "status_counts": counts,
    "result_sha256": sha_file(csv),
    "run_status": (
        "PASS"
        if errors == 0
        else "COMPLETED_WITH_ERRORS"
    ),
    "finished_at_utc":
        datetime.now(timezone.utc).isoformat()
}

(
    OUT / "summary.json"
).write_text(
    json.dumps(
        summary,
        ensure_ascii=False,
        indent=2
    ),
    encoding="utf-8"
)

# 9. Ringkasan penelitian
print("\n=== STEP 11E: TRUSTMARK-ONLY ===")

print("Total gambar:", len(result))

print(
    "Kode cocok:",
    summary["matched_count"]
)

print(
    "Tidak terdeteksi:",
    summary["not_detected_count"]
)

print("Error:", errors)

print("\n=== HASIL PER AI, EDIT DAN SPLIT ===")

display(
    pd.crosstab(
        [
            result["split"],
            result["provider"],
            result["edit_type"]
        ],
        result["status"]
    )
)

print("\nSTATUS:", summary["run_status"])
print("Hasil:", csv)
print("Ringkasan:", OUT / "summary.json")
print("BELUM FUSION")
```

```python
# ==================================================
# JATIKARYA NEW - STEP 11F
# DINOV2-ONLY | 240 GAMBAR x 40 SUMBER
# Belum Fusion, belum kalibrasi
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from PIL import Image, ImageOps
from torchvision import transforms
from IPython.display import display
from google.colab import drive
import torch
import torch.nn.functional as F
import numpy as np
import pandas as pd
import hashlib
import json
import time

# 1. Sambungkan Google Drive
if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

ROOT = Path(
    "/content/drive/MyDrive/"
    "PROJECT JURNAL - JATIKARYA/JATIKARYA_NEW"
)

EXP = ROOT / "02_EXPERIMENTS"

MAN = (
    EXP / "STEP_11D_FINAL_REUPLOAD"
    / "generative_final_manifest_240.csv"
)

REF = (
    EXP / "STEP_11B_WM_DINOV2_BASELINES"
    / "20261108T083639291045Z_5876dadc"
    / "outputs/source_baseline_manifest.csv"
)

TM = (
    EXP / "STEP_11E_TRUSTMARK_GENERATIVE"
    / "RUN_d5328ba208c93a5e3b40"
    / "outputs/trustmark_240_results.csv"
)

DATA_ROOT = (
    ROOT / "01_DATASET/GENERATIVE_EDITS_FINAL"
).resolve()

ORIG_ROOT = (
    ROOT / "01_DATASET/INDEPENDENT_ORIGINALS"
).resolve()

EMB_ROOT = (REF.parent / "embeddings").resolve()

def sha(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for chunk in iter(
            lambda: f.read(1024 * 1024), b""
        ):
            h.update(chunk)
    return h.hexdigest()

def readj(path):
    return json.loads(
        Path(path).read_text(encoding="utf-8")
    )

# 2. Verifikasi checkpoint sebelumnya
assert ROOT.is_dir()
assert MAN.is_file()
assert REF.is_file()
assert TM.is_file()

assert (
    readj(REF.parent / "summary.json")[
        "manifest_sha256"
    ] == sha(REF)
)

assert readj(
    REF.parent.parent / "metadata.json"
)["status"] == "PASS"

tm_summary = readj(TM.parent / "summary.json")

assert tm_summary["result_sha256"] == sha(TM)
assert tm_summary["run_status"] == "PASS"

sources = pd.read_csv(
    REF, dtype=str, keep_default_na=False
)

queries = pd.read_csv(
    MAN, dtype=str, keep_default_na=False
)

prev_tm = pd.read_csv(
    TM, dtype=str, keep_default_na=False
)

assert len(sources) == 40
assert len(queries) == 240
assert len(prev_tm) == 240
assert sources.source_number.is_unique

assert sources.split.value_counts().to_dict() == {
    "CALIBRATION": 20,
    "FINAL_TEST": 20
}

assert queries.provider.value_counts().to_dict() == {
    "chatgpt": 120,
    "gemini": 120
}

assert queries.split.value_counts().to_dict() == {
    "CALIBRATION": 120,
    "FINAL_TEST": 120
}

KEY = ["provider", "source_number", "edit_type"]

assert not queries.duplicated(KEY).any()
assert not prev_tm.duplicated(KEY).any()

expected = {
    (ai, f"{n:02d}", edit)
    for ai in ("chatgpt", "gemini")
    for n in range(1, 41)
    for edit in (
        "removal", "replacement", "background"
    )
}

assert set(
    map(tuple, queries[KEY].to_numpy())
) == expected

assert set(
    map(tuple, prev_tm[KEY].to_numpy())
) == expected

# 3. Periksa identitas dan integritas 240 gambar
ref = sources.set_index(
    "source_number"
).to_dict("index")

for r in queries.itertuples(index=False):

    s = ref[r.source_number]

    assert r.source_id == s["source_id"]
    assert r.split == s["split"]

    assert (
        r.parent_watermarked_sha256
        == s["watermarked_sha256"]
    )

    p = (ROOT / r.stored_path).resolve()

    assert p.is_relative_to(DATA_ROOT)
    assert p.is_file()
    assert sha(p) == r.sha256

# Pastikan sama persis dengan input TrustMark 11E
check = queries[
    KEY + ["sha256"]
].merge(
    prev_tm[
        KEY + ["image_sha256", "status"]
    ],
    on=KEY,
    validate="one_to_one"
)

assert (check.sha256 == check.image_sha256).all()
assert not check.status.eq("ERROR").any()

print("Validasi gambar: 240/240 PASS")

# 4. Baca 40 embedding referensi asli
ids = sorted(sources.source_number)
reference = []
proof = []

for n in ids:

    s = ref[n]

    p_orig = (
        ROOT / s["source_path"]
    ).resolve()

    p_emb = (
        ROOT / s["embedding_path"]
    ).resolve()

    assert p_orig.is_relative_to(ORIG_ROOT)
    assert p_orig.is_file()
    assert sha(p_orig) == s["source_sha256"]

    assert p_emb.is_relative_to(EMB_ROOT)
    assert p_emb.is_file()
    assert sha(p_emb) == s["embedding_sha256"]

    v = np.load(
        p_emb,
        allow_pickle=False
    )

    assert v.shape == (384,)
    assert np.isfinite(v).all()
    assert np.isclose(
        np.linalg.norm(v), 1, atol=1e-3
    )

    reference.append(
        v.astype(np.float32)
    )

    proof.append((
        n,
        s["source_id"],
        s["source_sha256"],
        s["embedding_sha256"]
    ))

ref_matrix = np.stack(reference)

# 5. Identitas eksperimen
CONFIG = {
    "query_manifest_sha256": sha(MAN),
    "reference_manifest_sha256": sha(REF),
    "reference_embeddings": proof,
    "model": "dinov2_vits14",
    "preprocessing": "Resize518_CenterCrop518_ImageNet",
    "orientation": "EXIF_TRANSPOSE_RGB_FRAME0",
    "candidate_count": 40
}

fingerprint = hashlib.sha256(
    json.dumps(
        CONFIG, sort_keys=True
    ).encode()
).hexdigest()

RUN = (
    EXP / "STEP_11F_DINOV2_GENERATIVE"
    / ("RUN_" + fingerprint[:20])
)

OUT = RUN / "outputs"
RECORDS = OUT / "records"

RECORDS.mkdir(
    parents=True, exist_ok=True
)

config_path = RUN / "config.json"

if config_path.is_file():
    assert readj(config_path) == CONFIG
else:
    config_path.write_text(
        json.dumps(CONFIG, indent=2),
        encoding="utf-8"
    )

# 6. Siapkan model DINOv2
DEVICE = (
    "cuda" if torch.cuda.is_available()
    else "cpu"
)

if (
    "model05a" in globals()
    and getattr(model05a, "embed_dim", None) == 384
):
    model = model05a

else:
    model = torch.hub.load(
        "facebookresearch/dinov2",
        "dinov2_vits14",
        pretrained=True,
        trust_repo=True
    )

assert getattr(model, "embed_dim", None) == 384

model = model.to(DEVICE).eval()
model05a = model

prep = transforms.Compose([
    transforms.Resize(518),
    transforms.CenterCrop(518),
    transforms.ToTensor(),
    transforms.Normalize(
        [0.485, 0.456, 0.406],
        [0.229, 0.224, 0.225]
    )
])

def embedding(path):

    with Image.open(path) as im:
        im.seek(0)

        rgb = ImageOps.exif_transpose(
            im
        ).convert("RGB")

        x = prep(rgb).unsqueeze(0).to(DEVICE)

    with torch.inference_mode():
        output = F.normalize(
            model(x), p=2, dim=1
        )

    v = (
        output[0]
        .detach()
        .cpu()
        .numpy()
        .astype(np.float32)
    )

    assert v.shape == (384,)
    assert np.isfinite(v).all()

    return v

# 7. Pastikan model sama dengan referensi STEP 11B
first_num = ids[0]

assert np.allclose(
    embedding(
        ROOT / ref[first_num]["source_path"]
    ),
    ref_matrix[0],
    atol=2e-3,
    rtol=2e-3
), (
    "DINOv2 tidak sesuai referensi STEP 11B. "
    "Proses dihentikan."
)

print("\n=== STEP 11F: DINOV2-ONLY ===")
print("Gambar:", len(queries))
print("Kandidat sumber:", len(ids))
print("Device:", DEVICE)

# 8. Jalankan retrieval
results = []
ranking_rows = []

for i, r in enumerate(
    queries.sort_values(KEY).itertuples(index=False),
    1
):

    uid = (
        f"{r.provider}_"
        f"{r.source_number}_"
        f"{r.edit_type}"
    )

    record_path = RECORDS / f"{uid}.json"
    cached = None

    # Gunakan checkpoint bila valid
    if record_path.is_file():

        d = readj(record_path)

        if (
            d.get("status") == "PASS"
            and d.get("fingerprint") == fingerprint
            and d.get("image_sha256") == r.sha256
            and d.get("source_id") == r.source_id
            and len(d.get("rankings", [])) == 40
            and {
                t["candidate_source_id"]
                for t in d["rankings"]
            } == {
                ref[n]["source_id"]
                for n in ids
            }
        ):
            cached = d

    if cached is not None:
        record = cached

    else:
        t0 = time.perf_counter()

        record = {
            "fingerprint": fingerprint,
            "provider": r.provider,
            "source_number": r.source_number,
            "source_id": r.source_id,
            "split": r.split,
            "edit_type": r.edit_type,
            "image_sha256": r.sha256,
            "filename": r.original_filename,
            "status": "ERROR",
            "error": "",
            "rankings": []
        }

        try:
            q = embedding(
                ROOT / r.stored_path
            )

            # Bandingkan dengan 40 sumber asli
            scores = ref_matrix @ q

            assert len(scores) == 40
            assert np.isfinite(scores).all()

            assert np.all(
                np.abs(scores) <= 1.001
            )

            order = sorted(
                range(40),
                key=lambda j: (
                    -float(scores[j]),
                    ids[j]
                )
            )

            true_idx = ids.index(
                r.source_number
            )

            true_rank = (
                order.index(true_idx) + 1
            )

            ranks = [
                {
                    "candidate_source_number": ids[j],
                    "candidate_source_id":
                        ref[ids[j]]["source_id"],
                    "rank": rank,
                    "cosine_similarity":
                        float(scores[j]),
                    "is_true_source": j == true_idx
                }
                for rank, j in enumerate(order, 1)
            ]

            record.update({
                "status": "PASS",
                "predicted_source_number":
                    ids[order[0]],
                "predicted_source_id":
                    ref[ids[order[0]]]["source_id"],
                "true_source_rank": true_rank,
                "top1_correct": true_rank == 1,
                "similarity_top1":
                    float(scores[order[0]]),
                "similarity_top2":
                    float(scores[order[1]]),
                "margin_top1_top2":
                    float(
                        scores[order[0]]
                        - scores[order[1]]
                    ),
                "similarity_true_source":
                    float(scores[true_idx]),
                "rankings": ranks
            })

        except Exception as exc:
            record["error"] = (
                f"{type(exc).__name__}: {exc}"
            )

        record["seconds"] = round(
            time.perf_counter() - t0, 5
        )

        # Simpan checkpoint setiap gambar
        tmp = record_path.with_suffix(
            ".json.tmp"
        )

        tmp.write_text(
            json.dumps(
                record,
                ensure_ascii=False,
                indent=2
            ),
            encoding="utf-8"
        )

        tmp.replace(record_path)

    results.append({
        k: v for k, v in record.items()
        if k != "rankings"
    })

    for candidate in record["rankings"]:

        ranking_rows.append({
            "provider": record["provider"],
            "source_number": record["source_number"],
            "source_id": record["source_id"],
            "split": record["split"],
            "edit_type": record["edit_type"],
            "image_sha256": record["image_sha256"],
            **candidate
        })

    if i % 10 == 0 or record["status"] == "ERROR":

        print(
            f"[{i:03d}/240] "
            f"{r.provider} | "
            f"{r.source_number} | "
            f"{r.edit_type}: "
            f"{record['status']}"
            + (
                f" | rank {record['true_source_rank']}"
                if record["status"] == "PASS"
                else ""
            )
        )

# 9. Simpan seluruh hasil
result = pd.DataFrame(results)
rankings = pd.DataFrame(ranking_rows)

assert len(result) == 240

errors = int(
    result.status.eq("ERROR").sum()
)

if errors == 0:
    assert len(rankings) == 9600

result_path = (
    OUT / "dinov2_240_results.csv"
)

ranking_path = (
    OUT / "all_candidate_rankings_9600.csv"
)

result.to_csv(
    result_path, index=False
)

rankings.to_csv(
    ranking_path, index=False
)

good = result[
    result.status == "PASS"
]

summary = {
    "status": (
        "PASS" if errors == 0
        else "COMPLETED_WITH_ERRORS"
    ),
    "fingerprint": fingerprint,
    "images": len(result),
    "reference_sources": 40,
    "candidate_scores": len(rankings),
    "top1_correct": int(
        good.top1_correct.sum()
    ),
    "errors": errors,
    "result_sha256": sha(result_path),
    "rankings_sha256": sha(ranking_path),
    "finished_at_utc":
        datetime.now(timezone.utc).isoformat(),
    "fusion_used": False
}

(
    OUT / "summary.json"
).write_text(
    json.dumps(
        summary,
        indent=2
    ),
    encoding="utf-8"
)

# 10. Laporan
print("\n=== STEP 11F: HASIL DINOV2-ONLY ===")

print("Total gambar:", len(result))
print("Top-1 benar:", summary["top1_correct"])
print("Error:", errors)
print("Skor kandidat:", len(rankings))

if errors == 0:

    report = result.groupby(
        ["split", "provider", "edit_type"],
        as_index=False
    ).agg(
        gambar=("source_id", "size"),
        top1_benar=("top1_correct", "sum"),
        rata_similarity_top1=(
            "similarity_top1", "mean"
        ),
        rata_margin=(
            "margin_top1_top2", "mean"
        )
    ).round(5)

    display(report)

    wrong = result.loc[
        ~result.top1_correct,
        [
            "split",
            "provider",
            "source_number",
            "edit_type",
            "source_id",
            "predicted_source_id",
            "true_source_rank",
            "similarity_top1",
            "margin_top1_top2"
        ]
    ]

    print(
        "\nJumlah salah identifikasi:",
        len(wrong)
    )

    if len(wrong):
        display(wrong)

print("\nSTATUS:", summary["status"])
print("Hasil:", result_path)
print("Seluruh skor:", ranking_path)
print("BELUM FUSION")
```

```python
# ==================================================
# JATIKARYA NEW - STEP 11G
# NAIVE FUSION | 240 GAMBAR GENERATIF
# TRUSTMARK FIRST, ELSE DINOV2 TOP-1
# TANPA UPLOAD, TANPA MENJALANKAN MODEL ULANG
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from google.colab import drive
from IPython.display import display
import pandas as pd
import hashlib
import json

if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

ROOT = Path(
    "/content/drive/MyDrive/"
    "PROJECT JURNAL - JATIKARYA/JATIKARYA_NEW"
)

EXP = ROOT / "02_EXPERIMENTS"

MAN = (
    EXP / "STEP_11D_FINAL_REUPLOAD"
    / "generative_final_manifest_240.csv"
)

REF = (
    EXP / "STEP_11B_WM_DINOV2_BASELINES"
    / "20261108T083639291045Z_5876dadc"
    / "outputs/source_baseline_manifest.csv"
)

TM = (
    EXP / "STEP_11E_TRUSTMARK_GENERATIVE"
    / "RUN_d5328ba208c93a5e3b40"
    / "outputs/trustmark_240_results.csv"
)

DI = (
    EXP / "STEP_11F_DINOV2_GENERATIVE"
    / "RUN_010c9ee0adf230a3bf70"
    / "outputs/dinov2_240_results.csv"
)

RANK = DI.parent / "all_candidate_rankings_9600.csv"

def sha_file(path):
    h = hashlib.sha256()

    with Path(path).open("rb") as f:
        for part in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(part)

    return h.hexdigest()

def read_json(path):
    return json.loads(
        Path(path).read_text(encoding="utf-8")
    )

def read_csv(path):
    return pd.read_csv(
        path,
        dtype=str,
        keep_default_na=False
    )

# 1. Verifikasi semua input
for path in (MAN, REF, TM, DI, RANK):
    assert path.is_file(), (
        f"Berkas belum tersedia: {path}"
    )

assert (
    read_json(REF.parent / "summary.json")[
        "manifest_sha256"
    ] == sha_file(REF)
)

assert (
    read_json(
        REF.parent.parent / "metadata.json"
    )["status"] == "PASS"
)

tm_summary = read_json(
    TM.parent / "summary.json"
)

di_summary = read_json(
    DI.parent / "summary.json"
)

assert tm_summary["result_sha256"] == sha_file(TM)
assert tm_summary["run_status"] == "PASS"

assert di_summary["result_sha256"] == sha_file(DI)
assert di_summary["rankings_sha256"] == sha_file(RANK)
assert di_summary["status"] == "PASS"
assert di_summary["candidate_scores"] == 9600

queries = read_csv(MAN)
refs = read_csv(REF)
tm = read_csv(TM)
dino = read_csv(DI)

KEY = [
    "provider",
    "source_number",
    "edit_type"
]

expected = {
    (ai, f"{n:02d}", edit)
    for ai in ("chatgpt", "gemini")
    for n in range(1, 41)
    for edit in (
        "removal",
        "replacement",
        "background"
    )
}

for name, df in [
    ("Manifest", queries),
    ("TrustMark", tm),
    ("DINOv2", dino)
]:
    assert len(df) == 240, name
    assert not df.duplicated(KEY).any(), name
    assert set(
        map(tuple, df[KEY].to_numpy())
    ) == expected, name

assert len(refs) == 40
assert refs.source_id.is_unique
assert refs.source_number.is_unique

assert refs.split.value_counts().to_dict() == {
    "CALIBRATION": 20,
    "FINAL_TEST": 20
}

# 2. Cocokkan hasil berdasarkan identitas
# dan SHA-256 dari gambar yang sama
base = queries[
    KEY + [
        "sha256",
        "source_id",
        "split",
        "original_filename"
    ]
].rename(columns={
    "sha256": "image_sha256",
    "source_id": "true_source_id"
})

tm = tm[
    KEY + [
        "source_id",
        "split",
        "image_sha256",
        "detected",
        "decoded_payload",
        "mapped_source_id",
        "status",
        "error"
    ]
].rename(columns={
    "source_id": "tm_true_source_id",
    "split": "tm_split",
    "status": "tm_status",
    "error": "tm_error"
})

dino = dino[
    KEY + [
        "source_id",
        "split",
        "image_sha256",
        "predicted_source_id",
        "status",
        "error",
        "similarity_top1",
        "margin_top1_top2"
    ]
].rename(columns={
    "source_id": "dino_true_source_id",
    "split": "dino_split",
    "status": "dino_status",
    "error": "dino_error"
})

data = base.merge(
    tm,
    on=KEY + ["image_sha256"],
    how="inner",
    validate="one_to_one"
).merge(
    dino,
    on=KEY + ["image_sha256"],
    how="inner",
    validate="one_to_one"
)

assert len(data) == 240

assert data.true_source_id.eq(
    data.tm_true_source_id
).all()

assert data.true_source_id.eq(
    data.dino_true_source_id
).all()

assert data.split.eq(data.tm_split).all()
assert data.split.eq(data.dino_split).all()

assert data.tm_error.eq("").all()
assert data.dino_error.eq("").all()
assert data.dino_status.eq("PASS").all()

assert data.tm_status.isin([
    "KODE_COCOK",
    "TIDAK_TERDETEKSI",
    "SCHEMA_TIDAK_SESUAI",
    "KODE_SUMBER_LAIN",
    "KODE_TIDAK_COCOK"
]).all()

# Cocokkan dengan checkpoint terdahulu
assert data.tm_status.eq(
    "KODE_COCOK"
).sum() == 29

assert (
    data.predicted_source_id
    == data.true_source_id
).sum() == 238

# 3. Bangun katalog watermark
# dari SHA-256 sumber asli
payload_map = {}

for r in refs.itertuples(index=False):

    payload = bin(
        int(r.source_sha256, 16)
    )[2:].zfill(256)[:61]

    assert payload not in payload_map

    payload_map[payload] = r.source_id

known_ids = set(refs.source_id)

ref_map = refs.set_index(
    "source_number"
).to_dict("index")

# 4. Jalankan aturan Naive Fusion
rows = []

for r in data.itertuples(index=False):

    assert (
        ref_map[r.source_number]["source_id"]
        == r.true_source_id
    )

    assert r.predicted_source_id in known_ids

    detected = (
        r.detected.lower() == "true"
    )

    mapped_from_payload = (
        payload_map.get(
            r.decoded_payload, ""
        )
        if detected else ""
    )

    # TrustMark STEP 11E sudah memeriksa
    # skema decoding sebelum memetakan identitas.
    if r.mapped_source_id:

        assert (
            r.mapped_source_id
            == mapped_from_payload
        )

        assert r.tm_status in (
            "KODE_COCOK",
            "KODE_SUMBER_LAIN"
        )

    else:

        assert r.tm_status not in (
            "KODE_COCOK",
            "KODE_SUMBER_LAIN"
        )

    # INI ATURAN FUSION
    # Tidak menggunakan ground truth
    tm_choice = r.mapped_source_id

    if tm_choice:
        prediction = tm_choice
        decision = "TRUSTMARK"
    else:
        prediction = r.predicted_source_id
        decision = "DINOV2_FALLBACK"

    conflict = bool(
        tm_choice
        and tm_choice != r.predicted_source_id
    )

    # Kebenaran diperiksa setelah prediksi
    dino_correct = (
        r.predicted_source_id
        == r.true_source_id
    )

    naive_correct = (
        prediction == r.true_source_id
    )

    rows.append({
        "provider": r.provider,
        "source_number": r.source_number,
        "source_id_sebenarnya":
            r.true_source_id,
        "split": r.split,
        "edit_type": r.edit_type,
        "original_filename":
            r.original_filename,
        "image_sha256": r.image_sha256,
        "tm_status": r.tm_status,
        "tm_source_prediksi": tm_choice,
        "dino_source_prediksi":
            r.predicted_source_id,
        "similarity_top1":
            r.similarity_top1,
        "margin_top1_top2":
            r.margin_top1_top2,
        "naive_source_prediksi":
            prediction,
        "dasar_keputusan": decision,
        "konflik": conflict,
        "dino_benar": dino_correct,
        "naive_benar": naive_correct
    })

result = pd.DataFrame(rows).sort_values(
    KEY
).reset_index(drop=True)

assert len(result) == 240
assert result.dino_benar.sum() == 238

assert result.split.value_counts().to_dict() == {
    "CALIBRATION": 120,
    "FINAL_TEST": 120
}

# 5. Simpan secara reproducible
config = {
    "rule": "TM_FIRST_THEN_DINOV2_TOP1_V1",
    "manifest_sha256": sha_file(MAN),
    "ref_sha256": sha_file(REF),
    "trustmark_sha256": sha_file(TM),
    "dinov2_sha256": sha_file(DI)
}

fingerprint = hashlib.sha256(
    json.dumps(
        config, sort_keys=True
    ).encode()
).hexdigest()

RUN = (
    EXP / "STEP_11G_NAIVE_FUSION"
    / ("RUN_" + fingerprint[:20])
)

OUT = RUN / "outputs"
OUT.mkdir(parents=True, exist_ok=True)

config_path = RUN / "config.json"

if config_path.exists():
    assert read_json(config_path) == config
else:
    config_path.write_text(
        json.dumps(config, indent=2),
        encoding="utf-8"
    )

CSV = OUT / "naive_fusion_240_results.csv"
SUMMARY = OUT / "summary.json"

if SUMMARY.exists():

    previous = read_json(SUMMARY)

    assert CSV.is_file()
    assert (
        previous["results_sha256"]
        == sha_file(CSV)
    )

    assert previous["fingerprint"] == fingerprint

    print("Arsip valid ditemukan. Tidak disimpan ulang.")

else:

    if CSV.exists():

        earlier = read_csv(CSV)

        assert earlier.equals(
            result.astype(str)
        ), "Arsip sebelumnya berbeda."

    else:
        result.to_csv(
            CSV, index=False
        )

    summary = {
        "fingerprint": fingerprint,
        "total": len(result),
        "tm_mapped_count": int(
            result.tm_source_prediksi.ne("").sum()
        ),
        "dino_correct": int(
            result.dino_benar.sum()
        ),
        "naive_correct": int(
            result.naive_benar.sum()
        ),
        "conflicts": int(
            result.konflik.sum()
        ),
        "calibration_naive_correct": int(
            result.loc[
                result.split == "CALIBRATION",
                "naive_benar"
            ].sum()
        ),
        "final_test_naive_correct": int(
            result.loc[
                result.split == "FINAL_TEST",
                "naive_benar"
            ].sum()
        ),
        "results_sha256": sha_file(CSV),
        "status": "PASS",
        "created_at_utc":
            datetime.now(
                timezone.utc
            ).isoformat()
    }

    SUMMARY.write_text(
        json.dumps(
            summary,
            ensure_ascii=False,
            indent=2
        ),
        encoding="utf-8"
    )

# 6. Tampilkan hasil
print("\n=== STEP 11G: NAIVE FUSION ===")
print("Total gambar:", len(result))
print(
    "TrustMark terpetakan:",
    int(result.tm_source_prediksi.ne("").sum())
)
print(
    "DINOv2 benar:",
    int(result.dino_benar.sum())
)
print(
    "Naive Fusion benar:",
    int(result.naive_benar.sum())
)
print(
    "Konflik kedua metode:",
    int(result.konflik.sum())
)

print("\n=== HASIL PER AI, EDIT DAN SPLIT ===")

display(
    result.groupby(
        ["split", "provider", "edit_type"],
        as_index=False
    ).agg(
        gambar=("naive_benar", "size"),
        dino_benar=("dino_benar", "sum"),
        naive_benar=("naive_benar", "sum")
    )
)

print("\n=== KONFLIK TRUSTMARK VS DINOV2 ===")

display(
    result.loc[
        result.konflik,
        [
            "split",
            "provider",
            "source_number",
            "edit_type",
            "tm_source_prediksi",
            "dino_source_prediksi",
            "naive_source_prediksi",
            "naive_benar"
        ]
    ]
)

print("\n=== KASUS NAIVE FUSION SALAH ===")

display(
    result.loc[
        ~result.naive_benar,
        [
            "split",
            "provider",
            "source_number",
            "edit_type",
            "source_id_sebenarnya",
            "naive_source_prediksi",
            "dasar_keputusan",
            "similarity_top1",
            "margin_top1_top2"
        ]
    ]
)

print("\nSTATUS: PASS | BELUM CALIBRATED FUSION")
print("Hasil:", CSV)
```

```python

# ==================================================
# JATIKARYA NEW - STEP 11H-A
# KALIBRASI AWAL CALIBRATED FUSION
# 20 SUMBER CALIBRATION | FINAL_TEST TIDAK DIPAKAI
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from google.colab import drive
from IPython.display import display
from sklearn.model_selection import LeaveOneGroupOut
from sklearn.pipeline import make_pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import brier_score_loss
import pandas as pd
import numpy as np
import hashlib
import json
import sklearn

if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

ROOT = Path(
    "/content/drive/MyDrive/"
    "PROJECT JURNAL - JATIKARYA/JATIKARYA_NEW"
)

assert ROOT.is_dir()

EXP = ROOT / "02_EXPERIMENTS"

MAN = (
    EXP / "STEP_11D_FINAL_REUPLOAD"
    / "generative_final_manifest_240.csv"
)

assert MAN.is_file()

def sha11h(path):
    h = hashlib.sha256()

    with Path(path).open("rb") as f:
        for chunk in iter(
            lambda: f.read(1024 * 1024),
            b""
        ):
            h.update(chunk)

    return h.hexdigest()

def read11h(path):
    return json.loads(
        Path(path).read_text(encoding="utf-8")
    )

# 1. Temukan hasil STEP 11G yang valid
candidates = sorted(
    (
        EXP / "STEP_11G_NAIVE_FUSION"
    ).glob(
        "RUN_*/outputs/naive_fusion_240_results.csv"
    )
)

valid = []

for p in candidates:

    try:
        summary = read11h(
            p.parent / "summary.json"
        )

        config = read11h(
            p.parent.parent / "config.json"
        )

        if (
            summary.get("status") == "PASS"
            and summary.get("total") == 240
            and summary.get("results_sha256") == sha11h(p)
            and config.get("manifest_sha256") == sha11h(MAN)
        ):
            valid.append(p)

    except (OSError, ValueError, KeyError):
        continue

assert valid, "STEP 11G yang valid belum ditemukan."

assert len({
    sha11h(p) for p in valid
}) == 1, (
    "Terdapat arsip STEP 11G berbeda. "
    "Periksa sebelum melanjutkan."
)

SOURCE = valid[-1]

data = pd.read_csv(
    SOURCE,
    dtype=str,
    keep_default_na=False
)

assert len(data) == 240

assert data.split.value_counts().to_dict() == {
    "CALIBRATION": 120,
    "FINAL_TEST": 120
}

assert not data.duplicated([
    "provider", "source_number", "edit_type"
]).any()

# 2. Gunakan CALIBRATION saja
cal = data.loc[
    data.split == "CALIBRATION"
].copy()

assert len(cal) == 120
assert cal.source_id_sebenarnya.nunique() == 20

assert cal.groupby(
    "source_id_sebenarnya"
).size().eq(6).all()

cal["tm_available"] = (
    cal.tm_source_prediksi != ""
)

cal["conflict"] = (
    cal.tm_available
    & (
        cal.tm_source_prediksi
        != cal.dino_source_prediksi
    )
)

# Konflik pada data ini diketahui 0.
assert not cal.conflict.any()

# 3. Kalibrasi hanya kasus fallback DINOv2
fallback = cal.loc[
    ~cal.tm_available
].copy().reset_index(drop=True)

fallback["s1"] = pd.to_numeric(
    fallback.similarity_top1
)

fallback["margin"] = pd.to_numeric(
    fallback.margin_top1_top2
)

X = fallback[
    ["s1", "margin"]
].to_numpy(dtype=float)

y = (
    fallback.dino_source_prediksi
    == fallback.source_id_sebenarnya
).astype(int).to_numpy()

groups = (
    fallback.source_id_sebenarnya.to_numpy()
)

assert np.isfinite(X).all()
assert len(np.unique(groups)) == 20

assert int((y == 0).sum()) == 2, (
    "Jumlah kesalahan berbeda dari STEP 11G. "
    "Audit data terlebih dahulu."
)

# 4. Model estimasi keandalan
def make_model11h():

    return make_pipeline(
        StandardScaler(),
        LogisticRegression(
            C=0.5,
            solver="lbfgs",
            max_iter=3000
        )
    )

# 5. Prediksi out-of-fold berdasarkan sumber
oof = np.full(
    len(fallback), np.nan
)

logo = LeaveOneGroupOut()

for train_idx, val_idx in logo.split(
    X, y, groups
):

    assert len(np.unique(
        y[train_idx]
    )) == 2

    model = make_model11h()

    model.fit(
        X[train_idx],
        y[train_idx]
    )

    correct_idx = list(
        model.classes_
    ).index(1)

    oof[val_idx] = model.predict_proba(
        X[val_idx]
    )[:, correct_idx]

assert np.isfinite(oof).all()

# 6. Pilih ambang menggunakan OOF CALIBRATION
# Aturan eksploratif: nol kesalahan diterima
# pada CALIBRATION out-of-fold.
# BUKAN jaminan pada data baru.

wrong_max = float(
    oof[y == 0].max()
)

threshold = float(
    np.nextafter(
        wrong_max,
        np.inf
    )
)

accepted = oof >= threshold

assert int(
    ((y == 0) & accepted).sum()
) == 0

# 7. Fit model akhir pada seluruh CALIBRATION
final_model = make_model11h()

final_model.fit(X, y)

scaler = final_model.named_steps[
    "standardscaler"
]

regression = final_model.named_steps[
    "logisticregression"
]

assert list(regression.classes_) == [0, 1]

# 8. Simpan protokol tanpa membuka hasil FINAL_TEST
protocol = {
    "status":
        "LOCKED_PRELIMINARY_CALIBRATION",

    "source_step11g_sha256": sha11h(SOURCE),

    "training_split": "CALIBRATION",

    "evaluation_split": "FINAL_TEST",

    "grouping": "source_id_sebenarnya",

    "features": [
        "similarity_top1",
        "margin_top1_top2"
    ],

    "model":
        "regularized_logistic_regression",

    "sklearn_version": sklearn.__version__,

    "C": 0.5,

    "scaler_mean": scaler.mean_.tolist(),

    "scaler_scale": scaler.scale_.tolist(),

    "coefficients":
        regression.coef_[0].tolist(),

    "intercept":
        float(regression.intercept_[0]),

    "decision_threshold": threshold,

    "rules": {
        "tm_dino_agree":
            "ACCEPT",
        "tm_dino_conflict":
            "ABSTAIN",
        "tm_missing":
            "ACCEPT_DINO_IF_SCORE_GE_THRESHOLD",
        "insufficient_evidence":
            "ABSTAIN"
    },

    "oof_brier_score":
        float(brier_score_loss(y, oof)),

    "calibration_sources": 20,

    "calibration_images": 120,

    "fallback_images": len(fallback),

    "fallback_errors": int(
        (y == 0).sum()
    ),

    "oof_fallback_accepted": int(
        accepted.sum()
    ),

    "oof_wrong_accepted": int(
        ((y == 0) & accepted).sum()
    ),

    "limitations": (
        "Only two incorrect DINOv2 predictions. "
        "Reliability estimation and threshold "
        "are preliminary and statistically uncertain."
    )
}

fingerprint = hashlib.sha256(
    json.dumps(
        protocol,
        sort_keys=True
    ).encode()
).hexdigest()

RUN = (
    EXP / "STEP_11H_CALIBRATED_FUSION"
    / ("RUN_" + fingerprint[:20])
)

OUT = RUN / "outputs"

OUT.mkdir(
    parents=True,
    exist_ok=True
)

P_PROTOCOL = (
    OUT / "calibration_protocol_locked.json"
)

P_OOF = (
    OUT / "source_disjoint_oof_calibration.csv"
)

# 9. Simpan hasil kalibrasi
fallback["oof_reliability_score"] = oof

fallback["oof_accepted"] = accepted

fallback["dino_correct"] = y.astype(bool)

report = fallback[[
    "provider",
    "source_number",
    "source_id_sebenarnya",
    "edit_type",
    "image_sha256",
    "s1",
    "margin",
    "dino_correct",
    "oof_reliability_score",
    "oof_accepted"
]]

if P_PROTOCOL.exists() or P_OOF.exists():

    assert P_PROTOCOL.is_file()
    assert P_OOF.is_file()

    previous = read11h(P_PROTOCOL)

    assert previous == protocol, (
        "Protokol tersimpan berbeda. "
        "Tidak akan ditimpa."
    )

    print("Checkpoint ditemukan, tidak membuat duplikat.")

else:

    report.to_csv(
        P_OOF,
        index=False
    )

    P_PROTOCOL.write_text(
        json.dumps(
            protocol,
            ensure_ascii=False,
            indent=2
        ),
        encoding="utf-8"
    )

# 10. Ringkasan
print("\n=== STEP 11H-A: KALIBRASI AWAL ===")

print(
    "Sumber CALIBRATION:",
    cal.source_id_sebenarnya.nunique()
)

print("Gambar CALIBRATION:", len(cal))

print(
    "TrustMark terpetakan:",
    int(cal.tm_available.sum())
)

print("Kasus fallback:", len(fallback))

print(
    "Kesalahan fallback:",
    int((y == 0).sum())
)

print(
    "OOF Brier Score:",
    round(brier_score_loss(y, oof), 6)
)

print(
    "Ambang awal:",
    round(threshold, 6)
)

print(
    "Fallback OOF diterima:",
    int(accepted.sum()),
    "/", len(fallback)
)

print(
    "Kesalahan OOF diterima:",
    int(((y == 0) & accepted).sum())
)

print("Protokol:", P_PROTOCOL)

print(
    "STATUS: KALIBRASI AWAL TERKUNCI"
)
print(
    "CALIBRATED FUSION FINAL_TEST BELUM DIUJI"
)
```

```python
# ==================================================
# JATIKARYA NEW - STEP 11H-B
# EVALUASI CALIBRATED FUSION PADA FINAL_TEST
# Menggunakan model dan threshold STEP 11H-A
# Tanpa fitting ulang atau perubahan dataset
# ==================================================

from pathlib import Path
from datetime import datetime, timezone
from google.colab import drive
from IPython.display import display
import pandas as pd
import numpy as np
import hashlib
import json

if not Path("/content/drive/MyDrive").is_dir():
    drive.mount("/content/drive")

ROOT = Path(
    "/content/drive/MyDrive/"
    "PROJECT JURNAL - JATIKARYA/JATIKARYA_NEW"
)

EXP = ROOT / "02_EXPERIMENTS"

PROTO = (
    EXP / "STEP_11H_CALIBRATED_FUSION"
    / "RUN_35ae2435bcdee11241b2"
    / "outputs/calibration_protocol_locked.json"
)

MAN = (
    EXP / "STEP_11D_FINAL_REUPLOAD"
    / "generative_final_manifest_240.csv"
)

REF = (
    EXP / "STEP_11B_WM_DINOV2_BASELINES"
    / "20261108T083639291045Z_5876dadc"
    / "outputs/source_baseline_manifest.csv"
)

def sha(path):
    h = hashlib.sha256()
    with Path(path).open("rb") as f:
        for chunk in iter(
            lambda: f.read(1024 * 1024), b""
        ):
            h.update(chunk)
    return h.hexdigest()

def read_json(path):
    return json.loads(
        Path(path).read_text(encoding="utf-8")
    )

# 1. Verifikasi protokol kalibrasi terkunci
assert PROTO.is_file()
assert MAN.is_file()
assert REF.is_file()

proto = read_json(PROTO)

assert proto["status"] == (
    "LOCKED_PRELIMINARY_CALIBRATION"
)

assert proto["training_split"] == "CALIBRATION"
assert proto["evaluation_split"] == "FINAL_TEST"

assert proto["features"] == [
    "similarity_top1",
    "margin_top1_top2"
]

assert proto["rules"] == {
    "tm_dino_agree": "ACCEPT",
    "tm_dino_conflict": "ABSTAIN",
    "tm_missing":
        "ACCEPT_DINO_IF_SCORE_GE_THRESHOLD",
    "insufficient_evidence": "ABSTAIN"
}

# 2. Temukan hasil STEP 11G yang tepat
candidates = list(
    (
        EXP / "STEP_11G_NAIVE_FUSION"
    ).glob(
        "RUN_*/outputs/naive_fusion_240_results.csv"
    )
)

matching = []

for p in candidates:
    try:
        summ = read_json(
            p.parent / "summary.json"
        )
        config = read_json(
            p.parent.parent / "config.json"
        )

        if (
            sha(p) == proto["source_step11g_sha256"]
            and summ.get("status") == "PASS"
            and summ.get("results_sha256") == sha(p)
            and config.get("manifest_sha256") == sha(MAN)
        ):
            matching.append(p)

    except (OSError, ValueError, KeyError):
        continue

assert len(matching) == 1, (
    "Arsip STEP 11G tidak cocok atau ambigu."
)

G = matching[0]

# 3. Validasi katalog sumber
ref_summary = read_json(
    REF.parent / "summary.json"
)

assert (
    ref_summary["manifest_sha256"] == sha(REF)
)

refs = pd.read_csv(
    REF, dtype=str, keep_default_na=False
)

assert len(refs) == 40
assert refs.source_id.is_unique

known_sources = set(refs.source_id)

# 4. Ambil hanya 120 FINAL_TEST
all_rows = pd.read_csv(
    G, dtype=str, keep_default_na=False
)

assert len(all_rows) == 240

assert all_rows.split.value_counts().to_dict() == {
    "CALIBRATION": 120,
    "FINAL_TEST": 120
}

KEY = [
    "provider",
    "source_number",
    "edit_type",
    "image_sha256"
]

assert not all_rows.duplicated(KEY).any()

final = all_rows.loc[
    all_rows.split == "FINAL_TEST"
].copy()

assert len(final) == 120
assert final.source_id_sebenarnya.nunique() == 20

assert final.groupby(
    "source_id_sebenarnya"
).size().eq(6).all()

assert final.dino_source_prediksi.isin(
    known_sources
).all()

assert (
    final.tm_source_prediksi.eq("")
    | final.tm_source_prediksi.isin(known_sources)
).all()

# 5. Pulihkan parameter model hasil kalibrasi
# Tidak memanggil fit() atau fit_transform()
mu = np.asarray(
    proto["scaler_mean"], dtype=float
)

scale = np.asarray(
    proto["scaler_scale"], dtype=float
)

weights = np.asarray(
    proto["coefficients"], dtype=float
)

intercept = float(proto["intercept"])

threshold = float(
    proto["decision_threshold"]
)

assert mu.shape == (2,)
assert scale.shape == (2,)
assert weights.shape == (2,)

assert np.isfinite(
    np.concatenate([
        mu, scale, weights,
        [intercept, threshold]
    ])
).all()

assert (scale > 0).all()
assert 0 < threshold < 1

# 6. Siapkan fitur tanpa label jawaban benar
inference = final[
    KEY + [
        "split",
        "tm_source_prediksi",
        "dino_source_prediksi",
        "similarity_top1",
        "margin_top1_top2"
    ]
].copy().sort_values(KEY).reset_index(drop=True)

X = inference[
    ["similarity_top1", "margin_top1_top2"]
].apply(
    pd.to_numeric, errors="raise"
).to_numpy(dtype=float)

assert np.isfinite(X).all()

# Model logistic regression yang sudah dilatih
z = (
    ((X - mu) / scale) @ weights
    + intercept
)

# Sigmoid numerik stabil
z = np.clip(z, -700, 700)

reliability = 1 / (1 + np.exp(-z))

# 7. Keputusan Calibrated Fusion
pred = inference[KEY + ["split"]].copy()

pred["tm_source"] = (
    inference.tm_source_prediksi
)

pred["dinov2_source"] = (
    inference.dino_source_prediksi
)

pred["s1"] = X[:, 0]
pred["margin"] = X[:, 1]

pred["reliability_estimate"] = reliability
pred["threshold"] = threshold

pred["decision"] = ""
pred["predicted_source_id"] = ""

for i, row in inference.iterrows():

    tm_source = row.tm_source_prediksi
    dino_source = row.dino_source_prediksi

    if tm_source and tm_source != dino_source:

        decision = "ABSTAIN_CONFLICT"
        chosen = ""

    elif tm_source:

        decision = "ACCEPT_AGREEMENT"
        chosen = tm_source

    elif reliability[i] >= threshold:

        decision = "ACCEPT_DINOV2"
        chosen = dino_source

    else:

        decision = "ABSTAIN_LOW_EVIDENCE"
        chosen = ""

    pred.at[i, "decision"] = decision

    pred.at[
        i, "predicted_source_id"
    ] = chosen

assert pred.decision.ne("").all()
assert len(pred) == 120
assert not pred.duplicated(KEY).any()

# 8. Kunci prediksi sebelum mengevaluasi jawaban
config = {
    "calibration_protocol_sha256": sha(PROTO),
    "step11g_sha256": sha(G),
    "manifest_sha256": sha(MAN),
    "rule":
        "LOCKED_TM_CONFLICT_ABSTAIN_LOGREG_V1"
}

fingerprint = hashlib.sha256(
    json.dumps(
        config, sort_keys=True
    ).encode()
).hexdigest()

RUN = (
    EXP / "STEP_11H_B_FINAL_EVALUATION"
    / ("RUN_" + fingerprint[:20])
)

OUT = RUN / "outputs"
OUT.mkdir(parents=True, exist_ok=True)

PRED = (
    OUT / "final_test_predictions_locked.csv"
)

LOCK = OUT / "prediction_lock.json"

if LOCK.exists():

    locked = read_json(LOCK)

    assert PRED.exists()
    assert locked["config"] == config
    assert locked["prediction_sha256"] == sha(PRED)

    archived = pd.read_csv(
        PRED, dtype=str, keep_default_na=False
    )

    assert len(archived) == 120
    assert archived[KEY].equals(pred[KEY])

    print("Prediksi terkunci digunakan kembali.")

else:

    assert not PRED.exists(), (
        "Prediksi tersedia tanpa catatan lock."
    )

    pred.to_csv(PRED, index=False)

    LOCK.write_text(
        json.dumps({
            "config": config,
            "prediction_sha256": sha(PRED),
            "locked_at_utc":
                datetime.now(
                    timezone.utc
                ).isoformat()
        }, indent=2),
        encoding="utf-8"
    )

    print("Prediksi FINAL_TEST berhasil dikunci.")

# 9. Evaluasi setelah prediksi dikunci
frozen = pd.read_csv(
    PRED, dtype=str, keep_default_na=False
)

truth = final[
    KEY + [
        "source_id_sebenarnya",
        "dino_source_prediksi",
        "naive_source_prediksi"
    ]
].copy()

evaluation = frozen.merge(
    truth,
    on=KEY,
    validate="one_to_one"
)

assert len(evaluation) == 120

assert evaluation.dinov2_source.eq(
    evaluation.dino_source_prediksi
).all()

evaluation["accepted"] = (
    evaluation.predicted_source_id != ""
)

evaluation["correct_accepted"] = (
    evaluation.accepted
    & evaluation.predicted_source_id.eq(
        evaluation.source_id_sebenarnya
    )
)

evaluation["wrong_accepted"] = (
    evaluation.accepted
    & ~evaluation.predicted_source_id.eq(
        evaluation.source_id_sebenarnya
    )
)

evaluation["dino_correct"] = (
    evaluation.dino_source_prediksi
    == evaluation.source_id_sebenarnya
)

evaluation["naive_correct"] = (
    evaluation.naive_source_prediksi
    == evaluation.source_id_sebenarnya
)

n = len(evaluation)

accepted = int(
    evaluation.accepted.sum()
)

correct = int(
    evaluation.correct_accepted.sum()
)

wrong = int(
    evaluation.wrong_accepted.sum()
)

abstain = n - accepted

summary = {
    "status": "FINAL_TEST_EVALUATED",
    "calibration_protocol_sha256": sha(PROTO),
    "prediction_lock_sha256": sha(PRED),
    "total_final_test": n,
    "accepted": accepted,
    "correct_accepted": correct,
    "wrong_accepted": wrong,
    "abstained": abstain,
    "coverage": accepted / n,
    "selective_accuracy": (
        correct / accepted if accepted else None
    ),
    "overall_correct_fraction": correct / n,
    "dino_correct": int(
        evaluation.dino_correct.sum()
    ),
    "naive_correct": int(
        evaluation.naive_correct.sum()
    ),
    "note": (
        "Preliminary calibration. Threshold selected "
        "using CALIBRATION OOF scores. FINAL_TEST "
        "labels were previously observed in STEP 11F."
    )
}

EVAL = OUT / "final_test_evaluation.csv"
SUM = OUT / "summary.json"

if SUM.exists():

    previous = read_json(SUM)

    assert previous == summary
    assert EVAL.exists()

else:

    assert not EVAL.exists(), (
        "Hasil evaluasi tanpa ringkasan."
    )

    evaluation.to_csv(
        EVAL, index=False
    )

    SUM.write_text(
        json.dumps(
            summary,
            indent=2,
            ensure_ascii=False
        ),
        encoding="utf-8"
    )

# 10. Laporan akhir
print("\n=== STEP 11H-B: EVALUASI FINAL_TEST ===")

print("Gambar uji:", n)
print("DINOv2 benar:", summary["dino_correct"])
print("Naive Fusion benar:", summary["naive_correct"])

print("\n=== CALIBRATED FUSION ===")
print("Keputusan diterima:", accepted)
print("Benar diterima:", correct)
print("Salah diterima:", wrong)
print("Abstain:", abstain)

print("Coverage:", f"{accepted / n:.2%}")

if accepted:
    print(
        "Ketepatan keputusan diterima:",
        f"{correct / accepted:.2%}"
    )

print("\n=== KEPUTUSAN PER AI DAN EDIT ===")

display(
    pd.crosstab(
        [
            evaluation.provider,
            evaluation.edit_type
        ],
        evaluation.decision
    )
)

print("\n=== KASUS ABSTAIN / SALAH DITERIMA ===")

cases = evaluation.loc[
    (~evaluation.accepted)
    | evaluation.wrong_accepted,
    [
        "provider",
        "source_number",
        "edit_type",
        "decision",
        "s1",
        "margin",
        "reliability_estimate",
        "source_id_sebenarnya",
        "dinov2_source"
    ]
]

if len(cases):
    display(cases)
else:
    print("Tidak ada kasus.")

print("\nSTATUS:", summary["status"])
print("Hasil:", EVAL)
print("Prediksi terkunci:", PRED)
```
